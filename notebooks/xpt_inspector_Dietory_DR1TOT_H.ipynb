```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2013_2014"
    / "Dietary"
)

xpt_file = DATA_DIR / "DR1TOT_H.xpt"
csv_file = DATA_DIR / "DR1TOT_H_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Preserve completely untouched pandas-decoded source.
#
source_df = df.copy()


print("=" * 90)
print("NHANES 2013-2014 DR1TOT_H XPT -> CSV")
print("=" * 90)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "WTDRD1",
    "WTDR2D",
    "DR1DRSTZ",
    "DR1EXMER",
    "DRABF",
    "DRDINT",
    "DR1DBIH",
    "DR1DAY",
    "DR1LANG",
    "DR1MNRSP",
    "DR1HELPD",
    "DBQ095Z",
    "DBD100",
    "DRQSPREP",
    "DR1STY",
    "DR1SKY",
    "DRQSDIET",
    "DRQSDT1",
    "DRQSDT2",
    "DRQSDT3",
    "DRQSDT4",
    "DRQSDT5",
    "DRQSDT6",
    "DRQSDT7",
    "DRQSDT8",
    "DRQSDT9",
    "DRQSDT10",
    "DRQSDT11",
    "DRQSDT12",
    "DRQSDT91",
    "DR1TNUMF",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TCARB",
    "DR1TSUGR",
    "DR1TFIBE",
    "DR1TTFAT",
    "DR1TSFAT",
    "DR1TMFAT",
    "DR1TPFAT",
    "DR1TCHOL",
    "DR1TATOC",
    "DR1TATOA",
    "DR1TRET",
    "DR1TVARA",
    "DR1TACAR",
    "DR1TBCAR",
    "DR1TCRYP",
    "DR1TLYCO",
    "DR1TLZ",
    "DR1TVB1",
    "DR1TVB2",
    "DR1TNIAC",
    "DR1TVB6",
    "DR1TFOLA",
    "DR1TFA",
    "DR1TFF",
    "DR1TFDFE",
    "DR1TCHL",
    "DR1TVB12",
    "DR1TB12A",
    "DR1TVC",
    "DR1TVD",
    "DR1TVK",
    "DR1TCALC",
    "DR1TPHOS",
    "DR1TMAGN",
    "DR1TIRON",
    "DR1TZINC",
    "DR1TCOPP",
    "DR1TSODI",
    "DR1TPOTA",
    "DR1TSELE",
    "DR1TCAFF",
    "DR1TTHEO",
    "DR1TALCO",
    "DR1TMOIS",
    "DR1TS040",
    "DR1TS060",
    "DR1TS080",
    "DR1TS100",
    "DR1TS120",
    "DR1TS140",
    "DR1TS160",
    "DR1TS180",
    "DR1TM161",
    "DR1TM181",
    "DR1TM201",
    "DR1TM221",
    "DR1TP182",
    "DR1TP183",
    "DR1TP184",
    "DR1TP204",
    "DR1TP205",
    "DR1TP225",
    "DR1TP226",
    "DR1_300",
    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ",
    "DR1TWS",
    "DRD340",
    "DRD350A",
    "DRD350AQ",
    "DRD350B",
    "DRD350BQ",
    "DRD350C",
    "DRD350CQ",
    "DRD350D",
    "DRD350DQ",
    "DRD350E",
    "DRD350EQ",
    "DRD350F",
    "DRD350FQ",
    "DRD350G",
    "DRD350GQ",
    "DRD350H",
    "DRD350HQ",
    "DRD350I",
    "DRD350IQ",
    "DRD350J",
    "DRD350JQ",
    "DRD350K",
    "DRD360",
    "DRD370A",
    "DRD370AQ",
    "DRD370B",
    "DRD370BQ",
    "DRD370C",
    "DRD370CQ",
    "DRD370D",
    "DRD370DQ",
    "DRD370E",
    "DRD370EQ",
    "DRD370F",
    "DRD370FQ",
    "DRD370G",
    "DRD370GQ",
    "DRD370H",
    "DRD370HQ",
    "DRD370I",
    "DRD370IQ",
    "DRD370J",
    "DRD370JQ",
    "DRD370K",
    "DRD370KQ",
    "DRD370L",
    "DRD370LQ",
    "DRD370M",
    "DRD370MQ",
    "DRD370N",
    "DRD370NQ",
    "DRD370O",
    "DRD370OQ",
    "DRD370P",
    "DRD370PQ",
    "DRD370Q",
    "DRD370QQ",
    "DRD370R",
    "DRD370RQ",
    "DRD370S",
    "DRD370SQ",
    "DRD370T",
    "DRD370TQ",
    "DRD370U",
    "DRD370UQ",
    "DRD370V"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "DR1TOT_H columns/order differ from expected.\n"
        f"Expected {len(EXPECTED_COLUMNS)} columns.\n"
        f"Actual   {len(df.columns)} columns."
    )


if len(df) != 9813:

    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 168:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 9,813 rows x 168 columns."
)


# ============================================================
# 4. DEFINE GENUINE DIETARY DECIMAL COLUMNS
# ============================================================

#
# VERY IMPORTANT:
#
# These 50 variables contain genuine fractional values.
#
# They MUST remain floating-point/decimal fields.
#
# DO NOT:
#
#   - round them
#   - convert them to Int64
#   - globally strip ".0"
#   - use float_format in to_csv()
#

DECIMAL_COLUMNS = [
    "WTDRD1",
    "WTDR2D",
    "DR1TPROT",
    "DR1TCARB",
    "DR1TSUGR",
    "DR1TFIBE",
    "DR1TTFAT",
    "DR1TSFAT",
    "DR1TMFAT",
    "DR1TPFAT",
    "DR1TATOC",
    "DR1TATOA",
    "DR1TVB1",
    "DR1TVB2",
    "DR1TNIAC",
    "DR1TVB6",
    "DR1TCHL",
    "DR1TVB12",
    "DR1TB12A",
    "DR1TVC",
    "DR1TVD",
    "DR1TVK",
    "DR1TIRON",
    "DR1TZINC",
    "DR1TCOPP",
    "DR1TSELE",
    "DR1TALCO",
    "DR1TMOIS",
    "DR1TS040",
    "DR1TS060",
    "DR1TS080",
    "DR1TS100",
    "DR1TS120",
    "DR1TS140",
    "DR1TS160",
    "DR1TS180",
    "DR1TM161",
    "DR1TM181",
    "DR1TM201",
    "DR1TM221",
    "DR1TP182",
    "DR1TP183",
    "DR1TP184",
    "DR1TP204",
    "DR1TP205",
    "DR1TP225",
    "DR1TP226",
    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ"
]


INTEGER_COLUMNS = [
    col
    for col in EXPECTED_COLUMNS
    if col not in DECIMAL_COLUMNS
]


if len(DECIMAL_COLUMNS) != 50:

    raise ValueError(
        "Expected exactly 50 genuine decimal fields."
    )


if len(INTEGER_COLUMNS) != 118:

    raise ValueError(
        "Expected exactly 118 whole-number fields."
    )


print("\n--- FIELD TYPE PLAN ---")

print(
    "Genuine decimal fields:",
    len(DECIMAL_COLUMNS)
)

print(
    "Whole-number fields:",
    len(INTEGER_COLUMNS)
)


# ============================================================
# 5. CHECK KNOWN XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known SAS XPORT / pandas zero-decoding artifact:
#
#     5.397605346934028e-79
#
# NEVER perform:
#
#     df.replace(TINY_VALUE, 0)
#
# globally without first validating the exact pattern.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


numeric_cols = (
    df.select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)


if len(numeric_cols) != 168:

    raise ValueError(
        "Expected all 168 DR1TOT_H fields to load numerically."
    )


tiny_counts = (
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


true_zero_total = int(
    df[numeric_cols]
    .eq(0)
    .sum()
    .sum()
)


print("\n--- ORIGINAL XPT ZERO CHECK ---")

print(
    "Ordinary numeric zeros before correction:",
    f"{true_zero_total:,}"
)

print(
    "Tiny XPORT artifacts:",
    f"{tiny_total:,}"
)


print("\nTiny values by variable:")


print(
    tiny_counts[
        tiny_counts > 0
    ]
    .sort_values(
        ascending=False
    )
    .to_string()
)


# ============================================================
# 6. VALIDATE EXACT TINY-ZERO PATTERN
# ============================================================

#
# Exact pattern found in the attached DR1TOT_H.xpt.
#
# 55,707 tiny values across exactly 58 variables.
#

EXPECTED_TINY_COUNTS = {

    "WTDRD1": 1152,
    "WTDR2D": 1087,

    "DR1DBIH": 312,

    "DR1TPROT": 1,
    "DR1TFIBE": 77,
    "DR1TTFAT": 2,
    "DR1TSFAT": 2,
    "DR1TMFAT": 2,
    "DR1TPFAT": 2,

    "DR1TCHOL": 99,

    "DR1TATOC": 2,
    "DR1TATOA": 6892,

    "DR1TRET": 62,
    "DR1TVARA": 11,
    "DR1TACAR": 1256,
    "DR1TBCAR": 122,
    "DR1TCRYP": 896,
    "DR1TLYCO": 2263,
    "DR1TLZ": 98,

    "DR1TVB2": 1,

    "DR1TFOLA": 3,
    "DR1TFA": 251,
    "DR1TFF": 74,
    "DR1TFDFE": 3,

    "DR1TCHL": 2,

    "DR1TVB12": 19,
    "DR1TB12A": 5591,

    "DR1TVC": 49,
    "DR1TVD": 221,
    "DR1TVK": 5,

    "DR1TPHOS": 1,
    "DR1TMAGN": 1,
    "DR1TSELE": 1,

    "DR1TCAFF": 2263,
    "DR1TTHEO": 4401,

    "DR1TALCO": 7271,

    "DR1TS040": 326,
    "DR1TS060": 355,
    "DR1TS080": 80,
    "DR1TS100": 29,
    "DR1TS120": 15,
    "DR1TS140": 4,
    "DR1TS160": 2,
    "DR1TS180": 3,

    "DR1TM161": 34,
    "DR1TM181": 2,
    "DR1TM201": 57,
    "DR1TM221": 1288,

    "DR1TP182": 2,
    "DR1TP183": 3,
    "DR1TP184": 3953,
    "DR1TP204": 111,
    "DR1TP205": 1018,
    "DR1TP225": 739,
    "DR1TP226": 1250,

    "DR1_320Z": 1972,
    "DR1_330Z": 4676,
    "DR1BWATZ": 5293
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: Tiny-value pattern differs from "
        "validated DR1TOT_H attachment.\n"
        f"Expected variables: {len(EXPECTED_TINY_COUNTS)}\n"
        f"Actual variables:   {len(actual_tiny_counts)}\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{actual_tiny_counts}"
    )


if tiny_total != 55707:

    raise ValueError(
        f"Unexpected total tiny-value count: "
        f"{tiny_total:,}"
    )


if len(EXPECTED_TINY_COUNTS) != 58:

    raise ValueError(
        "Expected exactly 58 tiny-value variables."
    )


if true_zero_total != 0:

    raise ValueError(
        "Unexpected ordinary zeros already existed "
        "before XPORT correction."
    )


print(
    "\nPASS: Exact 55,707-value tiny-zero "
    "pattern across 58 variables validated."
)


# ============================================================
# 7. RESTORE ONLY VALIDATED ZEROS
# ============================================================

print("\n--- VALIDATED ZERO RESTORATION ---")


correction_total = 0


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    mask = (
        df[col]
        .eq(TINY_VALUE)
    )


    actual_count = int(
        mask.sum()
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: expected "
            f"{expected_count:,} tiny values, "
            f"found {actual_count:,}."
        )


    df.loc[
        mask,
        col
    ] = 0


    correction_total += actual_count


    print(
        f"{col}: "
        f"{actual_count:,} restored to 0"
    )


if correction_total != 55707:

    raise ValueError(
        "Incorrect total zero-restoration count."
    )


print(
    "\nPASS:",
    f"{correction_total:,}",
    "validated zero values restored."
)


# ============================================================
# 8. CREATE EXPECTED CORRECTED SOURCE COPY
# ============================================================

#
# expected_df is used for exact fidelity testing.
#
# The ONLY allowed modifications to the pandas-decoded
# source XPT are the validated tiny -> 0 corrections.
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col]
        .eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 9. CONFIRM NO TINY VALUES REMAIN
# ============================================================

remaining_tiny = int(
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print("\n--- POST-CORRECTION TINY CHECK ---")

print(
    "Tiny values remaining:",
    remaining_tiny
)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain after correction."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 10. DETECT GENUINE DECIMAL VARIABLES
# ============================================================

#
# This MUST occur before Int64 conversion.
#
# np.round() is used here ONLY to TEST whether a value
# is whole-number.
#
# It is NOT used to alter data.
#

print("\n--- GENUINE DECIMAL DETECTION ---")


detected_decimal_columns = []


for col in numeric_cols:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        continue


    fractional_mask = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    fractional_count = int(
        fractional_mask.sum()
    )


    if fractional_count > 0:

        detected_decimal_columns.append(
            col
        )


        print(
            f"{col}: "
            f"{fractional_count:,} fractional observations"
        )


if detected_decimal_columns != DECIMAL_COLUMNS:

    raise ValueError(
        "STOP: Genuine decimal-variable pattern changed.\n"
        f"Expected:\n{DECIMAL_COLUMNS}\n\n"
        f"Detected:\n{detected_decimal_columns}"
    )


print(
    "\nPASS: Exactly 50 genuine decimal variables detected."
)


# ============================================================
# 11. DECIMAL FIELD DETAILS
# ============================================================

print("\n--- DECIMAL FIELD DETAILS ---")


for col in DECIMAL_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    fractional_mask = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    print(
        f"{col:10s} "
        f"nonmissing={len(values):5,d}   "
        f"fractional={int(fractional_mask.sum()):5,d}   "
        f"whole-looking={int((~fractional_mask).sum()):5,d}"
    )


# ============================================================
# 12. CLEAN AND VALIDATE SEQN
# ============================================================

seqn = (
    df["SEQN"]
    .dropna()
)


if not np.isclose(
    seqn.to_numpy(
        dtype=float
    ),
    np.round(
        seqn.to_numpy(
            dtype=float
        )
    ),
    rtol=0,
    atol=1e-12
).all():

    raise ValueError(
        "SEQN contains unexpected fractional values."
    )


df["SEQN"] = (
    df["SEQN"]
    .astype("Int64")
)


print("\n--- SEQN ---")


print(
    "Minimum:",
    df["SEQN"].min()
)


print(
    "Maximum:",
    df["SEQN"].max()
)


print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)


print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if df["SEQN"].min() != 73557:

    raise ValueError(
        "Unexpected minimum SEQN."
    )


if df["SEQN"].max() != 83731:

    raise ValueError(
        "Unexpected maximum SEQN."
    )


if int(
    df["SEQN"]
    .isna()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains missing values."
    )


if int(
    df["SEQN"]
    .duplicated()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains duplicates."
    )


print(
    "PASS: SEQN 73557-83731 validated."
)


# ============================================================
# 13. CONVERT ONLY PROVEN WHOLE-NUMBER VARIABLES
# ============================================================

#
# There are 118 proven whole-number fields.
#
# These may safely lose unnecessary ".0".
#
# The 50 DECIMAL_COLUMNS remain untouched.
#
# NO rounding is used during conversion.
#

for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) > 0:

        whole_mask = np.isclose(
            values,
            np.round(values),
            rtol=0,
            atol=1e-12
        )


        if not whole_mask.all():

            raise ValueError(
                f"{col} unexpectedly contains "
                "genuine decimal values."
            )


    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: 118 proven whole-number variables "
    "converted to nullable Int64."
)


print(
    "PASS: All 50 Dietary decimal variables "
    "remain untouched."
)


# ============================================================
# 14. CDC CHECK - WTDRD1
# ============================================================

#
# CDC:
#
# Positive Day 1 weights = 8,661
# Zero = Day 1 recall not done/incomplete = 1,152
# Missing = 0
#
# WTDRD1 is a GENUINE DECIMAL variable.
#

s = df["WTDRD1"]


wtdrd1_zero = int(
    s.eq(0)
    .sum()
)


wtdrd1_positive = int(
    s.gt(0)
    .sum()
)


wtdrd1_missing = int(
    s.isna()
    .sum()
)


print("\n--- WTDRD1 ---")

print(
    "Positive:",
    wtdrd1_positive
)

print(
    "Zero:",
    wtdrd1_zero
)

print(
    "Missing:",
    wtdrd1_missing
)


if not (
    wtdrd1_positive == 8661
    and
    wtdrd1_zero == 1152
    and
    wtdrd1_missing == 0
):

    raise ValueError(
        "WTDRD1 does not match CDC release counts."
    )


print(
    "PASS: WTDRD1 matches CDC."
)


# ============================================================
# 15. CDC CHECK - WTDR2D
# ============================================================

#
# CDC:
#
# Positive two-day weights = 7,574
# Zero = Day 2 recall not done/incomplete = 1,087
# Missing = 1,152
#

s = df["WTDR2D"]


wtdr2d_positive = int(
    s.gt(0)
    .sum()
)


wtdr2d_zero = int(
    s.eq(0)
    .sum()
)


wtdr2d_missing = int(
    s.isna()
    .sum()
)


print("\n--- WTDR2D ---")

print(
    "Positive:",
    wtdr2d_positive
)

print(
    "Zero:",
    wtdr2d_zero
)

print(
    "Missing:",
    wtdr2d_missing
)


if not (
    wtdr2d_positive == 7574
    and
    wtdr2d_zero == 1087
    and
    wtdr2d_missing == 1152
):

    raise ValueError(
        "WTDR2D does not match CDC release counts."
    )


print(
    "PASS: WTDR2D matches CDC."
)


# ============================================================
# 16. DAY 1 / DAY 2 WEIGHT CONSISTENCY
# ============================================================

#
# Day 2 weight is missing for participants whose
# Day 1 recall was not complete.
#
# Therefore:
#
# WTDRD1 = 0
#
# should exactly match:
#
# WTDR2D missing
#

day1_zero_mask = (
    df["WTDRD1"]
    .eq(0)
    .to_numpy(
        dtype=bool
    )
)


day2_missing_mask = (
    df["WTDR2D"]
    .isna()
    .to_numpy(
        dtype=bool
    )
)


weight_routing_mismatches = int(
    np.sum(
        day1_zero_mask
        !=
        day2_missing_mask
    )
)


print("\n--- DIETARY WEIGHT ROUTING ---")

print(
    "WTDRD1=0 vs WTDR2D missing mismatches:",
    weight_routing_mismatches
)


if weight_routing_mismatches != 0:

    raise ValueError(
        "WTDRD1 zero positions do not exactly "
        "match WTDR2D missing positions."
    )


print(
    "PASS: Dietary weight routing validated."
)


# ============================================================
# 17. CDC CHECK - DR1DRSTZ
# ============================================================

#
# Dietary recall status.
#
# CDC:
#
# 1 Reliable/minimum criteria       8,531
# 2 Not reliable                      122
# 4 Reported breast milk              130
# 5 Not done                        1,030
#

expected_dr1drstz = {
    1: 8531,
    2: 122,
    4: 130,
    5: 1030
}


actual_dr1drstz = (
    df["DR1DRSTZ"]
    .value_counts()
    .sort_index()
    .to_dict()
)


print("\n--- DR1DRSTZ ---")

print(
    actual_dr1drstz
)


if actual_dr1drstz != expected_dr1drstz:

    raise ValueError(
        "DR1DRSTZ frequencies differ from CDC."
    )


if int(
    df["DR1DRSTZ"]
    .isna()
    .sum()
) != 0:

    raise ValueError(
        "Unexpected DR1DRSTZ missing values."
    )


print(
    "PASS: DR1DRSTZ matches CDC."
)


# ============================================================
# 18. CDC CHECK - DRDINT
# ============================================================

#
# Number of intake days.
#
# Day 1 only = 1,087
# Day 1 + Day 2 = 7,574
# Missing = 1,152
#

expected_drdint = {
    1: 1087,
    2: 7574
}


actual_drdint = (
    df["DRDINT"]
    .value_counts()
    .sort_index()
    .to_dict()
)


if actual_drdint != expected_drdint:

    raise ValueError(
        "DRDINT frequencies differ from CDC."
    )


if int(
    df["DRDINT"]
    .isna()
    .sum()
) != 1152:

    raise ValueError(
        "Unexpected DRDINT missing count."
    )


print(
    "PASS: DRDINT matches CDC."
)


# ============================================================
# 19. WTDR2D / DRDINT CONSISTENCY
# ============================================================

#
# DRDINT = 2 should correspond to positive WTDR2D.
#
# DRDINT = 1 should correspond to WTDR2D = 0.
#

two_day_mask = (
    df["DRDINT"]
    .eq(2)
    .fillna(False)
    .to_numpy(
        dtype=bool
    )
)


positive_two_day_weight = (
    df["WTDR2D"]
    .gt(0)
    .fillna(False)
    .to_numpy(
        dtype=bool
    )
)


if int(
    np.sum(
        two_day_mask
        !=
        positive_two_day_weight
    )
) != 0:

    raise ValueError(
        "DRDINT=2 does not exactly match "
        "positive WTDR2D."
    )


day1_only_mask = (
    df["DRDINT"]
    .eq(1)
    .fillna(False)
    .to_numpy(
        dtype=bool
    )
)


zero_two_day_weight = (
    df["WTDR2D"]
    .eq(0)
    .fillna(False)
    .to_numpy(
        dtype=bool
    )
)


if int(
    np.sum(
        day1_only_mask
        !=
        zero_two_day_weight
    )
) != 0:

    raise ValueError(
        "DRDINT=1 does not exactly match "
        "WTDR2D=0."
    )


print(
    "PASS: DRDINT and two-day weights are consistent."
)


# ============================================================
# 20. CDC CHECK - DR1DBIH
# ============================================================

#
# Days between dietary intake and household interview.
#
# CDC released range:
#
#     -31 through 74
#
# The attached XPT contains 312 valid zeros.
#

s = df["DR1DBIH"]


if int(
    s.notna()
    .sum()
) != 8706:

    raise ValueError(
        "Unexpected DR1DBIH nonmissing count."
    )


if int(
    s.isna()
    .sum()
) != 1107:

    raise ValueError(
        "Unexpected DR1DBIH missing count."
    )


if int(
    s.eq(0)
    .sum()
) != 312:

    raise ValueError(
        "Unexpected DR1DBIH zero count."
    )


if s.dropna().min() != -31:

    raise ValueError(
        "Unexpected DR1DBIH minimum."
    )


if s.dropna().max() != 74:

    raise ValueError(
        "Unexpected DR1DBIH maximum."
    )


print(
    "PASS: DR1DBIH matches CDC range and "
    "312 validated zero-day records."
)


# ============================================================
# 21. NUTRIENT COMPLETENESS CHECK
# ============================================================

#
# The 65 total nutrient/component variables from
# DR1TKCAL through DR1TP226 have:
#
#     8,531 nonmissing
#     1,282 missing
#
# in the released file.
#

start_idx = (
    EXPECTED_COLUMNS
    .index("DR1TKCAL")
)


end_idx = (
    EXPECTED_COLUMNS
    .index("DR1_300")
)


NUTRIENT_COLUMNS = (
    EXPECTED_COLUMNS[
        start_idx:end_idx
    ]
)


if len(NUTRIENT_COLUMNS) != 65:

    raise ValueError(
        "Expected 65 nutrient/component columns."
    )


nutrient_missing_failures = {}


for col in NUTRIENT_COLUMNS:

    nonmissing_count = int(
        df[col]
        .notna()
        .sum()
    )

    missing_count = int(
        df[col]
        .isna()
        .sum()
    )


    if (
        nonmissing_count != 8531
        or
        missing_count != 1282
    ):

        nutrient_missing_failures[col] = {
            "nonmissing": nonmissing_count,
            "missing": missing_count
        }


if nutrient_missing_failures:

    raise ValueError(
        "Unexpected nutrient missing pattern:\n"
        + str(nutrient_missing_failures)
    )


print(
    "PASS: All 65 nutrient/component fields have "
    "8,531 nonmissing and 1,282 missing records."
)


# ============================================================
# 22. KEY ZERO-BEARING NUTRIENT CHECKS
# ============================================================

#
# These large zero counts provide additional validation
# of the XPORT correction.
#

KEY_ZERO_COUNTS = {
    "DR1TALCO": 7271,
    "DR1TATOA": 6892,
    "DR1TB12A": 5591,
    "DR1TTHEO": 4401,
    "DR1TP184": 3953,
    "DR1TCAFF": 2263,
    "DR1TLYCO": 2263,
    "DR1TM221": 1288,
    "DR1TACAR": 1256,
    "DR1TP226": 1250
}


print("\n--- KEY NUTRIENT ZERO CHECK ---")


for col, expected_zero_count in KEY_ZERO_COUNTS.items():

    actual_zero_count = int(
        df[col]
        .eq(0)
        .sum()
    )


    print(
        f"{col}: {actual_zero_count:,}"
    )


    if actual_zero_count != expected_zero_count:

        raise ValueError(
            f"{col}: expected "
            f"{expected_zero_count:,} zeros, "
            f"found {actual_zero_count:,}."
        )


print(
    "PASS: Key nutrient zero counts validated."
)


# ============================================================
# 23. WATER VARIABLE CHECKS
# ============================================================

#
# CDC:
#
# DR1_320Z - total plain water
# DR1_330Z - total tap water
# DR1BWATZ - total bottled water
#
# Each has 8,661 records and 1,152 missing.
#
# The valid ranges include zero.
#

WATER_ZERO_COUNTS = {
    "DR1_320Z": 1972,
    "DR1_330Z": 4676,
    "DR1BWATZ": 5293
}


WATER_MAXIMUMS = {
    "DR1_320Z": 13440,
    "DR1_330Z": 13440,
    "DR1BWATZ": 11310
}


print("\n--- WATER CHECK ---")


for col in WATER_ZERO_COUNTS:

    nonmissing_count = int(
        df[col]
        .notna()
        .sum()
    )

    missing_count = int(
        df[col]
        .isna()
        .sum()
    )

    zero_count = int(
        df[col]
        .eq(0)
        .sum()
    )

    minimum = float(
        df[col]
        .dropna()
        .min()
    )

    maximum = float(
        df[col]
        .dropna()
        .max()
    )


    print(
        f"{col}: "
        f"nonmissing={nonmissing_count:,}, "
        f"zero={zero_count:,}, "
        f"min={minimum}, "
        f"max={maximum}"
    )


    if nonmissing_count != 8661:

        raise ValueError(
            f"{col}: unexpected nonmissing count."
        )


    if missing_count != 1152:

        raise ValueError(
            f"{col}: unexpected missing count."
        )


    if zero_count != WATER_ZERO_COUNTS[col]:

        raise ValueError(
            f"{col}: unexpected zero count."
        )


    if minimum != 0:

        raise ValueError(
            f"{col}: expected minimum zero."
        )


    if maximum != WATER_MAXIMUMS[col]:

        raise ValueError(
            f"{col}: unexpected maximum."
        )


print(
    "PASS: Plain/tap/bottled water variables validated."
)


# ============================================================
# 24. CDC CHECK - DR1_300
# ============================================================

#
# Amount eaten yesterday compared with usual.
#

expected_dr1_300 = {
    1: 830,
    2: 6210,
    3: 1673,
    9: 70
}


actual_dr1_300 = (
    df["DR1_300"]
    .value_counts()
    .sort_index()
    .to_dict()
)


if actual_dr1_300 != expected_dr1_300:

    raise ValueError(
        "DR1_300 frequencies differ from CDC."
    )


if int(
    df["DR1_300"]
    .isna()
    .sum()
) != 1030:

    raise ValueError(
        "Unexpected DR1_300 missing count."
    )


print(
    "PASS: DR1_300 matches CDC."
)


# ============================================================
# 25. CDC CHECK - DR1TWS
# ============================================================

#
# Tap water source.
#

expected_dr1tws = {
    1: 6033,
    2: 570,
    3: 124,
    4: 1648,
    99: 408
}


actual_dr1tws = (
    df["DR1TWS"]
    .value_counts()
    .sort_index()
    .to_dict()
)


if actual_dr1tws != expected_dr1tws:

    raise ValueError(
        "DR1TWS frequencies differ from CDC."
    )


if int(
    df["DR1TWS"]
    .isna()
    .sum()
) != 1030:

    raise ValueError(
        "Unexpected DR1TWS missing count."
    )


print(
    "PASS: DR1TWS matches CDC."
)


# ============================================================
# 26. FINAL VALIDATED ZERO PATTERN
# ============================================================

#
# The raw pandas-decoded attachment contained:
#
#     ordinary zeros = 0
#
# Therefore every final numeric zero should correspond
# exactly to one of the 55,707 validated restorations.
#

final_zero_counts = {}


for col in EXPECTED_COLUMNS:

    zero_count = int(
        df[col]
        .eq(0)
        .sum()
    )


    if zero_count > 0:

        final_zero_counts[col] = zero_count


print("\n--- FINAL ZERO PATTERN ---")

print(
    "Variables containing zero:",
    len(final_zero_counts)
)

print(
    "Total zeros:",
    f"{sum(final_zero_counts.values()):,}"
)


if final_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final zero pattern differs from "
        "the validated attachment pattern.\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{final_zero_counts}"
    )


print(
    "PASS: Final zero pattern exactly matches "
    "the 55,707 validated corrections."
)


# ============================================================
# 27. VALUE FIDELITY BEFORE EXPORT
# ============================================================

#
# This confirms that NOTHING other than:
#
#     TINY_VALUE -> 0
#
# was numerically changed.
#

print("\n--- VALUE FIDELITY BEFORE EXPORT ---")


fidelity_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    current_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )


    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        fidelity_failures.append(
            col
        )


if fidelity_failures:

    raise ValueError(
        "Unexpected numeric changes before export:\n"
        + str(fidelity_failures)
    )


print(
    "PASS: All 168 variables match the exact "
    "expected corrected XPT values."
)


# ============================================================
# 28. EXPORT CSV
# ============================================================

#
# MOST IMPORTANT DIETARY EXPORT RULE:
#
# DO NOT:
#
#     float_format=
#
# DO NOT:
#
#     round()
#
# DO NOT:
#
#     format decimal columns manually
#
# Pandas writes the genuine floating values naturally.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print("\nCSV created:")

print(
    csv_file
)


# ============================================================
# 29. RE-READ WITH ROUND-TRIP FLOAT PARSER
# ============================================================

#
# float_precision="round_trip" is important for exact
# floating-point fidelity testing.
#

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != df.shape:

    raise ValueError(
        "CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV columns/order changed."
    )


print(
    "PASS: CSV structure preserved."
)


# ============================================================
# 30. EXACT CORRECTED XPT -> CSV NUMERIC FIDELITY
# ============================================================

print(
    "\n--- EXACT CORRECTED XPT -> CSV FIDELITY ---"
)


csv_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "Corrected XPT -> CSV differences found:\n"
        + str(csv_failures)
    )


print(
    "PASS: All 168 variables round-trip to "
    "the exact corrected XPT numeric values."
)


# ============================================================
# 31. MISSING VALUE FIDELITY
# ============================================================

missing_mismatches = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )


    exported_missing = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )


    missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            exported_missing
        )
    )


print("\n--- MISSING VALUE FIDELITY ---")

print(
    "Missing-position mismatches:",
    missing_mismatches
)


if missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed."
    )


print(
    "PASS: Missing positions preserved exactly."
)


# ============================================================
# 32. READ FINAL CSV AS RAW TEXT
# ============================================================

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


# ============================================================
# 33. FINAL TINY-VALUE CHECK
# ============================================================

tiny_csv_count = 0


for col in EXPECTED_COLUMNS:

    numeric_values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    tiny_csv_count += int(
        numeric_values
        .eq(TINY_VALUE)
        .sum()
    )


print("\n--- FINAL TINY VALUE CHECK ---")

print(
    "Tiny values remaining:",
    tiny_csv_count
)


if tiny_csv_count != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain in final CSV."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 34. .0 CHECK - INTEGER FIELDS ONLY
# ============================================================

#
# CRITICAL FOR DIETARY DATA:
#
# DO NOT search all 168 columns for ".0".
#
# There are 50 genuine decimal columns.
#
# Legitimate examples include:
#
#     WTDRD1   = 0.0
#     WTDR2D   = 0.0
#     DR1TALCO = 0.0
#     DR1TATOA = 0.0
#     DR1_320Z = 0.0
#
# These are NOT formatting errors.
#
# Only the 118 proven integer/code fields are checked.
#

print("\n--- INTEGER COLUMN .0 CHECK ---")


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = raw_csv_df[col]


    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )


    count = int(
        bad_mask.sum()
    )


    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unexpected '.0' formatting in "
        "whole-number fields:\n"
        + str(dot_zero_errors)
    )


print(
    "PASS: No unnecessary '.0' in "
    "the 118 whole-number fields."
)


# ============================================================
# 35. LEGITIMATE DECIMAL .0 TOKENS
# ============================================================

#
# INFORMATIONAL ONLY.
#
# Do not fail simply because a genuine Dietary
# decimal column ends in ".0".
#

print(
    "\n--- LEGITIMATE DIETARY DECIMAL .0 TOKENS ---"
)


decimal_dot_zero_counts = {}


for col in DECIMAL_COLUMNS:

    count = int(
        (
            raw_csv_df[col]
            .ne("")
            &
            raw_csv_df[col]
            .str.endswith(".0")
        )
        .sum()
    )


    decimal_dot_zero_counts[col] = count


    print(
        f"{col}: {count:,}"
    )


total_decimal_dot_zero = sum(
    decimal_dot_zero_counts.values()
)


print(
    "\nTotal legitimate '.0' tokens "
    "inside decimal fields:",
    f"{total_decimal_dot_zero:,}"
)


print(
    "PASS: Dietary decimal fields were "
    "not globally reformatted."
)


# ============================================================
# 36. EXACT DECIMAL FIDELITY
# ============================================================

#
# Validate the 50 genuine decimal columns separately.
#

print("\n--- EXACT DIETARY DECIMAL FIDELITY ---")


decimal_failures = []


for col in DECIMAL_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        decimal_failures.append(
            col
        )


if decimal_failures:

    raise ValueError(
        "Dietary decimal fidelity failed:\n"
        + str(decimal_failures)
    )


print(
    "PASS: All 50 genuine Dietary decimal "
    "variables preserved exactly."
)


# ============================================================
# 37. FINAL CSV ZERO-PATTERN CHECK
# ============================================================

csv_zero_counts = {}


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    zero_count = int(
        values.eq(0)
        .sum()
    )


    if zero_count > 0:

        csv_zero_counts[col] = zero_count


print("\n--- FINAL CSV ZERO PATTERN ---")

print(
    "Variables containing zero:",
    len(csv_zero_counts)
)

print(
    "Total zeros:",
    f"{sum(csv_zero_counts.values()):,}"
)


if csv_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final CSV zero pattern differs.\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{csv_zero_counts}"
    )


print(
    "PASS: Final CSV zero pattern exactly matches "
    "the validated 55,707 corrections."
)


# ============================================================
# 38. SCIENTIFIC NOTATION CHECK
# ============================================================

with open(
    csv_file,
    "r",
    encoding="utf-8"
) as f:

    csv_text = f.read()


scientific_count = (
    csv_text.lower().count("e-")
    +
    csv_text.lower().count("e+")
)


print(
    "\n--- SCIENTIFIC NOTATION CHECK ---"
)

print(
    "Scientific notation occurrences:",
    scientific_count
)


if scientific_count != 0:

    raise ValueError(
        "Unexpected scientific notation found."
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 39. FINAL SEQN CHECK
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if final_seqn.min() != 73557:

    raise ValueError(
        "Final CSV minimum SEQN is incorrect."
    )


if final_seqn.max() != 83731:

    raise ValueError(
        "Final CSV maximum SEQN is incorrect."
    )


if int(
    final_seqn.duplicated()
    .sum()
) != 0:

    raise ValueError(
        "Final CSV contains duplicate SEQN."
    )


if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "Final CSV SEQN still contains '.0'."
    )


print(
    "PASS: Final CSV SEQN valid "
    "and contains no '.0'."
)


# ============================================================
# 40. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 90
)


print(
    "DR1TOT_H CONVERSION COMPLETE"
)


print(
    "=" * 90
)


print(
    "Cycle: 2013-2014"
)


print(
    "Rows:",
    f"{len(df):,}"
)


print(
    "Columns:",
    len(df.columns)
)


print(
    "SEQN range:",
    f"{df['SEQN'].min()}-{df['SEQN'].max()}"
)


print(
    "Original tiny-value artifacts:",
    f"{tiny_total:,}"
)


print(
    "Variables containing tiny artifacts:",
    len(EXPECTED_TINY_COUNTS)
)


print(
    "Validated zero restorations:",
    f"{correction_total:,}"
)


print(
    "Genuine Dietary decimal variables:",
    len(DECIMAL_COLUMNS)
)


print(
    "Whole-number/code variables:",
    len(INTEGER_COLUMNS)
)


print(
    "WTDRD1 zero restorations:",
    EXPECTED_TINY_COUNTS["WTDRD1"]
)


print(
    "WTDR2D zero restorations:",
    EXPECTED_TINY_COUNTS["WTDR2D"]
)


print(
    "DR1TALCO zero restorations:",
    EXPECTED_TINY_COUNTS["DR1TALCO"]
)


print(
    "DR1TATOA zero restorations:",
    EXPECTED_TINY_COUNTS["DR1TATOA"]
)


print(
    "DR1TB12A zero restorations:",
    EXPECTED_TINY_COUNTS["DR1TB12A"]
)


print(
    "DR1_320Z zero restorations:",
    EXPECTED_TINY_COUNTS["DR1_320Z"]
)


print(
    "DR1_330Z zero restorations:",
    EXPECTED_TINY_COUNTS["DR1_330Z"]
)


print(
    "DR1BWATZ zero restorations:",
    EXPECTED_TINY_COUNTS["DR1BWATZ"]
)


print(
    "XPT -> CSV unexpected numeric differences: "
    "0 expected"
)


print(
    "Missing-position differences: "
    "0 expected"
)


print(
    "Dietary decimal differences: "
    "0 expected"
)


print(
    "Tiny values remaining: "
    "0 expected"
)


print(
    "Unnecessary '.0' in integer fields: "
    "0 expected"
)


print(
    "Legitimate '.0' in decimal fields: "
    "ALLOWED"
)


print(
    "Scientific notation: "
    "0 expected"
)
```
