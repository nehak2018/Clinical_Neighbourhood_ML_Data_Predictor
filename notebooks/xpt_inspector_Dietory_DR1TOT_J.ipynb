```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2017_2018"
    / "Dietary"
)

xpt_file = DATA_DIR / "DR1TOT_J.xpt"
csv_file = DATA_DIR / "DR1TOT_J_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Preserve completely untouched pandas-decoded source.
#
source_df = df.copy()


print("=" * 100)
print("NHANES 2017-2018 DR1TOT_J XPT -> CSV")
print("=" * 100)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "WTDRD1",
    "WTDR2D",
    "DR1DRSTZ",
    "DR1EXMER",
    "DRABF",
    "DRDINT",
    "DR1DBIH",
    "DR1DAY",
    "DR1LANG",
    "DR1MRESP",
    "DR1HELP",
    "DBQ095Z",
    "DBD100",
    "DRQSPREP",
    "DR1STY",
    "DR1SKY",
    "DRQSDIET",
    "DRQSDT1",
    "DRQSDT2",
    "DRQSDT3",
    "DRQSDT4",
    "DRQSDT5",
    "DRQSDT6",
    "DRQSDT7",
    "DRQSDT8",
    "DRQSDT9",
    "DRQSDT10",
    "DRQSDT11",
    "DRQSDT12",
    "DRQSDT91",
    "DR1TNUMF",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TCARB",
    "DR1TSUGR",
    "DR1TFIBE",
    "DR1TTFAT",
    "DR1TSFAT",
    "DR1TMFAT",
    "DR1TPFAT",
    "DR1TCHOL",
    "DR1TATOC",
    "DR1TATOA",
    "DR1TRET",
    "DR1TVARA",
    "DR1TACAR",
    "DR1TBCAR",
    "DR1TCRYP",
    "DR1TLYCO",
    "DR1TLZ",
    "DR1TVB1",
    "DR1TVB2",
    "DR1TNIAC",
    "DR1TVB6",
    "DR1TFOLA",
    "DR1TFA",
    "DR1TFF",
    "DR1TFDFE",
    "DR1TCHL",
    "DR1TVB12",
    "DR1TB12A",
    "DR1TVC",
    "DR1TVD",
    "DR1TVK",
    "DR1TCALC",
    "DR1TPHOS",
    "DR1TMAGN",
    "DR1TIRON",
    "DR1TZINC",
    "DR1TCOPP",
    "DR1TSODI",
    "DR1TPOTA",
    "DR1TSELE",
    "DR1TCAFF",
    "DR1TTHEO",
    "DR1TALCO",
    "DR1TMOIS",
    "DR1TS040",
    "DR1TS060",
    "DR1TS080",
    "DR1TS100",
    "DR1TS120",
    "DR1TS140",
    "DR1TS160",
    "DR1TS180",
    "DR1TM161",
    "DR1TM181",
    "DR1TM201",
    "DR1TM221",
    "DR1TP182",
    "DR1TP183",
    "DR1TP184",
    "DR1TP204",
    "DR1TP205",
    "DR1TP225",
    "DR1TP226",
    "DR1_300",
    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ",
    "DR1TWSZ",
    "DRD340",
    "DRD350A",
    "DRD350AQ",
    "DRD350B",
    "DRD350BQ",
    "DRD350C",
    "DRD350CQ",
    "DRD350D",
    "DRD350DQ",
    "DRD350E",
    "DRD350EQ",
    "DRD350F",
    "DRD350FQ",
    "DRD350G",
    "DRD350GQ",
    "DRD350H",
    "DRD350HQ",
    "DRD350I",
    "DRD350IQ",
    "DRD350J",
    "DRD350JQ",
    "DRD350K",
    "DRD360",
    "DRD370A",
    "DRD370AQ",
    "DRD370B",
    "DRD370BQ",
    "DRD370C",
    "DRD370CQ",
    "DRD370D",
    "DRD370DQ",
    "DRD370E",
    "DRD370EQ",
    "DRD370F",
    "DRD370FQ",
    "DRD370G",
    "DRD370GQ",
    "DRD370H",
    "DRD370HQ",
    "DRD370I",
    "DRD370IQ",
    "DRD370J",
    "DRD370JQ",
    "DRD370K",
    "DRD370KQ",
    "DRD370L",
    "DRD370LQ",
    "DRD370M",
    "DRD370MQ",
    "DRD370N",
    "DRD370NQ",
    "DRD370O",
    "DRD370OQ",
    "DRD370P",
    "DRD370PQ",
    "DRD370Q",
    "DRD370QQ",
    "DRD370R",
    "DRD370RQ",
    "DRD370S",
    "DRD370SQ",
    "DRD370T",
    "DRD370TQ",
    "DRD370U",
    "DRD370UQ",
    "DRD370V"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "DR1TOT_J columns/order differ from expected.\n\n"
        f"Expected count: {len(EXPECTED_COLUMNS)}\n"
        f"Actual count:   {len(df.columns)}"
    )


if len(df) != 8704:

    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 168:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 8,704 rows x 168 columns."
)


# ============================================================
# 4. FIELD-TYPE / PRESERVATION PLAN
# ============================================================

#
# CRITICAL DIETARY POLICY
# ------------------------------------------------------------
#
# We do NOT decide whether a dietary measurement is an
# integer variable simply because the current observations
# happen to have no fractional part.
#
# For example:
#
#     DR1TKCAL
#     DR1TCHOL
#     DR1TCALC
#     DR1TSODI
#
# may contain whole-number values in this release, but they
# are still Dietary MEASUREMENTS.
#
# They are therefore preserved as numeric measurement fields
# rather than converted to Int64.
#
#
# We similarly preserve:
#
#     dietary weights
#     nutrient totals
#     fatty acid totals
#     water amounts
#
# AS-IS numerically.
#

PRESERVE_AS_FLOAT_COLUMNS = [

    # Dietary survey weights
    "WTDRD1",
    "WTDR2D",

    # Total nutrient / intake measurements
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TCARB",
    "DR1TSUGR",
    "DR1TFIBE",
    "DR1TTFAT",
    "DR1TSFAT",
    "DR1TMFAT",
    "DR1TPFAT",
    "DR1TCHOL",
    "DR1TATOC",
    "DR1TATOA",
    "DR1TRET",
    "DR1TVARA",
    "DR1TACAR",
    "DR1TBCAR",
    "DR1TCRYP",
    "DR1TLYCO",
    "DR1TLZ",
    "DR1TVB1",
    "DR1TVB2",
    "DR1TNIAC",
    "DR1TVB6",
    "DR1TFOLA",
    "DR1TFA",
    "DR1TFF",
    "DR1TFDFE",
    "DR1TCHL",
    "DR1TVB12",
    "DR1TB12A",
    "DR1TVC",
    "DR1TVD",
    "DR1TVK",
    "DR1TCALC",
    "DR1TPHOS",
    "DR1TMAGN",
    "DR1TIRON",
    "DR1TZINC",
    "DR1TCOPP",
    "DR1TSODI",
    "DR1TPOTA",
    "DR1TSELE",
    "DR1TCAFF",
    "DR1TTHEO",
    "DR1TALCO",
    "DR1TMOIS",
    "DR1TS040",
    "DR1TS060",
    "DR1TS080",
    "DR1TS100",
    "DR1TS120",
    "DR1TS140",
    "DR1TS160",
    "DR1TS180",
    "DR1TM161",
    "DR1TM181",
    "DR1TM201",
    "DR1TM221",
    "DR1TP182",
    "DR1TP183",
    "DR1TP184",
    "DR1TP204",
    "DR1TP205",
    "DR1TP225",
    "DR1TP226",

    # Water measurements
    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ"
]


#
# Everything else is an identifier, categorical code,
# whole-number count, recall variable, fish/shellfish code,
# or fish/shellfish frequency count.
#

INTEGER_COLUMNS = [
    col
    for col in EXPECTED_COLUMNS
    if col not in PRESERVE_AS_FLOAT_COLUMNS
]


if len(PRESERVE_AS_FLOAT_COLUMNS) != 70:

    raise ValueError(
        "Expected exactly 70 preserved "
        "Dietary measurement columns."
    )


if len(INTEGER_COLUMNS) != 98:

    raise ValueError(
        "Expected exactly 98 integer/code/count fields."
    )


print("\n--- FIELD PRESERVATION PLAN ---")

print(
    "Dietary measurement fields preserved:",
    len(PRESERVE_AS_FLOAT_COLUMNS)
)

print(
    "Proposed integer/code/count fields:",
    len(INTEGER_COLUMNS)
)


# ============================================================
# 5. VERIFY SOURCE IS NUMERIC
# ============================================================

numeric_columns = (
    df.select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)


if numeric_columns != EXPECTED_COLUMNS:

    raise ValueError(
        "Expected all DR1TOT_J fields to "
        "load numerically."
    )


print(
    "PASS: All 168 fields loaded numerically."
)


# ============================================================
# 6. KNOWN XPORT TINY-ZERO VALUE
# ============================================================

#
# Known SAS XPORT / pandas decoded zero:
#
#     5.397605346934028e-79
#
# NEVER perform a blanket dataframe replacement without
# first verifying the exact file-specific pattern.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


tiny_counts = (
    df[EXPECTED_COLUMNS]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


ordinary_zero_counts = (
    df[EXPECTED_COLUMNS]
    .eq(0)
    .sum()
)


ordinary_zero_total = int(
    ordinary_zero_counts.sum()
)


print(
    "\n--- ORIGINAL XPT ZERO CHECK ---"
)

print(
    "Ordinary zeros before repair:",
    f"{ordinary_zero_total:,}"
)

print(
    "Tiny XPORT values:",
    f"{tiny_total:,}"
)


print(
    "\nTiny values by variable:"
)

print(
    tiny_counts[
        tiny_counts > 0
    ]
    .to_string()
)


# ============================================================
# 7. VALIDATED DR1TOT_J TINY-ZERO PATTERN
# ============================================================

#
# Directly validated against attached DR1TOT_J.xpt.
#
# CDC confirms:
#
# WTDRD1:
#     1,063 legitimate zero weights
#
# WTDR2D:
#     1,002 legitimate zero weights
#
# Nutrient/water codebook ranges explicitly include 0.
#
# DR1DBIH:
#     published range -37 to 75, so zero is legitimate.
#

EXPECTED_TINY_COUNTS = {

    "WTDRD1": 1063,
    "WTDR2D": 1002,

    "DR1DBIH": 593,

    "DR1TNUMF": 1,
    "DR1TKCAL": 1,
    "DR1TPROT": 4,
    "DR1TCARB": 1,
    "DR1TSUGR": 2,
    "DR1TFIBE": 73,
    "DR1TTFAT": 4,
    "DR1TSFAT": 5,
    "DR1TMFAT": 5,
    "DR1TPFAT": 5,
    "DR1TCHOL": 102,
    "DR1TATOC": 7,
    "DR1TATOA": 6261,
    "DR1TRET": 55,
    "DR1TVARA": 18,
    "DR1TACAR": 1034,
    "DR1TBCAR": 98,
    "DR1TCRYP": 698,
    "DR1TLYCO": 1980,
    "DR1TLZ": 81,
    "DR1TVB1": 4,
    "DR1TVB2": 5,
    "DR1TNIAC": 4,
    "DR1TVB6": 5,
    "DR1TFOLA": 6,
    "DR1TFA": 250,
    "DR1TFF": 69,
    "DR1TFDFE": 6,
    "DR1TCHL": 6,
    "DR1TVB12": 39,
    "DR1TB12A": 5284,
    "DR1TVC": 50,
    "DR1TVD": 228,
    "DR1TVK": 8,
    "DR1TCALC": 1,
    "DR1TPHOS": 3,
    "DR1TMAGN": 2,
    "DR1TIRON": 4,
    "DR1TZINC": 1,
    "DR1TCOPP": 2,
    "DR1TSODI": 1,
    "DR1TPOTA": 1,
    "DR1TSELE": 4,
    "DR1TCAFF": 1817,
    "DR1TTHEO": 3747,
    "DR1TALCO": 6481,
    "DR1TMOIS": 1,

    "DR1TS040": 263,
    "DR1TS060": 244,
    "DR1TS080": 71,
    "DR1TS100": 26,
    "DR1TS120": 19,
    "DR1TS140": 8,
    "DR1TS160": 5,
    "DR1TS180": 6,

    "DR1TM161": 19,
    "DR1TM181": 5,
    "DR1TM201": 32,
    "DR1TM221": 987,

    "DR1TP182": 5,
    "DR1TP183": 8,
    "DR1TP184": 3367,
    "DR1TP204": 43,
    "DR1TP205": 807,
    "DR1TP225": 488,
    "DR1TP226": 1219,

    "DR1_320Z": 1504,
    "DR1_330Z": 4494,
    "DR1BWATZ": 4034
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: DR1TOT_J tiny-value pattern differs "
        "from the validated attachment.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{actual_tiny_counts}"
    )


if tiny_total != 48776:

    raise ValueError(
        f"Unexpected total tiny-value count: "
        f"{tiny_total:,}"
    )


if ordinary_zero_total != 0:

    raise ValueError(
        "Unexpected ordinary numeric zeros existed "
        "before XPORT repair."
    )


print(
    "PASS: Exact 48,776-value tiny-zero "
    "pattern validated."
)


# ============================================================
# 8. RESTORE ONLY THE VALIDATED ZEROS
# ============================================================

#
# IMPORTANT:
#
# No:
#
#     df.replace(TINY_VALUE, 0)
#
# across the whole dataframe.
#
# Each validated field is corrected independently.
#

correction_counts = {}


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    mask = (
        df[col]
        .eq(TINY_VALUE)
    )


    actual_count = int(
        mask.sum()
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: unexpected tiny count.\n"
            f"Expected: {expected_count:,}\n"
            f"Actual:   {actual_count:,}"
        )


    df.loc[
        mask,
        col
    ] = 0


    correction_counts[col] = actual_count


total_corrections = sum(
    correction_counts.values()
)


print(
    "\n--- VALIDATED ZERO RESTORATION ---"
)

print(
    "Variables repaired:",
    len(correction_counts)
)

print(
    "Total zero restorations:",
    f"{total_corrections:,}"
)


if total_corrections != 48776:

    raise ValueError(
        "Total zero-restoration count incorrect."
    )


print(
    "PASS: 48,776 validated zeros restored."
)


# ============================================================
# 9. EXPECTED CORRECTED SOURCE
# ============================================================

#
# This represents exactly what the final dataset is
# allowed to contain.
#
# ONLY:
#
#     validated tiny XPORT value -> 0
#
# may differ from source_df.
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col]
        .eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 10. CONFIRM NO TINY VALUES REMAIN
# ============================================================

remaining_tiny = int(
    df[EXPECTED_COLUMNS]
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print(
    "\nTiny artifacts remaining:",
    remaining_tiny
)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny XPORT values remain after correction."
    )


print(
    "PASS: No tiny XPORT values remain."
)


# ============================================================
# 11. VALIDATE SEQN
# ============================================================

seqn_values = (
    df["SEQN"]
    .dropna()
    .to_numpy(
        dtype=float
    )
)


if not np.isclose(
    seqn_values,
    np.round(seqn_values),
    rtol=0,
    atol=1e-12
).all():

    raise ValueError(
        "SEQN contains fractional values."
    )


print(
    "\n--- SEQN ---"
)

print(
    "Minimum:",
    df["SEQN"].min()
)

print(
    "Maximum:",
    df["SEQN"].max()
)

print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)

print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if df["SEQN"].min() != 93703:

    raise ValueError(
        "Unexpected minimum SEQN."
    )


if df["SEQN"].max() != 102956:

    raise ValueError(
        "Unexpected maximum SEQN."
    )


if int(
    df["SEQN"]
    .isna()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains missing values."
    )


if int(
    df["SEQN"]
    .duplicated()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains duplicate values."
    )


print(
    "PASS: SEQN 93703-102956 validated."
)


# ============================================================
# 12. CDC WEIGHT VALIDATION
# ============================================================

#
# WTDRD1
#
# CDC:
#
#     Positive Day-1 weights = 7,641
#     Zero weights          = 1,063
#     Missing               = 0
#
#
# WTDR2D
#
#     Positive two-day weights = 6,639
#     Zero weights             = 1,002
#     Missing                  = 1,063
#

print(
    "\n--- CDC WEIGHT CHECK ---"
)


# WTDRD1

w1_positive = (
    df["WTDRD1"] > 0
)


w1_positive_count = int(
    w1_positive.sum()
)


w1_zero_count = int(
    df["WTDRD1"]
    .eq(0)
    .sum()
)


w1_missing = int(
    df["WTDRD1"]
    .isna()
    .sum()
)


print(
    "WTDRD1 positive:",
    w1_positive_count
)

print(
    "WTDRD1 zero:",
    w1_zero_count
)

print(
    "WTDRD1 missing:",
    w1_missing
)


if w1_positive_count != 7641:

    raise ValueError(
        "WTDRD1 positive-weight count differs."
    )


if w1_zero_count != 1063:

    raise ValueError(
        "WTDRD1 zero-weight count differs."
    )


if w1_missing != 0:

    raise ValueError(
        "WTDRD1 unexpected missing values."
    )


# WTDR2D

w2_positive = (
    df["WTDR2D"] > 0
)


w2_positive_count = int(
    w2_positive.sum()
)


w2_zero_count = int(
    df["WTDR2D"]
    .eq(0)
    .sum()
)


w2_missing = int(
    df["WTDR2D"]
    .isna()
    .sum()
)


print(
    "WTDR2D positive:",
    w2_positive_count
)

print(
    "WTDR2D zero:",
    w2_zero_count
)

print(
    "WTDR2D missing:",
    w2_missing
)


if w2_positive_count != 6639:

    raise ValueError(
        "WTDR2D positive-weight count differs."
    )


if w2_zero_count != 1002:

    raise ValueError(
        "WTDR2D zero-weight count differs."
    )


if w2_missing != 1063:

    raise ValueError(
        "WTDR2D missing count differs."
    )


print(
    "PASS: Dietary sample weights match CDC."
)


# ============================================================
# 13. WEIGHT RANGE CHECK
# ============================================================

w1_positive_values = (
    df.loc[
        df["WTDRD1"] > 0,
        "WTDRD1"
    ]
)


w2_positive_values = (
    df.loc[
        df["WTDR2D"] > 0,
        "WTDR2D"
    ]
)


print(
    "\n--- DIETARY WEIGHT RANGES ---"
)

print(
    "WTDRD1:",
    repr(
        float(w1_positive_values.min())
    ),
    "to",
    repr(
        float(w1_positive_values.max())
    )
)

print(
    "WTDR2D:",
    repr(
        float(w2_positive_values.min())
    ),
    "to",
    repr(
        float(w2_positive_values.max())
    )
)


if not np.isclose(
    float(w1_positive_values.min()),
    1478.156428,
    rtol=0,
    atol=0
):

    raise ValueError(
        "Unexpected WTDRD1 minimum."
    )


if not np.isclose(
    float(w1_positive_values.max()),
    691139.546883,
    rtol=0,
    atol=0
):

    raise ValueError(
        "Unexpected WTDRD1 maximum."
    )


if not np.isclose(
    float(w2_positive_values.min()),
    1232.691132,
    rtol=0,
    atol=0
):

    raise ValueError(
        "Unexpected WTDR2D minimum."
    )


if not np.isclose(
    float(w2_positive_values.max()),
    901995.118861,
    rtol=0,
    atol=0
):

    raise ValueError(
        "Unexpected WTDR2D maximum."
    )


print(
    "PASS: Dietary weight ranges validated."
)


# ============================================================
# 14. DIETARY RECALL STATUS CHECK
# ============================================================

#
# CDC distribution:
#
# 1 = Reliable / minimum criteria met       7,484
# 2 = Not reliable / criteria not met          87
# 4 = Reported consuming breast milk          157
# 5 = Not done                                976
#

expected_recall_status = {
    1: 7484,
    2: 87,
    4: 157,
    5: 976
}


actual_recall_status = (
    df["DR1DRSTZ"]
    .value_counts()
    .sort_index()
    .to_dict()
)


print(
    "\n--- DR1DRSTZ CHECK ---"
)

print(
    actual_recall_status
)


if actual_recall_status != expected_recall_status:

    raise ValueError(
        "DR1DRSTZ distribution differs from CDC."
    )


print(
    "PASS: Dietary recall status matches CDC."
)


# ============================================================
# 15. DR1DBIH CHECK
# ============================================================

#
# CDC:
#
#     range -37 to 75
#     nonmissing 7,458
#     missing 1,246
#
# Zero is legitimate and was decoded as the tiny value
# 593 times in this attachment.
#

dbih_nonmissing = int(
    df["DR1DBIH"]
    .notna()
    .sum()
)


dbih_missing = int(
    df["DR1DBIH"]
    .isna()
    .sum()
)


dbih_zero = int(
    df["DR1DBIH"]
    .eq(0)
    .sum()
)


print(
    "\n--- DR1DBIH CHECK ---"
)

print(
    "Nonmissing:",
    dbih_nonmissing
)

print(
    "Missing:",
    dbih_missing
)

print(
    "Zero:",
    dbih_zero
)

print(
    "Range:",
    df["DR1DBIH"].min(),
    "to",
    df["DR1DBIH"].max()
)


if dbih_nonmissing != 7458:

    raise ValueError(
        "DR1DBIH nonmissing count differs."
    )


if dbih_missing != 1246:

    raise ValueError(
        "DR1DBIH missing count differs."
    )


if df["DR1DBIH"].min() != -37:

    raise ValueError(
        "Unexpected DR1DBIH minimum."
    )


if df["DR1DBIH"].max() != 75:

    raise ValueError(
        "Unexpected DR1DBIH maximum."
    )


if dbih_zero != 593:

    raise ValueError(
        "Unexpected DR1DBIH zero count."
    )


print(
    "PASS: DR1DBIH validated."
)


# ============================================================
# 16. DR1TNUMF CHECK
# ============================================================

#
# Number of foods/beverages reported.
#
# CDC:
#
#     range 0-46
#     nonmissing 7,641
#     missing 1,063
#

tnumf_nonmissing = int(
    df["DR1TNUMF"]
    .notna()
    .sum()
)


tnumf_missing = int(
    df["DR1TNUMF"]
    .isna()
    .sum()
)


if tnumf_nonmissing != 7641:

    raise ValueError(
        "DR1TNUMF nonmissing count differs."
    )


if tnumf_missing != 1063:

    raise ValueError(
        "DR1TNUMF missing count differs."
    )


if df["DR1TNUMF"].min() != 0:

    raise ValueError(
        "Unexpected DR1TNUMF minimum."
    )


if df["DR1TNUMF"].max() != 46:

    raise ValueError(
        "Unexpected DR1TNUMF maximum."
    )


print(
    "PASS: DR1TNUMF range/count validated."
)


# ============================================================
# 17. NUTRIENT COMPLETENESS CHECK
# ============================================================

#
# All 65 nutrient-measurement fields from DR1TKCAL through
# DR1TP226 have:
#
#     7,484 populated
#     1,220 missing
#
# in the attached file / CDC release.
#

NUTRIENT_COLUMNS = [
    col
    for col in EXPECTED_COLUMNS[
        EXPECTED_COLUMNS.index("DR1TKCAL"):
        EXPECTED_COLUMNS.index("DR1TP226") + 1
    ]
]


if len(NUTRIENT_COLUMNS) != 65:

    raise ValueError(
        "Unexpected nutrient-column count."
    )


nutrient_count_failures = {}


for col in NUTRIENT_COLUMNS:

    nonmissing = int(
        df[col]
        .notna()
        .sum()
    )

    missing = int(
        df[col]
        .isna()
        .sum()
    )


    if (
        nonmissing != 7484
        or
        missing != 1220
    ):

        nutrient_count_failures[col] = {
            "nonmissing": nonmissing,
            "missing": missing
        }


if nutrient_count_failures:

    raise ValueError(
        "Nutrient completeness differs:\n"
        + str(nutrient_count_failures)
    )


print(
    "PASS: All 65 nutrient fields have "
    "7,484 populated / 1,220 missing."
)


# ============================================================
# 18. WATER MEASUREMENT CHECKS
# ============================================================

WATER_COLUMNS = [
    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ"
]


EXPECTED_WATER_RANGES = {

    "DR1_320Z": (
        0,
        15360
    ),

    "DR1_330Z": (
        0,
        15360
    ),

    "DR1BWATZ": (
        0,
        10800
    )
}


print(
    "\n--- WATER MEASUREMENT CHECKS ---"
)


for col in WATER_COLUMNS:

    nonmissing = int(
        df[col]
        .notna()
        .sum()
    )

    missing = int(
        df[col]
        .isna()
        .sum()
    )

    expected_min, expected_max = (
        EXPECTED_WATER_RANGES[col]
    )


    print(
        f"{col}:",
        df[col].min(),
        "to",
        df[col].max()
    )


    if nonmissing != 7641:

        raise ValueError(
            f"{col} nonmissing count differs."
        )


    if missing != 1063:

        raise ValueError(
            f"{col} missing count differs."
        )


    if not np.isclose(
        df[col].min(),
        expected_min,
        rtol=0,
        atol=0
    ):

        raise ValueError(
            f"{col} minimum differs."
        )


    if not np.isclose(
        df[col].max(),
        expected_max,
        rtol=0,
        atol=0
    ):

        raise ValueError(
            f"{col} maximum differs."
        )


print(
    "PASS: Water totals match expected "
    "counts/ranges."
)


# ============================================================
# 19. DISCOVER FRACTIONAL VALUES
# ============================================================

#
# IMPORTANT:
#
# This is a diagnostic/proof step.
#
# np.round() is used ONLY as a comparison.
#
# It DOES NOT transform the data.
#

print(
    "\n--- FRACTIONAL VALUE DISCOVERY ---"
)


fractional_counts = {}


for col in EXPECTED_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        fractional_count = 0

    else:

        fractional_count = int(
            (
                ~np.isclose(
                    values,
                    np.round(values),
                    rtol=0,
                    atol=1e-12
                )
            )
            .sum()
        )


    fractional_counts[col] = (
        fractional_count
    )


#
# Print only variables with fractional observations.
#

for col, count in fractional_counts.items():

    if count > 0:

        print(
            f"{col}: {count:,}"
        )


# ============================================================
# 20. PROVE INTEGER/CODE/COUNT FIELDS ARE INTEGRAL
# ============================================================

#
# We do NOT infer this from variable name alone.
#
# Every proposed integer field must contain only
# whole-number values after the validated zero repair.
#

integer_fraction_failures = {}


for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        continue


    whole_mask = np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    failure_count = int(
        (~whole_mask)
        .sum()
    )


    if failure_count > 0:

        integer_fraction_failures[col] = (
            failure_count
        )


if integer_fraction_failures:

    raise ValueError(
        "STOP: Proposed integer fields contain "
        "fractional values:\n"
        + str(integer_fraction_failures)
    )


print(
    "PASS: All 98 ID/code/count fields "
    "are proven integral."
)


# ============================================================
# 21. DO NOT INTEGERIZE DIETARY MEASUREMENTS
# ============================================================

#
# A Dietary measurement is preserved even if the currently
# observed values happen to all be whole numbers.
#
# Examples in this release include variables such as:
#
#     DR1TKCAL
#     DR1TCHOL
#     DR1TCALC
#     DR1TSODI
#
# This prevents datatype decisions from changing merely
# because of the particular values observed in one cycle.
#

print(
    "\n--- DIETARY MEASUREMENT PROTECTION ---"
)

print(
    "Measurement fields protected:",
    len(PRESERVE_AS_FLOAT_COLUMNS)
)


for col in PRESERVE_AS_FLOAT_COLUMNS:

    if col not in df.columns:

        raise ValueError(
            f"Missing protected measurement field: {col}"
        )


print(
    "PASS: All Dietary measurement fields protected."
)


# ============================================================
# 22. PRE-CONVERSION EXACT NUMERIC FIDELITY
# ============================================================

#
# Compare current corrected dataframe to the expected
# corrected source BEFORE any Int64 conversion.
#

pre_conversion_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    actual_values = (
        df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        actual_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        pre_conversion_failures.append(
            col
        )


if pre_conversion_failures:

    raise ValueError(
        "Unexpected numeric modifications before "
        "Int64 conversion:\n"
        + str(pre_conversion_failures)
    )


print(
    "PASS: Corrected dataframe exactly matches "
    "expected corrected XPT values."
)


# ============================================================
# 23. FINAL ZERO PATTERN BEFORE TYPE CONVERSION
# ============================================================

final_zero_counts = {}


for col in EXPECTED_COLUMNS:

    count = int(
        df[col]
        .eq(0)
        .sum()
    )


    if count > 0:

        final_zero_counts[col] = count


if final_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final zero pattern differs from validated "
        "DR1TOT_J pattern.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{final_zero_counts}"
    )


print(
    "PASS: Final zero pattern exactly matches "
    "validated XPORT corrections."
)


# ============================================================
# 24. CONVERT ONLY PROVEN INTEGER/CODE/COUNT FIELDS
# ============================================================

#
# PRESERVE_AS_FLOAT_COLUMNS are deliberately NOT touched.
#

for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) > 0:

        whole_mask = np.isclose(
            values,
            np.round(values),
            rtol=0,
            atol=1e-12
        )


        if not whole_mask.all():

            raise ValueError(
                f"{col} cannot safely become Int64."
            )


    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: 98 ID/code/count fields converted "
    "to nullable Int64."
)

print(
    "PASS: 70 Dietary measurement fields "
    "left as numeric measurement values."
)


# ============================================================
# 25. CONFIRM MEASUREMENT VALUES DID NOT CHANGE
# ============================================================

measurement_failures = []


for col in PRESERVE_AS_FLOAT_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    actual_values = (
        df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        actual_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        measurement_failures.append(
            col
        )


if measurement_failures:

    raise ValueError(
        "Dietary measurement values changed:\n"
        + str(measurement_failures)
    )


print(
    "PASS: All 70 Dietary measurement fields "
    "remain numerically unchanged."
)


# ============================================================
# 26. COMPLETE PRE-EXPORT FIDELITY
# ============================================================

pre_export_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    actual_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )


    same = np.allclose(
        expected_values,
        actual_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        pre_export_failures.append(
            col
        )


if pre_export_failures:

    raise ValueError(
        "Pre-export numeric fidelity failures:\n"
        + str(pre_export_failures)
    )


print(
    "PASS: All 168 fields match exact "
    "expected corrected values."
)


# ============================================================
# 27. MISSING-POSITION FIDELITY
# ============================================================

missing_position_errors = 0


for col in EXPECTED_COLUMNS:

    original_missing = (
        source_df[col]
        .isna()
        .to_numpy()
    )


    current_missing = (
        df[col]
        .isna()
        .to_numpy()
    )


    missing_position_errors += int(
        (
            original_missing
            !=
            current_missing
        )
        .sum()
    )


print(
    "\nMissing-position differences:",
    missing_position_errors
)


if missing_position_errors != 0:

    raise ValueError(
        "Missing positions changed."
    )


print(
    "PASS: Missing positions preserved exactly."
)


# ============================================================
# 28. EXPORT CSV
# ============================================================

#
# CRITICAL DIETARY FIDELITY RULE
# ------------------------------------------------------------
#
# DO NOT:
#
#     df.round(...)
#
# DO NOT:
#
#     float_format=
#
# DO NOT:
#
#     force decimal places
#
# DO NOT:
#
#     convert Dietary measurements to integers
#
#
# This allows pandas to serialize each measurement from its
# existing floating-point value.
#
# This is especially important for values such as:
#
#     0.028999999999...
#
# where forcing a precision could alter the value.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print(
    "\nCSV created:"
)

print(
    csv_file
)


# ============================================================
# 29. RE-READ WITH ROUND-TRIP FLOAT PARSER
# ============================================================

#
# Exact float fidelity MUST use:
#
#     float_precision="round_trip"
#
# This prevents false failures caused by generic float
# string reparsing.
#

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != (
    8704,
    168
):

    raise ValueError(
        "CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV column order changed."
    )


print(
    "PASS: CSV structure preserved."
)


# ============================================================
# 30. EXACT CORRECTED XPT -> CSV NUMERIC FIDELITY
# ============================================================

print(
    "\n--- EXACT CORRECTED XPT -> CSV FIDELITY ---"
)


roundtrip_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        roundtrip_failures.append(
            col
        )


if roundtrip_failures:

    raise ValueError(
        "Corrected XPT -> CSV numeric fidelity failures:\n"
        + str(roundtrip_failures)
    )


print(
    "PASS: All 168 variables round-trip exactly."
)


# ============================================================
# 31. SPECIFIC DIETARY MEASUREMENT ROUND-TRIP CHECK
# ============================================================

print(
    "\n--- DIETARY MEASUREMENT EXACT FIDELITY ---"
)


dietary_roundtrip_failures = []


for col in PRESERVE_AS_FLOAT_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    final_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        final_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        dietary_roundtrip_failures.append(
            col
        )


if dietary_roundtrip_failures:

    raise ValueError(
        "Dietary measurement round-trip failures:\n"
        + str(dietary_roundtrip_failures)
    )


print(
    "PASS: All Dietary measurements preserve "
    "their exact corrected XPT numeric values."
)


# ============================================================
# 32. CSV MISSING-POSITION FIDELITY
# ============================================================

csv_missing_errors = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )


    exported_missing = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )


    csv_missing_errors += int(
        (
            expected_missing
            !=
            exported_missing
        )
        .sum()
    )


if csv_missing_errors != 0:

    raise ValueError(
        "CSV missing positions changed."
    )


print(
    "PASS: CSV missing positions preserved exactly."
)


# ============================================================
# 33. READ FINAL CSV AS RAW TEXT
# ============================================================

#
# Raw-text dataframe is for FORMAT checks only.
#
# DO NOT use this for exact float equality.
#

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


if raw_csv_df.shape != (
    8704,
    168
):

    raise ValueError(
        "Raw CSV dimensions changed."
    )


if (
    raw_csv_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "Raw CSV column order changed."
    )


# ============================================================
# 34. INTEGER/CODE/COUNT ".0" CHECK
# ============================================================

#
# ONLY the 98 proven integer/code/count fields are checked.
#
# Dietary measurement variables are deliberately excluded.
#

print(
    "\n--- INTEGER/CODE/COUNT .0 CHECK ---"
)


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = (
        raw_csv_df[col]
    )


    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )


    count = int(
        bad_mask.sum()
    )


    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unnecessary '.0' found in proven "
        "integer/code/count fields:\n"
        + str(dot_zero_errors)
    )


print(
    "PASS: No unnecessary '.0' in "
    "ID/code/count fields."
)


# ============================================================
# 35. SEQN RAW-TEXT CHECK
# ============================================================

if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "SEQN still contains '.0'."
    )


print(
    "PASS: SEQN contains no '.0'."
)


# ============================================================
# 36. DIETARY MEASUREMENT FORMAT POLICY
# ============================================================

#
# We deliberately do NOT remove ".0" from measurement fields.
#
# Example:
#
#     DR1TKCAL = 2000.0
#
# is acceptable because DR1TKCAL is a measurement variable.
#
# A whole-valued observation does not change the variable's
# conceptual type.
#
# The important test is the exact round-trip test above.
#

print(
    "\n--- DIETARY MEASUREMENT RAW-TEXT CHECK ---"
)


for col in PRESERVE_AS_FLOAT_COLUMNS:

    populated = int(
        raw_csv_df[col]
        .ne("")
        .sum()
    )


    dot_zero = int(
        (
            raw_csv_df[col]
            .ne("")
            &
            raw_csv_df[col]
            .str.endswith(".0")
        )
        .sum()
    )


    print(
        f"{col}: "
        f"populated={populated:,}, "
        f"tokens ending .0={dot_zero:,}"
    )


print(
    "PASS: Measurement fields were not subjected "
    "to cosmetic integer formatting."
)


# ============================================================
# 37. FINAL TINY-VALUE CHECK
# ============================================================

#
# Search raw CSV text numerically for the known tiny value.
#
# This is a classification check, not an exact float
# fidelity comparison.
#

tiny_csv_total = 0


for col in EXPECTED_COLUMNS:

    numeric_values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    tiny_csv_total += int(
        numeric_values
        .eq(TINY_VALUE)
        .sum()
    )


print(
    "\n--- FINAL TINY-VALUE CHECK ---"
)

print(
    "Tiny values remaining:",
    tiny_csv_total
)


if tiny_csv_total != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain in CSV."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 38. FINAL ZERO PATTERN
# ============================================================

#
# Because the original XPT had no ordinary numeric zeros,
# every final zero should correspond exactly to a validated
# tiny-zero restoration.
#

csv_zero_counts = {}


for col in EXPECTED_COLUMNS:

    numeric_values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    count = int(
        numeric_values
        .eq(0)
        .sum()
    )


    if count > 0:

        csv_zero_counts[col] = count


if csv_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final CSV zero pattern differs.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{csv_zero_counts}"
    )


print(
    "PASS: Final zero pattern exactly matches "
    "the 48,776 validated restorations."
)


# ============================================================
# 39. SCIENTIFIC NOTATION CHECK
# ============================================================

#
# The known XPORT tiny artifact should never appear as:
#
#     5.397605e-79
#
# in the final CSV.
#

scientific_notation_errors = {}


for col in EXPECTED_COLUMNS:

    tokens = raw_csv_df[col]


    count = int(
        tokens
        .str.contains(
            r"[eE][+-]\d+",
            regex=True
        )
        .sum()
    )


    if count > 0:

        scientific_notation_errors[col] = count


if scientific_notation_errors:

    raise ValueError(
        "Unexpected scientific notation found:\n"
        + str(scientific_notation_errors)
    )


print(
    "PASS: No scientific notation in final CSV."
)


# ============================================================
# 40. FINAL SEQN VALIDATION
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if final_seqn.min() != 93703:

    raise ValueError(
        "Final minimum SEQN incorrect."
    )


if final_seqn.max() != 102956:

    raise ValueError(
        "Final maximum SEQN incorrect."
    )


if int(
    final_seqn.isna()
    .sum()
) != 0:

    raise ValueError(
        "Final CSV contains missing SEQN."
    )


if int(
    final_seqn.duplicated()
    .sum()
) != 0:

    raise ValueError(
        "Final CSV contains duplicate SEQN."
    )


print(
    "PASS: Final SEQN 93703-102956 validated."
)


# ============================================================
# 41. ROUND-TRIP MAX DIFFERENCE DIAGNOSTIC
# ============================================================

#
# Expected:
#
#     0.0
#
# for every field.
#

print(
    "\n--- ROUND-TRIP MAXIMUM ABSOLUTE DIFFERENCES ---"
)


max_difference_failures = {}


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    final_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    valid = (
        ~np.isnan(expected_values)
        &
        ~np.isnan(final_values)
    )


    if valid.any():

        max_diff = float(
            np.max(
                np.abs(
                    expected_values[valid]
                    -
                    final_values[valid]
                )
            )
        )

    else:

        max_diff = 0.0


    if max_diff != 0:

        max_difference_failures[col] = (
            max_diff
        )


if max_difference_failures:

    print(
        "Non-zero maximum differences:",
        max_difference_failures
    )

    raise ValueError(
        "Round-trip differences detected."
    )


print(
    "PASS: Maximum numeric difference = 0 "
    "for all 168 variables."
)


# ============================================================
# 42. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 100
)

print(
    "DR1TOT_J CONVERSION COMPLETE"
)

print(
    "=" * 100
)

print(
    "Cycle: 2017-2018"
)

print(
    "Component: Dietary Interview - "
    "Total Nutrient Intakes, First Day"
)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)

print(
    "SEQN range:",
    f"{df['SEQN'].min()}-{df['SEQN'].max()}"
)

print(
    "Original tiny XPORT artifacts:",
    f"{tiny_total:,}"
)

print(
    "Variables containing tiny artifacts:",
    len(EXPECTED_TINY_COUNTS)
)

print(
    "Validated zero restorations:",
    f"{total_corrections:,}"
)

print(
    "Integer/code/count fields converted to Int64:",
    len(INTEGER_COLUMNS)
)

print(
    "Dietary measurement fields preserved:",
    len(PRESERVE_AS_FLOAT_COLUMNS)
)

print(
    "WTDRD1 positive weights:",
    w1_positive_count
)

print(
    "WTDRD1 zero weights:",
    w1_zero_count
)

print(
    "WTDR2D positive weights:",
    w2_positive_count
)

print(
    "WTDR2D zero weights:",
    w2_zero_count
)

print(
    "Nutrient fields:",
    len(NUTRIENT_COLUMNS)
)

print(
    "Nutrient records with reliable intake:",
    7484
)

print(
    "Exact Dietary measurement round-trip failures:",
    0
)

print(
    "Exact all-column round-trip failures:",
    0
)

print(
    "Missing-position differences:",
    0
)

print(
    "Tiny values remaining:",
    0
)

print(
    "Unnecessary '.0' in ID/code/count fields:",
    0
)

print(
    "Scientific notation:",
    0
)

print(
    "=" * 100
)
```
