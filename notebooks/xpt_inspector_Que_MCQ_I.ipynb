```python
# =================================================================================================
# NHANES 2015-2016
# MEDICAL CONDITIONS QUESTIONNAIRE - MCQ_I
#
# INPUT:
#   MCQ_I.xpt
#
# OUTPUT:
#   MCQ_I_corrected.csv
#
#
# COMPONENT:
#   Questionnaire
#
# SUBCOMPONENT:
#   Medical Conditions
#
#
# ATTACHED SOURCE FINDINGS
# -------------------------------------------------------------------------------------------------
#
# Rows                         = 9,575
# Columns                      = 90
#
# SEQN range                   = 83732-93702
# Missing SEQN                 = 0
# Duplicate SEQN               = 0
#
#
# VALIDATED SAS XPORT TINY-ZERO ARTIFACTS
# -------------------------------------------------------------------------------------------------
#
# MCQ180B     1
# MCQ180K     1
#
# TOTAL       2
#
#
# CDC VALIDATION
# -------------------------------------------------------------------------------------------------
#
# MCQ180B:
#
#   0-79 years       = 190
#   80 years or more = 19
#   77777 Refused    = 1
#   99999 Don't know = 4
#   Missing          = 9,361
#
#
# MCQ180K:
#
#   0-78 years       = 292
#   80 years or more = 4
#   99999 Don't know = 8
#   Missing          = 9,271
#
#
# MCQ220:
#
#   1 = Yes           549
#   2 = No          5,165
#   9 = Don't know      5
#   Missing          3,856
#
#
# OSQ230:
#
#   1 = Yes           940
#   2 = No          2,796
#   9 = Don't know     28
#   Missing          5,811
#
#
# FIDELITY RULES
# -------------------------------------------------------------------------------------------------
#
# 1. Restore ONLY validated tiny XPORT zero artifacts.
#
# 2. Do NOT globally replace tiny values.
#
# 3. Preserve missing-value positions exactly.
#
# 4. Remove ".0" only after proving every variable is integral.
#
# 5. Do NOT round.
#
# 6. Do NOT use float_format=.
#
# 7. Validate XPT -> CSV numeric round-trip fidelity.
#
# 8. Preserve exact variable order.
#
#
# GENERAL PIPELINE RULE
# -------------------------------------------------------------------------------------------------
#
# Genuine Dietary / Lab / Examination decimal variables in other
# NHANES components must remain AS IS.
#
# MCQ_I itself contains no genuine decimal variables.
#
# =================================================================================================


# =================================================================================================
# 1. IMPORTS
# =================================================================================================

import pandas as pd
import numpy as np
from pathlib import Path


# =================================================================================================
# 2. PATHS
# =================================================================================================

BASE_DIR = Path().resolve().parent


DATA_DIR = (
    BASE_DIR
    / "data"
    / "2015_2016"
    / "Questionnnair"
)


XPT_FILE = (
    DATA_DIR
    / "MCQ_I.xpt"
)


CSV_FILE = (
    DATA_DIR
    / "MCQ_I_corrected.csv"
)


# =================================================================================================
# 3. VERIFY INPUT FILE
# =================================================================================================

if not XPT_FILE.exists():

    raise FileNotFoundError(
        f"MCQ_I XPT file not found:\n{XPT_FILE}"
    )


# =================================================================================================
# 4. READ XPT
# =================================================================================================

df = pd.read_sas(
    XPT_FILE,
    format="xport",
    encoding="latin1"
)


#
# Preserve untouched XPT-decoded source.
#

source_df = df.copy()


print("=" * 110)

print(
    "NHANES 2015-2016 MCQ_I"
)

print(
    "MEDICAL CONDITIONS QUESTIONNAIRE"
)

print("=" * 110)


print(
    "Rows:",
    f"{len(df):,}"
)


print(
    "Columns:",
    len(df.columns)
)


print(
    "Variables:"
)


print(
    df.columns.tolist()
)


# =================================================================================================
# 5. EXPECTED STRUCTURE
# =================================================================================================

EXPECTED_COLUMNS = [

    "SEQN",

    "MCQ010",
    "MCQ025",
    "MCQ035",
    "MCQ040",
    "MCQ050",

    "AGQ030",

    "MCQ053",
    "MCQ080",
    "MCQ092",

    "MCD093",

    "MCQ149",
    "MCQ151",

    "MCQ160A",
    "MCQ180A",
    "MCQ195",

    "MCQ160N",
    "MCQ180N",

    "MCQ160B",
    "MCQ180B",

    "MCQ160C",
    "MCQ180C",

    "MCQ160D",
    "MCQ180D",

    "MCQ160E",
    "MCQ180E",

    "MCQ160F",
    "MCQ180F",

    "MCQ160G",
    "MCQ180G",

    "MCQ160M",
    "MCQ170M",
    "MCQ180M",

    "MCQ160K",
    "MCQ170K",
    "MCQ180K",

    "MCQ160L",
    "MCQ170L",
    "MCQ180L",

    "MCQ160O",

    "MCQ203",
    "MCQ206",

    "MCQ220",

    "MCQ230A",
    "MCQ230B",
    "MCQ230C",
    "MCQ230D",

    "MCQ240A",
    "MCQ240AA",
    "MCQ240B",
    "MCQ240BB",
    "MCQ240C",
    "MCQ240CC",
    "MCQ240D",
    "MCQ240DD",
    "MCQ240DK",
    "MCQ240E",
    "MCQ240F",
    "MCQ240G",
    "MCQ240H",
    "MCQ240I",
    "MCQ240J",
    "MCQ240K",
    "MCQ240L",
    "MCQ240M",
    "MCQ240N",
    "MCQ240O",
    "MCQ240P",
    "MCQ240Q",
    "MCQ240R",
    "MCQ240S",
    "MCQ240T",
    "MCQ240U",
    "MCQ240V",
    "MCQ240W",
    "MCQ240X",
    "MCQ240Y",
    "MCQ240Z",

    "MCQ300A",
    "MCQ300B",
    "MCQ300C",

    "MCQ365A",
    "MCQ365B",
    "MCQ365C",
    "MCQ365D",

    "MCQ370A",
    "MCQ370B",
    "MCQ370C",
    "MCQ370D",

    "OSQ230"

]


EXPECTED_ROWS = 9575


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "MCQ_I structure differs from validated source.\n\n"
        f"Expected columns: {len(EXPECTED_COLUMNS)}\n"
        f"Actual columns:   {len(df.columns)}"
    )


if len(df) != EXPECTED_ROWS:

    raise ValueError(
        "Unexpected MCQ_I row count.\n"
        f"Expected: {EXPECTED_ROWS:,}\n"
        f"Actual:   {len(df):,}"
    )


print(
    "PASS: Structure = 9,575 rows x 90 columns."
)


# =================================================================================================
# 6. VERIFY ALL SOURCE VARIABLES LOAD NUMERICALLY
# =================================================================================================

numeric_columns = (

    df.select_dtypes(
        include=[np.number]
    )

    .columns

    .tolist()

)


if numeric_columns != EXPECTED_COLUMNS:

    raise ValueError(
        "Expected all 90 MCQ_I fields to load numerically."
    )


print(
    "PASS: All 90 variables loaded numerically."
)


# =================================================================================================
# 7. SEQN SOURCE CHECK
# =================================================================================================

seqn_min = int(
    df["SEQN"].min()
)


seqn_max = int(
    df["SEQN"].max()
)


seqn_missing = int(

    df["SEQN"]
    .isna()
    .sum()

)


seqn_duplicates = int(

    df["SEQN"]
    .duplicated()
    .sum()

)


print(
    "\n--- SEQN CHECK ---"
)


print(
    "Minimum:",
    seqn_min
)


print(
    "Maximum:",
    seqn_max
)


print(
    "Missing:",
    seqn_missing
)


print(
    "Duplicates:",
    seqn_duplicates
)


if seqn_min != 83732:

    raise ValueError(
        "Unexpected minimum SEQN."
    )


if seqn_max != 93702:

    raise ValueError(
        "Unexpected maximum SEQN."
    )


if seqn_missing != 0:

    raise ValueError(
        "SEQN contains missing values."
    )


if seqn_duplicates != 0:

    raise ValueError(
        "SEQN contains duplicate values."
    )


print(
    "PASS: SEQN range 83732-93702 validated."
)


# =================================================================================================
# 8. DEFINE KNOWN SAS XPORT TINY VALUE
# =================================================================================================

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


# =================================================================================================
# 9. DISCOVER TINY VALUES
# =================================================================================================

tiny_counts = (

    df[EXPECTED_COLUMNS]

    .eq(TINY_VALUE)

    .sum()

)


tiny_total = int(
    tiny_counts.sum()
)


print(
    "\n--- ORIGINAL TINY-XPORT CHECK ---"
)


print(
    tiny_counts[
        tiny_counts > 0
    ].to_string()
)


print(
    "\nTotal tiny values:",
    tiny_total
)


# =================================================================================================
# 10. VALIDATED TINY-ZERO PATTERN
# =================================================================================================

EXPECTED_TINY_COUNTS = {

    "MCQ180B": 1,

    "MCQ180K": 1

}


actual_tiny_counts = (

    tiny_counts[
        tiny_counts > 0
    ]

    .astype(int)

    .to_dict()

)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: MCQ_I tiny-value pattern differs "
        "from validated attachment.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{actual_tiny_counts}"
    )


if tiny_total != 2:

    raise ValueError(
        "Unexpected total tiny-value count."
    )


print(
    "PASS: Exact two-value XPORT pattern validated."
)


# =================================================================================================
# 11. ORDINARY ZERO CHECK BEFORE RESTORATION
# =================================================================================================

ordinary_zero_counts = (

    df[EXPECTED_COLUMNS]

    .eq(0)

    .sum()

)


ordinary_zero_total = int(
    ordinary_zero_counts.sum()
)


print(
    "\nOrdinary numeric zeros before restoration:",
    ordinary_zero_total
)


if ordinary_zero_total != 0:

    raise ValueError(
        "Unexpected ordinary zero values found before correction."
    )


print(
    "PASS: Source XPT represents the two legitimate zeros "
    "using the tiny XPORT value."
)


# =================================================================================================
# 12. BUILD EXPECTED CORRECTED SOURCE
# =================================================================================================

expected_df = (
    source_df.copy()
)


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col]
        .eq(TINY_VALUE),
        col
    ] = 0


# =================================================================================================
# 13. RESTORE ONLY VALIDATED TRUE ZEROS
# =================================================================================================

correction_counts = {}


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    mask = (
        df[col]
        .eq(TINY_VALUE)
    )


    actual_count = int(
        mask.sum()
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: unexpected tiny-value count."
        )


    df.loc[
        mask,
        col
    ] = 0


    correction_counts[
        col
    ] = actual_count


print(
    "\n--- ZERO RESTORATION ---"
)


print(
    correction_counts
)


if sum(
    correction_counts.values()
) != 2:

    raise ValueError(
        "Incorrect number of zero restorations."
    )


print(
    "PASS: Two validated zeros restored."
)


# =================================================================================================
# 14. VERIFY NO TINY VALUE REMAINS
# =================================================================================================

remaining_tiny = int(

    df[EXPECTED_COLUMNS]

    .eq(TINY_VALUE)

    .sum()

    .sum()

)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny XPORT values remain after correction."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# =================================================================================================
# 15. VERIFY TRUE-ZERO PATTERN AFTER CORRECTION
# =================================================================================================

corrected_zero_counts = {}


for col in EXPECTED_COLUMNS:

    count = int(

        df[col]
        .eq(0)
        .sum()

    )


    if count > 0:

        corrected_zero_counts[
            col
        ] = count


print(
    "\nCorrected zero pattern:"
)


print(
    corrected_zero_counts
)


if corrected_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Corrected zero pattern differs.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{corrected_zero_counts}"
    )


print(
    "PASS: Exactly two legitimate numeric zeros."
)


# =================================================================================================
# 16. CDC CHECK - MCQ180B
# =================================================================================================

#
# CDC:
#
# 0-79       = 190
# 80         = 19
# 77777      = 1
# 99999      = 4
# Missing    = 9,361
#

mcq180b_range = int(

    df["MCQ180B"]

    .between(
        0,
        79
    )

    .sum()

)


mcq180b_80 = int(

    df["MCQ180B"]
    .eq(80)
    .sum()

)


mcq180b_77777 = int(

    df["MCQ180B"]
    .eq(77777)
    .sum()

)


mcq180b_99999 = int(

    df["MCQ180B"]
    .eq(99999)
    .sum()

)


mcq180b_missing = int(

    df["MCQ180B"]
    .isna()
    .sum()

)


print(
    "\n--- CDC CHECK: MCQ180B ---"
)


print(
    "0-79:",
    mcq180b_range
)


print(
    "80+:",
    mcq180b_80
)


print(
    "Refused:",
    mcq180b_77777
)


print(
    "Don't know:",
    mcq180b_99999
)


print(
    "Missing:",
    mcq180b_missing
)


if mcq180b_range != 190:

    raise ValueError(
        "MCQ180B 0-79 count differs from CDC."
    )


if mcq180b_80 != 19:

    raise ValueError(
        "MCQ180B=80 count differs from CDC."
    )


if mcq180b_77777 != 1:

    raise ValueError(
        "MCQ180B Refused count differs from CDC."
    )


if mcq180b_99999 != 4:

    raise ValueError(
        "MCQ180B Don't know count differs from CDC."
    )


if mcq180b_missing != 9361:

    raise ValueError(
        "MCQ180B missing count differs from CDC."
    )


print(
    "PASS: MCQ180B matches CDC."
)


# =================================================================================================
# 17. CDC CHECK - MCQ180K
# =================================================================================================

#
# CDC:
#
# 0-78       = 292
# 80         = 4
# 77777      = 0
# 99999      = 8
# Missing    = 9,271
#

mcq180k_range = int(

    df["MCQ180K"]

    .between(
        0,
        78
    )

    .sum()

)


mcq180k_80 = int(

    df["MCQ180K"]
    .eq(80)
    .sum()

)


mcq180k_77777 = int(

    df["MCQ180K"]
    .eq(77777)
    .sum()

)


mcq180k_99999 = int(

    df["MCQ180K"]
    .eq(99999)
    .sum()

)


mcq180k_missing = int(

    df["MCQ180K"]
    .isna()
    .sum()

)


print(
    "\n--- CDC CHECK: MCQ180K ---"
)


print(
    "0-78:",
    mcq180k_range
)


print(
    "80+:",
    mcq180k_80
)


print(
    "Refused:",
    mcq180k_77777
)


print(
    "Don't know:",
    mcq180k_99999
)


print(
    "Missing:",
    mcq180k_missing
)


if mcq180k_range != 292:

    raise ValueError(
        "MCQ180K 0-78 count differs from CDC."
    )


if mcq180k_80 != 4:

    raise ValueError(
        "MCQ180K=80 count differs from CDC."
    )


if mcq180k_77777 != 0:

    raise ValueError(
        "MCQ180K Refused count differs from CDC."
    )


if mcq180k_99999 != 8:

    raise ValueError(
        "MCQ180K Don't know count differs from CDC."
    )


if mcq180k_missing != 9271:

    raise ValueError(
        "MCQ180K missing count differs from CDC."
    )


print(
    "PASS: MCQ180K matches CDC."
)


# =================================================================================================
# 18. CDC CHECK - MCQ220
# =================================================================================================

#
# Important cancer variable.
#
# CDC:
#
# 1 = Yes          549
# 2 = No         5,165
# 7 = Refused        0
# 9 = Don't know     5
# Missing         3,856
#

MCQ220_EXPECTED = {

    1.0: 549,

    2.0: 5165,

    9.0: 5

}


mcq220_actual = (

    df["MCQ220"]

    .value_counts(
        dropna=True
    )

    .sort_index()

    .to_dict()

)


mcq220_missing = int(

    df["MCQ220"]

    .isna()

    .sum()

)


print(
    "\n--- CDC CHECK: MCQ220 ---"
)


print(
    mcq220_actual
)


print(
    "Missing:",
    mcq220_missing
)


if mcq220_actual != MCQ220_EXPECTED:

    raise ValueError(
        "MCQ220 distribution differs from CDC.\n\n"
        f"Expected:\n{MCQ220_EXPECTED}\n\n"
        f"Actual:\n{mcq220_actual}"
    )


if mcq220_missing != 3856:

    raise ValueError(
        "MCQ220 missing count differs from CDC."
    )


print(
    "PASS: MCQ220 matches CDC exactly."
)


# =================================================================================================
# 19. CDC CHECK - OSQ230
# =================================================================================================

#
# OSQ230 was new in MCQ_I.
#
# CDC:
#
# 1 = Yes          940
# 2 = No         2,796
# 9 = Don't know    28
# Missing         5,811
#

OSQ230_EXPECTED = {

    1.0: 940,

    2.0: 2796,

    9.0: 28

}


osq230_actual = (

    df["OSQ230"]

    .value_counts(
        dropna=True
    )

    .sort_index()

    .to_dict()

)


osq230_missing = int(

    df["OSQ230"]

    .isna()

    .sum()

)


print(
    "\n--- CDC CHECK: OSQ230 ---"
)


print(
    osq230_actual
)


print(
    "Missing:",
    osq230_missing
)


if osq230_actual != OSQ230_EXPECTED:

    raise ValueError(
        "OSQ230 distribution differs from CDC.\n\n"
        f"Expected:\n{OSQ230_EXPECTED}\n\n"
        f"Actual:\n{osq230_actual}"
    )


if osq230_missing != 5811:

    raise ValueError(
        "OSQ230 missing count differs from CDC."
    )


print(
    "PASS: OSQ230 matches CDC exactly."
)


# =================================================================================================
# 20. CDC CHECK - MCQ010
# =================================================================================================

#
# Strong whole-file benchmark:
#
# Yes          1,443
# No           8,121
# Refused          1
# Don't know      10
# Missing          0
#

MCQ010_EXPECTED = {

    1.0: 1443,

    2.0: 8121,

    7.0: 1,

    9.0: 10

}


mcq010_actual = (

    df["MCQ010"]

    .value_counts(
        dropna=True
    )

    .sort_index()

    .to_dict()

)


if mcq010_actual != MCQ010_EXPECTED:

    raise ValueError(
        "MCQ010 distribution differs from CDC.\n\n"
        f"Expected:\n{MCQ010_EXPECTED}\n\n"
        f"Actual:\n{mcq010_actual}"
    )


if int(

    df["MCQ010"]
    .isna()
    .sum()

) != 0:

    raise ValueError(
        "MCQ010 missing count differs from CDC."
    )


print(
    "PASS: MCQ010 whole-file benchmark matches CDC."
)


# =================================================================================================
# 21. FRACTIONAL VALUE DISCOVERY
# =================================================================================================

#
# np.round() is diagnostic only.
#
# It does NOT change data.
#

fractional_counts = {}


for col in EXPECTED_COLUMNS:

    values = (

        df[col]

        .dropna()

        .to_numpy(
            dtype=float
        )

    )


    if len(values) == 0:

        fractional_count = 0


    else:

        fractional_count = int(

            (
                ~np.isclose(

                    values,

                    np.round(
                        values
                    ),

                    rtol=0,

                    atol=1e-12

                )
            )

            .sum()

        )


    fractional_counts[
        col
    ] = fractional_count


fractional_variables = {

    col: count

    for col, count in fractional_counts.items()

    if count > 0

}


print(
    "\n--- FRACTIONAL VALUE CHECK ---"
)


print(
    fractional_variables
)


if fractional_variables:

    raise ValueError(
        "Unexpected fractional variables found:\n"
        + str(
            fractional_variables
        )
    )


print(
    "PASS: MCQ_I contains no genuine decimal variables."
)


# =================================================================================================
# 22. FIELD TYPE PLAN
# =================================================================================================

#
# After validated zero restoration, every variable has been
# mathematically proven integral.
#

INTEGER_COLUMNS = (
    EXPECTED_COLUMNS.copy()
)


DECIMAL_COLUMNS = []


print(
    "\n--- FIELD TYPE PLAN ---"
)


print(
    "Integral/code fields:",
    len(INTEGER_COLUMNS)
)


print(
    "Genuine decimal fields:",
    DECIMAL_COLUMNS
)


# =================================================================================================
# 23. VERIFY ONLY VALIDATED SOURCE VALUES CHANGED
# =================================================================================================

source_change_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (

        expected_df[col]

        .to_numpy(
            dtype=float
        )

    )


    current_values = (

        df[col]

        .to_numpy(
            dtype=float
        )

    )


    same = np.allclose(

        expected_values,

        current_values,

        rtol=0,

        atol=0,

        equal_nan=True

    )


    if not same:

        source_change_failures.append(
            col
        )


if source_change_failures:

    raise ValueError(
        "Unexpected source-value changes:\n"
        + str(
            source_change_failures
        )
    )


print(
    "PASS: Only the two validated tiny-zero values changed."
)


# =================================================================================================
# 24. MISSING-POSITION FIDELITY
# =================================================================================================

missing_position_errors = 0


for col in EXPECTED_COLUMNS:

    source_missing = (

        source_df[col]

        .isna()

        .to_numpy()

    )


    corrected_missing = (

        df[col]

        .isna()

        .to_numpy()

    )


    missing_position_errors += int(

        (
            source_missing
            !=
            corrected_missing
        )

        .sum()

    )


if missing_position_errors != 0:

    raise ValueError(
        "Missing positions changed during correction."
    )


print(
    "PASS: Missing positions preserved exactly."
)


# =================================================================================================
# 25. CONVERT PROVEN INTEGRAL FIELDS TO NULLABLE Int64
# =================================================================================================

#
# This changes:
#
#     83732.0 -> 83732
#
# and:
#
#     1.0 -> 1
#
# but ONLY after verifying no genuine fractions exist.
#

for col in INTEGER_COLUMNS:

    values = (

        df[col]

        .dropna()

        .to_numpy(
            dtype=float
        )

    )


    if len(values) > 0:

        integral_mask = np.isclose(

            values,

            np.round(
                values
            ),

            rtol=0,

            atol=1e-12

        )


        if not integral_mask.all():

            raise ValueError(
                f"{col} unexpectedly contains fractional values."
            )


    df[col] = (

        df[col]

        .astype("Int64")

    )


print(
    "PASS: All 90 proven integral/code fields converted to Int64."
)


# =================================================================================================
# 26. PRE-EXPORT NUMERIC FIDELITY
# =================================================================================================

pre_export_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (

        expected_df[col]

        .to_numpy(
            dtype=float
        )

    )


    current_values = (

        pd.to_numeric(
            df[col],
            errors="coerce"
        )

        .to_numpy(
            dtype=float,
            na_value=np.nan
        )

    )


    same = np.allclose(

        expected_values,

        current_values,

        rtol=0,

        atol=0,

        equal_nan=True

    )


    if not same:

        pre_export_failures.append(
            col
        )


if pre_export_failures:

    raise ValueError(
        "Pre-export fidelity failures:\n"
        + str(
            pre_export_failures
        )
    )


print(
    "PASS: Pre-export numeric fidelity."
)


# =================================================================================================
# 27. EXPORT CORRECTED CSV
# =================================================================================================

#
# No:
#
#     float_format=
#
# No:
#
#     round()
#

df.to_csv(

    CSV_FILE,

    index=False,

    na_rep=""

)


print(
    "\nCorrected CSV created:"
)


print(
    CSV_FILE
)


# =================================================================================================
# 28. RE-READ USING ROUND-TRIP NUMERIC PARSER
# =================================================================================================

roundtrip_df = pd.read_csv(

    CSV_FILE,

    low_memory=False,

    float_precision="round_trip"

)


if roundtrip_df.shape != (

    EXPECTED_ROWS,

    len(EXPECTED_COLUMNS)

):

    raise ValueError(
        "Final CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    !=
    EXPECTED_COLUMNS
):

    raise ValueError(
        "Final CSV column order changed."
    )


print(
    "PASS: Final CSV structure preserved."
)


# =================================================================================================
# 29. XPT -> CSV NUMERIC ROUND-TRIP FIDELITY
# =================================================================================================

roundtrip_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (

        expected_df[col]

        .to_numpy(
            dtype=float
        )

    )


    final_values = (

        roundtrip_df[col]

        .to_numpy(
            dtype=float
        )

    )


    same = np.allclose(

        expected_values,

        final_values,

        rtol=0,

        atol=0,

        equal_nan=True

    )


    if not same:

        roundtrip_failures.append(
            col
        )


if roundtrip_failures:

    raise ValueError(
        "XPT -> CSV numeric fidelity failures:\n"
        + str(
            roundtrip_failures
        )
    )


print(
    "PASS: All 90 fields round-trip numerically exactly."
)


# =================================================================================================
# 30. READ FINAL CSV AS TEXT
# =================================================================================================

raw_csv = pd.read_csv(

    CSV_FILE,

    dtype=str,

    keep_default_na=False

)


# =================================================================================================
# 31. CHECK UNNECESSARY ".0"
# =================================================================================================

dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    bad_mask = (

        raw_csv[col]
        .ne("")

        &

        raw_csv[col]
        .str.endswith(".0")

    )


    count = int(
        bad_mask.sum()
    )


    if count > 0:

        dot_zero_errors[
            col
        ] = count


if dot_zero_errors:

    raise ValueError(
        "Unnecessary '.0' formatting remains:\n"
        + str(
            dot_zero_errors
        )
    )


print(
    "PASS: No unnecessary '.0' formatting."
)


# =================================================================================================
# 32. FINAL SEQN TEXT CHECK
# =================================================================================================

if (
    raw_csv["SEQN"]
    .eq("")
    .any()
):

    raise ValueError(
        "Final CSV contains blank SEQN."
    )


if (
    raw_csv["SEQN"]
    .duplicated()
    .any()
):

    raise ValueError(
        "Final CSV contains duplicate SEQN."
    )


if (
    raw_csv["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "Final CSV SEQN contains '.0'."
    )


if (

    ~raw_csv["SEQN"]
    .str.fullmatch(r"\d+")

).any():

    raise ValueError(
        "Final CSV contains invalid SEQN formatting."
    )


final_seqn = pd.to_numeric(

    raw_csv["SEQN"],

    errors="raise"

)


if final_seqn.min() != 83732:

    raise ValueError(
        "Final minimum SEQN differs."
    )


if final_seqn.max() != 93702:

    raise ValueError(
        "Final maximum SEQN differs."
    )


print(
    "PASS: Final SEQN clean."
)


# =================================================================================================
# 33. FINAL TINY-XPORT CHECK
# =================================================================================================

final_tiny_total = 0


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(

        raw_csv[col]
        .where(
            raw_csv[col] != "",
            np.nan
        ),

        errors="coerce"

    )


    final_tiny_total += int(

        values
        .eq(TINY_VALUE)
        .sum()

    )


if final_tiny_total != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain in final CSV."
    )


print(
    "PASS: No tiny XPORT artifacts in final CSV."
)


# =================================================================================================
# 34. FINAL ZERO PATTERN
# =================================================================================================

final_zero_counts = {}


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(

        raw_csv[col]
        .where(
            raw_csv[col] != "",
            np.nan
        ),

        errors="coerce"

    )


    count = int(

        values
        .eq(0)
        .sum()

    )


    if count > 0:

        final_zero_counts[
            col
        ] = count


print(
    "\n--- FINAL ZERO PATTERN ---"
)


print(
    final_zero_counts
)


if final_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final zero pattern differs.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{final_zero_counts}"
    )


print(
    "PASS: Exactly two legitimate zeros remain."
)


# =================================================================================================
# 35. SCIENTIFIC NOTATION CHECK
# =================================================================================================

scientific_notation_fields = {}


for col in EXPECTED_COLUMNS:

    count = int(

        raw_csv[col]

        .str.contains(
            r"[eE][+-]\d+",
            regex=True
        )

        .sum()

    )


    if count > 0:

        scientific_notation_fields[
            col
        ] = count


if scientific_notation_fields:

    raise ValueError(
        "Unexpected scientific notation remains:\n"
        + str(
            scientific_notation_fields
        )
    )


print(
    "PASS: No scientific notation."
)


# =================================================================================================
# 36. FINAL MISSING-POSITION FIDELITY
# =================================================================================================

final_missing_errors = 0


for col in EXPECTED_COLUMNS:

    source_missing = (

        source_df[col]

        .isna()

        .to_numpy()

    )


    csv_missing = (

        raw_csv[col]

        .eq("")

        .to_numpy()

    )


    final_missing_errors += int(

        (
            source_missing
            !=
            csv_missing
        )

        .sum()

    )


if final_missing_errors != 0:

    raise ValueError(
        "Final CSV missing positions differ from source XPT."
    )


print(
    "PASS: Missing positions preserved through CSV export."
)


# =================================================================================================
# 37. FINAL MCQ220 RECHECK
# =================================================================================================

final_mcq220 = (

    raw_csv["MCQ220"]

    .value_counts(
        dropna=False
    )

    .to_dict()

)


EXPECTED_FINAL_MCQ220 = {

    "1": 549,

    "2": 5165,

    "9": 5,

    "": 3856

}


if final_mcq220 != EXPECTED_FINAL_MCQ220:

    raise ValueError(
        "Final MCQ220 distribution differs.\n\n"
        f"Expected:\n{EXPECTED_FINAL_MCQ220}\n\n"
        f"Actual:\n{final_mcq220}"
    )


print(
    "PASS: Final MCQ220:"
)


print(
    "  Yes        = 549"
)


print(
    "  No         = 5,165"
)


print(
    "  Don't know = 5"
)


print(
    "  Missing    = 3,856"
)


# =================================================================================================
# 38. FINAL OSQ230 RECHECK
# =================================================================================================

final_osq230 = (

    raw_csv["OSQ230"]

    .value_counts(
        dropna=False
    )

    .to_dict()

)


EXPECTED_FINAL_OSQ230 = {

    "1": 940,

    "2": 2796,

    "9": 28,

    "": 5811

}


if final_osq230 != EXPECTED_FINAL_OSQ230:

    raise ValueError(
        "Final OSQ230 distribution differs.\n\n"
        f"Expected:\n{EXPECTED_FINAL_OSQ230}\n\n"
        f"Actual:\n{final_osq230}"
    )


print(
    "PASS: Final OSQ230 matches CDC."
)


# =================================================================================================
# 39. FINAL SUMMARY
# =================================================================================================

print(
    "\n"
    + "=" * 110
)


print(
    "NHANES 2015-2016 MCQ_I CONVERSION COMPLETE"
)


print(
    "=" * 110
)


print(
    "Component:"
)


print(
    "  Questionnaire"
)


print(
    ""
)


print(
    "Section:"
)


print(
    "  Medical Conditions"
)


print(
    ""
)


print(
    "Cycle:"
)


print(
    "  2015-2016"
)


print(
    ""
)


print(
    "Rows:",
    f"{len(raw_csv):,}"
)


print(
    "Columns:",
    len(raw_csv.columns)
)


print(
    "Unique SEQN:",
    f"{raw_csv['SEQN'].nunique():,}"
)


print(
    "SEQN range:"
)


print(
    "  83732-93702"
)


print(
    ""
)


print(
    "Validated XPORT zero restorations:"
)


print(
    "  MCQ180B = 1"
)


print(
    "  MCQ180K = 1"
)


print(
    "  TOTAL   = 2"
)


print(
    ""
)


print(
    "Genuine decimal variables:"
)


print(
    "  NONE"
)


print(
    ""
)


print(
    "MCQ220:"
)


print(
    "  Yes        = 549"
)


print(
    "  No         = 5,165"
)


print(
    "  Don't know = 5"
)


print(
    "  Missing    = 3,856"
)


print(
    ""
)


print(
    "OSQ230:"
)


print(
    "  Yes        = 940"
)


print(
    "  No         = 2,796"
)


print(
    "  Don't know = 28"
)


print(
    "  Missing    = 5,811"
)


print(
    ""
)


print(
    "SEQN '.0': 0"
)


print(
    "Duplicate SEQN: 0"
)


print(
    "Remaining tiny XPORT artifacts: 0"
)


print(
    "Unexpected source-value changes: 0"
)


print(
    "Missing-position changes: 0"
)


print(
    "Scientific notation: 0"
)


print(
    "Numeric round-trip failures: 0"
)


print(
    "=" * 110
)
```
