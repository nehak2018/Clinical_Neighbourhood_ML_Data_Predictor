```python
import pandas as pd
import numpy as np
import csv
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2009_2010"
    / "Dietary"
)

xpt_file = DATA_DIR / "DR1TOT_F.xpt"
csv_file = DATA_DIR / "DR1TOT_F_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Keep untouched source copy.
#
source_df = df.copy()


print("=" * 75)
print("NHANES 2009-2010 DR1TOT_F XPT -> CSV")
print("=" * 75)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "WTDRD1",
    "WTDR2D",
    "DR1DRSTZ",
    "DR1EXMER",
    "DRABF",
    "DRDINT",
    "DR1DBIH",
    "DR1DAY",
    "DR1LANG",
    "DR1MNRSP",
    "DR1HELPD",
    "DBQ095Z",
    "DBD100",
    "DRQSPREP",
    "DRQSDIET",
    "DRQSDT1",
    "DRQSDT2",
    "DRQSDT3",
    "DRQSDT4",
    "DRQSDT5",
    "DRQSDT6",
    "DRQSDT7",
    "DRQSDT8",
    "DRQSDT9",
    "DRQSDT10",
    "DRQSDT11",
    "DRQSDT12",
    "DRQSDT91",
    "DR1TNUMF",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TCARB",
    "DR1TSUGR",
    "DR1TFIBE",
    "DR1TTFAT",
    "DR1TSFAT",
    "DR1TMFAT",
    "DR1TPFAT",
    "DR1TCHOL",
    "DR1TATOC",
    "DR1TATOA",
    "DR1TRET",
    "DR1TVARA",
    "DR1TACAR",
    "DR1TBCAR",
    "DR1TCRYP",
    "DR1TLYCO",
    "DR1TLZ",
    "DR1TVB1",
    "DR1TVB2",
    "DR1TNIAC",
    "DR1TVB6",
    "DR1TFOLA",
    "DR1TFA",
    "DR1TFF",
    "DR1TFDFE",
    "DR1TCHL",
    "DR1TVB12",
    "DR1TB12A",
    "DR1TVC",
    "DR1TVD",
    "DR1TVK",
    "DR1TCALC",
    "DR1TPHOS",
    "DR1TMAGN",
    "DR1TIRON",
    "DR1TZINC",
    "DR1TCOPP",
    "DR1TSODI",
    "DR1TPOTA",
    "DR1TSELE",
    "DR1TCAFF",
    "DR1TTHEO",
    "DR1TALCO",
    "DR1TMOIS",
    "DR1TS040",
    "DR1TS060",
    "DR1TS080",
    "DR1TS100",
    "DR1TS120",
    "DR1TS140",
    "DR1TS160",
    "DR1TS180",
    "DR1TM161",
    "DR1TM181",
    "DR1TM201",
    "DR1TM221",
    "DR1TP182",
    "DR1TP183",
    "DR1TP184",
    "DR1TP204",
    "DR1TP205",
    "DR1TP225",
    "DR1TP226",
    "DR1_300",
    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ",
    "DR1TWS",
    "DRD340",
    "DRD350A",
    "DRD350AQ",
    "DRD350B",
    "DRD350BQ",
    "DRD350C",
    "DRD350CQ",
    "DRD350D",
    "DRD350DQ",
    "DRD350E",
    "DRD350EQ",
    "DRD350F",
    "DRD350FQ",
    "DRD350G",
    "DRD350GQ",
    "DRD350H",
    "DRD350HQ",
    "DRD350I",
    "DRD350IQ",
    "DRD350J",
    "DRD350JQ",
    "DRD350K",
    "DRD360",
    "DRD370A",
    "DRD370AQ",
    "DRD370B",
    "DRD370BQ",
    "DRD370C",
    "DRD370CQ",
    "DRD370D",
    "DRD370DQ",
    "DRD370E",
    "DRD370EQ",
    "DRD370F",
    "DRD370FQ",
    "DRD370G",
    "DRD370GQ",
    "DRD370H",
    "DRD370HQ",
    "DRD370I",
    "DRD370IQ",
    "DRD370J",
    "DRD370JQ",
    "DRD370K",
    "DRD370KQ",
    "DRD370L",
    "DRD370LQ",
    "DRD370M",
    "DRD370MQ",
    "DRD370N",
    "DRD370NQ",
    "DRD370O",
    "DRD370OQ",
    "DRD370P",
    "DRD370PQ",
    "DRD370Q",
    "DRD370QQ",
    "DRD370R",
    "DRD370RQ",
    "DRD370S",
    "DRD370SQ",
    "DRD370T",
    "DRD370TQ",
    "DRD370U",
    "DRD370UQ",
    "DRD370V"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "DR1TOT_F column names/order differ from expected.\n"
        f"Actual: {df.columns.tolist()}"
    )


if len(df) != 10253:

    raise ValueError(
        f"Unexpected row count: {len(df)}"
    )


if len(df.columns) != 166:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 10,253 rows x 166 columns."
)


# ============================================================
# 4. CHECK FOR KNOWN XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known SAS XPORT / pandas zero decoding artifact.
#
# DO NOT globally replace until the exact pattern
# has been validated.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


numeric_cols = df.select_dtypes(
    include=[np.number]
).columns


tiny_counts = (
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


true_zero_total = int(
    df[numeric_cols]
    .eq(0)
    .sum()
    .sum()
)


print("\n--- ORIGINAL XPT ZERO CHECK ---")


print(
    "True numeric zeros:",
    f"{true_zero_total:,}"
)


print(
    "Tiny artifact occurrences:",
    f"{tiny_total:,}"
)


if tiny_total > 0:

    print(
        "\nTiny values by variable:"
    )

    print(
        tiny_counts[
            tiny_counts > 0
        ]
        .sort_values(
            ascending=False
        )
        .to_string()
    )


# ============================================================
# 5. VALIDATE EXACT TINY-VALUE PATTERN
# ============================================================

#
# This pattern was verified against the attached
# DR1TOT_F.xpt.
#
# These exact tiny values represent numeric zero.
#
# The script STOPS if the XPT pattern differs.
#

EXPECTED_TINY_COUNTS = {

    "DR1TALCO": 8106,
    "DR1TATOA": 7864,
    "DR1TB12A": 6166,
    "DR1BWATZ": 6134,
    "DR1_330Z": 5746,
    "DR1TTHEO": 4738,
    "DR1TP184": 4083,
    "DR1TM221": 3282,
    "DR1_320Z": 2648,
    "DR1TLYCO": 2392,
    "DR1TCAFF": 2268,
    "DR1TP205": 1279,
    "DR1TP226": 1276,
    "DR1TACAR": 1221,
    "DR1TCRYP": 1127,
    "DR1TP225": 988,
    "DR1TS060": 501,
    "DR1TS040": 413,
    "DR1DBIH": 378,
    "DR1TP204": 233,
    "DR1TS080": 188,
    "DR1TFA": 164,
    "DR1TBCAR": 137,
    "DR1TVD": 118,
    "DR1TLZ": 116,
    "DR1TM201": 100,
    "DR1TFIBE": 96,
    "DR1TFF": 88,
    "DR1TRET": 78,
    "DR1TCHOL": 57,
    "DR1TS100": 52,
    "DR1TVARA": 25,
    "DR1TS120": 23,
    "DR1TVC": 22,
    "DR1TM161": 18,
    "DR1TVB12": 12,
    "DR1TVK": 1
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: Tiny-value pattern differs from "
        "validated DR1TOT_F pattern.\n"
        f"Expected: {EXPECTED_TINY_COUNTS}\n"
        f"Actual:   {actual_tiny_counts}"
    )


if tiny_total != 62138:

    raise ValueError(
        f"Unexpected total tiny-value count: {tiny_total}"
    )


print(
    "\nPASS: Tiny-value pattern matches "
    "validated DR1TOT_F pattern."
)


# ============================================================
# 6. RESTORE VALIDATED ZERO VALUES
# ============================================================

#
# IMPORTANT:
#
# Restore ONLY:
#
#     exact TINY_VALUE
#
# in the exact validated columns above.
#
# Legitimate non-zero dietary decimal values are
# NOT modified.
#

print("\n--- ZERO RESTORATION ---")


correction_total = 0


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    tiny_mask = (
        df[col]
        .eq(TINY_VALUE)
    )


    actual_count = int(
        tiny_mask.sum()
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: unexpected tiny-value count."
        )


    df.loc[
        tiny_mask,
        col
    ] = 0


    correction_total += actual_count


    print(
        f"{col}: "
        f"{actual_count:,} restored to 0"
    )


if correction_total != 62138:

    raise ValueError(
        "Total zero restoration count is incorrect."
    )


print(
    "\nPASS:",
    f"{correction_total:,}",
    "validated zero values restored."
)


# ============================================================
# 7. CREATE EXPECTED CORRECTED SOURCE
# ============================================================

#
# expected_df differs from source_df ONLY where
# validated tiny-zero artifacts were restored.
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col].eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 8. IDENTIFY GENUINE DIETARY DECIMAL VARIABLES
# ============================================================

#
# CRITICAL:
#
# Dietary decimal values MUST remain unchanged.
#
# No rounding.
# No conversion to Int64.
# No decimal truncation.
#

print("\n--- GENUINE DECIMAL CHECK ---")


fractional_columns = []


for col in numeric_cols:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        continue


    fractional_mask = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    if fractional_mask.any():

        fractional_columns.append(
            col
        )


# ============================================================
# 9. EXPECTED GENUINE DECIMAL VARIABLES
# ============================================================

#
# These 50 variables contain genuine fractional
# values in the attached DR1TOT_F.
#
# They MUST remain floating point.
#

DECIMAL_COLUMNS = [

    "WTDRD1",
    "WTDR2D",

    "DR1TPROT",
    "DR1TCARB",
    "DR1TSUGR",
    "DR1TFIBE",
    "DR1TTFAT",
    "DR1TSFAT",
    "DR1TMFAT",
    "DR1TPFAT",

    "DR1TATOC",
    "DR1TATOA",

    "DR1TVB1",
    "DR1TVB2",
    "DR1TNIAC",
    "DR1TVB6",

    "DR1TCHL",
    "DR1TVB12",
    "DR1TB12A",
    "DR1TVC",
    "DR1TVD",
    "DR1TVK",

    "DR1TIRON",
    "DR1TZINC",
    "DR1TCOPP",
    "DR1TSELE",

    "DR1TALCO",
    "DR1TMOIS",

    "DR1TS040",
    "DR1TS060",
    "DR1TS080",
    "DR1TS100",
    "DR1TS120",
    "DR1TS140",
    "DR1TS160",
    "DR1TS180",

    "DR1TM161",
    "DR1TM181",
    "DR1TM201",
    "DR1TM221",

    "DR1TP182",
    "DR1TP183",
    "DR1TP184",
    "DR1TP204",
    "DR1TP205",
    "DR1TP225",
    "DR1TP226",

    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ"
]


if fractional_columns != DECIMAL_COLUMNS:

    raise ValueError(
        "STOP: Genuine decimal-variable pattern differs "
        "from validated DR1TOT_F.\n"
        f"Expected:\n{DECIMAL_COLUMNS}\n\n"
        f"Actual:\n{fractional_columns}"
    )


print(
    "PASS: Genuine decimal variables:",
    len(DECIMAL_COLUMNS)
)


print(
    "\nThese variables will NOT be rounded "
    "or converted to integers."
)


# ============================================================
# 10. CLEAN AND VALIDATE SEQN
# ============================================================

seqn = (
    df["SEQN"]
    .dropna()
)


if not np.all(
    np.isclose(
        seqn,
        np.round(seqn),
        rtol=0,
        atol=1e-12
    )
):

    raise ValueError(
        "SEQN contains unexpected fractional values."
    )


df["SEQN"] = (
    df["SEQN"]
    .astype("Int64")
)


print("\n--- SEQN ---")


print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)


print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


print(
    "Minimum:",
    df["SEQN"].min()
)


print(
    "Maximum:",
    df["SEQN"].max()
)


if (
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
    != 0
    or
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
    != 0
    or
    df["SEQN"].min()
    != 51624
    or
    df["SEQN"].max()
    != 62160
):

    raise ValueError(
        "SEQN validation failed."
    )


print(
    "PASS: SEQN validated."
)


# ============================================================
# 11. FORMAT ONLY WHOLE-NUMBER VARIABLES
# ============================================================

#
# IMPORTANT:
#
# This is different from questionnaire files.
#
# DO NOT convert DECIMAL_COLUMNS to Int64.
#
# Only columns proven to contain whole-number
# values are converted.
#
# NO .round() is used.
#

INTEGER_COLUMNS = [
    col
    for col in df.columns
    if col not in DECIMAL_COLUMNS
]


for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) > 0:

        if not np.all(
            np.isclose(
                values,
                np.round(values),
                rtol=0,
                atol=1e-12
            )
        ):

            raise ValueError(
                f"{col} contains genuine decimal values "
                "and cannot safely be converted to Int64."
            )


    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: Whole-number fields formatted "
    "without unnecessary .0."
)


print(
    "PASS: Dietary decimal fields remain unchanged."
)


# ============================================================
# 12. CDC CHECK - DR1DRSTZ
# ============================================================

#
# CDC:
#
# 1 = Reliable                      9,623
# 2 = Not reliable                     63
# 4 = Breast milk                     131
# 5 = Not done                        436
#

EXPECTED_DR1DRSTZ = {
    1: 9623,
    2: 63,
    4: 131,
    5: 436
}


actual = (
    df["DR1DRSTZ"]
    .value_counts()
    .sort_index()
    .to_dict()
)


print("\n--- DR1DRSTZ ---")


if actual != EXPECTED_DR1DRSTZ:

    raise ValueError(
        "DR1DRSTZ differs from CDC.\n"
        f"Expected: {EXPECTED_DR1DRSTZ}\n"
        f"Actual: {actual}"
    )


print(
    "PASS: DR1DRSTZ matches CDC."
)


# ============================================================
# 13. CDC CHECK - DR1DBIH
# ============================================================

#
# CDC:
#
# -36 to 76 = 9,775
# Missing   =   478
#
# The attached XPT has 378 tiny values representing zero.
#

s = df["DR1DBIH"]


print("\n--- DR1DBIH ---")


print(
    "-36 to 76:",
    int(
        s.between(
            -36,
            76
        ).sum()
    )
)


print(
    "Zero:",
    int(
        s.eq(0).sum()
    )
)


print(
    "Missing:",
    int(
        s.isna().sum()
    )
)


if not (
    int(
        s.between(
            -36,
            76
        ).sum()
    )
    == 9775
    and
    int(
        s.eq(0).sum()
    )
    == 378
    and
    int(
        s.isna().sum()
    )
    == 478
):

    raise ValueError(
        "DR1DBIH differs from CDC / "
        "validated zero restoration."
    )


print(
    "PASS: DR1DBIH matches CDC."
)


# ============================================================
# 14. CDC CHECK - DR1TNUMF
# ============================================================

#
# CDC:
#
# 1-52 foods = 9,754
# Missing    =   499
#

s = df["DR1TNUMF"]


if not (
    int(
        s.between(
            1,
            52
        ).sum()
    )
    == 9754
    and
    int(
        s.isna().sum()
    )
    == 499
):

    raise ValueError(
        "DR1TNUMF differs from CDC."
    )


print(
    "PASS: DR1TNUMF matches CDC."
)


# ============================================================
# 15. CDC CHECK - DR1TALCO
# ============================================================

#
# IMPORTANT:
#
# DR1TALCO is a REAL DECIMAL VARIABLE.
#
# CDC:
#
# 0 to 335.3 gm = 9,623
# Missing       =   630
#
# The 8,106 tiny values represent dietary alcohol = 0.
#

s = df["DR1TALCO"]


print("\n--- DR1TALCO ---")


print(
    "Zero:",
    int(
        s.eq(0).sum()
    )
)


print(
    "0-335.3:",
    int(
        s.between(
            0,
            335.3
        ).sum()
    )
)


print(
    "Missing:",
    int(
        s.isna().sum()
    )
)


if not (
    int(
        s.eq(0).sum()
    )
    == 8106
    and
    int(
        s.between(
            0,
            335.3
        ).sum()
    )
    == 9623
    and
    int(
        s.isna().sum()
    )
    == 630
):

    raise ValueError(
        "DR1TALCO differs from CDC / "
        "validated zero restoration."
    )


print(
    "PASS: DR1TALCO matches CDC."
)


print(
    "PASS: Non-zero DR1TALCO decimals "
    "were not changed."
)


# ============================================================
# 16. CDC CHECK - WATER VARIABLES
# ============================================================

#
# DR1_320Z:
# Total plain water drank yesterday
#
# CDC range:
# 0 to 16116
# 9,754 records
# 499 missing
#

s = df["DR1_320Z"]


if not (
    int(
        s.between(
            0,
            16116
        ).sum()
    )
    == 9754
    and
    int(
        s.isna().sum()
    )
    == 499
):

    raise ValueError(
        "DR1_320Z differs from CDC."
    )


print(
    "PASS: DR1_320Z matches CDC."
)


#
# DR1_330Z:
# Total tap water drank yesterday
#
# CDC range:
# 0 to 16116
#

s = df["DR1_330Z"]


if not (
    int(
        s.between(
            0,
            16116
        ).sum()
    )
    == 9754
    and
    int(
        s.isna().sum()
    )
    == 499
):

    raise ValueError(
        "DR1_330Z differs from CDC."
    )


print(
    "PASS: DR1_330Z matches CDC."
)


#
# DR1BWATZ:
# Total bottled water drank yesterday
#
# CDC range:
# 0 to 9537.75
#

s = df["DR1BWATZ"]


if not (
    int(
        s.between(
            0,
            9537.75
        ).sum()
    )
    == 9754
    and
    int(
        s.isna().sum()
    )
    == 499
):

    raise ValueError(
        "DR1BWATZ differs from CDC."
    )


print(
    "PASS: DR1BWATZ matches CDC."
)


# ============================================================
# 17. VALUE FIDELITY BEFORE EXPORT
# ============================================================

#
# Expected changes:
#
# exact TINY_VALUE -> 0
#
# in the validated 37 columns ONLY.
#
# NOTHING ELSE may change.
#
# Decimal dietary values must remain numerically exact.
#

print("\n--- VALUE FIDELITY BEFORE EXPORT ---")


failures = []


for col in expected_df.columns:

    expected_values = (
        expected_df[col]
        .to_numpy()
    )


    if col in DECIMAL_COLUMNS:

        current_values = (
            df[col]
            .to_numpy(
                dtype=float
            )
        )

    else:

        current_values = (
            df[col]
            .astype("Float64")
            .to_numpy(
                dtype=float,
                na_value=np.nan
            )
        )


    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        failures.append(
            col
        )


if failures:

    raise ValueError(
        "Unexpected value changes before export:\n"
        + str(failures)
    )


print(
    "PASS: All 166 variables match expected "
    "corrected XPT values exactly."
)


# ============================================================
# 18. EXPORT CSV
# ============================================================

#
# CRITICAL:
#
# NO float_format
# NO rounding
#
# Decimal dietary variables remain floats.
#
# Missing values become blank.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print("\nCSV created:")

print(
    csv_file
)


# ============================================================
# 19. RE-READ EXPORTED CSV
# ============================================================

#
# round_trip is used so floating-point values
# can be checked exactly after CSV export.
#

csv_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


# ============================================================
# 20. EXACT CORRECTED XPT -> CSV FIDELITY
# ============================================================

print("\n--- XPT -> CSV FIDELITY ---")


csv_failures = []


for col in expected_df.columns:

    expected_values = (
        expected_df[col]
        .to_numpy()
    )


    exported_values = (
        csv_df[col]
        .to_numpy()
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "Corrected XPT -> CSV differences found in:\n"
        + str(csv_failures)
    )


print(
    "PASS: All 166 CSV columns match "
    "expected corrected XPT values."
)


# ============================================================
# 21. MISSING VALUE FIDELITY
# ============================================================

missing_mismatches = 0


for col in expected_df.columns:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )


    exported_missing = (
        csv_df[col]
        .isna()
        .to_numpy()
    )


    missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            exported_missing
        )
    )


print("\n--- MISSING VALUE CHECK ---")


print(
    "Missing-position mismatches:",
    missing_mismatches
)


if missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed."
    )


print(
    "PASS: Missing values preserved exactly."
)


# ============================================================
# 22. TINY-VALUE CHECK AFTER EXPORT
# ============================================================

tiny_csv_count = 0


for col in csv_df.select_dtypes(
    include=[np.number]
).columns:

    tiny_csv_count += int(
        csv_df[col]
        .eq(TINY_VALUE)
        .sum()
    )


print("\n--- FINAL TINY VALUE CHECK ---")


print(
    "Tiny values remaining:",
    tiny_csv_count
)


if tiny_csv_count != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain in CSV."
    )


print(
    "PASS: No tiny-value artifacts remain."
)


# ============================================================
# 23. WHOLE-NUMBER .0 FORMAT CHECK
# ============================================================

#
# IMPORTANT:
#
# DO NOT search the entire Dietary CSV for ".0".
#
# Decimal variables are allowed to contain values
# such as:
#
#     12.0
#     0.0
#     335.3
#
# We check ONLY columns that are supposed to be
# whole-number variables.
#

print("\n--- WHOLE-NUMBER FORMAT CHECK ---")


dot_zero_errors = []


with open(
    csv_file,
    "r",
    encoding="utf-8",
    newline=""
) as f:

    reader = csv.DictReader(
        f
    )


    for row_number, row in enumerate(
        reader,
        start=2
    ):

        for col in INTEGER_COLUMNS:

            value = row[col]


            if (
                value != ""
                and
                value.endswith(".0")
            ):

                dot_zero_errors.append(
                    (
                        row_number,
                        col,
                        value
                    )
                )


                if len(
                    dot_zero_errors
                ) >= 20:

                    break


        if len(
            dot_zero_errors
        ) >= 20:

            break


print(
    "Unnecessary '.0' errors:",
    len(
        dot_zero_errors
    )
)


if dot_zero_errors:

    print(
        dot_zero_errors
    )

    raise ValueError(
        "Unexpected .0 remains in "
        "whole-number columns."
    )


print(
    "PASS: SEQN and whole-number fields "
    "have no unnecessary .0."
)


# ============================================================
# 24. EXACT DIETARY DECIMAL FIDELITY
# ============================================================

#
# Explicitly verify every genuine decimal column.
#
# These must survive:
#
# XPT -> correction -> CSV -> pandas
#
# with no numeric change whatsoever.
#

print("\n--- DIETARY DECIMAL FIDELITY ---")


decimal_failures = []


for col in DECIMAL_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy()
    )


    exported_values = (
        csv_df[col]
        .to_numpy()
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        decimal_failures.append(
            col
        )


if decimal_failures:

    raise ValueError(
        "Dietary decimal values changed in:\n"
        + str(decimal_failures)
    )


print(
    "PASS: All",
    len(DECIMAL_COLUMNS),
    "genuine decimal variables preserved exactly."
)


# ============================================================
# 25. SCIENTIFIC NOTATION CHECK
# ============================================================

#
# After tiny-zero correction there should be no
# scientific notation in this CSV.
#

with open(
    csv_file,
    "r",
    encoding="utf-8"
) as f:

    csv_text = f.read()


scientific_count = (
    csv_text.lower().count("e-")
    +
    csv_text.lower().count("e+")
)


print("\n--- SCIENTIFIC NOTATION CHECK ---")


print(
    "Scientific notation occurrences:",
    scientific_count
)


if scientific_count != 0:

    raise ValueError(
        "Unexpected scientific notation found."
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 26. FINAL CSV SEQN CHECK
# ============================================================

print("\n--- FINAL CSV SEQN CHECK ---")


if (
    csv_df["SEQN"].min()
    != 51624
    or
    csv_df["SEQN"].max()
    != 62160
    or
    int(
        csv_df["SEQN"]
        .isna()
        .sum()
    )
    != 0
    or
    int(
        csv_df["SEQN"]
        .duplicated()
        .sum()
    )
    != 0
):

    raise ValueError(
        "Final CSV SEQN validation failed."
    )


print(
    "PASS: Final CSV SEQN is valid."
)


# ============================================================
# 27. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 75
)


print(
    "DR1TOT_F CONVERSION COMPLETE"
)


print(
    "=" * 75
)


print(
    "Rows:",
    f"{len(df):,}"
)


print(
    "Columns:",
    len(df.columns)
)


print(
    "Original tiny-value artifacts:",
    f"{tiny_total:,}"
)


print(
    "Validated zero restorations:",
    f"{correction_total:,}"
)


print(
    "Genuine decimal variables preserved:",
    len(DECIMAL_COLUMNS)
)


print(
    "XPT -> CSV unexpected differences: 0 expected"
)


print(
    "Missing-value differences: 0 expected"
)


print(
    "Tiny values remaining: 0 expected"
)


print(
    "Unnecessary .0 in whole-number fields: 0 expected"
)


print(
    "Dietary decimal values changed: 0 expected"
)
```
