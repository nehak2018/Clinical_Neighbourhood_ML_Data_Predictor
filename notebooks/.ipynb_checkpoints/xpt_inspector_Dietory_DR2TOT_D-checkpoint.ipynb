```python
import pandas as pd
import numpy as np
import csv
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2005_2006"
    / "Dietary"
)

xpt_file = DATA_DIR / "DR2TOT_D.xpt"
csv_file = DATA_DIR / "DR2TOT_D_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

# Keep untouched source for final fidelity checking
source_df = df.copy()


print("=" * 80)
print("NHANES 2005-2006 DR2TOT_D XPT -> CSV")
print("=" * 80)

print("Rows:", f"{len(df):,}")
print("Columns:", len(df.columns))


# ============================================================
# 3. STRUCTURE CHECK
# ============================================================

if len(df) != 9950:
    raise ValueError(
        f"Unexpected row count: {len(df)}"
    )

if len(df.columns) != 81:
    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )

if df.columns.duplicated().any():
    raise ValueError(
        "Duplicate column names found."
    )


REQUIRED_COLUMNS = [
    "SEQN",
    "WTDRD1",
    "WTDR2D",
    "DR2DRSTZ",
    "DR2EXMER",
    "DRABF",
    "DRDINT",
    "DR2DAY",
    "DR2LANG",
    "DR2MNRSP",
    "DR2HELPD",
    "DR2TNUMF",
    "DR2TKCAL",
    "DR2TPROT",
    "DR2TP204",
    "DR2TP226",
    "DR2_300",
    "DR2_320Z",
    "DR2_330Z",
    "DR2BWATZ",
    "DR2TWS"
]


missing_columns = [
    col
    for col in REQUIRED_COLUMNS
    if col not in df.columns
]


if missing_columns:
    raise ValueError(
        f"Required columns missing: {missing_columns}"
    )


print(
    "PASS: DR2TOT_D structure = "
    "9,950 rows x 81 columns."
)


# ============================================================
# 4. DEFINE DIETARY MEASUREMENT FIELDS
# ============================================================

#
# These MUST remain exactly as read from XPT.
#
# NO rounding.
# NO recalculation.
# NO decimal trimming.
#

nutrient_start = df.columns.get_loc(
    "DR2TKCAL"
)

nutrient_end = df.columns.get_loc(
    "DR2TP226"
)


NUTRIENT_COLUMNS = (
    df.columns[
        nutrient_start:
        nutrient_end + 1
    ]
    .tolist()
)


MEASUREMENT_COLUMNS = [
    "WTDRD1",
    "WTDR2D",
] + NUTRIENT_COLUMNS + [
    "DR2_320Z",
    "DR2_330Z",
    "DR2BWATZ"
]


print(
    "Nutrient fields:",
    len(NUTRIENT_COLUMNS)
)

print(
    "Total measurement/weight fields:",
    len(MEASUREMENT_COLUMNS)
)


# ============================================================
# 5. KNOWN PANDAS/XPORT ZERO ARTIFACT
# ============================================================

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


numeric_cols = df.select_dtypes(
    include=[np.number]
).columns


tiny_counts = (
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
)


true_zero_count = int(
    df[numeric_cols]
    .eq(0)
    .sum()
    .sum()
)


print("\n--- ORIGINAL XPT ZERO CHECK ---")

print(
    "True numeric zeros:",
    f"{true_zero_count:,}"
)

print(
    "Tiny artifact occurrences:",
    f"{int(tiny_counts.sum()):,}"
)

print(
    "Variables containing tiny values:",
    int((tiny_counts > 0).sum())
)


print("\nTiny values by variable:")

print(
    tiny_counts[
        tiny_counts > 0
    ]
    .sort_values(
        ascending=False
    )
    .to_string()
)


# ============================================================
# 6. EXACT TINY COUNTS FOUND IN THIS XPT
# ============================================================

#
# These are the ONLY variables where tiny -> 0
# is allowed for this source.
#

EXPECTED_TINY_COUNTS = {

    "DR2TKCAL": 2,
    "DR2TPROT": 3,
    "DR2TCARB": 3,
    "DR2TSUGR": 3,
    "DR2TFIBE": 87,

    "DR2TTFAT": 2,
    "DR2TSFAT": 2,
    "DR2TMFAT": 2,
    "DR2TPFAT": 2,

    "DR2TCHOL": 42,
    "DR2TATOC": 3,
    "DR2TATOA": 7551,

    "DR2TRET": 66,
    "DR2TVARA": 16,
    "DR2TACAR": 892,
    "DR2TBCAR": 96,
    "DR2TCRYP": 811,
    "DR2TLYCO": 2370,
    "DR2TLZ": 104,

    "DR2TVB1": 2,
    "DR2TVB2": 2,
    "DR2TNIAC": 2,
    "DR2TVB6": 2,

    "DR2TFOLA": 3,
    "DR2TFA": 162,
    "DR2TFF": 71,
    "DR2TFDFE": 3,

    "DR2TCHL": 3,
    "DR2TVB12": 22,
    "DR2TB12A": 5133,

    "DR2TVC": 31,
    "DR2TVK": 7,

    "DR2TPHOS": 2,
    "DR2TIRON": 2,
    "DR2TPOTA": 2,
    "DR2TSELE": 5,

    "DR2TCAFF": 2521,
    "DR2TTHEO": 4478,
    "DR2TALCO": 7408,

    "DR2TS040": 515,
    "DR2TS060": 432,
    "DR2TS080": 276,
    "DR2TS100": 114,
    "DR2TS120": 30,
    "DR2TS140": 6,
    "DR2TS160": 2,
    "DR2TS180": 6,

    "DR2TM161": 27,
    "DR2TM181": 4,
    "DR2TM201": 105,
    "DR2TM221": 3148,

    "DR2TP182": 2,
    "DR2TP183": 5,
    "DR2TP184": 5754,
    "DR2TP204": 343,
    "DR2TP205": 2662,
    "DR2TP225": 3680,
    "DR2TP226": 2094,

    "DR2_320Z": 2183,
    "DR2_330Z": 4441,
    "DR2BWATZ": 5631
}


ZERO_ALLOWED_COLUMNS = list(
    EXPECTED_TINY_COUNTS.keys()
)


# ============================================================
# 7. SAFETY CHECK
# ============================================================

actual_tiny_columns = set(
    tiny_counts[
        tiny_counts > 0
    ].index
)


approved_tiny_columns = set(
    ZERO_ALLOWED_COLUMNS
)


unexpected = (
    actual_tiny_columns
    -
    approved_tiny_columns
)


if unexpected:

    raise ValueError(
        "STOP: Tiny values found in unexpected "
        "DR2TOT_D fields:\n"
        + str(sorted(unexpected))
    )


missing_expected = (
    approved_tiny_columns
    -
    actual_tiny_columns
)


if missing_expected:

    raise ValueError(
        "STOP: Expected tiny-value fields missing:\n"
        + str(sorted(missing_expected))
    )


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    actual_count = int(
        tiny_counts[col]
    )

    if actual_count != expected_count:

        raise ValueError(
            f"{col}: expected {expected_count} "
            f"tiny values, found {actual_count}."
        )


if int(tiny_counts.sum()) != 63378:

    raise ValueError(
        "Unexpected total tiny-value count."
    )


print(
    "\nPASS: All 63,378 tiny values occur only "
    "in the 61 validated dietary measurement fields."
)


# ============================================================
# 8. RESTORE VALIDATED ZEROS
# ============================================================

print("\n--- ZERO RESTORATION ---")


for col in ZERO_ALLOWED_COLUMNS:

    mask = (
        df[col]
        .eq(TINY_VALUE)
    )

    count = int(
        mask.sum()
    )


    df.loc[
        mask,
        col
    ] = 0.0


    print(
        f"{col:<12}: "
        f"{count:,} tiny -> 0"
    )


# ============================================================
# 9. VERIFY ZERO COUNTS
# ============================================================

print("\n--- RESTORED ZERO CHECK ---")


zero_failures = []


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    actual_zero = int(
        df[col]
        .eq(0)
        .sum()
    )


    if actual_zero != expected_count:

        zero_failures.append(
            (
                col,
                expected_count,
                actual_zero
            )
        )


if zero_failures:

    raise ValueError(
        "Zero restoration differences:\n"
        + str(zero_failures)
    )


print(
    "PASS: All 63,378 validated zeros restored."
)


# ============================================================
# 10. DEFINE CODE / COUNT FIELDS
# ============================================================

#
# Anything numeric that is NOT:
#
# SEQN
# or measurement field
#
# should contain only whole-number codes/counts.
#

CODE_COLUMNS = [
    col
    for col in numeric_cols
    if col not in (
        ["SEQN"]
        + MEASUREMENT_COLUMNS
    )
]


print(
    "\nWhole-number code/count fields:",
    len(CODE_COLUMNS)
)


# ============================================================
# 11. VERIFY CODE FIELDS HAVE NO GENUINE DECIMALS
# ============================================================

print("\n--- CODE FIELD FRACTION CHECK ---")


fraction_failures = []


for col in ["SEQN"] + CODE_COLUMNS:

    values = (
        df[col]
        .dropna()
    )


    if values.empty:
        continue


    fractional = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    if fractional.any():

        fraction_failures.append(
            col
        )

        print(
            f"\nUnexpected decimal values in {col}:"
        )

        print(
            values[
                fractional
            ].head(20)
        )


if fraction_failures:

    raise ValueError(
        "STOP: Genuine decimals detected in "
        "identifier/code/count fields:\n"
        + str(fraction_failures)
    )


print(
    "PASS: SEQN and all code/count fields "
    "contain whole-number values only."
)


# ============================================================
# 12. IDENTIFY GENUINE DIETARY FRACTIONS
# ============================================================

print("\n--- DIETARY DECIMAL CHECK ---")


fractional_measurements = []


for col in MEASUREMENT_COLUMNS:

    values = (
        df[col]
        .dropna()
    )


    if values.empty:
        continue


    fractional = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    if fractional.any():

        fractional_measurements.append(
            col
        )


print(
    "Measurement fields containing genuine fractions:",
    len(fractional_measurements)
)

print(
    fractional_measurements
)


if len(fractional_measurements) != 49:

    raise ValueError(
        "Unexpected number of fractional "
        "measurement fields."
    )


print(
    "PASS: Genuine dietary decimal fields identified. "
    "They will remain untouched."
)


# ============================================================
# 13. CLEAN SEQN
# ============================================================

seqn = (
    df["SEQN"]
    .dropna()
)


if not np.all(
    np.isclose(
        seqn,
        np.round(seqn),
        rtol=0,
        atol=1e-12
    )
):

    raise ValueError(
        "SEQN contains unexpected fractions."
    )


df["SEQN"] = (
    df["SEQN"]
    .astype("Int64")
)


print("\n--- SEQN ---")

print(
    "Missing:",
    int(df["SEQN"].isna().sum())
)

print(
    "Duplicates:",
    int(df["SEQN"].duplicated().sum())
)

print(
    "Minimum:",
    df["SEQN"].min()
)

print(
    "Maximum:",
    df["SEQN"].max()
)


if (
    int(df["SEQN"].isna().sum()) != 0
    or
    int(df["SEQN"].duplicated().sum()) != 0
    or
    df["SEQN"].min() != 31127
    or
    df["SEQN"].max() != 41474
):

    raise ValueError(
        "SEQN validation failed."
    )


print(
    "PASS: SEQN validated."
)


# ============================================================
# 14. FORMAT ONLY CODE / COUNT FIELDS
# ============================================================

#
# NO round()
#
# Dietary measurements are NOT converted.
#

for col in CODE_COLUMNS:

    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: Whole-number code/count fields formatted "
    "without unnecessary .0."
)


# ============================================================
# 15. CDC CHECK - RECALL STATUS
# ============================================================

EXPECTED_DR2DRSTZ = {
    1: 8264,
    2: 87,
    4: 165,
    5: 1434
}


actual = (
    df["DR2DRSTZ"]
    .value_counts()
    .sort_index()
    .to_dict()
)


if actual != EXPECTED_DR2DRSTZ:

    raise ValueError(
        "DR2DRSTZ differs from CDC.\n"
        f"Actual: {actual}"
    )


print(
    "\nPASS: DR2DRSTZ matches CDC."
)


# ============================================================
# 16. CDC CHECK - NUMBER OF FOODS
# ============================================================

s = df["DR2TNUMF"]


if not (
    int(
        s.between(
            1,
            49
        ).sum()
    )
    ==
    8429

    and

    int(
        s.isna()
        .sum()
    )
    ==
    1521

    and

    int(
        s.eq(0)
        .sum()
    )
    ==
    0
):

    raise ValueError(
        "DR2TNUMF differs from CDC."
    )


print(
    "PASS: DR2TNUMF = 1-49, "
    "8,429 valid, 1,521 missing."
)


# ============================================================
# 17. SAMPLE WEIGHT CHECK
# ============================================================

if not (
    int(
        df["WTDRD1"]
        .notna()
        .sum()
    )
    ==
    9349

    and

    int(
        df["WTDRD1"]
        .isna()
        .sum()
    )
    ==
    601
):

    raise ValueError(
        "WTDRD1 completeness differs from CDC."
    )


if not (
    int(
        df["WTDR2D"]
        .notna()
        .sum()
    )
    ==
    8429

    and

    int(
        df["WTDR2D"]
        .isna()
        .sum()
    )
    ==
    1521
):

    raise ValueError(
        "WTDR2D completeness differs from CDC."
    )


print(
    "PASS: Dietary sample-weight completeness validated."
)


# ============================================================
# 18. NUTRIENT COMPLETENESS CHECK
# ============================================================

#
# Day 2 reliable recalls = 8,264
#
# Therefore every nutrient total:
#
# Valid   = 8,264
# Missing = 1,686
#

nutrient_failures = []


for col in NUTRIENT_COLUMNS:

    valid = int(
        df[col]
        .notna()
        .sum()
    )

    missing = int(
        df[col]
        .isna()
        .sum()
    )


    if (
        valid != 8264
        or
        missing != 1686
    ):

        nutrient_failures.append(
            (
                col,
                valid,
                missing
            )
        )


if nutrient_failures:

    raise ValueError(
        "Nutrient completeness differs:\n"
        + str(nutrient_failures)
    )


print(
    "PASS: All 64 nutrient fields have "
    "8,264 valid and 1,686 missing."
)


# ============================================================
# 19. WATER TOTAL CHECKS
# ============================================================

WATER_CHECKS = {

    "DR2_320Z": {
        "min": 0,
        "max": 11376,
        "valid": 8429,
        "missing": 1521,
        "zeros": 2183
    },

    "DR2_330Z": {
        "min": 0,
        "max": 11376,
        "valid": 8429,
        "missing": 1521,
        "zeros": 4441
    },

    "DR2BWATZ": {
        "min": 0,
        "max": 8217.77,
        "valid": 8429,
        "missing": 1521,
        "zeros": 5631
    }
}


print("\n--- WATER TOTAL CHECKS ---")


for col, spec in WATER_CHECKS.items():

    s = df[col]


    if not (
        int(s.notna().sum())
        ==
        spec["valid"]

        and

        int(s.isna().sum())
        ==
        spec["missing"]

        and

        np.isclose(
            s.min(),
            spec["min"]
        )

        and

        np.isclose(
            s.max(),
            spec["max"]
        )

        and

        int(
            s.eq(0)
            .sum()
        )
        ==
        spec["zeros"]
    ):

        raise ValueError(
            f"{col} differs from CDC/source."
        )


    print(
        f"PASS: {col} "
        f"range={s.min()}-{s.max()} "
        f"zeros={int(s.eq(0).sum()):,}"
    )


# ============================================================
# 20. CDC CHECK - DR2_300
# ============================================================

EXPECTED_DR2_300 = {
    1: 517,
    2: 6938,
    3: 1031,
    7: 3,
    9: 27
}


actual = (
    df["DR2_300"]
    .value_counts()
    .sort_index()
    .to_dict()
)


if not (
    actual == EXPECTED_DR2_300
    and
    int(
        df["DR2_300"]
        .isna()
        .sum()
    )
    ==
    1434
):

    raise ValueError(
        "DR2_300 differs from CDC."
    )


print(
    "PASS: DR2_300 matches CDC."
)


# ============================================================
# 21. BUILD EXPECTED SOURCE
# ============================================================

#
# Expected source =
#
# original XPT
# +
# ONLY the 61 validated tiny -> 0 restorations.
#

expected_df = source_df.copy()


for col in ZERO_ALLOWED_COLUMNS:

    expected_df.loc[
        expected_df[col]
        .eq(TINY_VALUE),
        col
    ] = 0.0


# ============================================================
# 22. VALUE FIDELITY BEFORE EXPORT
# ============================================================

print("\n--- VALUE FIDELITY BEFORE EXPORT ---")


failures = []


for col in expected_df.columns:

    expected_values = (
        expected_df[col]
        .to_numpy()
    )


    if (
        col == "SEQN"
        or
        col in CODE_COLUMNS
    ):

        current_values = (
            df[col]
            .astype("Float64")
            .to_numpy(
                dtype=float,
                na_value=np.nan
            )
        )

    else:

        current_values = (
            df[col]
            .to_numpy()
        )


    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        failures.append(
            col
        )


if failures:

    raise ValueError(
        "Unexpected value changes in:\n"
        + str(failures)
    )


print(
    "PASS: All 81 variables preserved exactly "
    "after validated zero restoration."
)


# ============================================================
# 23. FINAL TINY CHECK
# ============================================================

remaining_tiny = int(
    df.select_dtypes(
        include=[np.number]
    )
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print(
    "\nRemaining tiny values:",
    remaining_tiny
)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny-value artifacts remain."
    )


print(
    "PASS: No tiny-value artifact remains."
)


# ============================================================
# 24. EXPORT CSV
# ============================================================

#
# CRITICAL:
#
# NO float_format
# NO round()
# NO recalculation
# NO decimal trimming
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print("\nCSV created:")
print(csv_file)


# ============================================================
# 25. RE-READ CSV
# ============================================================

csv_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


# ============================================================
# 26. EXACT XPT -> CSV NUMERIC FIDELITY
# ============================================================

print("\n--- XPT -> CSV NUMERIC FIDELITY ---")


csv_failures = []


for col in expected_df.columns:

    expected_values = (
        expected_df[col]
        .to_numpy()
    )

    exported_values = (
        csv_df[col]
        .to_numpy()
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "XPT -> CSV differences found in:\n"
        + str(csv_failures)
    )


print(
    "PASS: All 81 CSV columns numerically "
    "match expected XPT values."
)


# ============================================================
# 27. MISSING VALUE FIDELITY
# ============================================================

missing_mismatches = 0


for col in expected_df.columns:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )

    exported_missing = (
        csv_df[col]
        .isna()
        .to_numpy()
    )


    missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            exported_missing
        )
    )


print(
    "\nMissing-position mismatches:",
    missing_mismatches
)


if missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed."
    )


print(
    "PASS: Missing values preserved exactly."
)


# ============================================================
# 28. RAW CSV TOKEN CHECK
# ============================================================

with open(
    csv_file,
    "r",
    encoding="utf-8",
    newline=""
) as f:

    reader = csv.DictReader(f)
    raw_rows = list(reader)


# ============================================================
# 29. SEQN / CODE .0 CHECK
# ============================================================

bad_code_dot_zero = 0


for row in raw_rows:

    for col in ["SEQN"] + CODE_COLUMNS:

        value = row[col]

        if value.endswith(".0"):

            bad_code_dot_zero += 1


print(
    "\nSEQN/code fields ending '.0':",
    bad_code_dot_zero
)


if bad_code_dot_zero != 0:

    raise ValueError(
        "Unnecessary .0 remains in "
        "identifier/code/count fields."
    )


print(
    "PASS: SEQN/code/count fields have "
    "no unnecessary .0."
)


# ============================================================
# 30. RAW DIETARY MEASUREMENT FIDELITY
# ============================================================

#
# Compare CSV text directly against:
#
# str(float(expected XPT value))
#
# This intentionally preserves values such as:
#
# 0.029
# 0.028999999999999998
# 15.0
# 0.0
#
# exactly as pandas decoded the XPT.
#

measurement_mismatches = {
    col: 0
    for col in MEASUREMENT_COLUMNS
}


for i, row in enumerate(raw_rows):

    for col in MEASUREMENT_COLUMNS:

        xpt_value = (
            expected_df.iloc[i][col]
        )


        expected_text = (
            ""
            if pd.isna(xpt_value)
            else str(float(xpt_value))
        )


        if row[col] != expected_text:

            measurement_mismatches[col] += 1


bad_measurements = {
    col: count
    for col, count
    in measurement_mismatches.items()
    if count > 0
}


print(
    "Dietary measurement raw-text mismatches:",
    bad_measurements
)


if bad_measurements:

    raise ValueError(
        "Dietary measurement formatting changed."
    )


print(
    "PASS: All dietary measurements and weights "
    "appear AS IS in CSV."
)


# ============================================================
# 31. SCIENTIFIC NOTATION CHECK
# ============================================================

with open(
    csv_file,
    "r",
    encoding="utf-8"
) as f:

    csv_text = f.read()


scientific_count = (
    csv_text.lower().count("e-")
    +
    csv_text.lower().count("e+")
)


print(
    "Scientific notation occurrences:",
    scientific_count
)


if scientific_count != 0:

    raise ValueError(
        "Unexpected scientific notation found."
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 32. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("DR2TOT_D CONVERSION COMPLETE")
print("=" * 80)

print("Rows:", f"{len(df):,}")
print("Columns:", len(df.columns))
print("Validated tiny -> zero:", "63,378")
print("Tiny-value variables:", 61)
print("Nutrient fields:", len(NUTRIENT_COLUMNS))
print("Measurement fields:", len(MEASUREMENT_COLUMNS))

print(
    "Measurement fields containing real fractions:",
    len(fractional_measurements)
)

print("XPT -> CSV differences: 0 expected")
print("Missing differences: 0 expected")
print("SEQN/code .0: 0 expected")
```
