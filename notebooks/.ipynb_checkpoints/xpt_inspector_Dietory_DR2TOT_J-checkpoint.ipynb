```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2017_2018"
    / "Dietary"
)

xpt_file = DATA_DIR / "DR2TOT_J.xpt"
csv_file = DATA_DIR / "DR2TOT_J_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Preserve untouched pandas-decoded source.
#
source_df = df.copy()


print("=" * 100)
print("NHANES 2017-2018 DR2TOT_J XPT -> CSV")
print("=" * 100)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "WTDRD1",
    "WTDR2D",
    "DR2DRSTZ",
    "DR2EXMER",
    "DRABF",
    "DRDINT",
    "DR2DBIH",
    "DR2DAY",
    "DR2LANG",
    "DR2MRESP",
    "DR2HELP",
    "DR2TNUMF",
    "DR2STY",
    "DR2SKY",
    "DR2TKCAL",
    "DR2TPROT",
    "DR2TCARB",
    "DR2TSUGR",
    "DR2TFIBE",
    "DR2TTFAT",
    "DR2TSFAT",
    "DR2TMFAT",
    "DR2TPFAT",
    "DR2TCHOL",
    "DR2TATOC",
    "DR2TATOA",
    "DR2TRET",
    "DR2TVARA",
    "DR2TACAR",
    "DR2TBCAR",
    "DR2TCRYP",
    "DR2TLYCO",
    "DR2TLZ",
    "DR2TVB1",
    "DR2TVB2",
    "DR2TNIAC",
    "DR2TVB6",
    "DR2TFOLA",
    "DR2TFA",
    "DR2TFF",
    "DR2TFDFE",
    "DR2TCHL",
    "DR2TVB12",
    "DR2TB12A",
    "DR2TVC",
    "DR2TVD",
    "DR2TVK",
    "DR2TCALC",
    "DR2TPHOS",
    "DR2TMAGN",
    "DR2TIRON",
    "DR2TZINC",
    "DR2TCOPP",
    "DR2TSODI",
    "DR2TPOTA",
    "DR2TSELE",
    "DR2TCAFF",
    "DR2TTHEO",
    "DR2TALCO",
    "DR2TMOIS",
    "DR2TS040",
    "DR2TS060",
    "DR2TS080",
    "DR2TS100",
    "DR2TS120",
    "DR2TS140",
    "DR2TS160",
    "DR2TS180",
    "DR2TM161",
    "DR2TM181",
    "DR2TM201",
    "DR2TM221",
    "DR2TP182",
    "DR2TP183",
    "DR2TP184",
    "DR2TP204",
    "DR2TP205",
    "DR2TP225",
    "DR2TP226",
    "DR2_300",
    "DR2_320Z",
    "DR2_330Z",
    "DR2BWATZ",
    "DR2TWSZ"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "DR2TOT_J columns/order differ from expected.\n\n"
        f"Expected:\n{EXPECTED_COLUMNS}\n\n"
        f"Actual:\n{df.columns.tolist()}"
    )


if len(df) != 8704:

    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 85:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 8,704 rows x 85 columns."
)


# ============================================================
# 4. DIETARY FIELD PRESERVATION PLAN
# ============================================================

#
# CRITICAL DIETARY RULE
# ------------------------------------------------------------
#
# Measurement variables are preserved as measurement
# variables even if a particular variable happens to have
# whole-number observations in this cycle.
#
# We preserve:
#
#     dietary weights
#     nutrient totals
#     fatty acids
#     water measurements
#
# as float/numeric measurement fields.
#
# We do NOT round them.
#
# We do NOT convert them to Int64.
#

PRESERVE_AS_FLOAT_COLUMNS = [

    # Survey weights
    "WTDRD1",
    "WTDR2D",

    # Daily nutrient totals
    "DR2TKCAL",
    "DR2TPROT",
    "DR2TCARB",
    "DR2TSUGR",
    "DR2TFIBE",
    "DR2TTFAT",
    "DR2TSFAT",
    "DR2TMFAT",
    "DR2TPFAT",
    "DR2TCHOL",
    "DR2TATOC",
    "DR2TATOA",
    "DR2TRET",
    "DR2TVARA",
    "DR2TACAR",
    "DR2TBCAR",
    "DR2TCRYP",
    "DR2TLYCO",
    "DR2TLZ",
    "DR2TVB1",
    "DR2TVB2",
    "DR2TNIAC",
    "DR2TVB6",
    "DR2TFOLA",
    "DR2TFA",
    "DR2TFF",
    "DR2TFDFE",
    "DR2TCHL",
    "DR2TVB12",
    "DR2TB12A",
    "DR2TVC",
    "DR2TVD",
    "DR2TVK",
    "DR2TCALC",
    "DR2TPHOS",
    "DR2TMAGN",
    "DR2TIRON",
    "DR2TZINC",
    "DR2TCOPP",
    "DR2TSODI",
    "DR2TPOTA",
    "DR2TSELE",
    "DR2TCAFF",
    "DR2TTHEO",
    "DR2TALCO",
    "DR2TMOIS",

    # Fatty acids
    "DR2TS040",
    "DR2TS060",
    "DR2TS080",
    "DR2TS100",
    "DR2TS120",
    "DR2TS140",
    "DR2TS160",
    "DR2TS180",
    "DR2TM161",
    "DR2TM181",
    "DR2TM201",
    "DR2TM221",
    "DR2TP182",
    "DR2TP183",
    "DR2TP184",
    "DR2TP204",
    "DR2TP205",
    "DR2TP225",
    "DR2TP226",

    # Water measurements
    "DR2_320Z",
    "DR2_330Z",
    "DR2BWATZ"
]


INTEGER_COLUMNS = [
    col
    for col in EXPECTED_COLUMNS
    if col not in PRESERVE_AS_FLOAT_COLUMNS
]


if len(PRESERVE_AS_FLOAT_COLUMNS) != 70:

    raise ValueError(
        "Expected exactly 70 Dietary measurement fields."
    )


if len(INTEGER_COLUMNS) != 15:

    raise ValueError(
        "Expected exactly 15 ID/code/count fields."
    )


print("\n--- FIELD PRESERVATION PLAN ---")

print(
    "Dietary measurement fields preserved:",
    len(PRESERVE_AS_FLOAT_COLUMNS)
)

print(
    "Integer/code/count fields:",
    INTEGER_COLUMNS
)


# ============================================================
# 5. VERIFY ALL SOURCE VARIABLES ARE NUMERIC
# ============================================================

numeric_columns = (
    df.select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)


if numeric_columns != EXPECTED_COLUMNS:

    raise ValueError(
        "Expected all 85 DR2TOT_J variables "
        "to load numerically."
    )


print(
    "PASS: All 85 variables loaded numerically."
)


# ============================================================
# 6. KNOWN XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known SAS XPORT / pandas decoded zero:
#
#     5.397605346934028e-79
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


tiny_counts = (
    df[EXPECTED_COLUMNS]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


ordinary_zero_counts = (
    df[EXPECTED_COLUMNS]
    .eq(0)
    .sum()
)


ordinary_zero_total = int(
    ordinary_zero_counts.sum()
)


print(
    "\n--- ORIGINAL XPT ZERO CHECK ---"
)

print(
    "Ordinary numeric zeros before repair:",
    f"{ordinary_zero_total:,}"
)

print(
    "Tiny XPORT values:",
    f"{tiny_total:,}"
)


print(
    "\nTiny values by variable:"
)

print(
    tiny_counts[
        tiny_counts > 0
    ]
    .to_string()
)


# ============================================================
# 7. EXACT VALIDATED TINY-ZERO PATTERN
# ============================================================

#
# Directly inspected from attached DR2TOT_J.xpt.
#
# Total:
#
#     44,105
#
# across:
#
#     66 variables
#
# CDC confirms zeros are valid for these fields.
#

EXPECTED_TINY_COUNTS = {

    "WTDRD1": 1063,
    "WTDR2D": 1002,

    "DR2DBIH": 16,

    "DR2TKCAL": 2,
    "DR2TPROT": 3,
    "DR2TCARB": 2,
    "DR2TSUGR": 3,
    "DR2TFIBE": 74,
    "DR2TTFAT": 3,
    "DR2TSFAT": 7,
    "DR2TMFAT": 6,
    "DR2TPFAT": 5,
    "DR2TCHOL": 94,
    "DR2TATOC": 7,
    "DR2TATOA": 5543,
    "DR2TRET": 64,
    "DR2TVARA": 19,
    "DR2TACAR": 947,
    "DR2TBCAR": 98,
    "DR2TCRYP": 712,
    "DR2TLYCO": 1984,
    "DR2TLZ": 78,
    "DR2TVB1": 3,
    "DR2TVB2": 3,
    "DR2TNIAC": 4,
    "DR2TVB6": 5,
    "DR2TFOLA": 6,
    "DR2TFA": 285,
    "DR2TFF": 61,
    "DR2TFDFE": 6,
    "DR2TCHL": 5,
    "DR2TVB12": 33,
    "DR2TB12A": 4471,
    "DR2TVC": 48,
    "DR2TVD": 214,
    "DR2TVK": 11,

    "DR2TPHOS": 2,
    "DR2TIRON": 4,
    "DR2TZINC": 1,
    "DR2TPOTA": 2,
    "DR2TSELE": 4,

    "DR2TCAFF": 1934,
    "DR2TTHEO": 3600,
    "DR2TALCO": 5878,

    "DR2TS040": 254,
    "DR2TS060": 282,
    "DR2TS080": 93,
    "DR2TS100": 37,
    "DR2TS120": 29,
    "DR2TS140": 11,
    "DR2TS160": 7,
    "DR2TS180": 9,

    "DR2TM161": 16,
    "DR2TM181": 6,
    "DR2TM201": 25,
    "DR2TM221": 1006,

    "DR2TP182": 5,
    "DR2TP183": 7,
    "DR2TP184": 3108,
    "DR2TP204": 40,
    "DR2TP205": 729,
    "DR2TP225": 478,
    "DR2TP226": 1076,

    "DR2_320Z": 1199,
    "DR2_330Z": 3792,
    "DR2BWATZ": 3584
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: DR2TOT_J tiny-value pattern differs "
        "from validated attachment.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{actual_tiny_counts}"
    )


if tiny_total != 44105:

    raise ValueError(
        f"Unexpected tiny-value total: "
        f"{tiny_total:,}"
    )


if len(actual_tiny_counts) != 66:

    raise ValueError(
        "Unexpected number of variables "
        "containing tiny artifacts."
    )


if ordinary_zero_total != 0:

    raise ValueError(
        "Unexpected ordinary numeric zeros existed "
        "before XPORT correction."
    )


print(
    "PASS: Exact 44,105-value tiny-zero "
    "pattern validated."
)


# ============================================================
# 8. RESTORE ONLY VALIDATED ZEROS
# ============================================================

#
# DO NOT use:
#
#     df.replace(TINY_VALUE, 0)
#
# globally.
#

correction_counts = {}


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    mask = (
        df[col]
        .eq(TINY_VALUE)
    )


    actual_count = int(
        mask.sum()
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: tiny count differs.\n"
            f"Expected: {expected_count:,}\n"
            f"Actual:   {actual_count:,}"
        )


    df.loc[
        mask,
        col
    ] = 0


    correction_counts[col] = actual_count


total_corrections = sum(
    correction_counts.values()
)


print(
    "\n--- VALIDATED ZERO RESTORATION ---"
)

print(
    "Variables corrected:",
    len(correction_counts)
)

print(
    "Total zero restorations:",
    f"{total_corrections:,}"
)


if total_corrections != 44105:

    raise ValueError(
        "Total zero-restoration count incorrect."
    )


print(
    "PASS: 44,105 validated zeros restored."
)


# ============================================================
# 9. BUILD EXPECTED CORRECTED SOURCE
# ============================================================

#
# Only tiny -> zero corrections are permitted.
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col]
        .eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 10. CONFIRM NO TINY VALUES REMAIN
# ============================================================

remaining_tiny = int(
    df[EXPECTED_COLUMNS]
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print(
    "\nTiny artifacts remaining:",
    remaining_tiny
)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain after repair."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 11. VALIDATE SEQN
# ============================================================

seqn_values = (
    df["SEQN"]
    .dropna()
    .to_numpy(
        dtype=float
    )
)


if not np.isclose(
    seqn_values,
    np.round(seqn_values),
    rtol=0,
    atol=1e-12
).all():

    raise ValueError(
        "SEQN contains unexpected fractional values."
    )


print(
    "\n--- SEQN ---"
)

print(
    "Minimum:",
    df["SEQN"].min()
)

print(
    "Maximum:",
    df["SEQN"].max()
)

print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)

print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if df["SEQN"].min() != 93703:

    raise ValueError(
        "Unexpected minimum SEQN."
    )


if df["SEQN"].max() != 102956:

    raise ValueError(
        "Unexpected maximum SEQN."
    )


if int(
    df["SEQN"]
    .isna()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains missing values."
    )


if int(
    df["SEQN"]
    .duplicated()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains duplicate values."
    )


print(
    "PASS: SEQN 93703-102956 validated."
)


# ============================================================
# 12. CDC DIETARY WEIGHT CHECK
# ============================================================

#
# WTDRD1:
#
#     positive = 7641
#     zero     = 1063
#     missing  = 0
#
#
# WTDR2D:
#
#     positive = 6639
#     zero     = 1002
#     missing  = 1063
#

print(
    "\n--- CDC WEIGHT CHECK ---"
)


w1_positive_count = int(
    df["WTDRD1"]
    .gt(0)
    .sum()
)

w1_zero_count = int(
    df["WTDRD1"]
    .eq(0)
    .sum()
)

w1_missing_count = int(
    df["WTDRD1"]
    .isna()
    .sum()
)


print(
    "WTDRD1 positive:",
    w1_positive_count
)

print(
    "WTDRD1 zero:",
    w1_zero_count
)

print(
    "WTDRD1 missing:",
    w1_missing_count
)


if w1_positive_count != 7641:

    raise ValueError(
        "WTDRD1 positive count differs."
    )


if w1_zero_count != 1063:

    raise ValueError(
        "WTDRD1 zero count differs."
    )


if w1_missing_count != 0:

    raise ValueError(
        "WTDRD1 missing count differs."
    )


w2_positive_count = int(
    df["WTDR2D"]
    .gt(0)
    .sum()
)

w2_zero_count = int(
    df["WTDR2D"]
    .eq(0)
    .sum()
)

w2_missing_count = int(
    df["WTDR2D"]
    .isna()
    .sum()
)


print(
    "WTDR2D positive:",
    w2_positive_count
)

print(
    "WTDR2D zero:",
    w2_zero_count
)

print(
    "WTDR2D missing:",
    w2_missing_count
)


if w2_positive_count != 6639:

    raise ValueError(
        "WTDR2D positive count differs."
    )


if w2_zero_count != 1002:

    raise ValueError(
        "WTDR2D zero count differs."
    )


if w2_missing_count != 1063:

    raise ValueError(
        "WTDR2D missing count differs."
    )


print(
    "PASS: Dietary sample weights match CDC."
)


# ============================================================
# 13. EXACT ATTACHMENT WEIGHT RANGE CHECK
# ============================================================

w1_positive = (
    df.loc[
        df["WTDRD1"] > 0,
        "WTDRD1"
    ]
)


w2_positive = (
    df.loc[
        df["WTDR2D"] > 0,
        "WTDR2D"
    ]
)


w1_min = float(
    w1_positive.min()
)

w1_max = float(
    w1_positive.max()
)

w2_min = float(
    w2_positive.min()
)

w2_max = float(
    w2_positive.max()
)


print(
    "\n--- WEIGHT RANGE CHECK ---"
)

print(
    "WTDRD1:",
    repr(w1_min),
    "to",
    repr(w1_max)
)

print(
    "WTDR2D:",
    repr(w2_min),
    "to",
    repr(w2_max)
)


if not np.isclose(
    w1_min,
    1478.156428,
    rtol=0,
    atol=0
):

    raise ValueError(
        "Unexpected WTDRD1 minimum."
    )


if not np.isclose(
    w1_max,
    691139.546883,
    rtol=0,
    atol=0
):

    raise ValueError(
        "Unexpected WTDRD1 maximum."
    )


if not np.isclose(
    w2_min,
    1232.691132,
    rtol=0,
    atol=0
):

    raise ValueError(
        "Unexpected WTDR2D minimum."
    )


if not np.isclose(
    w2_max,
    901995.118861,
    rtol=0,
    atol=0
):

    raise ValueError(
        "Unexpected WTDR2D maximum."
    )


print(
    "PASS: Exact attachment weight ranges validated."
)


# ============================================================
# 14. DAY 2 RECALL STATUS
# ============================================================

EXPECTED_DR2DRSTZ = {
    1: 6502,
    2: 75,
    4: 137,
    5: 1990
}


actual_status = (
    df["DR2DRSTZ"]
    .value_counts()
    .sort_index()
    .to_dict()
)


print(
    "\n--- DR2DRSTZ CHECK ---"
)

print(
    actual_status
)


if actual_status != EXPECTED_DR2DRSTZ:

    raise ValueError(
        "DR2DRSTZ distribution differs."
    )


print(
    "PASS: Day 2 recall status matches CDC."
)


# ============================================================
# 15. DR2DBIH CHECK
# ============================================================

#
# CDC:
#
#     range -33 to 137
#     nonmissing 6499
#     missing 2205
#
# Zero is valid.
#

dbih_nonmissing = int(
    df["DR2DBIH"]
    .notna()
    .sum()
)

dbih_missing = int(
    df["DR2DBIH"]
    .isna()
    .sum()
)

dbih_zero = int(
    df["DR2DBIH"]
    .eq(0)
    .sum()
)


print(
    "\n--- DR2DBIH CHECK ---"
)

print(
    "Nonmissing:",
    dbih_nonmissing
)

print(
    "Missing:",
    dbih_missing
)

print(
    "Zero:",
    dbih_zero
)

print(
    "Range:",
    df["DR2DBIH"].min(),
    "to",
    df["DR2DBIH"].max()
)


if dbih_nonmissing != 6499:

    raise ValueError(
        "DR2DBIH nonmissing count differs."
    )


if dbih_missing != 2205:

    raise ValueError(
        "DR2DBIH missing count differs."
    )


if df["DR2DBIH"].min() != -33:

    raise ValueError(
        "Unexpected DR2DBIH minimum."
    )


if df["DR2DBIH"].max() != 137:

    raise ValueError(
        "Unexpected DR2DBIH maximum."
    )


if dbih_zero != 16:

    raise ValueError(
        "Unexpected DR2DBIH zero count."
    )


print(
    "PASS: DR2DBIH validated."
)


# ============================================================
# 16. DR2TNUMF CHECK
# ============================================================

#
# CDC:
#
#     range 1-39
#     nonmissing 6639
#     missing 2065
#

tnumf_nonmissing = int(
    df["DR2TNUMF"]
    .notna()
    .sum()
)

tnumf_missing = int(
    df["DR2TNUMF"]
    .isna()
    .sum()
)


if tnumf_nonmissing != 6639:

    raise ValueError(
        "DR2TNUMF nonmissing count differs."
    )


if tnumf_missing != 2065:

    raise ValueError(
        "DR2TNUMF missing count differs."
    )


if df["DR2TNUMF"].min() != 1:

    raise ValueError(
        "Unexpected DR2TNUMF minimum."
    )


if df["DR2TNUMF"].max() != 39:

    raise ValueError(
        "Unexpected DR2TNUMF maximum."
    )


print(
    "PASS: DR2TNUMF 1-39 validated."
)


# ============================================================
# 17. SELECTED CDC CODE FREQUENCIES
# ============================================================

def check_frequency(
    column,
    expected_values,
    expected_missing
):

    actual_values = (
        df[column]
        .value_counts()
        .sort_index()
        .to_dict()
    )

    actual_missing = int(
        df[column]
        .isna()
        .sum()
    )


    print(
        f"\n--- CHECK: {column} ---"
    )

    print(
        actual_values
    )

    print(
        "Missing:",
        actual_missing
    )


    if actual_values != expected_values:

        raise ValueError(
            f"{column} frequencies differ.\n\n"
            f"Expected:\n{expected_values}\n\n"
            f"Actual:\n{actual_values}"
        )


    if actual_missing != expected_missing:

        raise ValueError(
            f"{column} missing count differs."
        )


check_frequency(
    "DR2STY",
    {
        1: 1045,
        2: 5650,
        9: 19
    },
    1990
)


check_frequency(
    "DR2SKY",
    {
        1: 980,
        2: 24,
        3: 23,
        99: 18
    },
    7659
)


check_frequency(
    "DR2_300",
    {
        1: 417,
        2: 5281,
        3: 986,
        9: 30
    },
    1990
)


check_frequency(
    "DR2TWSZ",
    {
        1: 3989,
        4: 1794,
        91: 625,
        99: 306
    },
    1990
)


print(
    "PASS: Key Day 2 code frequencies validated."
)


# ============================================================
# 18. NUTRIENT COMPLETENESS CHECK
# ============================================================

#
# DR2TKCAL through DR2TP226:
#
#     65 nutrient / food component variables
#
# Each has:
#
#     6502 nonmissing
#     2202 missing
#

NUTRIENT_COLUMNS = EXPECTED_COLUMNS[
    EXPECTED_COLUMNS.index("DR2TKCAL"):
    EXPECTED_COLUMNS.index("DR2TP226") + 1
]


if len(NUTRIENT_COLUMNS) != 65:

    raise ValueError(
        "Unexpected nutrient-column count."
    )


nutrient_completeness_failures = {}


for col in NUTRIENT_COLUMNS:

    nonmissing = int(
        df[col]
        .notna()
        .sum()
    )

    missing = int(
        df[col]
        .isna()
        .sum()
    )


    if (
        nonmissing != 6502
        or
        missing != 2202
    ):

        nutrient_completeness_failures[col] = {
            "nonmissing": nonmissing,
            "missing": missing
        }


if nutrient_completeness_failures:

    raise ValueError(
        "Nutrient completeness differs:\n"
        + str(nutrient_completeness_failures)
    )


print(
    "PASS: All 65 nutrient variables have "
    "6,502 populated / 2,202 missing."
)


# ============================================================
# 19. WATER MEASUREMENT CHECK
# ============================================================

EXPECTED_WATER_RANGES = {

    "DR2_320Z": (
        0,
        11835
    ),

    "DR2_330Z": (
        0,
        11835
    ),

    "DR2BWATZ": (
        0,
        8400
    )
}


print(
    "\n--- WATER MEASUREMENT CHECK ---"
)


for col, (
    expected_min,
    expected_max
) in EXPECTED_WATER_RANGES.items():

    nonmissing = int(
        df[col]
        .notna()
        .sum()
    )

    missing = int(
        df[col]
        .isna()
        .sum()
    )


    print(
        f"{col}:",
        df[col].min(),
        "to",
        df[col].max()
    )


    if nonmissing != 6639:

        raise ValueError(
            f"{col} nonmissing count differs."
        )


    if missing != 2065:

        raise ValueError(
            f"{col} missing count differs."
        )


    if not np.isclose(
        df[col].min(),
        expected_min,
        rtol=0,
        atol=0
    ):

        raise ValueError(
            f"{col} minimum differs."
        )


    if not np.isclose(
        df[col].max(),
        expected_max,
        rtol=0,
        atol=0
    ):

        raise ValueError(
            f"{col} maximum differs."
        )


print(
    "PASS: Water measurements validated."
)


# ============================================================
# 20. FRACTIONAL VALUE DISCOVERY
# ============================================================

#
# DIAGNOSTIC ONLY.
#
# np.round() is only used to test whether a value is
# numerically integral.
#
# No source value is changed.
#

print(
    "\n--- FRACTIONAL VALUE DISCOVERY ---"
)


fractional_counts = {}


for col in EXPECTED_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        fractional_count = 0

    else:

        fractional_count = int(
            (
                ~np.isclose(
                    values,
                    np.round(values),
                    rtol=0,
                    atol=1e-12
                )
            )
            .sum()
        )


    fractional_counts[col] = (
        fractional_count
    )


for col, count in fractional_counts.items():

    if count > 0:

        print(
            f"{col}: {count:,}"
        )


# ============================================================
# 21. PROVE THE 15 INTEGER/CODE/COUNT FIELDS
# ============================================================

integer_fraction_failures = {}


for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        continue


    whole_mask = np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    failure_count = int(
        (~whole_mask)
        .sum()
    )


    if failure_count > 0:

        integer_fraction_failures[col] = (
            failure_count
        )


if integer_fraction_failures:

    raise ValueError(
        "STOP: Proposed integer fields contain "
        "fractional values:\n"
        + str(integer_fraction_failures)
    )


print(
    "PASS: All 15 ID/code/count variables "
    "are proven integral."
)


# ============================================================
# 22. PROTECT DIETARY MEASUREMENT FIELDS
# ============================================================

#
# We keep all 70 measurement fields as their current
# numeric representation.
#
# A field can remain a Dietary measurement even if its
# observations in this cycle happen to be whole-number.
#

print(
    "\n--- DIETARY MEASUREMENT PROTECTION ---"
)

print(
    "Protected measurement fields:",
    len(PRESERVE_AS_FLOAT_COLUMNS)
)


for col in PRESERVE_AS_FLOAT_COLUMNS:

    if col not in df.columns:

        raise ValueError(
            f"Missing protected variable: {col}"
        )


print(
    "PASS: All 70 Dietary measurements protected."
)


# ============================================================
# 23. PRE-CONVERSION EXACT FIDELITY
# ============================================================

pre_conversion_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    actual_values = (
        df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        actual_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        pre_conversion_failures.append(
            col
        )


if pre_conversion_failures:

    raise ValueError(
        "Unexpected numeric modifications before "
        "Int64 conversion:\n"
        + str(pre_conversion_failures)
    )


print(
    "PASS: Corrected dataframe exactly matches "
    "expected corrected source."
)


# ============================================================
# 24. FINAL ZERO PATTERN BEFORE TYPE CONVERSION
# ============================================================

final_zero_counts = {}


for col in EXPECTED_COLUMNS:

    zero_count = int(
        df[col]
        .eq(0)
        .sum()
    )


    if zero_count > 0:

        final_zero_counts[col] = (
            zero_count
        )


if final_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final zero pattern differs from validated "
        "DR2TOT_J pattern.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{final_zero_counts}"
    )


print(
    "PASS: Final zero pattern matches all "
    "44,105 validated restorations."
)


# ============================================================
# 25. CONVERT ONLY PROVEN INTEGER/CODE/COUNT FIELDS
# ============================================================

#
# Only these 15 fields become Int64.
#
# All 70 measurement fields remain untouched.
#

for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) > 0:

        whole_mask = np.isclose(
            values,
            np.round(values),
            rtol=0,
            atol=1e-12
        )


        if not whole_mask.all():

            raise ValueError(
                f"{col} cannot safely become Int64."
            )


    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: 15 ID/code/count fields converted "
    "to nullable Int64."
)

print(
    "PASS: 70 Dietary measurement fields "
    "remain untouched."
)


# ============================================================
# 26. MEASUREMENT FIDELITY BEFORE EXPORT
# ============================================================

measurement_failures = []


for col in PRESERVE_AS_FLOAT_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    actual_values = (
        df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        actual_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        measurement_failures.append(
            col
        )


if measurement_failures:

    raise ValueError(
        "Dietary measurement fields changed:\n"
        + str(measurement_failures)
    )


print(
    "PASS: All 70 Dietary measurement values "
    "remain unchanged."
)


# ============================================================
# 27. COMPLETE PRE-EXPORT NUMERIC FIDELITY
# ============================================================

pre_export_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    actual_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )


    same = np.allclose(
        expected_values,
        actual_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        pre_export_failures.append(
            col
        )


if pre_export_failures:

    raise ValueError(
        "Pre-export numeric fidelity failures:\n"
        + str(pre_export_failures)
    )


print(
    "PASS: All 85 variables match exact "
    "expected corrected values."
)


# ============================================================
# 28. MISSING-POSITION FIDELITY
# ============================================================

missing_position_errors = 0


for col in EXPECTED_COLUMNS:

    source_missing = (
        source_df[col]
        .isna()
        .to_numpy()
    )


    current_missing = (
        df[col]
        .isna()
        .to_numpy()
    )


    missing_position_errors += int(
        (
            source_missing
            !=
            current_missing
        )
        .sum()
    )


print(
    "\nMissing-position differences:",
    missing_position_errors
)


if missing_position_errors != 0:

    raise ValueError(
        "Missing positions changed."
    )


print(
    "PASS: Missing positions preserved exactly."
)


# ============================================================
# 29. EXPORT CSV
# ============================================================

#
# CRITICAL DIETARY FIDELITY POLICY
# ------------------------------------------------------------
#
# DO NOT:
#
#     df.round(...)
#
# DO NOT:
#
#     float_format=
#
# DO NOT:
#
#     force a fixed number of decimals
#
# DO NOT:
#
#     convert Dietary measurement fields to Int64
#
#
# Examples such as:
#
#     0.028999999999...
#     160.67299999999997
#     903.1999999999999
#
# must not be deliberately rewritten.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print(
    "\nCSV created:"
)

print(
    csv_file
)


# ============================================================
# 30. RE-READ CSV USING ROUND-TRIP FLOAT PARSER
# ============================================================

#
# Exact numeric fidelity MUST use:
#
#     float_precision="round_trip"
#
# Raw CSV string parsing should NOT be used for exact
# IEEE-754 float equality.
#

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != (
    8704,
    85
):

    raise ValueError(
        "CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV column order changed."
    )


print(
    "PASS: CSV structure preserved."
)


# ============================================================
# 31. EXACT CORRECTED XPT -> CSV FIDELITY
# ============================================================

print(
    "\n--- EXACT CORRECTED XPT -> CSV FIDELITY ---"
)


roundtrip_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        roundtrip_failures.append(
            col
        )


if roundtrip_failures:

    raise ValueError(
        "Corrected XPT -> CSV fidelity failures:\n"
        + str(roundtrip_failures)
    )


print(
    "PASS: All 85 variables round-trip exactly."
)


# ============================================================
# 32. EXACT DIETARY MEASUREMENT ROUND-TRIP CHECK
# ============================================================

dietary_roundtrip_failures = []


for col in PRESERVE_AS_FLOAT_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    final_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        final_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        dietary_roundtrip_failures.append(
            col
        )


if dietary_roundtrip_failures:

    raise ValueError(
        "Dietary measurement round-trip failures:\n"
        + str(dietary_roundtrip_failures)
    )


print(
    "PASS: All 70 Dietary measurements "
    "round-trip exactly."
)


# ============================================================
# 33. CSV MISSING-POSITION FIDELITY
# ============================================================

csv_missing_errors = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )


    exported_missing = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )


    csv_missing_errors += int(
        (
            expected_missing
            !=
            exported_missing
        )
        .sum()
    )


if csv_missing_errors != 0:

    raise ValueError(
        "CSV missing-value positions changed."
    )


print(
    "PASS: CSV missing positions preserved exactly."
)


# ============================================================
# 34. READ CSV AS RAW TEXT
# ============================================================

#
# Used ONLY for formatting checks.
#
# Not for exact float fidelity.
#

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


if raw_csv_df.shape != (
    8704,
    85
):

    raise ValueError(
        "Raw CSV dimensions changed."
    )


if (
    raw_csv_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "Raw CSV column order changed."
    )


# ============================================================
# 35. INTEGER/CODE/COUNT ".0" CHECK
# ============================================================

#
# ONLY the 15 proven integer/code/count variables.
#
# Measurement variables are intentionally excluded.
#

print(
    "\n--- INTEGER/CODE/COUNT .0 CHECK ---"
)


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = (
        raw_csv_df[col]
    )


    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )


    count = int(
        bad_mask.sum()
    )


    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unnecessary '.0' found in "
        "integer/code/count fields:\n"
        + str(dot_zero_errors)
    )


print(
    "PASS: No unnecessary '.0' in "
    "integer/code/count fields."
)


# ============================================================
# 36. FINAL SEQN TEXT CHECK
# ============================================================

if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "SEQN still contains '.0'."
    )


print(
    "PASS: SEQN contains no '.0'."
)


# ============================================================
# 37. MEASUREMENT FORMAT POLICY
# ============================================================

#
# Do NOT remove ".0" from Dietary measurement fields.
#
# A measurement such as:
#
#     DR2TKCAL = 2000.0
#
# remains a measurement variable even though that one
# observation is whole-valued.
#
# Exact numeric fidelity was already established by
# roundtrip_df.
#

print(
    "\n--- DIETARY MEASUREMENT RAW-TEXT CHECK ---"
)


for col in PRESERVE_AS_FLOAT_COLUMNS:

    populated = int(
        raw_csv_df[col]
        .ne("")
        .sum()
    )


    dot_zero = int(
        (
            raw_csv_df[col]
            .ne("")
            &
            raw_csv_df[col]
            .str.endswith(".0")
        )
        .sum()
    )


    print(
        f"{col}: "
        f"populated={populated:,}, "
        f"tokens ending .0={dot_zero:,}"
    )


print(
    "PASS: Dietary measurements were not "
    "subjected to cosmetic integer formatting."
)


# ============================================================
# 38. FINAL TINY-VALUE CHECK
# ============================================================

#
# pd.to_numeric is safe here because this is simply
# detecting whether the known tiny artifact remains.
#
# It is NOT being used for exact float fidelity.
#

tiny_csv_total = 0


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    tiny_csv_total += int(
        values
        .eq(TINY_VALUE)
        .sum()
    )


print(
    "\n--- FINAL TINY-VALUE CHECK ---"
)

print(
    "Tiny values remaining:",
    tiny_csv_total
)


if tiny_csv_total != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain in CSV."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 39. FINAL ZERO PATTERN
# ============================================================

csv_zero_counts = {}


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    zero_count = int(
        values
        .eq(0)
        .sum()
    )


    if zero_count > 0:

        csv_zero_counts[col] = (
            zero_count
        )


if csv_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final CSV zero pattern differs.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{csv_zero_counts}"
    )


print(
    "PASS: Final zero pattern exactly matches "
    "the 44,105 validated restorations."
)


# ============================================================
# 40. SCIENTIFIC NOTATION CHECK
# ============================================================

scientific_notation_errors = {}


for col in EXPECTED_COLUMNS:

    tokens = raw_csv_df[col]


    count = int(
        tokens
        .str.contains(
            r"[eE][+-]\d+",
            regex=True
        )
        .sum()
    )


    if count > 0:

        scientific_notation_errors[col] = (
            count
        )


if scientific_notation_errors:

    raise ValueError(
        "Unexpected scientific notation found:\n"
        + str(scientific_notation_errors)
    )


print(
    "PASS: No scientific notation in final CSV."
)


# ============================================================
# 41. FINAL SEQN VALIDATION
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if final_seqn.min() != 93703:

    raise ValueError(
        "Final minimum SEQN incorrect."
    )


if final_seqn.max() != 102956:

    raise ValueError(
        "Final maximum SEQN incorrect."
    )


if int(
    final_seqn.isna()
    .sum()
) != 0:

    raise ValueError(
        "Final SEQN contains missing values."
    )


if int(
    final_seqn.duplicated()
    .sum()
) != 0:

    raise ValueError(
        "Final SEQN contains duplicates."
    )


print(
    "PASS: Final SEQN 93703-102956 validated."
)


# ============================================================
# 42. ROUND-TRIP MAXIMUM ABSOLUTE DIFFERENCE
# ============================================================

#
# Expected:
#
#     0.0
#
# for every field.
#

print(
    "\n--- ROUND-TRIP MAXIMUM ABSOLUTE DIFFERENCES ---"
)


difference_failures = {}


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    final_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    valid_mask = (
        ~np.isnan(expected_values)
        &
        ~np.isnan(final_values)
    )


    if valid_mask.any():

        max_difference = float(
            np.max(
                np.abs(
                    expected_values[valid_mask]
                    -
                    final_values[valid_mask]
                )
            )
        )

    else:

        max_difference = 0.0


    print(
        f"{col}: {max_difference}"
    )


    if max_difference != 0:

        difference_failures[col] = (
            max_difference
        )


if difference_failures:

    raise ValueError(
        "Round-trip numeric differences detected:\n"
        + str(difference_failures)
    )


print(
    "PASS: Maximum numeric difference = 0 "
    "for every variable."
)


# ============================================================
# 43. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 100
)

print(
    "DR2TOT_J CONVERSION COMPLETE"
)

print(
    "=" * 100
)

print(
    "Cycle: 2017-2018"
)

print(
    "Component: Dietary Interview - "
    "Total Nutrient Intakes, Second Day"
)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)

print(
    "SEQN range:",
    f"{df['SEQN'].min()}-{df['SEQN'].max()}"
)

print(
    "Original tiny XPORT artifacts:",
    f"{tiny_total:,}"
)

print(
    "Variables containing tiny artifacts:",
    len(EXPECTED_TINY_COUNTS)
)

print(
    "Validated zero restorations:",
    f"{total_corrections:,}"
)

print(
    "Integer/code/count fields converted to Int64:",
    len(INTEGER_COLUMNS)
)

print(
    "Dietary measurement fields preserved:",
    len(PRESERVE_AS_FLOAT_COLUMNS)
)

print(
    "WTDRD1 positive:",
    w1_positive_count
)

print(
    "WTDRD1 zero:",
    w1_zero_count
)

print(
    "WTDR2D positive:",
    w2_positive_count
)

print(
    "WTDR2D zero:",
    w2_zero_count
)

print(
    "WTDR2D missing:",
    w2_missing_count
)

print(
    "Day 2 reliable recalls:",
    6502
)

print(
    "Nutrient fields:",
    len(NUTRIENT_COLUMNS)
)

print(
    "Nutrient nonmissing per field:",
    6502
)

print(
    "Exact Dietary measurement round-trip failures:",
    0
)

print(
    "Exact all-column round-trip failures:",
    0
)

print(
    "Missing-position differences:",
    0
)

print(
    "Tiny values remaining:",
    0
)

print(
    "Unnecessary '.0' in integer/code fields:",
    0
)

print(
    "Scientific notation:",
    0
)

print(
    "=" * 100
)
```
