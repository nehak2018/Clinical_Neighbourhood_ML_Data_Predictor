```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2017_2018"
    / "Labs"
)

xpt_file = DATA_DIR / "TCHOL_J.xpt"
csv_file = DATA_DIR / "TCHOL_J_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Keep completely untouched pandas-decoded source.
#
# TCHOL_J requires NO numeric correction.
#
source_df = df.copy()


print("=" * 90)
print("NHANES 2017-2018 TCHOL_J XPT -> CSV")
print("=" * 90)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "LBXTC",
    "LBDTCSI"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "TCHOL_J column names/order differ from expected.\n\n"
        f"Expected:\n{EXPECTED_COLUMNS}\n\n"
        f"Actual:\n{df.columns.tolist()}"
    )


if len(df) != 7435:

    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 3:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 7,435 rows x 3 columns."
)


# ============================================================
# 4. FIELD TYPE PLAN
# ============================================================

#
# TCHOL_J contains:
#
# SEQN
#     Respondent identifier.
#     Whole-number field.
#
# LBXTC
#     Total cholesterol in mg/dL.
#     The released values in this file are whole-number.
#
# LBDTCSI
#     Total cholesterol in mmol/L.
#     GENUINE DECIMAL variable.
#
# IMPORTANT:
#
# LBDTCSI must NOT be converted to Int64.
#
# It must NOT be rounded.
#
# It must NOT be reformatted using float_format.
#
# This is the same rule that protects genuine decimal
# values in Dietary, Laboratory, Examination, weights,
# nutrient, and exposure datasets.
#

INTEGER_COLUMNS = [
    "SEQN",
    "LBXTC"
]


DECIMAL_COLUMNS = [
    "LBDTCSI"
]


if len(INTEGER_COLUMNS) != 2:

    raise ValueError(
        "Expected exactly two whole-number fields."
    )


if len(DECIMAL_COLUMNS) != 1:

    raise ValueError(
        "Expected exactly one genuine decimal field."
    )


print("\n--- FIELD TYPE PLAN ---")

print(
    "Whole-number fields:",
    INTEGER_COLUMNS
)

print(
    "Genuine decimal fields:",
    DECIMAL_COLUMNS
)


# ============================================================
# 5. VERIFY NUMERIC SOURCE STRUCTURE
# ============================================================

numeric_cols = (
    df.select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)


if numeric_cols != EXPECTED_COLUMNS:

    raise ValueError(
        "Expected all three TCHOL_J variables "
        "to load numerically.\n\n"
        f"Actual numeric columns:\n{numeric_cols}"
    )


print(
    "PASS: All 3 fields loaded numerically."
)


# ============================================================
# 6. CHECK KNOWN XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known pandas/SAS XPORT zero-decoding artifact:
#
#     5.397605346934028e-79
#
# Some NHANES files contain this value where zero
# was intended.
#
# Direct inspection of attached TCHOL_J:
#
#     tiny values = 0
#
# Therefore NO zero restoration should occur.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


tiny_counts = (
    df[EXPECTED_COLUMNS]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


print("\n--- ORIGINAL XPT TINY-VALUE CHECK ---")

print(
    tiny_counts.to_string()
)

print(
    "Total tiny XPORT artifacts:",
    tiny_total
)


if tiny_total != 0:

    raise ValueError(
        "STOP: Unexpected tiny XPORT values "
        "found in TCHOL_J.\n\n"
        + tiny_counts[
            tiny_counts > 0
        ].to_string()
    )


print(
    "PASS: No tiny XPORT artifacts."
)


# ============================================================
# 7. CHECK ORIGINAL NUMERIC ZEROS
# ============================================================

zero_counts = (
    df[EXPECTED_COLUMNS]
    .eq(0)
    .sum()
)


zero_total = int(
    zero_counts.sum()
)


print("\n--- ORIGINAL NUMERIC ZERO CHECK ---")

print(
    zero_counts.to_string()
)

print(
    "Total ordinary zeros:",
    zero_total
)


if zero_total != 0:

    raise ValueError(
        "Unexpected ordinary numeric zeros "
        "found in attached TCHOL_J."
    )


print(
    "PASS: Original TCHOL_J contains no "
    "numeric zero values."
)


# ============================================================
# 8. EXPECTED CORRECTED SOURCE
# ============================================================

#
# No XPORT repair is necessary.
#
# Therefore the expected corrected source is exactly
# the untouched source.
#

expected_df = source_df.copy()


# ============================================================
# 9. DISCOVER FRACTIONAL VARIABLES
# ============================================================

#
# CRITICAL:
#
# This occurs BEFORE converting anything to Int64.
#
# np.round() below is used ONLY as a comparison test.
#
# It DOES NOT change any value.
#

print("\n--- FRACTIONAL VALUE CHECK ---")


fractional_counts = {}


for col in EXPECTED_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )

    if len(values) == 0:

        fractional_count = 0

    else:

        fractional_mask = ~np.isclose(
            values,
            np.round(values),
            rtol=0,
            atol=1e-12
        )

        fractional_count = int(
            fractional_mask.sum()
        )

    print(
        f"{col}: "
        f"{fractional_count:,} fractional observations"
    )

    if fractional_count > 0:

        fractional_counts[col] = fractional_count


#
# Attached TCHOL_J:
#
# LBDTCSI has 6,696 fractional observations.
#
# The other 42 nonmissing LBDTCSI values happen to
# equal whole numbers such as 4.0 or 5.0 numerically,
# but the COLUMN itself remains a genuine decimal variable.
#

EXPECTED_FRACTIONAL_COUNTS = {
    "LBDTCSI": 6696
}


if fractional_counts != EXPECTED_FRACTIONAL_COUNTS:

    raise ValueError(
        "Unexpected fractional-variable pattern.\n\n"
        f"Expected:\n{EXPECTED_FRACTIONAL_COUNTS}\n\n"
        f"Actual:\n{fractional_counts}"
    )


print(
    "PASS: LBDTCSI is the only genuine "
    "decimal variable."
)


# ============================================================
# 10. PROVE SEQN AND LBXTC ARE WHOLE-NUMBER
# ============================================================

print("\n--- INTEGER FIELD VALIDATION ---")


for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )

    whole_mask = np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )

    fractional_count = int(
        (~whole_mask).sum()
    )

    print(
        f"{col}:",
        f"{fractional_count:,} fractional values"
    )

    if fractional_count != 0:

        raise ValueError(
            f"{col} unexpectedly contains "
            "fractional values."
        )


print(
    "PASS: SEQN and LBXTC are proven "
    "whole-number fields."
)


# ============================================================
# 11. VALIDATE SEQN
# ============================================================

print("\n--- SEQN ---")

print(
    "Minimum:",
    df["SEQN"].min()
)

print(
    "Maximum:",
    df["SEQN"].max()
)

print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)

print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if df["SEQN"].min() != 93705:

    raise ValueError(
        "Unexpected minimum SEQN."
    )


if df["SEQN"].max() != 102956:

    raise ValueError(
        "Unexpected maximum SEQN."
    )


if int(
    df["SEQN"]
    .isna()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains missing values."
    )


if int(
    df["SEQN"]
    .duplicated()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains duplicate values."
    )


print(
    "PASS: SEQN 93705-102956 validated."
)


# ============================================================
# 12. CDC CHECK - LBXTC
# ============================================================

#
# CDC:
#
# Total Cholesterol (mg/dL)
#
# Range:
#     76 - 446
#
# Nonmissing:
#     6,738
#
# Missing:
#     697
#

lbxtc_nonmissing = int(
    df["LBXTC"]
    .notna()
    .sum()
)


lbxtc_missing = int(
    df["LBXTC"]
    .isna()
    .sum()
)


lbxtc_min = df["LBXTC"].min()
lbxtc_max = df["LBXTC"].max()


print("\n--- CDC CHECK: LBXTC ---")

print(
    "Nonmissing:",
    f"{lbxtc_nonmissing:,}"
)

print(
    "Missing:",
    f"{lbxtc_missing:,}"
)

print(
    "Minimum:",
    lbxtc_min
)

print(
    "Maximum:",
    lbxtc_max
)


if lbxtc_nonmissing != 6738:

    raise ValueError(
        "LBXTC nonmissing count differs from CDC."
    )


if lbxtc_missing != 697:

    raise ValueError(
        "LBXTC missing count differs from CDC."
    )


if lbxtc_min != 76:

    raise ValueError(
        "LBXTC minimum differs from CDC."
    )


if lbxtc_max != 446:

    raise ValueError(
        "LBXTC maximum differs from CDC."
    )


print(
    "PASS: LBXTC matches CDC."
)


# ============================================================
# 13. CDC CHECK - LBDTCSI
# ============================================================

#
# CDC:
#
# Total Cholesterol (mmol/L)
#
# Range:
#     1.97 - 11.53
#
# Nonmissing:
#     6,738
#
# Missing:
#     697
#

lbdtcsi_nonmissing = int(
    df["LBDTCSI"]
    .notna()
    .sum()
)


lbdtcsi_missing = int(
    df["LBDTCSI"]
    .isna()
    .sum()
)


lbdtcsi_min = df["LBDTCSI"].min()
lbdtcsi_max = df["LBDTCSI"].max()


print("\n--- CDC CHECK: LBDTCSI ---")

print(
    "Nonmissing:",
    f"{lbdtcsi_nonmissing:,}"
)

print(
    "Missing:",
    f"{lbdtcsi_missing:,}"
)

print(
    "Minimum:",
    lbdtcsi_min
)

print(
    "Maximum:",
    lbdtcsi_max
)


if lbdtcsi_nonmissing != 6738:

    raise ValueError(
        "LBDTCSI nonmissing count differs from CDC."
    )


if lbdtcsi_missing != 697:

    raise ValueError(
        "LBDTCSI missing count differs from CDC."
    )


if not np.isclose(
    lbdtcsi_min,
    1.97,
    rtol=0,
    atol=0
):

    raise ValueError(
        "LBDTCSI minimum differs from CDC."
    )


if not np.isclose(
    lbdtcsi_max,
    11.53,
    rtol=0,
    atol=0
):

    raise ValueError(
        "LBDTCSI maximum differs from CDC."
    )


print(
    "PASS: LBDTCSI matches CDC."
)


# ============================================================
# 14. MISSING-POSITION RELATIONSHIP
# ============================================================

#
# LBXTC and LBDTCSI describe the same cholesterol
# measurement in two units.
#
# Therefore their missing positions should be identical.
#

lbxtc_missing_mask = (
    df["LBXTC"]
    .isna()
    .to_numpy()
)


lbdtcsi_missing_mask = (
    df["LBDTCSI"]
    .isna()
    .to_numpy()
)


missing_position_mismatch = int(
    np.sum(
        lbxtc_missing_mask
        !=
        lbdtcsi_missing_mask
    )
)


print(
    "\n--- LBXTC / LBDTCSI MISSING ALIGNMENT ---"
)

print(
    "Missing-position mismatches:",
    missing_position_mismatch
)


if missing_position_mismatch != 0:

    raise ValueError(
        "LBXTC and LBDTCSI missing positions differ."
    )


print(
    "PASS: LBXTC and LBDTCSI missing positions "
    "match exactly."
)


# ============================================================
# 15. CDC CONVERSION CONSISTENCY
# ============================================================

#
# CDC documentation:
#
#     LBDTCSI = LBXTC * 0.02586
#
# The released mmol/L field is represented to the
# hundredth.
#
# IMPORTANT:
#
# We DO NOT recalculate LBDTCSI.
#
# We only verify that the supplied CDC values are
# consistent with the documented conversion.
#
# This avoids changing original decimal values.
#

conversion_df = (
    df[
        [
            "LBXTC",
            "LBDTCSI"
        ]
    ]
    .dropna()
)


expected_unrounded = (
    conversion_df["LBXTC"]
    *
    0.02586
)


conversion_difference = (
    conversion_df["LBDTCSI"]
    -
    expected_unrounded
).abs()


max_conversion_difference = float(
    conversion_difference.max()
)


print(
    "\n--- CDC UNIT CONVERSION CONSISTENCY ---"
)

print(
    "Records checked:",
    f"{len(conversion_df):,}"
)

print(
    "Maximum absolute difference "
    "from LBXTC * 0.02586:",
    max_conversion_difference
)


#
# A value released to 2 decimal places may differ from
# the unrounded conversion by up to 0.005.
#

if max_conversion_difference > (
    0.005
    +
    1e-12
):

    raise ValueError(
        "LBDTCSI is inconsistent with "
        "CDC conversion factor."
    )


print(
    "PASS: LBDTCSI is consistent with "
    "LBXTC * 0.02586."
)


# ============================================================
# 16. VERIFY LBDTCSI DECIMAL PRECISION GRID
# ============================================================

#
# The released LBDTCSI measurements are represented
# on a hundredth-unit grid.
#
# This is a CHECK only.
#
# Nothing is rounded or rewritten.
#

decimal_values = (
    df["LBDTCSI"]
    .dropna()
    .to_numpy(
        dtype=float
    )
)


hundredth_grid_mask = np.isclose(
    decimal_values * 100,
    np.round(
        decimal_values * 100
    ),
    rtol=0,
    atol=1e-10
)


hundredth_grid_failures = int(
    (~hundredth_grid_mask)
    .sum()
)


print(
    "\n--- LBDTCSI DECIMAL GRID CHECK ---"
)

print(
    "Values not on hundredth grid:",
    hundredth_grid_failures
)


if hundredth_grid_failures != 0:

    raise ValueError(
        "Unexpected LBDTCSI decimal precision."
    )


print(
    "PASS: All LBDTCSI values are consistent "
    "with released hundredth-unit precision."
)


# ============================================================
# 17. CONVERT ONLY PROVEN INTEGER FIELDS
# ============================================================

#
# Convert:
#
#     SEQN
#     LBXTC
#
# to nullable Int64.
#
# DO NOT TOUCH:
#
#     LBDTCSI
#

for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )

    whole_mask = np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )

    if not whole_mask.all():

        raise ValueError(
            f"{col} cannot safely be converted "
            "to Int64."
        )

    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: SEQN and LBXTC converted "
    "to nullable Int64."
)

print(
    "PASS: LBDTCSI left untouched "
    "as genuine decimal."
)


# ============================================================
# 18. CONFIRM LBDTCSI WAS NOT ALTERED
# ============================================================

original_decimal = (
    source_df["LBDTCSI"]
    .to_numpy(
        dtype=float
    )
)


current_decimal = (
    df["LBDTCSI"]
    .to_numpy(
        dtype=float
    )
)


if not np.allclose(
    original_decimal,
    current_decimal,
    rtol=0,
    atol=0,
    equal_nan=True
):

    raise ValueError(
        "LBDTCSI changed during cleaning."
    )


print(
    "PASS: LBDTCSI unchanged before export."
)


# ============================================================
# 19. COMPLETE PRE-EXPORT NUMERIC FIDELITY
# ============================================================

#
# TCHOL_J permitted numeric modifications:
#
#     NONE
#
# Int64 conversion changes representation only.
#
# Every numerical value must still exactly equal the
# original XPT-derived value.
#

print(
    "\n--- COMPLETE PRE-EXPORT VALUE FIDELITY ---"
)


fidelity_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )

    current_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )

    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )

    if not same:

        fidelity_failures.append(
            col
        )


if fidelity_failures:

    raise ValueError(
        "Unexpected numeric changes before export:\n"
        + str(fidelity_failures)
    )


print(
    "PASS: All 3 variables exactly match "
    "their original XPT numeric values."
)


# ============================================================
# 20. MISSING-VALUE FIDELITY BEFORE EXPORT
# ============================================================

missing_mismatches = 0


for col in EXPECTED_COLUMNS:

    source_missing = (
        source_df[col]
        .isna()
        .to_numpy()
    )

    current_missing = (
        df[col]
        .isna()
        .to_numpy()
    )

    missing_mismatches += int(
        np.sum(
            source_missing
            !=
            current_missing
        )
    )


print(
    "\nMissing-position differences before export:",
    missing_mismatches
)


if missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed."
    )


print(
    "PASS: Missing positions preserved exactly."
)


# ============================================================
# 21. EXPORT CSV
# ============================================================

#
# CRITICAL FIDELITY POLICY
# ------------------------------------------------------------
#
# NO:
#
#     df.round(...)
#
# NO:
#
#     float_format=
#
# NO:
#
#     conversion of LBDTCSI to Int64
#
# NO:
#
#     recalculation of LBDTCSI
#
# This allows pandas to serialize the original
# floating-point values without imposing a new
# decimal precision.
#
# For example:
#
#     4.06 remains numerically 4.06
#
#     5.40 may serialize as 5.4
#
# Those represent the EXACT SAME numeric value.
#
# We deliberately do not force 5.40 because doing so
# would be output formatting rather than preservation
# of the underlying XPT numeric value.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print("\nCSV created:")

print(
    csv_file
)


# ============================================================
# 22. RE-READ CSV USING ROUND-TRIP FLOAT PARSER
# ============================================================

#
# float_precision="round_trip" ensures the parser
# reconstructs exported floating-point values as
# faithfully as possible.
#

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != df.shape:

    raise ValueError(
        "CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV columns/order changed."
    )


print(
    "PASS: CSV structure preserved."
)


# ============================================================
# 23. EXACT XPT -> CSV NUMERIC FIDELITY
# ============================================================

print(
    "\n--- EXACT XPT -> CSV NUMERIC FIDELITY ---"
)


csv_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        source_df[col]
        .to_numpy(
            dtype=float
        )
    )

    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )

    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )

    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "XPT -> CSV numeric fidelity differences:\n"
        + str(csv_failures)
    )


print(
    "PASS: All 3 CSV fields round-trip "
    "to exact original XPT numeric values."
)


# ============================================================
# 24. CSV MISSING-VALUE FIDELITY
# ============================================================

csv_missing_mismatches = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        source_df[col]
        .isna()
        .to_numpy()
    )

    exported_missing = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )

    csv_missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            exported_missing
        )
    )


print(
    "\n--- CSV MISSING VALUE FIDELITY ---"
)

print(
    "Missing-position mismatches:",
    csv_missing_mismatches
)


if csv_missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed "
        "during CSV export."
    )


print(
    "PASS: Missing positions preserved exactly."
)


# ============================================================
# 25. READ FINAL CSV AS RAW TEXT
# ============================================================

#
# Used to inspect literal CSV formatting.
#

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


if raw_csv_df.shape != (
    7435,
    3
):

    raise ValueError(
        "Raw CSV dimensions changed."
    )


if (
    raw_csv_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "Raw CSV column order changed."
    )


# ============================================================
# 26. INTEGER-FIELD .0 CHECK
# ============================================================

#
# ONLY check:
#
#     SEQN
#     LBXTC
#
# DO NOT include LBDTCSI.
#
# LBDTCSI is a genuine decimal field.
#

print("\n--- INTEGER-FIELD .0 CHECK ---")


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = raw_csv_df[col]

    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )

    count = int(
        bad_mask.sum()
    )

    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unexpected '.0' formatting in "
        "integer fields:\n"
        + str(dot_zero_errors)
    )


print(
    "PASS: SEQN and LBXTC contain no "
    "unnecessary '.0'."
)


# ============================================================
# 27. DECIMAL FIELD RAW CSV CHECK
# ============================================================

#
# LBDTCSI is allowed to contain:
#
#     4.06
#     3.83
#     5.4
#     etc.
#
# We DO NOT require or remove ".0".
#
# We validate numerical fidelity instead.
#

decimal_tokens = (
    raw_csv_df["LBDTCSI"]
)


decimal_nonblank = (
    decimal_tokens[
        decimal_tokens != ""
    ]
)


print(
    "\n--- LBDTCSI RAW DECIMAL CHECK ---"
)

print(
    "Populated decimal tokens:",
    f"{len(decimal_nonblank):,}"
)


if len(decimal_nonblank) != 6738:

    raise ValueError(
        "LBDTCSI populated CSV count changed."
    )


print(
    "PASS: LBDTCSI remains a decimal field."
)


# ============================================================
# 28. FINAL DECIMAL NUMERIC FIDELITY
# ============================================================

final_decimal_values = pd.to_numeric(
    raw_csv_df["LBDTCSI"]
    .where(
        raw_csv_df["LBDTCSI"] != "",
        np.nan
    ),
    errors="raise"
).to_numpy(
    dtype=float
)


if not np.allclose(
    source_df["LBDTCSI"]
    .to_numpy(
        dtype=float
    ),
    final_decimal_values,
    rtol=0,
    atol=0,
    equal_nan=True
):

    raise ValueError(
        "LBDTCSI decimal values changed "
        "during CSV export."
    )


print(
    "PASS: LBDTCSI decimal values preserved exactly."
)


# ============================================================
# 29. FINAL TINY-VALUE CHECK
# ============================================================

tiny_csv_count = 0


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )

    tiny_csv_count += int(
        values.eq(TINY_VALUE)
        .sum()
    )


print("\n--- FINAL TINY VALUE CHECK ---")

print(
    "Tiny values remaining:",
    tiny_csv_count
)


if tiny_csv_count != 0:

    raise ValueError(
        "Tiny XPORT artifacts found in final CSV."
    )


print(
    "PASS: No tiny XPORT artifacts."
)


# ============================================================
# 30. FINAL ZERO CHECK
# ============================================================

csv_zero_counts = {}


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )

    zero_count = int(
        values.eq(0)
        .sum()
    )

    if zero_count > 0:

        csv_zero_counts[col] = zero_count


print("\n--- FINAL CSV ZERO PATTERN ---")

print(
    csv_zero_counts
)


if csv_zero_counts != {}:

    raise ValueError(
        "Unexpected numeric zeros found "
        "in final CSV:\n"
        + str(csv_zero_counts)
    )


print(
    "PASS: Final CSV contains no numeric zeros."
)


# ============================================================
# 31. SCIENTIFIC NOTATION CHECK
# ============================================================

scientific_errors = {}


for col in EXPECTED_COLUMNS:

    tokens = raw_csv_df[col]

    count = int(
        tokens.str.contains(
            r"[eE][+-]\d+",
            regex=True
        )
        .sum()
    )

    if count > 0:

        scientific_errors[col] = count


print("\n--- SCIENTIFIC NOTATION CHECK ---")

print(
    "Occurrences:",
    scientific_errors
)


if scientific_errors:

    raise ValueError(
        "Unexpected scientific notation:\n"
        + str(scientific_errors)
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 32. FINAL SEQN CHECK
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if final_seqn.min() != 93705:

    raise ValueError(
        "Final minimum SEQN incorrect."
    )


if final_seqn.max() != 102956:

    raise ValueError(
        "Final maximum SEQN incorrect."
    )


if int(
    final_seqn.isna()
    .sum()
) != 0:

    raise ValueError(
        "Final CSV contains missing SEQN."
    )


if int(
    final_seqn.duplicated()
    .sum()
) != 0:

    raise ValueError(
        "Final CSV contains duplicate SEQN."
    )


if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "Final CSV SEQN still contains '.0'."
    )


print(
    "PASS: Final SEQN valid and contains no '.0'."
)


# ============================================================
# 33. FINAL COMPLETE SOURCE -> CSV FIDELITY
# ============================================================

final_fidelity_failures = []


for col in EXPECTED_COLUMNS:

    original_values = (
        source_df[col]
        .to_numpy(
            dtype=float
        )
    )

    final_values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    ).to_numpy(
        dtype=float
    )

    same = np.allclose(
        original_values,
        final_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )

    if not same:

        final_fidelity_failures.append(
            col
        )


if final_fidelity_failures:

    raise ValueError(
        "Final source -> CSV fidelity failures:\n"
        + str(final_fidelity_failures)
    )


print(
    "PASS: Final CSV preserves all original "
    "TCHOL_J numeric values exactly."
)


# ============================================================
# 34. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 90
)

print(
    "TCHOL_J CONVERSION COMPLETE"
)

print(
    "=" * 90
)

print(
    "Cycle: 2017-2018"
)

print(
    "Component: Laboratory - Total Cholesterol"
)

print(
    "Eligible population: Examined participants age 6+"
)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)

print(
    "SEQN range:",
    f"{df['SEQN'].min()}-{df['SEQN'].max()}"
)

print(
    "Original tiny-value artifacts:",
    tiny_total
)

print(
    "Numeric zero restorations:",
    0
)

print(
    "Whole-number fields:",
    INTEGER_COLUMNS
)

print(
    "Genuine decimal fields:",
    DECIMAL_COLUMNS
)

print(
    "LBXTC nonmissing:",
    lbxtc_nonmissing
)

print(
    "LBXTC range:",
    f"{lbxtc_min}-{lbxtc_max}"
)

print(
    "LBDTCSI nonmissing:",
    lbdtcsi_nonmissing
)

print(
    "LBDTCSI range:",
    f"{lbdtcsi_min}-{lbdtcsi_max}"
)

print(
    "LBDTCSI fractional observations:",
    fractional_counts["LBDTCSI"]
)

print(
    "XPT -> CSV unexpected numeric differences: "
    "0 expected"
)

print(
    "Missing-position differences: "
    "0 expected"
)

print(
    "Tiny values remaining: "
    "0 expected"
)

print(
    "Unnecessary '.0' in integer fields: "
    "0 expected"
)

print(
    "Scientific notation: "
    "0 expected"
)

print(
    "=" * 90
)
```
