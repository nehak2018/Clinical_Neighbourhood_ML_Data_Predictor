```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2011_2012"
    / "Questionnnair"
)

xpt_file = DATA_DIR / "DIQ_G.xpt"
csv_file = DATA_DIR / "DIQ_G_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Completely untouched source copy.
#
source_df = df.copy()


print("=" * 80)
print("NHANES 2011-2012 DIQ_G XPT -> CSV")
print("=" * 80)

print("Rows:", f"{len(df):,}")
print("Columns:", len(df.columns))


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "DIQ010",
    "DID040",
    "DIQ160",
    "DIQ170",
    "DIQ172",
    "DIQ175A",
    "DIQ175B",
    "DIQ175C",
    "DIQ175D",
    "DIQ175E",
    "DIQ175F",
    "DIQ175G",
    "DIQ175H",
    "DIQ175I",
    "DIQ175J",
    "DIQ175K",
    "DIQ175L",
    "DIQ175M",
    "DIQ175N",
    "DIQ175O",
    "DIQ175P",
    "DIQ175Q",
    "DIQ175R",
    "DIQ175S",
    "DIQ175T",
    "DIQ175U",
    "DIQ175V",
    "DIQ175W",
    "DIQ180",
    "DIQ050",
    "DID060",
    "DIQ060U",
    "DIQ070",
    "DIQ230",
    "DIQ240",
    "DID250",
    "DID260",
    "DIQ260U",
    "DIQ275",
    "DIQ280",
    "DIQ291",
    "DIQ300S",
    "DIQ300D",
    "DID310S",
    "DID310D",
    "DID320",
    "DID330",
    "DID341",
    "DID350",
    "DIQ350U",
    "DIQ360",
    "DIQ080"
]


if df.columns.tolist() != EXPECTED_COLUMNS:
    raise ValueError(
        "DIQ_G column names/order differ from expected.\n"
        f"Expected: {EXPECTED_COLUMNS}\n"
        f"Actual:   {df.columns.tolist()}"
    )


if len(df) != 9364:
    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 53:
    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 9,364 rows x 53 columns."
)


# ============================================================
# 4. DEFINE DECIMAL VS WHOLE-NUMBER VARIABLES
# ============================================================

#
# CRITICAL:
#
# DIQ280 = self-reported last A1C level.
#
# This is a genuine decimal questionnaire measurement.
#
# Examples may include:
#
#     5.8
#     6.5
#     7.2
#
# DO NOT:
#
#     round()
#     convert DIQ280 to Int64
#     use float_format during CSV export
#

DECIMAL_COLUMNS = [
    "DIQ280"
]


INTEGER_COLUMNS = [
    col
    for col in EXPECTED_COLUMNS
    if col not in DECIMAL_COLUMNS
]


if len(DECIMAL_COLUMNS) != 1:
    raise ValueError(
        "Expected exactly one genuine decimal field."
    )


if len(INTEGER_COLUMNS) != 52:
    raise ValueError(
        "Expected exactly 52 whole-number fields."
    )


print("\n--- VARIABLE TYPE PLAN ---")

print(
    "Genuine decimal variables:",
    DECIMAL_COLUMNS
)

print(
    "Whole-number variables:",
    len(INTEGER_COLUMNS)
)


# ============================================================
# 5. CHECK KNOWN XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known pandas / SAS XPORT zero-decoding artifact:
#
#     5.397605346934028e-79
#
# NEVER globally replace this before validating
# the exact variables and counts.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


numeric_cols = (
    df.select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)


tiny_counts = (
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


true_zero_total = int(
    df[numeric_cols]
    .eq(0)
    .sum()
    .sum()
)


print("\n--- ORIGINAL XPT ZERO CHECK ---")

print(
    "Ordinary numeric zeros before correction:",
    f"{true_zero_total:,}"
)

print(
    "Tiny XPORT artifacts:",
    f"{tiny_total:,}"
)


print("\nTiny values by variable:")

print(
    tiny_counts[
        tiny_counts > 0
    ]
    .sort_values(
        ascending=False
    )
    .to_string()
)


# ============================================================
# 6. VALIDATE EXACT TINY-ZERO PATTERN
# ============================================================

#
# Exact pattern found in attached DIQ_G.xpt:
#
# DID250 = 13
# DID260 = 131
# DID341 = 210
# DID350 = 134
#
# TOTAL = 488
#
#
# CDC explicitly documents:
#
# DID250
#     0 = None
#
# DID260
#     0 = Never
#
# DID341
#     0 = None
#
# DID350
#     0 = None
#

EXPECTED_TINY_COUNTS = {
    "DID250": 13,
    "DID260": 131,
    "DID341": 210,
    "DID350": 134
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:
    raise ValueError(
        "STOP: Tiny-value pattern differs from "
        "validated DIQ_G pattern.\n"
        f"Expected: {EXPECTED_TINY_COUNTS}\n"
        f"Actual:   {actual_tiny_counts}"
    )


if tiny_total != 488:
    raise ValueError(
        f"Unexpected total tiny count: {tiny_total:,}"
    )


if len(EXPECTED_TINY_COUNTS) != 4:
    raise ValueError(
        "Expected tiny artifacts in exactly 4 variables."
    )


print(
    "\nPASS: Exact 488-value tiny-zero "
    "pattern across four variables validated."
)


# ============================================================
# 7. RESTORE ONLY VALIDATED ZERO VALUES
# ============================================================

#
# DO NOT:
#
#     df.replace(TINY_VALUE, 0)
#
# across the complete dataset.
#
# Change only the four validated variables.
#

print("\n--- VALIDATED ZERO RESTORATION ---")


correction_total = 0


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    mask = df[col].eq(TINY_VALUE)

    actual_count = int(
        mask.sum()
    )


    if actual_count != expected_count:
        raise ValueError(
            f"{col}: expected {expected_count:,} tiny values, "
            f"found {actual_count:,}."
        )


    df.loc[
        mask,
        col
    ] = 0


    correction_total += actual_count


    print(
        f"{col}: {actual_count:,} restored to 0"
    )


if correction_total != 488:
    raise ValueError(
        "Incorrect zero-restoration total."
    )


print(
    "\nPASS:",
    f"{correction_total:,}",
    "validated zeros restored."
)


# ============================================================
# 8. CREATE EXPECTED CORRECTED SOURCE COPY
# ============================================================

#
# expected_df represents exactly what the corrected
# numeric source should contain.
#
# ONLY validated TINY_VALUE -> 0 positions change.
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col].eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 9. CONFIRM NO TINY VALUES REMAIN
# ============================================================

remaining_tiny = int(
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print(
    "\nTiny values remaining:",
    remaining_tiny
)


if remaining_tiny != 0:
    raise ValueError(
        "Tiny XPORT values remain after correction."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 10. DETECT GENUINE DECIMAL VARIABLES
# ============================================================

#
# Run BEFORE any Int64 conversion.
#
# Attached DIQ_G contains genuine fractional
# observations only in:
#
#     DIQ280
#

print("\n--- GENUINE DECIMAL CHECK ---")


detected_decimal_columns = []


for col in numeric_cols:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:
        continue


    fractional_mask = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    fractional_count = int(
        fractional_mask.sum()
    )


    if fractional_count > 0:

        detected_decimal_columns.append(
            col
        )

        print(
            f"{col}: "
            f"{fractional_count:,} fractional observations"
        )


if detected_decimal_columns != DECIMAL_COLUMNS:
    raise ValueError(
        "STOP: Decimal-variable pattern changed.\n"
        f"Expected: {DECIMAL_COLUMNS}\n"
        f"Actual:   {detected_decimal_columns}"
    )


print(
    "\nPASS: DIQ280 is the only genuine "
    "decimal variable."
)


# ============================================================
# 11. DIQ280 FRACTIONAL OBSERVATION CHECK
# ============================================================

diq280_values = (
    df["DIQ280"]
    .dropna()
    .to_numpy(
        dtype=float
    )
)


diq280_fractional = ~np.isclose(
    diq280_values,
    np.round(diq280_values),
    rtol=0,
    atol=1e-12
)


diq280_fractional_count = int(
    diq280_fractional.sum()
)


print(
    "\nDIQ280 fractional observations:",
    diq280_fractional_count
)


if diq280_fractional_count != 126:
    raise ValueError(
        "Unexpected DIQ280 fractional-value count."
    )


print(
    "PASS: 126 genuine fractional DIQ280 "
    "observations confirmed."
)


# ============================================================
# 12. CLEAN AND VALIDATE SEQN
# ============================================================

seqn = (
    df["SEQN"]
    .dropna()
)


if not np.all(
    np.isclose(
        seqn,
        np.round(seqn),
        rtol=0,
        atol=1e-12
    )
):
    raise ValueError(
        "SEQN contains unexpected fractional values."
    )


df["SEQN"] = (
    df["SEQN"]
    .astype("Int64")
)


print("\n--- SEQN ---")

print(
    "Minimum:",
    df["SEQN"].min()
)

print(
    "Maximum:",
    df["SEQN"].max()
)

print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)

print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if not (
    df["SEQN"].min() == 62161
    and
    df["SEQN"].max() == 71916
    and
    int(
        df["SEQN"]
        .isna()
        .sum()
    ) == 0
    and
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    ) == 0
):
    raise ValueError(
        "SEQN validation failed."
    )


print(
    "PASS: SEQN validated."
)


# ============================================================
# 13. FORMAT ONLY PROVEN WHOLE-NUMBER VARIABLES
# ============================================================

#
# Convert 52 fields to nullable Int64.
#
# DIQ280 IS EXCLUDED.
#
# NO .round() is used for conversion.
#

for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) > 0:

        whole_mask = np.isclose(
            values,
            np.round(values),
            rtol=0,
            atol=1e-12
        )


        if not whole_mask.all():

            raise ValueError(
                f"{col} unexpectedly contains "
                "genuine decimal values."
            )


    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: 52 proven whole-number variables "
    "converted to nullable Int64."
)

print(
    "PASS: DIQ280 remains decimal."
)


# ============================================================
# 14. HELPER - CDC FREQUENCY CHECK
# ============================================================

def check_frequency(
    column,
    expected_values,
    expected_missing
):

    actual_values = (
        df[column]
        .value_counts()
        .sort_index()
        .to_dict()
    )


    actual_missing = int(
        df[column]
        .isna()
        .sum()
    )


    if actual_values != expected_values:

        raise ValueError(
            f"{column} frequencies differ.\n"
            f"Expected: {expected_values}\n"
            f"Actual:   {actual_values}"
        )


    if actual_missing != expected_missing:

        raise ValueError(
            f"{column} missing count differs.\n"
            f"Expected: {expected_missing}\n"
            f"Actual:   {actual_missing}"
        )


    print(
        f"PASS: {column} matches CDC."
    )


# ============================================================
# 15. CDC CHECK - DIQ010
# ============================================================

#
# Doctor told participant they have diabetes
#
# CDC:
#
# 1 Yes          708
# 2 No          8524
# 3 Borderline   125
# 7 Refused        1
# 9 Don't know     5
# Missing          1
#

check_frequency(
    "DIQ010",
    {
        1: 708,
        2: 8524,
        3: 125,
        7: 1,
        9: 5
    },
    1
)


# ============================================================
# 16. CDC CHECK - DID040
# ============================================================

#
# Age when first told had diabetes
#
# CDC:
#
# 3-78 = 684
# 80   = 12
# 666  = Less than 1 year = 5
# 999  = 7
# Missing = 8656
#

s = df["DID040"]


if not (
    int(
        s.between(
            3,
            78
        ).sum()
    ) == 684

    and int(
        s.eq(80).sum()
    ) == 12

    and int(
        s.eq(666).sum()
    ) == 5

    and int(
        s.eq(777).sum()
    ) == 0

    and int(
        s.eq(999).sum()
    ) == 7

    and int(
        s.isna().sum()
    ) == 8656
):
    raise ValueError(
        "DID040 differs from CDC."
    )


print(
    "PASS: DID040 matches CDC."
)


# ============================================================
# 17. CDC CHECKS - PREDIABETES / RISK
# ============================================================

check_frequency(
    "DIQ160",
    {
        1: 245,
        2: 5748,
        7: 1,
        9: 6
    },
    3364
)


check_frequency(
    "DIQ170",
    {
        1: 772,
        2: 5335,
        7: 1,
        9: 16
    },
    3240
)


check_frequency(
    "DIQ172",
    {
        1: 1562,
        2: 4494,
        7: 1,
        9: 67
    },
    3240
)


# ============================================================
# 18. CHECK NEW DIQ175 RISK-FACTOR VARIABLES
# ============================================================

#
# These were new in the 2011-2012 survey.
#
# Each variable uses its category number when selected
# and missing otherwise.
#

EXPECTED_DIQ175 = {

    "DIQ175A": {
        10: 1165,
        99: 2
    },

    "DIQ175B": {
        11: 458
    },

    "DIQ175C": {
        12: 103
    },

    "DIQ175D": {
        13: 296
    },

    "DIQ175E": {
        14: 114
    },

    "DIQ175F": {
        15: 23
    },

    "DIQ175G": {
        16: 174
    },

    "DIQ175H": {
        17: 156
    },

    "DIQ175I": {
        18: 52
    },

    "DIQ175J": {
        19: 95
    },

    "DIQ175K": {
        20: 20
    },

    "DIQ175L": {
        21: 52
    },

    "DIQ175M": {
        22: 88
    },

    "DIQ175N": {
        23: 58
    },

    "DIQ175O": {
        24: 81
    },

    "DIQ175P": {
        25: 79
    },

    "DIQ175Q": {
        26: 95
    },

    "DIQ175R": {
        27: 28
    },

    "DIQ175S": {
        28: 21
    },

    "DIQ175T": {
        29: 38
    },

    "DIQ175U": {
        30: 35
    },

    "DIQ175V": {
        31: 15
    },

    "DIQ175W": {
        32: 7
    }
}


print("\n--- DIQ175 RISK FACTOR CHECKS ---")


for col, expected_values in EXPECTED_DIQ175.items():

    actual_values = (
        df[col]
        .dropna()
        .value_counts()
        .sort_index()
        .to_dict()
    )


    if actual_values != expected_values:

        raise ValueError(
            f"{col} differs from expected release pattern.\n"
            f"Expected: {expected_values}\n"
            f"Actual:   {actual_values}"
        )


print(
    "PASS: DIQ175A-DIQ175W release patterns validated."
)


# ============================================================
# 19. CDC CHECK - BLOOD TEST / INSULIN / PILLS
# ============================================================

check_frequency(
    "DIQ180",
    {
        1: 2575,
        2: 3381,
        7: 1,
        9: 167
    },
    3240
)


check_frequency(
    "DIQ050",
    {
        1: 213,
        2: 9149,
        7: 1
    },
    1
)


#
# DID060:
#
# 1-56 = 207
# 666 = Less than 1 month = 2
# 999 = 4
# missing = 9151
#

s = df["DID060"]


if not (
    int(
        s.between(
            1,
            56
        ).sum()
    ) == 207

    and int(
        s.eq(666).sum()
    ) == 2

    and int(
        s.eq(777).sum()
    ) == 0

    and int(
        s.eq(999).sum()
    ) == 4

    and int(
        s.isna().sum()
    ) == 9151
):
    raise ValueError(
        "DID060 differs from CDC."
    )


print(
    "PASS: DID060 matches CDC."
)


check_frequency(
    "DIQ060U",
    {
        1: 26,
        2: 181
    },
    9157
)


check_frequency(
    "DIQ070",
    {
        1: 546,
        2: 529,
        9: 3
    },
    8286
)


# ============================================================
# 20. CDC CHECK - DIABETES CARE
# ============================================================

check_frequency(
    "DIQ240",
    {
        1: 535,
        2: 173
    },
    8656
)


# ============================================================
# 21. CDC CHECK - DID250
# ============================================================

#
# Past year number of times seen doctor.
#
# CDC:
#
# 1-200 = 522
# 0 None = 13
# Missing = 8829
#

s = df["DID250"]


print("\n--- DID250 ---")

print(
    "Zeros:",
    int(
        s.eq(0).sum()
    )
)


if not (
    int(
        s.between(
            1,
            200
        ).sum()
    ) == 522

    and int(
        s.eq(0).sum()
    ) == 13

    and int(
        s.eq(7777).sum()
    ) == 0

    and int(
        s.eq(9999).sum()
    ) == 0

    and int(
        s.isna().sum()
    ) == 8829
):
    raise ValueError(
        "DID250 differs from CDC / "
        "validated zero restoration."
    )


print(
    "PASS: DID250 matches CDC."
)


# ============================================================
# 22. CDC CHECK - DID260
# ============================================================

#
# How often checks blood glucose/sugar.
#
# CDC:
#
# 1-30 = 571
# 0 Never = 131
# 999 Don't know = 5
# Missing = 8657
#

s = df["DID260"]


print("\n--- DID260 ---")

print(
    "Zeros:",
    int(
        s.eq(0).sum()
    )
)


if not (
    int(
        s.between(
            1,
            30
        ).sum()
    ) == 571

    and int(
        s.eq(0).sum()
    ) == 131

    and int(
        s.eq(777).sum()
    ) == 0

    and int(
        s.eq(999).sum()
    ) == 5

    and int(
        s.isna().sum()
    ) == 8657
):
    raise ValueError(
        "DID260 differs from CDC / "
        "validated zero restoration."
    )


print(
    "PASS: DID260 matches CDC."
)


check_frequency(
    "DIQ260U",
    {
        1: 352,
        2: 142,
        3: 47,
        4: 30
    },
    8793
)


# ============================================================
# 23. CDC CHECK - A1C
# ============================================================

check_frequency(
    "DIQ275",
    {
        1: 434,
        2: 168,
        9: 106
    },
    8656
)


#
# DIQ280 is SPECIAL.
#
# Genuine decimal A1C values:
#
#     2.1 - 36
#
# Count:
#
#     249
#
# Special codes:
#
#     777 = Refused      3
#     999 = Don't know 182
#
# Missing:
#
#     8930
#
# DO NOT convert this field to Int64.
#

s = df["DIQ280"]


print("\n--- DIQ280 ---")

print(
    "Actual A1C values:",
    int(
        s.between(
            2.1,
            36
        ).sum()
    )
)

print(
    "Refused 777:",
    int(
        s.eq(777).sum()
    )
)

print(
    "Don't know 999:",
    int(
        s.eq(999).sum()
    )
)

print(
    "Missing:",
    int(
        s.isna().sum()
    )
)


if not (
    int(
        s.between(
            2.1,
            36
        ).sum()
    ) == 249

    and np.isclose(
        s[
            s.between(
                2.1,
                36
            )
        ].min(),
        2.1,
        rtol=0,
        atol=0
    )

    and np.isclose(
        s[
            s.between(
                2.1,
                36
            )
        ].max(),
        36.0,
        rtol=0,
        atol=0
    )

    and int(
        s.eq(777).sum()
    ) == 3

    and int(
        s.eq(999).sum()
    ) == 182

    and int(
        s.isna().sum()
    ) == 8930
):
    raise ValueError(
        "DIQ280 differs from CDC."
    )


print(
    "PASS: DIQ280 matches CDC and "
    "remains a decimal field."
)


check_frequency(
    "DIQ291",
    {
        1: 131,
        2: 109,
        3: 16,
        4: 6,
        5: 1,
        6: 83,
        99: 88
    },
    8930
)


# ============================================================
# 24. CDC CHECK - DID341
# ============================================================

#
# Past year times doctor checked feet.
#
# CDC:
#
# 1-365 = 492
# 0 None = 210
# 9999 Don't know = 3
# Missing = 8659
#

s = df["DID341"]


print("\n--- DID341 ---")

print(
    "Zeros:",
    int(
        s.eq(0).sum()
    )
)


if not (
    int(
        s.between(
            1,
            365
        ).sum()
    ) == 492

    and int(
        s.eq(0).sum()
    ) == 210

    and int(
        s.eq(7777).sum()
    ) == 0

    and int(
        s.eq(9999).sum()
    ) == 3

    and int(
        s.isna().sum()
    ) == 8659
):
    raise ValueError(
        "DID341 differs from CDC / "
        "validated zero restoration."
    )


print(
    "PASS: DID341 matches CDC."
)


# ============================================================
# 25. CDC CHECK - DID350
# ============================================================

#
# How often participant checks feet.
#
# CDC:
#
# 1-30 = 570
# 0 None = 134
# 9999 Don't know = 1
# Missing = 8659
#

s = df["DID350"]


print("\n--- DID350 ---")

print(
    "Zeros:",
    int(
        s.eq(0).sum()
    )
)


if not (
    int(
        s.between(
            1,
            30
        ).sum()
    ) == 570

    and int(
        s.eq(0).sum()
    ) == 134

    and int(
        s.eq(7777).sum()
    ) == 0

    and int(
        s.eq(9999).sum()
    ) == 1

    and int(
        s.isna().sum()
    ) == 8659
):
    raise ValueError(
        "DID350 differs from CDC / "
        "validated zero restoration."
    )


print(
    "PASS: DID350 matches CDC."
)


# ============================================================
# 26. CDC CHECK - DIQ360 / DIQ080
# ============================================================

check_frequency(
    "DIQ360",
    {
        1: 79,
        2: 364,
        3: 109,
        4: 98,
        5: 54,
        9: 4
    },
    8656
)


check_frequency(
    "DIQ080",
    {
        1: 140,
        2: 563,
        9: 5
    },
    8656
)


# ============================================================
# 27. VALIDATED ZERO COUNT CHECK
# ============================================================

#
# Raw attached XPT contained no ordinary numeric zeros.
#
# Therefore each corrected variable's final zero count
# must exactly equal the validated tiny count.
#

print("\n--- VALIDATED ZERO COUNTS ---")


for col, expected_zero_count in EXPECTED_TINY_COUNTS.items():

    actual_zero_count = int(
        df[col]
        .eq(0)
        .sum()
    )


    print(
        f"{col}: {actual_zero_count:,}"
    )


    if actual_zero_count != expected_zero_count:

        raise ValueError(
            f"{col}: expected "
            f"{expected_zero_count:,} zeros, "
            f"found {actual_zero_count:,}."
        )


print(
    "PASS: All four restored zero counts match."
)


# ============================================================
# 28. VALUE FIDELITY BEFORE EXPORT
# ============================================================

#
# The ONLY numeric differences allowed from the
# raw pandas-decoded XPT are:
#
# DID250:
#     13 tiny -> 0
#
# DID260:
#     131 tiny -> 0
#
# DID341:
#     210 tiny -> 0
#
# DID350:
#     134 tiny -> 0
#
# DIQ280 MUST remain otherwise untouched.
#

print("\n--- VALUE FIDELITY BEFORE EXPORT ---")


fidelity_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    current_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )


    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        fidelity_failures.append(
            col
        )


if fidelity_failures:

    raise ValueError(
        "Unexpected numeric changes before export:\n"
        + str(fidelity_failures)
    )


print(
    "PASS: All 53 variables match the expected "
    "corrected XPT exactly."
)


# ============================================================
# 29. EXPORT CSV
# ============================================================

#
# CRITICAL:
#
# NO rounding.
#
# NO:
#
#     float_format=
#
# DIQ280 must serialize naturally as a genuine
# decimal measurement column.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print("\nCSV created:")

print(
    csv_file
)


# ============================================================
# 30. RE-READ WITH ROUND-TRIP FLOAT PARSER
# ============================================================

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != df.shape:

    raise ValueError(
        "CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV column names/order changed."
    )


print(
    "PASS: CSV structure preserved."
)


# ============================================================
# 31. EXACT CORRECTED XPT -> CSV FIDELITY
# ============================================================

print("\n--- EXACT XPT -> CSV FIDELITY ---")


csv_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "Corrected XPT -> CSV differences found:\n"
        + str(csv_failures)
    )


print(
    "PASS: All 53 CSV variables round-trip "
    "to exact corrected XPT numeric values."
)


# ============================================================
# 32. MISSING VALUE FIDELITY
# ============================================================

missing_mismatches = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )


    exported_missing = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )


    missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            exported_missing
        )
    )


print("\n--- MISSING VALUE FIDELITY ---")


print(
    "Missing-position mismatches:",
    missing_mismatches
)


if missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed."
    )


print(
    "PASS: Missing values preserved exactly."
)


# ============================================================
# 33. READ RAW CSV AS TEXT
# ============================================================

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


# ============================================================
# 34. FINAL TINY-VALUE CHECK
# ============================================================

tiny_csv_count = 0


for col in EXPECTED_COLUMNS:

    numeric_values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    tiny_csv_count += int(
        numeric_values
        .eq(TINY_VALUE)
        .sum()
    )


print("\n--- FINAL TINY VALUE CHECK ---")


print(
    "Tiny values remaining:",
    tiny_csv_count
)


if tiny_csv_count != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain in final CSV."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 35. .0 CHECK - ONLY WHOLE-NUMBER FIELDS
# ============================================================

#
# VERY IMPORTANT:
#
# DO NOT globally search the complete DIQ_G CSV
# for ".0".
#
# DIQ280 is a genuine decimal field and legitimate
# values can include:
#
#     5.0
#     6.0
#     7.0
#     36.0
#
# Also its special codes are stored in the same
# float column.
#
# Check .0 ONLY in the 52 proven integer columns.
#

print("\n--- INTEGER COLUMN .0 CHECK ---")


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = raw_csv_df[col]


    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )


    count = int(
        bad_mask.sum()
    )


    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unexpected .0 formatting in "
        "whole-number fields:\n"
        + str(dot_zero_errors)
    )


print(
    "PASS: No unnecessary .0 in the "
    "52 whole-number variables."
)


# ============================================================
# 36. DIQ280 .0 VALUES ARE ALLOWED
# ============================================================

#
# Informational only.
#
# DIQ280 is kept as the released decimal field.
#
# During testing of this attached XPT, pandas'
# normal CSV serialization produces 308 populated
# DIQ280 tokens ending in ".0".
#
# That is NOT a fidelity problem.
#
# Do not fail based on this count because textual
# formatting can vary slightly by pandas version.
#

diq280_dot_zero_count = int(
    (
        raw_csv_df["DIQ280"].ne("")
        &
        raw_csv_df["DIQ280"]
        .str.endswith(".0")
    )
    .sum()
)


print(
    "\nDIQ280 values ending in legitimate '.0':",
    diq280_dot_zero_count
)


print(
    "PASS: DIQ280 decimal representation was "
    "not globally stripped."
)


# ============================================================
# 37. EXACT DIQ280 DECIMAL FIDELITY
# ============================================================

expected_diq280 = (
    expected_df["DIQ280"]
    .to_numpy(
        dtype=float
    )
)


exported_diq280 = (
    roundtrip_df["DIQ280"]
    .to_numpy(
        dtype=float
    )
)


if not np.allclose(
    expected_diq280,
    exported_diq280,
    rtol=0,
    atol=0,
    equal_nan=True
):

    raise ValueError(
        "DIQ280 decimal fidelity failed."
    )


print(
    "PASS: DIQ280 numeric values preserved exactly."
)


# ============================================================
# 38. SCIENTIFIC NOTATION CHECK
# ============================================================

with open(
    csv_file,
    "r",
    encoding="utf-8"
) as f:

    csv_text = f.read()


scientific_count = (
    csv_text.lower().count("e-")
    +
    csv_text.lower().count("e+")
)


print(
    "\nScientific notation occurrences:",
    scientific_count
)


if scientific_count != 0:

    raise ValueError(
        "Unexpected scientific notation found."
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 39. FINAL SEQN CHECK
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if not (
    final_seqn.min() == 62161
    and
    final_seqn.max() == 71916
    and
    int(
        final_seqn.duplicated()
        .sum()
    ) == 0
):

    raise ValueError(
        "Final CSV SEQN validation failed."
    )


if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "Final CSV SEQN still contains .0."
    )


print(
    "PASS: Final CSV SEQN valid "
    "and contains no .0."
)


# ============================================================
# 40. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 80
)

print(
    "DIQ_G CONVERSION COMPLETE"
)

print(
    "=" * 80
)


print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)

print(
    "Original tiny-value artifacts:",
    f"{tiny_total:,}"
)

print(
    "Variables containing tiny artifacts:",
    len(EXPECTED_TINY_COUNTS)
)

print(
    "DID250 zero restorations:",
    EXPECTED_TINY_COUNTS["DID250"]
)

print(
    "DID260 zero restorations:",
    EXPECTED_TINY_COUNTS["DID260"]
)

print(
    "DID341 zero restorations:",
    EXPECTED_TINY_COUNTS["DID341"]
)

print(
    "DID350 zero restorations:",
    EXPECTED_TINY_COUNTS["DID350"]
)

print(
    "Total validated zero restorations:",
    f"{correction_total:,}"
)

print(
    "Genuine decimal variables:",
    DECIMAL_COLUMNS
)

print(
    "DIQ280 fractional observations:",
    diq280_fractional_count
)

print(
    "Whole-number variables:",
    len(INTEGER_COLUMNS)
)

print(
    "XPT -> CSV unexpected numeric differences: "
    "0 expected"
)

print(
    "Missing-position differences: "
    "0 expected"
)

print(
    "Tiny values remaining: "
    "0 expected"
)

print(
    "Unnecessary .0 in whole-number fields: "
    "0 expected"
)

print(
    "DIQ280 decimal values altered: "
    "0 expected"
)

print(
    "Scientific notation: "
    "0 expected"
)
```
