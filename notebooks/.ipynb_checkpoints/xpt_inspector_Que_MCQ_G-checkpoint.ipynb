```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2011_2012"
    / "Questionnnair"
)

xpt_file = DATA_DIR / "MCQ_G.xpt"
csv_file = DATA_DIR / "MCQ_G_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Preserve untouched pandas-decoded XPT.
#
source_df = df.copy()


print("=" * 80)
print("NHANES 2011-2012 MCQ_G XPT -> CSV")
print("=" * 80)

print("Rows:", f"{len(df):,}")
print("Columns:", len(df.columns))


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "MCQ010",
    "MCQ025",
    "MCQ035",
    "MCQ040",
    "MCQ050",
    "MCQ051",
    "MCQ053",
    "MCQ070",
    "MCQ075",
    "MCQ080",
    "MCQ082",
    "MCQ084",
    "MCQ086",
    "MCQ092",
    "MCD093",
    "MCQ140",
    "MCQ149",
    "MCQ160A",
    "MCQ180A",
    "MCQ195",
    "MCQ160N",
    "MCQ180N",
    "MCQ160B",
    "MCQ180B",
    "MCQ160C",
    "MCQ180C",
    "MCQ160D",
    "MCQ180D",
    "MCQ160E",
    "MCQ180E",
    "MCQ160F",
    "MCQ180F",
    "MCQ160G",
    "MCQ180G",
    "MCQ160M",
    "MCQ170M",
    "MCQ180M",
    "MCQ160K",
    "MCQ170K",
    "MCQ180K",
    "MCQ160L",
    "MCQ170L",
    "MCQ180L",
    "MCQ220",
    "MCQ230A",
    "MCQ230B",
    "MCQ230C",
    "MCQ230D",
    "MCQ240A",
    "MCQ240AA",
    "MCQ240B",
    "MCQ240BB",
    "MCQ240C",
    "MCQ240CC",
    "MCQ240D",
    "MCQ240DD",
    "MCQ240DK",
    "MCQ240E",
    "MCQ240F",
    "MCQ240G",
    "MCQ240H",
    "MCQ240I",
    "MCQ240J",
    "MCQ240K",
    "MCQ240L",
    "MCQ240M",
    "MCQ240N",
    "MCQ240O",
    "MCQ240P",
    "MCQ240Q",
    "MCQ240R",
    "MCQ240S",
    "MCQ240T",
    "MCQ240U",
    "MCQ240V",
    "MCQ240W",
    "MCQ240X",
    "MCQ240Y",
    "MCQ240Z",
    "MCQ300A",
    "MCQ300B",
    "MCQ300C",
    "MCQ365A",
    "MCQ365B",
    "MCQ365C",
    "MCQ365D",
    "MCQ370A",
    "MCQ370B",
    "MCQ370C",
    "MCQ370D",
    "MCQ380"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "MCQ_G column names/order differ from expected.\n"
        f"Expected: {EXPECTED_COLUMNS}\n"
        f"Actual:   {df.columns.tolist()}"
    )


if len(df) != 9364:

    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 92:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 9,364 rows x 92 columns."
)


# ============================================================
# 4. CHECK KNOWN XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known pandas / SAS XPORT zero-decoding artifact:
#
#     5.397605346934028e-79
#
# NEVER globally replace this value before validating
# the exact variable/count pattern.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


numeric_cols = (
    df.select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)


tiny_counts = (
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


true_zero_counts = (
    df[numeric_cols]
    .eq(0)
    .sum()
)


true_zero_total = int(
    true_zero_counts.sum()
)


print("\n--- ORIGINAL XPT ZERO CHECK ---")

print(
    "Ordinary numeric zeros before correction:",
    f"{true_zero_total:,}"
)

print(
    "Tiny XPORT artifacts:",
    f"{tiny_total:,}"
)


print("\nTiny values by variable:")

print(
    tiny_counts[
        tiny_counts > 0
    ]
    .sort_values(
        ascending=False
    )
    .to_string()
)


# ============================================================
# 5. VALIDATE EXACT TINY-ZERO PATTERN
# ============================================================

#
# Attached MCQ_G.xpt:
#
# MCQ380  = 1,012
# MCQ180A =     1
# MCQ180B =     1
# MCQ180F =     1
# MCQ180K =     1
#
# TOTAL = 1,016
#
#
# CDC:
#
# MCQ380
#     0 = Never
#     count = 1,012
#
#
# CDC age ranges include zero for:
#
# MCQ180A
#     0-79
#
# MCQ180B
#     0-79
#
# MCQ180F
#     0-79
#
# MCQ180K
#     0-79
#

EXPECTED_TINY_COUNTS = {
    "MCQ180A": 1,
    "MCQ180B": 1,
    "MCQ180F": 1,
    "MCQ180K": 1,
    "MCQ380": 1012
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: Tiny-value pattern differs from "
        "validated MCQ_G pattern.\n"
        f"Expected: {EXPECTED_TINY_COUNTS}\n"
        f"Actual:   {actual_tiny_counts}"
    )


if tiny_total != 1016:

    raise ValueError(
        f"Unexpected total tiny-value count: "
        f"{tiny_total:,}"
    )


if true_zero_total != 0:

    raise ValueError(
        "Unexpected ordinary zero values existed "
        "before correction."
    )


print(
    "\nPASS: Exact 1,016-value tiny-zero "
    "pattern validated."
)


# ============================================================
# 6. RESTORE ONLY VALIDATED ZEROS
# ============================================================

#
# DO NOT:
#
#     df.replace(TINY_VALUE, 0)
#
# globally.
#
# Change only the five validated variables.
#

print("\n--- VALIDATED ZERO RESTORATION ---")


correction_total = 0


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    mask = (
        df[col]
        .eq(TINY_VALUE)
    )


    actual_count = int(
        mask.sum()
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: expected "
            f"{expected_count:,} tiny values, "
            f"found {actual_count:,}."
        )


    df.loc[
        mask,
        col
    ] = 0


    correction_total += actual_count


    print(
        f"{col}: "
        f"{actual_count:,} restored to 0"
    )


if correction_total != 1016:

    raise ValueError(
        "Incorrect total zero-restoration count."
    )


print(
    "\nPASS:",
    f"{correction_total:,}",
    "validated zero values restored."
)


# ============================================================
# 7. CREATE EXPECTED CORRECTED SOURCE COPY
# ============================================================

#
# expected_df represents exactly what the corrected
# source should contain.
#
# ONLY the validated tiny -> zero locations change.
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col]
        .eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 8. CONFIRM NO TINY VALUES REMAIN
# ============================================================

remaining_tiny = int(
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print(
    "\nTiny values remaining after correction:",
    remaining_tiny
)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain "
        "after correction."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 9. CHECK FOR GENUINE DECIMAL VARIABLES
# ============================================================

#
# IMPORTANT:
#
# Always do this BEFORE converting to Int64.
#
# Attached MCQ_G:
#
#     genuine fractional variables = NONE
#
# All released values are identifiers, codes,
# ages, or whole-number questionnaire responses.
#
# If a future/local copy unexpectedly contains
# fractional values, STOP instead of rounding.
#

print("\n--- FRACTIONAL VALUE CHECK ---")


fractional_columns = []


for col in numeric_cols:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        continue


    fractional_mask = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    fractional_count = int(
        fractional_mask.sum()
    )


    if fractional_count > 0:

        fractional_columns.append(
            col
        )

        print(
            f"{col}: "
            f"{fractional_count:,} fractional observations"
        )


if fractional_columns:

    raise ValueError(
        "STOP: Genuine decimal values unexpectedly "
        "found in MCQ_G. They will NOT be rounded:\n"
        + str(fractional_columns)
    )


print(
    "PASS: No genuine fractional variables "
    "in MCQ_G."
)


# ============================================================
# 10. CLEAN AND VALIDATE SEQN
# ============================================================

seqn = (
    df["SEQN"]
    .dropna()
)


if not np.all(
    np.isclose(
        seqn,
        np.round(seqn),
        rtol=0,
        atol=1e-12
    )
):

    raise ValueError(
        "SEQN contains unexpected fractional values."
    )


df["SEQN"] = (
    df["SEQN"]
    .astype("Int64")
)


print("\n--- SEQN ---")

print(
    "Minimum:",
    df["SEQN"].min()
)

print(
    "Maximum:",
    df["SEQN"].max()
)

print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)

print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if not (
    df["SEQN"].min() == 62161
    and
    df["SEQN"].max() == 71916
    and
    int(
        df["SEQN"]
        .isna()
        .sum()
    ) == 0
    and
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    ) == 0
):

    raise ValueError(
        "SEQN validation failed."
    )


print(
    "PASS: SEQN validated."
)


# ============================================================
# 11. FORMAT ALL PROVEN WHOLE-NUMBER VARIABLES
# ============================================================

#
# MCQ_G has NO genuine decimal variables.
#
# All 92 fields can therefore become nullable Int64.
#
# IMPORTANT:
#
# NO .round() is used for conversion.
#

INTEGER_COLUMNS = EXPECTED_COLUMNS.copy()


for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) > 0:

        whole_mask = np.isclose(
            values,
            np.round(values),
            rtol=0,
            atol=1e-12
        )


        if not whole_mask.all():

            raise ValueError(
                f"{col} contains genuine decimal "
                "values and cannot safely become Int64."
            )


    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: All 92 MCQ_G variables converted "
    "to nullable Int64 without rounding."
)


# ============================================================
# 12. VALIDATE RESTORED ZERO COUNTS
# ============================================================

print("\n--- VALIDATED ZERO COUNTS ---")


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    actual_count = int(
        df[col]
        .eq(0)
        .sum()
    )


    print(
        f"{col}: "
        f"{actual_count:,}"
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: expected "
            f"{expected_count:,} zeros, "
            f"found {actual_count:,}."
        )


print(
    "PASS: All five restored zero counts match."
)


# ============================================================
# 13. CDC CHECK - MCQ180A
# ============================================================

#
# Age when told had arthritis.
#
# CDC:
#
# 0-79    = 1,280
# 80      = 46
# 77777   = 0
# 99999   = 39
# Missing = 7,999
#
# Attached XPT contains exactly one zero after
# tiny-value correction.
#

s = df["MCQ180A"]


if not (
    int(
        s.between(
            0,
            79
        ).sum()
    ) == 1280

    and int(
        s.eq(0).sum()
    ) == 1

    and int(
        s.eq(80).sum()
    ) == 46

    and int(
        s.eq(77777).sum()
    ) == 0

    and int(
        s.eq(99999).sum()
    ) == 39

    and int(
        s.isna().sum()
    ) == 7999
):

    raise ValueError(
        "MCQ180A differs from CDC / "
        "validated zero pattern."
    )


print(
    "PASS: MCQ180A matches CDC."
)


# ============================================================
# 14. CDC CHECK - MCQ180B
# ============================================================

#
# Age when told had congestive heart failure.
#
# CDC:
#
# 0-79    = 168
# 80      = 13
# 99999   = 6
# Missing = 9,177
#

s = df["MCQ180B"]


if not (
    int(
        s.between(
            0,
            79
        ).sum()
    ) == 168

    and int(
        s.eq(0).sum()
    ) == 1

    and int(
        s.eq(80).sum()
    ) == 13

    and int(
        s.eq(77777).sum()
    ) == 0

    and int(
        s.eq(99999).sum()
    ) == 6

    and int(
        s.isna().sum()
    ) == 9177
):

    raise ValueError(
        "MCQ180B differs from CDC / "
        "validated zero pattern."
    )


print(
    "PASS: MCQ180B matches CDC."
)


# ============================================================
# 15. CDC CHECK - MCQ180F
# ============================================================

#
# Age when told had stroke.
#
# CDC:
#
# 0-79    = 202
# 80      = 22
# 99999   = 5
# Missing = 9,135
#

s = df["MCQ180F"]


if not (
    int(
        s.between(
            0,
            79
        ).sum()
    ) == 202

    and int(
        s.eq(0).sum()
    ) == 1

    and int(
        s.eq(80).sum()
    ) == 22

    and int(
        s.eq(77777).sum()
    ) == 0

    and int(
        s.eq(99999).sum()
    ) == 5

    and int(
        s.isna().sum()
    ) == 9135
):

    raise ValueError(
        "MCQ180F differs from CDC / "
        "validated zero pattern."
    )


print(
    "PASS: MCQ180F matches CDC."
)


# ============================================================
# 16. CDC CHECK - MCQ180K
# ============================================================

#
# Age when told had chronic bronchitis.
#
# CDC:
#
# 0-79    = 282
# 80      = 0
# 99999   = 15
# Missing = 9,067
#

s = df["MCQ180K"]


if not (
    int(
        s.between(
            0,
            79
        ).sum()
    ) == 282

    and int(
        s.eq(0).sum()
    ) == 1

    and int(
        s.eq(80).sum()
    ) == 0

    and int(
        s.eq(77777).sum()
    ) == 0

    and int(
        s.eq(99999).sum()
    ) == 15

    and int(
        s.isna().sum()
    ) == 9067
):

    raise ValueError(
        "MCQ180K differs from CDC / "
        "validated zero pattern."
    )


print(
    "PASS: MCQ180K matches CDC."
)


# ============================================================
# 17. HELPER - CDC FREQUENCY CHECK
# ============================================================

def check_frequency(
    column,
    expected_values,
    expected_missing
):

    actual_values = (
        df[column]
        .value_counts()
        .sort_index()
        .to_dict()
    )


    actual_missing = int(
        df[column]
        .isna()
        .sum()
    )


    if actual_values != expected_values:

        raise ValueError(
            f"{column} frequencies differ from CDC.\n"
            f"Expected: {expected_values}\n"
            f"Actual:   {actual_values}"
        )


    if actual_missing != expected_missing:

        raise ValueError(
            f"{column} missing count differs from CDC.\n"
            f"Expected: {expected_missing}\n"
            f"Actual:   {actual_missing}"
        )


    print(
        f"PASS: {column} matches CDC."
    )


# ============================================================
# 18. CDC CHECK - MCQ010
# ============================================================

#
# Ever told had asthma
#
# CDC:
#
# 1 Yes          1,448
# 2 No           7,906
# 7 Refused          1
# 9 Don't know       8
# Missing            1
#

check_frequency(
    "MCQ010",
    {
        1: 1448,
        2: 7906,
        7: 1,
        9: 8
    },
    1
)


# ============================================================
# 19. CDC CHECK - MCQ160A
# ============================================================

#
# Doctor ever said participant had arthritis
#

check_frequency(
    "MCQ160A",
    {
        1: 1365,
        2: 4185,
        7: 1,
        9: 7
    },
    3806
)


# ============================================================
# 20. CDC CHECK - MCQ160B
# ============================================================

#
# Congestive heart failure
#

check_frequency(
    "MCQ160B",
    {
        1: 187,
        2: 5357,
        7: 1,
        9: 14
    },
    3805
)


# ============================================================
# 21. CDC CHECK - MCQ160F
# ============================================================

#
# Ever told participant had stroke
#

check_frequency(
    "MCQ160F",
    {
        1: 229,
        2: 5326,
        7: 1,
        9: 3
    },
    3805
)


# ============================================================
# 22. CDC CHECK - MCQ160K
# ============================================================

#
# Ever told participant had chronic bronchitis
#

check_frequency(
    "MCQ160K",
    {
        1: 297,
        2: 5253,
        7: 1,
        9: 8
    },
    3805
)


# ============================================================
# 23. CDC CHECK - MCQ220
# ============================================================

#
# Ever told participant had cancer/malignancy.
#
# Important variable for later NHANES analysis.
#
# CDC:
#
# 1 Yes          488
# 2 No         5,066
# 7 Refused        1
# 9 Don't know     4
# Missing      3,805
#

check_frequency(
    "MCQ220",
    {
        1: 488,
        2: 5066,
        7: 1,
        9: 4
    },
    3805
)


# ============================================================
# 24. CDC CHECK - MCQ300C
# ============================================================

#
# Close biological relative had diabetes
#

check_frequency(
    "MCQ300C",
    {
        1: 2191,
        2: 3265,
        7: 1,
        9: 102
    },
    3805
)


# ============================================================
# 25. CDC CHECK - MCQ365A
# ============================================================

#
# Doctor told participant to control/lose weight
#
# New in 2011-2012.
#

check_frequency(
    "MCQ365A",
    {
        1: 1537,
        2: 4636,
        7: 1
    },
    3190
)


# ============================================================
# 26. CDC CHECK - MCQ370A
# ============================================================

#
# Currently controlling/losing weight
#

check_frequency(
    "MCQ370A",
    {
        1: 3643,
        2: 2529,
        7: 1,
        9: 1
    },
    3190
)


# ============================================================
# 27. CDC CHECK - MCQ380
# ============================================================

#
# Past 7 days, trouble remembering where things
# were placed.
#
# CDC:
#
# 0 Never                 1,012
# 1 About once              364
# 2 Two or three times      265
# 3 Nearly every day         91
# 4 Several times/day        55
# 7 Refused                   1
# 9 Don't know                3
# Missing                 7,573
#
#
# Attached XPT originally decoded the 1,012 zeros
# as TINY_VALUE.
#

check_frequency(
    "MCQ380",
    {
        0: 1012,
        1: 364,
        2: 265,
        3: 91,
        4: 55,
        7: 1,
        9: 3
    },
    7573
)


if int(
    df["MCQ380"]
    .eq(0)
    .sum()
) != 1012:

    raise ValueError(
        "MCQ380 validated zero count changed."
    )


print(
    "PASS: MCQ380 has exactly "
    "1,012 validated zeros."
)


# ============================================================
# 28. AGE-QUESTION ROUTING CHECKS
# ============================================================

#
# The four tiny age values occur only where the
# corresponding condition question is Yes.
#
# This gives another useful consistency check.
#

ROUTING_CHECKS = {
    "MCQ180A": "MCQ160A",
    "MCQ180B": "MCQ160B",
    "MCQ180F": "MCQ160F",
    "MCQ180K": "MCQ160K"
}


print("\n--- AGE ROUTING CHECK ---")


for age_col, condition_col in ROUTING_CHECKS.items():

    zero_rows = (
        df[age_col]
        .eq(0)
    )


    invalid = int(
        (
            zero_rows
            &
            ~df[condition_col].eq(1)
        )
        .sum()
    )


    if invalid != 0:

        raise ValueError(
            f"{age_col}=0 occurs outside "
            f"{condition_col}=1."
        )


    print(
        f"PASS: {age_col}=0 occurs with "
        f"{condition_col}=1."
    )


# ============================================================
# 29. VALUE FIDELITY BEFORE EXPORT
# ============================================================

#
# The ONLY allowed numeric differences from raw
# pandas-decoded XPT are:
#
# MCQ180A: 1 tiny -> 0
# MCQ180B: 1 tiny -> 0
# MCQ180F: 1 tiny -> 0
# MCQ180K: 1 tiny -> 0
# MCQ380 : 1,012 tiny -> 0
#
# NOTHING else may change.
#

print("\n--- VALUE FIDELITY BEFORE EXPORT ---")


fidelity_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    current_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )


    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        fidelity_failures.append(
            col
        )


if fidelity_failures:

    raise ValueError(
        "Unexpected numeric changes before export:\n"
        + str(fidelity_failures)
    )


print(
    "PASS: All 92 variables match the expected "
    "corrected XPT exactly."
)


# ============================================================
# 30. EXPORT CSV
# ============================================================

#
# No rounding.
#
# No:
#
#     float_format=
#
# Missing values become blank.
#
# Even though MCQ_G has no real decimals, keeping the
# standard export pattern avoids creating a different
# workflow from your Dietary/Lab/Exam files.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print("\nCSV created:")

print(
    csv_file
)


# ============================================================
# 31. RE-READ CSV WITH ROUND-TRIP FLOAT PARSER
# ============================================================

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != df.shape:

    raise ValueError(
        "CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV column names/order changed."
    )


print(
    "PASS: CSV structure preserved."
)


# ============================================================
# 32. EXACT CORRECTED XPT -> CSV NUMERIC FIDELITY
# ============================================================

print("\n--- EXACT CORRECTED XPT -> CSV FIDELITY ---")


csv_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "Corrected XPT -> CSV differences found:\n"
        + str(csv_failures)
    )


print(
    "PASS: All 92 CSV variables round-trip "
    "to exact corrected XPT numeric values."
)


# ============================================================
# 33. MISSING VALUE FIDELITY
# ============================================================

missing_mismatches = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )


    exported_missing = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )


    missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            exported_missing
        )
    )


print("\n--- MISSING VALUE FIDELITY ---")


print(
    "Missing-position mismatches:",
    missing_mismatches
)


if missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed."
    )


print(
    "PASS: Missing values preserved exactly."
)


# ============================================================
# 34. READ RAW CSV AS TEXT
# ============================================================

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


# ============================================================
# 35. FINAL TINY-VALUE CHECK
# ============================================================

tiny_csv_count = 0


for col in EXPECTED_COLUMNS:

    numeric_values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    tiny_csv_count += int(
        numeric_values
        .eq(TINY_VALUE)
        .sum()
    )


print("\n--- FINAL TINY VALUE CHECK ---")


print(
    "Tiny values remaining:",
    tiny_csv_count
)


if tiny_csv_count != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain "
        "in final CSV."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 36. RAW CSV .0 FORMAT CHECK
# ============================================================

#
# MCQ_G contains NO genuine decimal variables.
#
# Therefore every populated variable in THIS FILE
# should be represented without unnecessary ".0".
#
# Examples:
#
#     62161.0 -> 62161
#     1.0     -> 1
#     80.0    -> 80
#     99999.0 -> 99999
#
# This global check is safe only because MCQ_G has
# zero genuine decimal variables.
#
# DO NOT copy this global rule to Dietary or any
# Lab/Exam/Questionnaire dataset containing genuine
# continuous decimals.
#

print("\n--- RAW CSV .0 CHECK ---")


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = raw_csv_df[col]


    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )


    count = int(
        bad_mask.sum()
    )


    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unexpected .0 formatting found:\n"
        + str(dot_zero_errors)
    )


print(
    "PASS: No unnecessary .0 formatting "
    "in any MCQ_G variable."
)


# ============================================================
# 37. FINAL ZERO-FIDELITY CHECK
# ============================================================

#
# Because raw pandas produced no ordinary zero values,
# the final CSV should contain exactly:
#
# MCQ180A = 1
# MCQ180B = 1
# MCQ180F = 1
# MCQ180K = 1
# MCQ380  = 1012
#
# and no zero in any other column.
#

print("\n--- FINAL ZERO-FIDELITY CHECK ---")


final_zero_counts = {}


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    count = int(
        values.eq(0).sum()
    )


    if count > 0:

        final_zero_counts[col] = count


if final_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final zero pattern differs from expected.\n"
        f"Expected: {EXPECTED_TINY_COUNTS}\n"
        f"Actual:   {final_zero_counts}"
    )


print(
    "PASS: Final zero pattern exactly matches "
    "the validated XPORT corrections."
)


# ============================================================
# 38. SCIENTIFIC NOTATION CHECK
# ============================================================

with open(
    csv_file,
    "r",
    encoding="utf-8"
) as f:

    csv_text = f.read()


scientific_count = (
    csv_text.lower().count("e-")
    +
    csv_text.lower().count("e+")
)


print(
    "\nScientific notation occurrences:",
    scientific_count
)


if scientific_count != 0:

    raise ValueError(
        "Unexpected scientific notation found."
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 39. FINAL SEQN CHECK
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if not (
    final_seqn.min() == 62161
    and
    final_seqn.max() == 71916
    and
    int(
        final_seqn.duplicated()
        .sum()
    ) == 0
):

    raise ValueError(
        "Final CSV SEQN validation failed."
    )


if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "Final CSV SEQN still contains .0."
    )


print(
    "PASS: Final CSV SEQN valid "
    "and contains no .0."
)


# ============================================================
# 40. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 80
)


print(
    "MCQ_G CONVERSION COMPLETE"
)


print(
    "=" * 80
)


print(
    "Rows:",
    f"{len(df):,}"
)


print(
    "Columns:",
    len(df.columns)
)


print(
    "Original tiny-value artifacts:",
    f"{tiny_total:,}"
)


print(
    "Variables containing tiny artifacts:",
    len(EXPECTED_TINY_COUNTS)
)


print(
    "MCQ180A zero restorations:",
    EXPECTED_TINY_COUNTS["MCQ180A"]
)


print(
    "MCQ180B zero restorations:",
    EXPECTED_TINY_COUNTS["MCQ180B"]
)


print(
    "MCQ180F zero restorations:",
    EXPECTED_TINY_COUNTS["MCQ180F"]
)


print(
    "MCQ180K zero restorations:",
    EXPECTED_TINY_COUNTS["MCQ180K"]
)


print(
    "MCQ380 zero restorations:",
    EXPECTED_TINY_COUNTS["MCQ380"]
)


print(
    "Total validated zero restorations:",
    f"{correction_total:,}"
)


print(
    "Genuine decimal variables: NONE"
)


print(
    "Whole-number variables:",
    len(INTEGER_COLUMNS)
)


print(
    "XPT -> CSV unexpected numeric differences: "
    "0 expected"
)


print(
    "Missing-position differences: "
    "0 expected"
)


print(
    "Tiny values remaining: "
    "0 expected"
)


print(
    "Unnecessary .0: "
    "0 expected"
)


print(
    "Scientific notation: "
    "0 expected"
)
```
