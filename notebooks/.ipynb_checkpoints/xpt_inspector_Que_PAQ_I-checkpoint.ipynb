```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

#
# Change "Questionnnair" -> "Questionnaire"
# if your actual folder uses the correct spelling.
#
DATA_DIR = (
    BASE_DIR
    / "data"
    / "2015_2016"
    / "Questionnnair"
)

xpt_file = DATA_DIR / "PAQ_I.xpt"
csv_file = DATA_DIR / "PAQ_I_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Preserve completely untouched pandas-decoded XPT.
#
source_df = df.copy()


print("=" * 100)
print("NHANES 2015-2016 PAQ_I XPT -> CSV")
print("=" * 100)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "PAQ605",
    "PAQ610",
    "PAD615",
    "PAQ620",
    "PAQ625",
    "PAD630",
    "PAQ635",
    "PAQ640",
    "PAD645",
    "PAQ650",
    "PAQ655",
    "PAD660",
    "PAQ665",
    "PAQ670",
    "PAD675",
    "PAD680",
    "PAQ706",
    "PAQ710",
    "PAQ715",
    "PAQ722",
    "PAQ724A",
    "PAQ724B",
    "PAQ724C",
    "PAQ724D",
    "PAQ724E",
    "PAQ724F",
    "PAQ724G",
    "PAQ724H",
    "PAQ724I",
    "PAQ724J",
    "PAQ724K",
    "PAQ724L",
    "PAQ724M",
    "PAQ724N",
    "PAQ724O",
    "PAQ724P",
    "PAQ724Q",
    "PAQ724R",
    "PAQ724S",
    "PAQ724T",
    "PAQ724U",
    "PAQ724V",
    "PAQ724W",
    "PAQ724X",
    "PAQ724Y",
    "PAQ724Z",
    "PAQ724AA",
    "PAQ724AB",
    "PAQ724AC",
    "PAQ724AD",
    "PAQ724AE",
    "PAQ724AF",
    "PAQ724CM",
    "PAQ731",
    "PAD733",
    "PAQ677",
    "PAQ678",
    "PAQ740",
    "PAQ742",
    "PAQ744",
    "PAQ746",
    "PAQ748",
    "PAQ755",
    "PAQ759A",
    "PAQ759B",
    "PAQ759C",
    "PAQ759D",
    "PAQ759E",
    "PAQ759F",
    "PAQ759G",
    "PAQ759H",
    "PAQ759I",
    "PAQ759J",
    "PAQ759K",
    "PAQ759L",
    "PAQ759M",
    "PAQ759N",
    "PAQ759O",
    "PAQ759P",
    "PAQ759Q",
    "PAQ759R",
    "PAQ759S",
    "PAQ759T",
    "PAQ759U",
    "PAQ762",
    "PAQ764",
    "PAQ766",
    "PAQ679",
    "PAQ750",
    "PAQ770",
    "PAQ772A",
    "PAQ772B",
    "PAQ772C"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "PAQ_I columns/order differ.\n\n"
        f"Expected:\n{EXPECTED_COLUMNS}\n\n"
        f"Actual:\n{df.columns.tolist()}"
    )


if len(df) != 9255:

    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 94:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 9,255 rows x 94 columns."
)


# ============================================================
# 4. FIELD TYPE PLAN
# ============================================================

#
# Direct inspection of attached PAQ_I:
#
# After restoring validated zero artifacts:
#
#     genuine decimal variables = NONE
#     string variables          = NONE
#
#
# All 94 fields contain only whole-number values:
#
#     IDs
#     categorical codes
#     number of days
#     number of minutes
#     number of hours/categories
#     multi-response activity codes
#
#
# Therefore all 94 can safely become nullable Int64
# AFTER the validated tiny-zero repair.
#
#
# IMPORTANT:
#
# This conclusion applies ONLY to PAQ_I.
#
# It does NOT change your broader NHANES rule:
#
#     Dietary decimals
#     laboratory continuous measurements
#     examination continuous measurements
#     sample weights
#
# must remain exactly AS IS.
#

DECIMAL_COLUMNS = []

STRING_COLUMNS = []

INTEGER_COLUMNS = EXPECTED_COLUMNS.copy()


if len(INTEGER_COLUMNS) != 94:

    raise ValueError(
        "Expected exactly 94 whole-number/code fields."
    )


print(
    "\n--- FIELD TYPE PLAN ---"
)

print(
    "Genuine decimal fields:",
    DECIMAL_COLUMNS
)

print(
    "Whole-number/code fields:",
    len(INTEGER_COLUMNS)
)


# ============================================================
# 5. CHECK KNOWN SAS XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known pandas/SAS XPORT representation of numeric zero:
#
#     5.397605346934028e-79
#
#
# NEVER globally replace this value before checking
# the exact variables and counts.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


numeric_cols = (
    df.select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)


if numeric_cols != EXPECTED_COLUMNS:

    raise ValueError(
        "Expected all 94 PAQ_I variables "
        "to load numerically."
    )


tiny_counts = (
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


tiny_variable_count = int(
    tiny_counts.gt(0)
    .sum()
)


true_zero_counts = (
    df[numeric_cols]
    .eq(0)
    .sum()
)


true_zero_total = int(
    true_zero_counts.sum()
)


print(
    "\n--- ORIGINAL XPT ZERO CHECK ---"
)

print(
    "Ordinary numeric zeros before correction:",
    f"{true_zero_total:,}"
)

print(
    "Tiny XPORT artifacts:",
    f"{tiny_total:,}"
)

print(
    "Variables containing tiny artifacts:",
    tiny_variable_count
)


print(
    "\nTiny values by variable:"
)

print(
    tiny_counts[
        tiny_counts > 0
    ]
    .sort_values(
        ascending=False
    )
    .to_string()
)


# ============================================================
# 6. VALIDATE EXACT TINY-ZERO PATTERN
# ============================================================

#
# Exact attached PAQ_I.xpt:
#
#     PAD680        2
#     PAQ706      163
#     PAQ710    1,173
#     PAQ715    2,001
#     PAQ731    1,535
#     PAQ677       85
#     PAQ678      165
#
# TOTAL        5,124
#
#
# CDC independently supports:
#
# PAQ706 = 0:
#     163 participants
#
# PAQ710 = 0:
#     1,173 participants
#
# PAQ715 = 0:
#     2,001 participants
#
# PAQ731 = 0:
#     1,535 participants
#
# PAQ677 = 0:
#     85 participants
#
# PAQ678 = 0:
#     165 participants
#
# PAD680:
#     valid range includes 0 minutes
#
#

EXPECTED_TINY_COUNTS = {
    "PAD680": 2,
    "PAQ706": 163,
    "PAQ710": 1173,
    "PAQ715": 2001,
    "PAQ731": 1535,
    "PAQ677": 85,
    "PAQ678": 165
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: Tiny-value pattern differs from "
        "validated PAQ_I attachment.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{actual_tiny_counts}"
    )


if tiny_total != 5124:

    raise ValueError(
        f"Unexpected total tiny-value count: "
        f"{tiny_total:,}"
    )


if tiny_variable_count != 7:

    raise ValueError(
        "Expected tiny values in exactly 7 variables."
    )


if true_zero_total != 0:

    raise ValueError(
        "Unexpected ordinary zeros existed "
        "before tiny-value correction."
    )


print(
    "\nPASS: Exact 5,124-value tiny-zero pattern "
    "across 7 variables validated."
)


# ============================================================
# 7. RESTORE ONLY VALIDATED ZEROS
# ============================================================

#
# DO NOT:
#
#     df.replace(TINY_VALUE, 0)
#
# globally.
#

print(
    "\n--- VALIDATED ZERO RESTORATION ---"
)


correction_total = 0


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    mask = (
        df[col]
        .eq(TINY_VALUE)
    )


    actual_count = int(
        mask.sum()
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: expected "
            f"{expected_count:,} tiny values, "
            f"found {actual_count:,}."
        )


    df.loc[
        mask,
        col
    ] = 0


    correction_total += actual_count


    print(
        f"{col}: "
        f"{actual_count:,} restored to 0"
    )


if correction_total != 5124:

    raise ValueError(
        "Incorrect total zero-restoration count."
    )


print(
    "\nPASS:",
    f"{correction_total:,}",
    "validated zeros restored."
)


# ============================================================
# 8. CREATE EXPECTED CORRECTED SOURCE COPY
# ============================================================

#
# expected_df becomes the strict numerical fidelity
# reference for all later checks.
#
# The ONLY permitted changes from the original XPT are:
#
#     validated tiny values -> real zero
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col]
        .eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 9. CONFIRM NO TINY VALUES REMAIN
# ============================================================

remaining_tiny = int(
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print(
    "\nTiny values remaining:",
    remaining_tiny
)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain after correction."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 10. DETECT GENUINE DECIMAL VARIABLES
# ============================================================

#
# This step MUST occur before Int64 conversion.
#
# np.round() is used only to DETECT fractional
# observations.
#
# It never overwrites the dataframe.
#

print(
    "\n--- GENUINE DECIMAL DETECTION ---"
)


fractional_columns = []

fractional_counts = {}


for col in numeric_cols:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        fractional_counts[col] = 0
        continue


    fractional_mask = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    fractional_count = int(
        fractional_mask.sum()
    )


    fractional_counts[col] = fractional_count


    if fractional_count > 0:

        fractional_columns.append(
            col
        )


print(
    "Variables containing fractional observations:",
    fractional_columns
)


if fractional_columns:

    raise ValueError(
        "STOP: Unexpected genuine decimal fields "
        "found in PAQ_I.\n\n"
        "These values will NOT be rounded:\n"
        + str(
            fractional_columns
        )
    )


print(
    "PASS: No genuine decimal variables "
    "found in PAQ_I."
)


# ============================================================
# 11. CLEAN AND VALIDATE SEQN
# ============================================================

seqn_values = (
    df["SEQN"]
    .dropna()
    .to_numpy(
        dtype=float
    )
)


if not np.isclose(
    seqn_values,
    np.round(seqn_values),
    rtol=0,
    atol=1e-12
).all():

    raise ValueError(
        "SEQN contains unexpected fractional values."
    )


df["SEQN"] = (
    df["SEQN"]
    .astype("Int64")
)


print(
    "\n--- SEQN ---"
)

print(
    "Minimum:",
    df["SEQN"].min()
)

print(
    "Maximum:",
    df["SEQN"].max()
)

print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)

print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if df["SEQN"].min() != 83732:

    raise ValueError(
        "Unexpected minimum SEQN."
    )


if df["SEQN"].max() != 93702:

    raise ValueError(
        "Unexpected maximum SEQN."
    )


if int(
    df["SEQN"]
    .isna()
    .sum()
) != 0:

    raise ValueError(
        "Missing SEQN found."
    )


if int(
    df["SEQN"]
    .duplicated()
    .sum()
) != 0:

    raise ValueError(
        "Duplicate SEQN found."
    )


print(
    "PASS: SEQN 83732-93702 validated."
)


# ============================================================
# 12. PROVE ALL 94 VARIABLES ARE WHOLE-NUMBER
# ============================================================

#
# This is done before Int64 conversion.
#
# NO value is rounded.
#

whole_number_failures = []


for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        continue


    whole_mask = np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    if not whole_mask.all():

        whole_number_failures.append(
            col
        )


if whole_number_failures:

    raise ValueError(
        "Unexpected fractional fields found:\n"
        + str(
            whole_number_failures
        )
    )


print(
    "\nPASS: All 94 PAQ_I variables are "
    "whole-number IDs/codes/counts."
)


# ============================================================
# 13. CONVERT ALL 94 FIELDS TO NULLABLE Int64
# ============================================================

#
# IMPORTANT:
#
# NO:
#
#     .round()
#
# is used during actual conversion.
#

for col in INTEGER_COLUMNS:

    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "PASS: All 94 variables converted "
    "to nullable Int64 without rounding."
)


# ============================================================
# 14. HELPER - EXACT FREQUENCY CHECK
# ============================================================

def check_frequency(
    column,
    expected_values,
    expected_missing
):

    actual_values = (
        df[column]
        .value_counts()
        .sort_index()
        .to_dict()
    )


    actual_missing = int(
        df[column]
        .isna()
        .sum()
    )


    print(
        f"\n--- {column} ---"
    )

    print(
        "Observed:",
        actual_values
    )

    print(
        "Missing:",
        f"{actual_missing:,}"
    )


    if actual_values != expected_values:

        raise ValueError(
            f"{column} frequency distribution differs.\n\n"
            f"Expected:\n{expected_values}\n\n"
            f"Actual:\n{actual_values}"
        )


    if actual_missing != expected_missing:

        raise ValueError(
            f"{column} missing count differs.\n"
            f"Expected: {expected_missing:,}\n"
            f"Actual:   {actual_missing:,}"
        )


    print(
        f"PASS: {column} matches CDC."
    )


# ============================================================
# 15. CDC CHECK - PAQ706
# ============================================================

#
# Days physically active >=60 minutes.
#
# CDC:
#
#     0 = 163
#     1 = 123
#     2 = 162
#     3 = 225
#     4 = 186
#     5 = 394
#     6 = 137
#     7 = 1486
#     77 = 1
#     99 = 8
#
# Missing = 6,370
#

check_frequency(
    "PAQ706",
    {
        0: 163,
        1: 123,
        2: 162,
        3: 225,
        4: 186,
        5: 394,
        6: 137,
        7: 1486,
        77: 1,
        99: 8
    },
    6370
)


# ============================================================
# 16. CDC CHECK - PAQ710
# ============================================================

#
# Average hours/day watching TV/videos.
#
# CDC:
#
#     0 = Less than 1 hour     1,173
#     1 = 1 hour               1,637
#     2 = 2 hours              2,400
#     3 = 3 hours              1,473
#     4 = 4 hours                854
#     5 = 5 hours or more      1,286
#     8 = Does not watch         369
#     77 = Refused                 1
#     99 = Don't know             16
#
# Missing = 46
#

check_frequency(
    "PAQ710",
    {
        0: 1173,
        1: 1637,
        2: 2400,
        3: 1473,
        4: 854,
        5: 1286,
        8: 369,
        77: 1,
        99: 16
    },
    46
)


# ============================================================
# 17. CDC CHECK - PAQ715
# ============================================================

#
# Average computer/game use outside school.
#
# CDC:
#
#     0 = Less than 1 hour       2,001
#     1 = 1 hour                 1,530
#     2 = 2 hours                1,316
#     3 = 3 hours                  574
#     4 = 4 hours                  350
#     5 = 5 hours or more          588
#     8 = Does not use computer  2,844
#     99 = Don't know                6
#
# Missing = 46
#

check_frequency(
    "PAQ715",
    {
        0: 2001,
        1: 1530,
        2: 1316,
        3: 574,
        4: 350,
        5: 588,
        8: 2844,
        99: 6
    },
    46
)


# ============================================================
# 18. CDC CHECK - PAQ731
# ============================================================

#
# Days played active video games.
#

check_frequency(
    "PAQ731",
    {
        0: 1535,
        1: 193,
        2: 166,
        3: 99,
        4: 38,
        5: 32,
        6: 3,
        7: 103,
        99: 3
    },
    7083
)


# ============================================================
# 19. CDC CHECK - PAQ677
# ============================================================

#
# Days cardiovascular exercise in past week.
#

check_frequency(
    "PAQ677",
    {
        0: 85,
        1: 63,
        2: 79,
        3: 79,
        4: 58,
        5: 173,
        6: 35,
        7: 60,
        99: 3
    },
    8620
)


# ============================================================
# 20. CDC CHECK - PAQ678
# ============================================================

#
# Days muscle-strengthening exercise.
#

check_frequency(
    "PAQ678",
    {
        0: 165,
        1: 80,
        2: 101,
        3: 98,
        4: 37,
        5: 105,
        6: 18,
        7: 31
    },
    8620
)


# ============================================================
# 21. CDC / ATTACHMENT CHECK - PAD680
# ============================================================

#
# Minutes sedentary activity.
#
# CDC:
#
#     0-1380 = 6,895
#     7777   = 3
#     9999   = 53
#     Missing = 2,304
#
#
# Direct attachment after restoration:
#
#     PAD680=0 = exactly 2
#
#
# CDC groups 0-1380 together instead of publishing
# the individual zero count, so:
#
#     2 zeros
#
# is an attachment-specific fidelity check.
#

pad680_range = int(
    df["PAD680"]
    .between(
        0,
        1380
    )
    .sum()
)


pad680_zero = int(
    df["PAD680"]
    .eq(0)
    .sum()
)


pad680_7777 = int(
    df["PAD680"]
    .eq(7777)
    .sum()
)


pad680_9999 = int(
    df["PAD680"]
    .eq(9999)
    .sum()
)


pad680_missing = int(
    df["PAD680"]
    .isna()
    .sum()
)


print(
    "\n--- PAD680 ---"
)

print(
    "0-1380:",
    f"{pad680_range:,}"
)

print(
    "Exact attachment zeros:",
    pad680_zero
)

print(
    "7777:",
    pad680_7777
)

print(
    "9999:",
    pad680_9999
)

print(
    "Missing:",
    f"{pad680_missing:,}"
)


if pad680_range != 6895:

    raise ValueError(
        "PAD680 0-1380 count differs."
    )


if pad680_zero != 2:

    raise ValueError(
        "PAD680 exact attachment zero count differs."
    )


if pad680_7777 != 3:

    raise ValueError(
        "PAD680 refused count differs."
    )


if pad680_9999 != 53:

    raise ValueError(
        "PAD680 don't-know count differs."
    )


if pad680_missing != 2304:

    raise ValueError(
        "PAD680 missing count differs."
    )


print(
    "PASS: PAD680 CDC range and exact "
    "attachment zero count validated."
)


# ============================================================
# 22. FINAL VALIDATED ZERO PATTERN
# ============================================================

#
# Original pandas-decoded XPT contained:
#
#     ordinary numeric zeros = 0
#
#
# Therefore every real zero after repair should correspond
# exactly to one of the validated tiny positions.
#

final_zero_counts = {}


for col in EXPECTED_COLUMNS:

    count = int(
        df[col]
        .eq(0)
        .sum()
    )


    if count > 0:

        final_zero_counts[col] = count


print(
    "\n--- FINAL ZERO PATTERN ---"
)

print(
    final_zero_counts
)


if final_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final zero pattern differs.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{final_zero_counts}"
    )


if sum(
    final_zero_counts.values()
) != 5124:

    raise ValueError(
        "Unexpected final zero total."
    )


print(
    "PASS: Final zero pattern exactly matches "
    "all 5,124 validated restorations."
)


# ============================================================
# 23. CREATE FULL COLUMN FIDELITY SNAPSHOT
# ============================================================

#
# Store nonmissing and missing counts for ALL 94 fields
# before export.
#
# This is useful because PAQ_I is a very wide file.
#

expected_nonmissing_counts = {
    col: int(
        df[col]
        .notna()
        .sum()
    )
    for col in EXPECTED_COLUMNS
}


expected_missing_counts = {
    col: int(
        df[col]
        .isna()
        .sum()
    )
    for col in EXPECTED_COLUMNS
}


print(
    "\nPASS: Stored full 94-column "
    "missing/nonmissing fidelity snapshot."
)


# ============================================================
# 24. VALUE FIDELITY BEFORE EXPORT
# ============================================================

#
# ONLY the seven validated tiny->zero repairs are allowed.
#
# Every other numeric value must remain EXACTLY equal
# to the corrected XPT reference.
#

print(
    "\n--- VALUE FIDELITY BEFORE EXPORT ---"
)


pre_export_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    current_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )


    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        pre_export_failures.append(
            col
        )


if pre_export_failures:

    raise ValueError(
        "Unexpected numeric changes before export:\n"
        + str(
            pre_export_failures
        )
    )


print(
    "PASS: All 94 PAQ_I variables match "
    "the exact corrected XPT values."
)


# ============================================================
# 25. EXPORT CSV
# ============================================================

#
# IMPORTANT:
#
# NO:
#
#     df.round(...)
#
# NO:
#
#     float_format=
#
#
# PAQ_I itself contains no true decimals.
#
# But this same export method protects genuine decimal
# values when you use the approach in Dietary/Lab/Exam files.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print(
    "\nCSV created:"
)

print(
    csv_file
)


# ============================================================
# 26. RE-READ CSV WITH ROUND-TRIP FLOAT PARSER
# ============================================================

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != df.shape:

    raise ValueError(
        "CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV columns/order changed."
    )


print(
    "PASS: CSV structure preserved."
)


# ============================================================
# 27. EXACT CORRECTED XPT -> CSV NUMERIC FIDELITY
# ============================================================

print(
    "\n--- EXACT CORRECTED XPT -> CSV FIDELITY ---"
)


csv_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        pd.to_numeric(
            roundtrip_df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "Corrected XPT -> CSV numeric differences:\n"
        + str(
            csv_failures
        )
    )


print(
    "PASS: 0 numeric differences across "
    "all 94 PAQ_I variables."
)


# ============================================================
# 28. MISSING-VALUE FIDELITY
# ============================================================

missing_position_mismatches = 0

missing_count_failures = []


for col in EXPECTED_COLUMNS:

    expected_missing_mask = (
        expected_df[col]
        .isna()
        .to_numpy()
    )


    exported_missing_mask = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )


    missing_position_mismatches += int(
        np.sum(
            expected_missing_mask
            !=
            exported_missing_mask
        )
    )


    actual_missing_count = int(
        roundtrip_df[col]
        .isna()
        .sum()
    )


    if (
        actual_missing_count
        !=
        expected_missing_counts[col]
    ):

        missing_count_failures.append(
            col
        )


print(
    "\n--- MISSING VALUE FIDELITY ---"
)

print(
    "Missing-position mismatches:",
    missing_position_mismatches
)


if missing_position_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed."
    )


if missing_count_failures:

    raise ValueError(
        "Missing counts changed in:\n"
        + str(
            missing_count_failures
        )
    )


print(
    "PASS: Missing positions/counts preserved "
    "exactly across all 94 variables."
)


# ============================================================
# 29. READ FINAL CSV AS RAW TEXT
# ============================================================

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


# ============================================================
# 30. FINAL TINY-VALUE CHECK
# ============================================================

tiny_csv_count = 0


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    tiny_csv_count += int(
        values
        .eq(TINY_VALUE)
        .sum()
    )


print(
    "\n--- FINAL TINY VALUE CHECK ---"
)

print(
    "Tiny values remaining:",
    tiny_csv_count
)


if tiny_csv_count != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain in CSV."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 31. RAW CSV .0 CHECK
# ============================================================

#
# PAQ_I contains NO genuine decimal fields.
#
# Therefore it is safe to verify that none of the
# 94 fields retains unnecessary ".0".
#
#
# IMPORTANT:
#
# Do NOT use this global check on mixed files such as
# Dietary, Lab or Examination datasets that contain
# real continuous decimals.
#

print(
    "\n--- RAW CSV .0 CHECK ---"
)


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = raw_csv_df[col]


    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )


    count = int(
        bad_mask.sum()
    )


    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unexpected '.0' formatting found:\n"
        + str(
            dot_zero_errors
        )
    )


print(
    "PASS: No unnecessary '.0' "
    "in any of the 94 PAQ_I fields."
)


# ============================================================
# 32. FINAL CSV ZERO PATTERN
# ============================================================

csv_zero_counts = {}


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    zero_count = int(
        values
        .eq(0)
        .sum()
    )


    if zero_count > 0:

        csv_zero_counts[col] = zero_count


print(
    "\n--- FINAL CSV ZERO PATTERN ---"
)

print(
    csv_zero_counts
)


if csv_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final CSV zero pattern differs.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{csv_zero_counts}"
    )


print(
    "PASS: Final CSV contains exactly "
    "the 5,124 validated zero values."
)


# ============================================================
# 33. FINAL NONMISSING COUNT CHECK - ALL 94 VARIABLES
# ============================================================

nonmissing_failures = []


for col in EXPECTED_COLUMNS:

    actual_nonmissing = int(
        raw_csv_df[col]
        .ne("")
        .sum()
    )


    if (
        actual_nonmissing
        !=
        expected_nonmissing_counts[col]
    ):

        nonmissing_failures.append(
            {
                "variable": col,
                "expected": expected_nonmissing_counts[col],
                "actual": actual_nonmissing
            }
        )


if nonmissing_failures:

    raise ValueError(
        "Nonmissing counts changed:\n"
        + str(
            nonmissing_failures
        )
    )


print(
    "\nPASS: Nonmissing counts preserved "
    "for all 94 variables."
)


# ============================================================
# 34. FINAL KEY ZERO FREQUENCY CHECKS
# ============================================================

FINAL_EXPECTED_ZEROS = {
    "PAD680": 2,
    "PAQ706": 163,
    "PAQ710": 1173,
    "PAQ715": 2001,
    "PAQ731": 1535,
    "PAQ677": 85,
    "PAQ678": 165
}


print(
    "\n--- FINAL KEY ZERO COUNTS ---"
)


for col, expected_count in FINAL_EXPECTED_ZEROS.items():

    values = pd.to_numeric(
        raw_csv_df[col],
        errors="coerce"
    )


    actual_count = int(
        values
        .eq(0)
        .sum()
    )


    print(
        f"{col}:",
        f"{actual_count:,}"
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: final zero count changed."
        )


print(
    "PASS: All validated zero frequencies preserved."
)


# ============================================================
# 35. FINAL PAQ710 FREQUENCY CHECK
# ============================================================

final_paq710 = (
    pd.to_numeric(
        raw_csv_df["PAQ710"],
        errors="coerce"
    )
    .dropna()
    .astype(int)
    .value_counts()
    .sort_index()
    .to_dict()
)


EXPECTED_PAQ710 = {
    0: 1173,
    1: 1637,
    2: 2400,
    3: 1473,
    4: 854,
    5: 1286,
    8: 369,
    77: 1,
    99: 16
}


if final_paq710 != EXPECTED_PAQ710:

    raise ValueError(
        "Final PAQ710 frequencies changed."
    )


print(
    "PASS: Final PAQ710 frequencies preserved."
)


# ============================================================
# 36. FINAL PAQ715 FREQUENCY CHECK
# ============================================================

final_paq715 = (
    pd.to_numeric(
        raw_csv_df["PAQ715"],
        errors="coerce"
    )
    .dropna()
    .astype(int)
    .value_counts()
    .sort_index()
    .to_dict()
)


EXPECTED_PAQ715 = {
    0: 2001,
    1: 1530,
    2: 1316,
    3: 574,
    4: 350,
    5: 588,
    8: 2844,
    99: 6
}


if final_paq715 != EXPECTED_PAQ715:

    raise ValueError(
        "Final PAQ715 frequencies changed."
    )


print(
    "PASS: Final PAQ715 frequencies preserved."
)


# ============================================================
# 37. SCIENTIFIC NOTATION CHECK
# ============================================================

scientific_count = 0


for col in EXPECTED_COLUMNS:

    scientific_count += int(
        raw_csv_df[col]
        .str.contains(
            r"[eE][+-]\d+",
            regex=True
        )
        .sum()
    )


print(
    "\n--- SCIENTIFIC NOTATION CHECK ---"
)

print(
    "Scientific notation occurrences:",
    scientific_count
)


if scientific_count != 0:

    raise ValueError(
        "Unexpected scientific notation found."
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 38. FINAL SEQN CHECK
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if final_seqn.min() != 83732:

    raise ValueError(
        "Final minimum SEQN incorrect."
    )


if final_seqn.max() != 93702:

    raise ValueError(
        "Final maximum SEQN incorrect."
    )


if int(
    final_seqn
    .duplicated()
    .sum()
) != 0:

    raise ValueError(
        "Duplicate SEQN found in final CSV."
    )


if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "Final CSV SEQN still contains '.0'."
    )


print(
    "PASS: Final SEQN valid "
    "and contains no '.0'."
)


# ============================================================
# 39. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 100
)

print(
    "PAQ_I CONVERSION COMPLETE"
)

print(
    "=" * 100
)


print(
    "Cycle: 2015-2016"
)


print(
    "Component: Questionnaire - Physical Activity"
)


print(
    "Rows:",
    f"{len(df):,}"
)


print(
    "Columns:",
    len(df.columns)
)


print(
    "SEQN range:",
    f"{df['SEQN'].min()}-{df['SEQN'].max()}"
)


print(
    "Original tiny-value artifacts:",
    f"{tiny_total:,}"
)


print(
    "Tiny artifact variables:",
    list(
        EXPECTED_TINY_COUNTS.keys()
    )
)


print(
    "Validated zero restorations:",
    f"{correction_total:,}"
)


print(
    "PAD680 zeros:",
    int(
        df["PAD680"]
        .eq(0)
        .sum()
    )
)


print(
    "PAQ706 zeros:",
    int(
        df["PAQ706"]
        .eq(0)
        .sum()
    )
)


print(
    "PAQ710 zeros:",
    f"{int(df['PAQ710'].eq(0).sum()):,}"
)


print(
    "PAQ715 zeros:",
    f"{int(df['PAQ715'].eq(0).sum()):,}"
)


print(
    "PAQ731 zeros:",
    f"{int(df['PAQ731'].eq(0).sum()):,}"
)


print(
    "PAQ677 zeros:",
    int(
        df["PAQ677"]
        .eq(0)
        .sum()
    )
)


print(
    "PAQ678 zeros:",
    int(
        df["PAQ678"]
        .eq(0)
        .sum()
    )
)


print(
    "Genuine decimal variables:",
    "NONE"
)


print(
    "Whole-number/code variables:",
    len(INTEGER_COLUMNS)
)


print(
    "XPT -> CSV numeric differences:",
    "0 expected"
)


print(
    "Missing-position differences:",
    "0 expected"
)


print(
    "Tiny values remaining:",
    "0 expected"
)


print(
    "Unnecessary '.0':",
    "0 expected"
)


print(
    "Scientific notation:",
    "0 expected"
)


print(
    "\nFINAL RULE:"
)


print(
    "PAQ_I contains no genuine decimal variables "
    "after the validated zero repair."
)


print(
    "Only the 5,124 validated XPORT tiny values "
    "are converted to real numeric zero."
)


print(
    "All 94 fields can then safely be exported "
    "as whole-number IDs/codes/counts."
)


print(
    "Your broader NHANES fidelity rule remains unchanged:"
)


print(
    "Genuine Dietary and other continuous decimal "
    "variables remain exactly AS IS."
)
```
