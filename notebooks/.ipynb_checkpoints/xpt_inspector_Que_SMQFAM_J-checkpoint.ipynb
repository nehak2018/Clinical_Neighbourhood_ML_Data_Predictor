```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2017_2018"
    / "Questionnnair"
)

xpt_file = DATA_DIR / "SMQFAM_J.xpt"
csv_file = DATA_DIR / "SMQFAM_J_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Preserve completely untouched pandas-decoded source.
#
source_df = df.copy()


print("=" * 90)
print("NHANES 2017-2018 SMQFAM_J XPT -> CSV")
print("=" * 90)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "SMD460",
    "SMD470",
    "SMD480"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "SMQFAM_J column names/order differ from expected.\n\n"
        f"Expected:\n{EXPECTED_COLUMNS}\n\n"
        f"Actual:\n{df.columns.tolist()}"
    )


if len(df) != 9254:

    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 4:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 9,254 rows x 4 columns."
)


# ============================================================
# 4. FIELD TYPE PLAN
# ============================================================

#
# Direct inspection of attached SMQFAM_J shows:
#
#     no genuine continuous decimal variables
#
# The four variables are:
#
#     SEQN   respondent identifier
#     SMD460 whole-number household smoker count/category
#     SMD470 whole-number inside-home smoker count/category
#     SMD480 whole-number number of days/category
#
# Therefore all 4 variables can become nullable Int64
# AFTER the tiny-zero artifacts have been validated and
# restored.
#
# IMPORTANT:
#
# This applies ONLY to SMQFAM_J.
#
# It must NOT be generalized to genuine decimal fields in:
#
#     Dietary
#     Laboratory
#     Examination
#     sample weights
#     nutrient values
#     exposure measurements
#
# Those values should remain untouched.
#

DECIMAL_COLUMNS = []

INTEGER_COLUMNS = EXPECTED_COLUMNS.copy()


print("\n--- FIELD TYPE PLAN ---")

print(
    "Genuine decimal variables:",
    DECIMAL_COLUMNS
)

print(
    "Whole-number/code variables:",
    len(INTEGER_COLUMNS)
)


# ============================================================
# 5. IDENTIFY KNOWN XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known SAS XPORT / pandas zero-decoding artifact:
#
#     5.397605346934028e-79
#
# NEVER blanket-replace this value without validating
# the exact variable and count first.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


numeric_cols = (
    df.select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)


if numeric_cols != EXPECTED_COLUMNS:

    raise ValueError(
        "Expected all 4 SMQFAM_J fields "
        "to load numerically."
    )


tiny_counts = (
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


true_zero_counts = (
    df[numeric_cols]
    .eq(0)
    .sum()
)


true_zero_total = int(
    true_zero_counts.sum()
)


print("\n--- ORIGINAL XPT ZERO CHECK ---")

print(
    "Ordinary numeric zeros before correction:",
    f"{true_zero_total:,}"
)

print(
    "Tiny XPORT artifacts:",
    f"{tiny_total:,}"
)

print("\nTiny values by variable:")

print(
    tiny_counts[
        tiny_counts > 0
    ]
    .sort_values(
        ascending=False
    )
    .to_string()
)


# ============================================================
# 6. VALIDATE EXACT TINY-ZERO PATTERN
# ============================================================

#
# Attached SMQFAM_J.xpt:
#
#     SMD460 = 6,092 tiny values
#     SMD470 = 1,717 tiny values
#     SMD480 =    81 tiny values
#
# TOTAL = 7,890
#
# CDC SMQFAM_J:
#
# SMD460:
#     0 = No one in household is a smoker
#     Count = 6,092
#
# SMD470:
#     0 = No one smokes inside the house
#     Count = 1,717
#
# SMD480:
#     0 = Zero days
#     Count = 81
#
# Therefore these three corrections are directly
# validated against CDC.
#

EXPECTED_TINY_COUNTS = {
    "SMD460": 6092,
    "SMD470": 1717,
    "SMD480": 81
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: Tiny-value pattern differs from "
        "validated SMQFAM_J attachment.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{actual_tiny_counts}"
    )


if tiny_total != 7890:

    raise ValueError(
        f"Unexpected total tiny-value count: "
        f"{tiny_total:,}"
    )


if true_zero_total != 0:

    raise ValueError(
        "Unexpected ordinary numeric zeros "
        "already existed before correction."
    )


print(
    "\nPASS: Exact 7,890-value SMQFAM_J "
    "tiny-zero pattern validated."
)


# ============================================================
# 7. RESTORE ONLY VALIDATED ZEROS
# ============================================================

#
# DO NOT:
#
#     df.replace(TINY_VALUE, 0)
#
# globally.
#
# Restore each independently validated variable.
#

correction_counts = {}


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    tiny_mask = (
        df[col]
        .eq(TINY_VALUE)
    )

    correction_count = int(
        tiny_mask.sum()
    )

    if correction_count != expected_count:

        raise ValueError(
            f"{col} unexpected tiny-value count.\n"
            f"Expected: {expected_count:,}\n"
            f"Actual:   {correction_count:,}"
        )

    df.loc[
        tiny_mask,
        col
    ] = 0

    correction_counts[col] = correction_count


print("\n--- VALIDATED ZERO RESTORATION ---")


for col, count in correction_counts.items():

    print(
        f"{col} restored to zero:",
        f"{count:,}"
    )


if int(
    df["SMD460"]
    .eq(0)
    .sum()
) != 6092:

    raise ValueError(
        "SMD460 zero restoration failed."
    )


if int(
    df["SMD470"]
    .eq(0)
    .sum()
) != 1717:

    raise ValueError(
        "SMD470 zero restoration failed."
    )


if int(
    df["SMD480"]
    .eq(0)
    .sum()
) != 81:

    raise ValueError(
        "SMD480 zero restoration failed."
    )


print(
    "PASS: All 7,890 validated zero values restored."
)


# ============================================================
# 8. CREATE EXPECTED CORRECTED SOURCE
# ============================================================

#
# expected_df represents exactly what the corrected
# source should contain.
#
# ONLY the independently validated tiny values may
# change to zero.
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col]
        .eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 9. CONFIRM NO TINY VALUES REMAIN
# ============================================================

remaining_tiny = int(
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print(
    "\nTiny values remaining:",
    remaining_tiny
)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain "
        "after correction."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 10. CHECK FOR GENUINE FRACTIONAL VARIABLES
# ============================================================

#
# Run BEFORE converting anything to Int64.
#
# np.round() here is ONLY a comparison test.
#
# It DOES NOT modify any values.
#

print("\n--- FRACTIONAL VALUE CHECK ---")


fractional_columns = []


for col in numeric_cols:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )

    if len(values) == 0:

        continue

    fractional_mask = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )

    fractional_count = int(
        fractional_mask.sum()
    )

    print(
        f"{col}: "
        f"{fractional_count:,} fractional observations"
    )

    if fractional_count > 0:

        fractional_columns.append(
            col
        )


if fractional_columns:

    raise ValueError(
        "STOP: Genuine decimal variables unexpectedly "
        "found in SMQFAM_J.\n\n"
        "They will NOT be rounded:\n"
        + str(fractional_columns)
    )


print(
    "PASS: No genuine fractional variables."
)


# ============================================================
# 11. VALIDATE SEQN
# ============================================================

seqn_values = (
    df["SEQN"]
    .dropna()
    .to_numpy(
        dtype=float
    )
)


if not np.isclose(
    seqn_values,
    np.round(seqn_values),
    rtol=0,
    atol=1e-12
).all():

    raise ValueError(
        "SEQN contains unexpected fractional values."
    )


print("\n--- SEQN ---")

print(
    "Minimum:",
    df["SEQN"].min()
)

print(
    "Maximum:",
    df["SEQN"].max()
)

print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)

print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if df["SEQN"].min() != 93703:

    raise ValueError(
        "Unexpected minimum SEQN."
    )


if df["SEQN"].max() != 102956:

    raise ValueError(
        "Unexpected maximum SEQN."
    )


if int(
    df["SEQN"]
    .isna()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains missing values."
    )


if int(
    df["SEQN"]
    .duplicated()
    .sum()
) != 0:

    raise ValueError(
        "SEQN contains duplicate values."
    )


print(
    "PASS: SEQN 93703-102956 validated."
)


# ============================================================
# 12. CONVERT PROVEN WHOLE-NUMBER FIELDS TO Int64
# ============================================================

#
# All four fields have now been proven to contain
# only whole-number values.
#
# NO .round() is used during conversion.
#

for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )

    if len(values) > 0:

        whole_mask = np.isclose(
            values,
            np.round(values),
            rtol=0,
            atol=1e-12
        )

        if not whole_mask.all():

            raise ValueError(
                f"{col} unexpectedly contains "
                "fractional values."
            )

    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: All 4 SMQFAM_J variables converted "
    "to nullable Int64 without rounding."
)


# ============================================================
# 13. EXACT CDC FREQUENCY CHECK HELPER
# ============================================================

def check_frequency(
    column,
    expected_values,
    expected_missing
):

    actual_values = (
        df[column]
        .value_counts()
        .sort_index()
        .to_dict()
    )

    actual_missing = int(
        df[column]
        .isna()
        .sum()
    )

    print(
        f"\n--- CDC CHECK: {column} ---"
    )

    print(
        "Observed:",
        actual_values
    )

    print(
        "Missing:",
        f"{actual_missing:,}"
    )

    if actual_values != expected_values:

        raise ValueError(
            f"{column} frequencies differ from CDC.\n\n"
            f"Expected:\n{expected_values}\n\n"
            f"Actual:\n{actual_values}"
        )

    if actual_missing != expected_missing:

        raise ValueError(
            f"{column} missing count differs from CDC.\n"
            f"Expected: {expected_missing:,}\n"
            f"Actual:   {actual_missing:,}"
        )

    print(
        f"PASS: {column} matches CDC."
    )


# ============================================================
# 14. CDC CHECK - SMD460
# ============================================================

#
# Number of people living in household who smoke.
#

check_frequency(
    "SMD460",
    {
        0: 6092,
        1: 1667,
        2: 867,
        3: 153,
        777: 2,
        999: 10
    },
    463
)


# ============================================================
# 15. CDC CHECK - SMD470
# ============================================================

#
# Number of household members who smoke inside home.
#

check_frequency(
    "SMD470",
    {
        0: 1717,
        1: 588,
        2: 314,
        3: 68
    },
    6567
)


# ============================================================
# 16. CDC CHECK - SMD480
# ============================================================

#
# Number of days tobacco was smoked inside home
# during past seven days.
#

check_frequency(
    "SMD480",
    {
        0: 81,
        1: 80,
        2: 96,
        3: 43,
        4: 16,
        5: 30,
        6: 7,
        7: 615,
        77: 2
    },
    8284
)


# ============================================================
# 17. ALLOWED CDC CODE CHECK
# ============================================================

ALLOWED_VALUES = {

    "SMD460": {
        0, 1, 2, 3, 777, 999
    },

    "SMD470": {
        0, 1, 2, 3, 777, 999
    },

    "SMD480": {
        0, 1, 2, 3, 4, 5, 6, 7, 77, 99
    }
}


print("\n--- ALLOWED CODE CHECK ---")


for col, allowed in ALLOWED_VALUES.items():

    observed = set(
        df[col]
        .dropna()
        .astype(int)
        .unique()
        .tolist()
    )

    unexpected = (
        observed
        -
        allowed
    )

    if unexpected:

        raise ValueError(
            f"{col} contains unexpected CDC codes: "
            f"{sorted(unexpected)}"
        )

    print(
        f"PASS: {col} contains only "
        "CDC-documented values."
    )


# ============================================================
# 18. SMD460 -> SMD470 ROUTING CHECK
# ============================================================

#
# CDC:
#
# If SMD460 is:
#
#     0   no smoker
#     777 refused
#     999 don't know
#
# interviewer skips to end of section.
#
# Therefore SMD470 is populated exactly for:
#
#     SMD460 = 1, 2, or 3
#
# Number expected:
#
#     1667 + 867 + 153 = 2687
#

smd470_expected = (
    df["SMD460"]
    .isin(
        [
            1,
            2,
            3
        ]
    )
    .to_numpy(
        dtype=bool
    )
)


smd470_present = (
    df["SMD470"]
    .notna()
    .to_numpy(
        dtype=bool
    )
)


smd470_route_mismatches = int(
    np.sum(
        smd470_expected
        !=
        smd470_present
    )
)


print(
    "\n--- SMD460 / SMD470 ROUTING ---"
)

print(
    "Expected SMD470 populated:",
    int(
        smd470_expected.sum()
    )
)

print(
    "Actually populated:",
    int(
        smd470_present.sum()
    )
)

print(
    "Routing mismatches:",
    smd470_route_mismatches
)


if smd470_route_mismatches != 0:

    raise ValueError(
        "SMD470 routing does not exactly match "
        "SMD460 = 1, 2, or 3."
    )


print(
    "PASS: All 2,687 eligible records "
    "route exactly to SMD470."
)


# ============================================================
# 19. SMD470 -> SMD480 ROUTING CHECK
# ============================================================

#
# CDC:
#
# SMD470 = 0
#     -> End section
#
# SMD470 = 1, 2, or 3
#     -> Ask SMD480
#
# Therefore SMD480 should have:
#
#     588 + 314 + 68 = 970
#
# populated records.
#

smd480_expected = (
    df["SMD470"]
    .isin(
        [
            1,
            2,
            3
        ]
    )
    .to_numpy(
        dtype=bool
    )
)


smd480_present = (
    df["SMD480"]
    .notna()
    .to_numpy(
        dtype=bool
    )
)


smd480_route_mismatches = int(
    np.sum(
        smd480_expected
        !=
        smd480_present
    )
)


print(
    "\n--- SMD470 / SMD480 ROUTING ---"
)

print(
    "Expected SMD480 populated:",
    int(
        smd480_expected.sum()
    )
)

print(
    "Actually populated:",
    int(
        smd480_present.sum()
    )
)

print(
    "Routing mismatches:",
    smd480_route_mismatches
)


if smd480_route_mismatches != 0:

    raise ValueError(
        "SMD480 routing does not exactly match "
        "SMD470 = 1, 2, or 3."
    )


print(
    "PASS: All 970 eligible records "
    "route exactly to SMD480."
)


# ============================================================
# 20. SMD470 <= SMD460 HARD EDIT CHECK
# ============================================================

#
# CDC documents:
#
# Number smoking inside home must be less than or
# equal to number of household smokers.
#
# Only compare the regular 0-3 response values.
#

regular_count_mask = (
    df["SMD460"]
    .isin(
        [
            0,
            1,
            2,
            3
        ]
    )
    &
    df["SMD470"]
    .isin(
        [
            0,
            1,
            2,
            3
        ]
    )
)


hard_edit_violations = int(
    (
        df.loc[
            regular_count_mask,
            "SMD470"
        ]
        >
        df.loc[
            regular_count_mask,
            "SMD460"
        ]
    )
    .sum()
)


print(
    "\n--- SMD470 <= SMD460 HARD EDIT ---"
)

print(
    "Hard-edit violations:",
    hard_edit_violations
)


if hard_edit_violations != 0:

    raise ValueError(
        "SMD470 exceeds SMD460 for one or "
        "more household records."
    )


print(
    "PASS: SMD470 never exceeds SMD460."
)


# ============================================================
# 21. FINAL VALIDATED ZERO PATTERN
# ============================================================

#
# After correction the ONLY expected zeros are:
#
#     SMD460 = 6,092
#     SMD470 = 1,717
#     SMD480 =    81
#

final_zero_counts = {}


for col in EXPECTED_COLUMNS:

    zero_count = int(
        df[col]
        .eq(0)
        .sum()
    )

    if zero_count > 0:

        final_zero_counts[col] = zero_count


print("\n--- FINAL ZERO PATTERN ---")

print(
    final_zero_counts
)


if final_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final zero pattern differs from "
        "validated SMQFAM_J correction pattern.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{final_zero_counts}"
    )


print(
    "PASS: Final zero pattern exactly matches CDC."
)


# ============================================================
# 22. VALUE FIDELITY BEFORE EXPORT
# ============================================================

#
# Permitted numeric modifications:
#
#     SMD460 tiny -> 0
#     SMD470 tiny -> 0
#     SMD480 tiny -> 0
#
# Nothing else is allowed to change.
#

print(
    "\n--- VALUE FIDELITY BEFORE EXPORT ---"
)


fidelity_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )

    current_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )

    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )

    if not same:

        fidelity_failures.append(
            col
        )


if fidelity_failures:

    raise ValueError(
        "Unexpected numeric changes before export:\n"
        + str(fidelity_failures)
    )


print(
    "PASS: All 4 variables match exact "
    "expected corrected XPT values."
)


# ============================================================
# 23. MISSING-VALUE FIDELITY BEFORE EXPORT
# ============================================================

missing_mismatches = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        source_df[col]
        .isna()
        .to_numpy()
    )

    current_missing = (
        df[col]
        .isna()
        .to_numpy()
    )

    missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            current_missing
        )
    )


print(
    "\nMissing-position mismatches before export:",
    missing_mismatches
)


if missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed "
        "before export."
    )


print(
    "PASS: Missing positions preserved exactly."
)


# ============================================================
# 24. EXPORT CSV
# ============================================================

#
# IMPORTANT FIDELITY POLICY
#
# NO:
#
#     df.round(...)
#
# NO:
#
#     float_format=
#
# NO:
#
#     blanket decimal formatting
#
# SMQFAM_J itself contains no genuine fractional
# variables.
#
# But keeping this export policy protects genuine
# decimals in Dietary/Lab/Exam/etc. files in your
# broader NHANES pipeline.
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print("\nCSV created:")

print(
    csv_file
)


# ============================================================
# 25. RE-READ CSV WITH ROUND-TRIP FLOAT PARSER
# ============================================================

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != df.shape:

    raise ValueError(
        "CSV dimensions changed."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV columns/order changed."
    )


print(
    "PASS: CSV structure preserved."
)


# ============================================================
# 26. EXACT CORRECTED XPT -> CSV NUMERIC FIDELITY
# ============================================================

print(
    "\n--- EXACT CORRECTED XPT -> CSV FIDELITY ---"
)


csv_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )

    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )

    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )

    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "Corrected XPT -> CSV numeric differences:\n"
        + str(csv_failures)
    )


print(
    "PASS: All 4 CSV variables round-trip "
    "to exact corrected XPT numeric values."
)


# ============================================================
# 27. CSV MISSING-VALUE FIDELITY
# ============================================================

csv_missing_mismatches = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )

    exported_missing = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )

    csv_missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            exported_missing
        )
    )


print("\n--- CSV MISSING VALUE FIDELITY ---")

print(
    "Missing-position mismatches:",
    csv_missing_mismatches
)


if csv_missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed "
        "during CSV export."
    )


print(
    "PASS: Missing positions preserved exactly."
)


# ============================================================
# 28. READ FINAL CSV AS RAW TEXT
# ============================================================

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


if raw_csv_df.shape != (
    9254,
    4
):

    raise ValueError(
        "Raw CSV dimensions changed."
    )


# ============================================================
# 29. FINAL TINY-VALUE CHECK
# ============================================================

tiny_csv_count = 0


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )

    tiny_csv_count += int(
        values.eq(TINY_VALUE)
        .sum()
    )


print("\n--- FINAL TINY VALUE CHECK ---")

print(
    "Tiny values remaining:",
    tiny_csv_count
)


if tiny_csv_count != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain "
        "in final CSV."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 30. RAW CSV .0 CHECK
# ============================================================

#
# SMQFAM_J has been proven to contain no genuine
# decimal variables.
#
# Therefore it is safe to check all four variables
# for unnecessary trailing ".0".
#
# Examples:
#
#     93703.0 -> 93703
#     0.0     -> 0
#     1.0     -> 1
#
# DO NOT apply this all-column rule to a file
# containing genuine decimal variables.
#

print("\n--- RAW CSV .0 CHECK ---")


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = raw_csv_df[col]

    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )

    count = int(
        bad_mask.sum()
    )

    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unexpected '.0' formatting found:\n"
        + str(dot_zero_errors)
    )


print(
    "PASS: No unnecessary '.0' "
    "in any SMQFAM_J field."
)


# ============================================================
# 31. FINAL CSV ZERO PATTERN
# ============================================================

csv_zero_counts = {}


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )

    zero_count = int(
        values.eq(0)
        .sum()
    )

    if zero_count > 0:

        csv_zero_counts[col] = zero_count


print("\n--- FINAL CSV ZERO PATTERN ---")

print(
    csv_zero_counts
)


if csv_zero_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "Final CSV zero pattern differs.\n\n"
        f"Expected:\n{EXPECTED_TINY_COUNTS}\n\n"
        f"Actual:\n{csv_zero_counts}"
    )


print(
    "PASS: Final CSV zero pattern exactly "
    "matches CDC."
)


# ============================================================
# 32. FINAL CDC FREQUENCY RECHECK FROM CSV
# ============================================================

FINAL_EXPECTED_FREQUENCIES = {

    "SMD460": (
        {
            0: 6092,
            1: 1667,
            2: 867,
            3: 153,
            777: 2,
            999: 10
        },
        463
    ),

    "SMD470": (
        {
            0: 1717,
            1: 588,
            2: 314,
            3: 68
        },
        6567
    ),

    "SMD480": (
        {
            0: 81,
            1: 80,
            2: 96,
            3: 43,
            4: 16,
            5: 30,
            6: 7,
            7: 615,
            77: 2
        },
        8284
    )
}


print(
    "\n--- FINAL CSV CDC FREQUENCY RECHECK ---"
)


for col, (
    expected_values,
    expected_missing
) in FINAL_EXPECTED_FREQUENCIES.items():

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )

    actual_values = (
        values
        .dropna()
        .astype(int)
        .value_counts()
        .sort_index()
        .to_dict()
    )

    actual_missing = int(
        values
        .isna()
        .sum()
    )

    if actual_values != expected_values:

        raise ValueError(
            f"Final CSV {col} frequencies differ "
            "from CDC.\n"
            f"Expected: {expected_values}\n"
            f"Actual:   {actual_values}"
        )

    if actual_missing != expected_missing:

        raise ValueError(
            f"Final CSV {col} missing count differs."
        )


print(
    "PASS: Final CSV frequencies match CDC exactly."
)


# ============================================================
# 33. SCIENTIFIC NOTATION CHECK
# ============================================================

scientific_count = 0


for col in EXPECTED_COLUMNS:

    tokens = raw_csv_df[col]

    scientific_count += int(
        tokens.str.contains(
            r"[eE][+-]\d+",
            regex=True
        )
        .sum()
    )


print("\n--- SCIENTIFIC NOTATION CHECK ---")

print(
    "Scientific notation occurrences:",
    scientific_count
)


if scientific_count != 0:

    raise ValueError(
        "Unexpected scientific notation found."
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 34. FINAL SEQN CHECK
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if final_seqn.min() != 93703:

    raise ValueError(
        "Final minimum SEQN incorrect."
    )


if final_seqn.max() != 102956:

    raise ValueError(
        "Final maximum SEQN incorrect."
    )


if int(
    final_seqn.isna()
    .sum()
) != 0:

    raise ValueError(
        "Final CSV contains missing SEQN."
    )


if int(
    final_seqn.duplicated()
    .sum()
) != 0:

    raise ValueError(
        "Final CSV contains duplicate SEQN."
    )


if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "Final CSV SEQN still contains '.0'."
    )


print(
    "PASS: Final CSV SEQN 93703-102956 valid "
    "and contains no '.0'."
)


# ============================================================
# 35. FINAL COMPLETE SOURCE -> CSV FIDELITY
# ============================================================

#
# Final all-variable numeric fidelity check against
# expected corrected XPT.
#

final_fidelity_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )

    final_values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    ).to_numpy(
        dtype=float
    )

    same = np.allclose(
        expected_values,
        final_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )

    if not same:

        final_fidelity_failures.append(
            col
        )


if final_fidelity_failures:

    raise ValueError(
        "Final source -> CSV fidelity failures:\n"
        + str(final_fidelity_failures)
    )


print(
    "PASS: Final CSV preserves exact corrected "
    "numeric values for all 4 variables."
)


# ============================================================
# 36. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 90
)

print(
    "SMQFAM_J CONVERSION COMPLETE"
)

print(
    "=" * 90
)

print(
    "Cycle: 2017-2018"
)

print(
    "Component: Questionnaire - "
    "Smoking / Household Smokers"
)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)

print(
    "SEQN range:",
    f"{df['SEQN'].min()}-{df['SEQN'].max()}"
)

print(
    "Original tiny-value artifacts:",
    f"{tiny_total:,}"
)

print(
    "SMD460 zero restorations:",
    correction_counts["SMD460"]
)

print(
    "SMD470 zero restorations:",
    correction_counts["SMD470"]
)

print(
    "SMD480 zero restorations:",
    correction_counts["SMD480"]
)

print(
    "Total zero restorations:",
    sum(
        correction_counts.values()
    )
)

print(
    "Genuine decimal variables: NONE"
)

print(
    "Whole-number/code variables:",
    len(INTEGER_COLUMNS)
)

print(
    "SMD460 household smokers = 0:",
    int(
        df["SMD460"]
        .eq(0)
        .sum()
    )
)

print(
    "SMD470 inside-home smokers = 0:",
    int(
        df["SMD470"]
        .eq(0)
        .sum()
    )
)

print(
    "SMD480 days smoked inside = 0:",
    int(
        df["SMD480"]
        .eq(0)
        .sum()
    )
)

print(
    "XPT -> CSV unexpected numeric differences: "
    "0 expected"
)

print(
    "Missing-position differences: "
    "0 expected"
)

print(
    "Tiny values remaining: "
    "0 expected"
)

print(
    "Unnecessary '.0': "
    "0 expected"
)

print(
    "Scientific notation: "
    "0 expected"
)

print(
    "=" * 90
)
```
