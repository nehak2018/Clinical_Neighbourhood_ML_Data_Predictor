```python
# =================================================================================================
# Raw data : This is Lab data from NHANES
# Purpose : Creating raw CSV file which converted from .xpt and generate ths Labs data.
# Specification : Each NHANE dataset has SEQN "Participent ID" which is uniqud id for labs. 
# Files Used :  DR1TOT_C.csv, DR2TOT_C.csv,DR1IFF_C.csv, DR2IFF_C.csv downloaded from Source 
# Data Dictinary : 
# Data Availability Matrix : 
# Source Site : 
# Developer Notes :  The notebook had created for windows desktop environment with Python3.
# File Creator : Neha Khairnar
# ===================================================================================================

# 📦 Step 1: Imports
import pandas as pd
from pathlib import Path
BASE_DIR = Path().resolve().parent
Day_1_total_intake_file ="../data/2003_2004/Dietary/DR1TOT_C_corrected.csv"
Day_2_total_intake_file = "../data/2003_2004/Dietary/DR2TOT_C_corrected.csv"
Day_1_total_intake = pd.read_csv(Day_1_total_intake_file,  dtype=str, keep_default_na=False)
Day_2_total_intake = pd.read_csv(Day_2_total_intake_file,  dtype=str, keep_default_na=False)

#Day_1_individual_intake_file = "../data/2003_2004/Dietary/DR1IFF_C.csv"
#Day_2_individual_intake_file = "../data/2003_2004/Dietary/DR2IFF_C.csv"
#Day_1_individual_intake = pd.read_csv(Day_1_individual_intake_file)
#Day_2_individual_intake = pd.read_csv(Day_2_individual_intake_file)

files = {
    "Day_1_total_intake": Day_1_total_intake_file,
    "Day_2_total_intake": Day_2_total_intake_file,
    #"Day_1_individual_intake": Day_1_individual_intake_file,
    #"Day_2_individual_intake" : Day_2_individual_intake_file
}

# =================================================================================================
# 📊 Summary container uses above JSON for inspecting file purpose which
# # ===============================================================================================
summary = []
for name, path in files.items():
    df = pd.read_csv(path)
    summary.append({
        "file": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": list(df.columns)
    })

summary_df = pd.DataFrame(summary)
print("📊 NHANES File Structure Summary:\n")
display(summary_df[["file", "rows", "columns", "column_names"]])
for i, row in summary_df.iterrows():
    print("\n" + "="*50)
    print("FILE:", row["file"])
    print("COLUMNS:", len(row["column_names"]))
    print(row["column_names"])

# =========================
# 🧹 Step 4: Standardize Column Names
# =========================
Day_1_total_intake.columns = Day_1_total_intake.columns.str.upper()
Day_2_total_intake.columns = Day_2_total_intake.columns.str.upper()
#Day_1_individual_intake.columns = Day_1_individual_intake.columns.str.upper()
#Day_2_individual_intake.columns = Day_2_individual_intake.columns.str.upper()

# ===================================================================================================
# 🔗 Step 5: Merge ALL columns (FULL OUTER JOIN) which will have unique sequence numbers with all values
# ===================================================================================================
df = Day_1_total_intake.merge(
        Day_2_total_intake,
        on="SEQN",
        how="outer",
        suffixes=("", "_Day_2_total_intake")
    )
    # .merge(
    #     Day_1_individual_intake,
    #     on="SEQN",
    #     how="outer",
    #     suffixes=("", "_Day_1_individual_intake")
    # ).merge(
    #     Day_2_individual_intake,
    #     on="SEQN",
    #     how="outer",
    #     suffixes=("", "_Day_2_individual_intake")
    # )

# ===================================================================================================
# 👀 Step 3: Validate result
# ===================================================================================================
print("📋 Merge ALL columns:")
print(df.columns.tolist())
print("\n📊 Shape:", df.shape)

# ===================================================================================================
# 💾 Step 14: Save Final Output, Need to change if you are on SCHARE
# ===================================================================================================
OUTPUT_DIR = BASE_DIR / "data" / "processed_v3"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_PATH = OUTPUT_DIR / "neighbourhood_nhanes_dietary_2003_2004_raw_corrected.csv"
df.to_csv(OUTPUT_PATH, index=False)
print("\n✅ Saved to:")
print(OUTPUT_PATH)
```
