```python
import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path().resolve().parent

DATA_DIR = (
    BASE_DIR
    / "data"
    / "2011_2012"
    / "Dietary"
)

xpt_file = DATA_DIR / "DR1TOT_G.xpt"
csv_file = DATA_DIR / "DR1TOT_G_corrected.csv"


# ============================================================
# 2. READ ORIGINAL XPT
# ============================================================

df = pd.read_sas(
    xpt_file,
    format="xport",
    encoding="latin1"
)

#
# Preserve completely untouched pandas-decoded XPT.
#
source_df = df.copy()


print("=" * 80)
print("NHANES 2011-2012 DR1TOT_G XPT -> CSV")
print("=" * 80)

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Columns:",
    len(df.columns)
)


# ============================================================
# 3. EXPECTED STRUCTURE
# ============================================================

EXPECTED_COLUMNS = [
    "SEQN",
    "WTDRD1",
    "WTDR2D",
    "DR1DRSTZ",
    "DR1EXMER",
    "DRABF",
    "DRDINT",
    "DR1DBIH",
    "DR1DAY",
    "DR1LANG",
    "DR1MNRSP",
    "DR1HELPD",
    "DBQ095Z",
    "DBD100",
    "DRQSPREP",
    "DRQSDIET",
    "DRQSDT1",
    "DRQSDT2",
    "DRQSDT3",
    "DRQSDT4",
    "DRQSDT5",
    "DRQSDT6",
    "DRQSDT7",
    "DRQSDT8",
    "DRQSDT9",
    "DRQSDT10",
    "DRQSDT11",
    "DRQSDT12",
    "DRQSDT91",
    "DR1TNUMF",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TCARB",
    "DR1TSUGR",
    "DR1TFIBE",
    "DR1TTFAT",
    "DR1TSFAT",
    "DR1TMFAT",
    "DR1TPFAT",
    "DR1TCHOL",
    "DR1TATOC",
    "DR1TATOA",
    "DR1TRET",
    "DR1TVARA",
    "DR1TACAR",
    "DR1TBCAR",
    "DR1TCRYP",
    "DR1TLYCO",
    "DR1TLZ",
    "DR1TVB1",
    "DR1TVB2",
    "DR1TNIAC",
    "DR1TVB6",
    "DR1TFOLA",
    "DR1TFA",
    "DR1TFF",
    "DR1TFDFE",
    "DR1TCHL",
    "DR1TVB12",
    "DR1TB12A",
    "DR1TVC",
    "DR1TVD",
    "DR1TVK",
    "DR1TCALC",
    "DR1TPHOS",
    "DR1TMAGN",
    "DR1TIRON",
    "DR1TZINC",
    "DR1TCOPP",
    "DR1TSODI",
    "DR1TPOTA",
    "DR1TSELE",
    "DR1TCAFF",
    "DR1TTHEO",
    "DR1TALCO",
    "DR1TMOIS",
    "DR1TS040",
    "DR1TS060",
    "DR1TS080",
    "DR1TS100",
    "DR1TS120",
    "DR1TS140",
    "DR1TS160",
    "DR1TS180",
    "DR1TM161",
    "DR1TM181",
    "DR1TM201",
    "DR1TM221",
    "DR1TP182",
    "DR1TP183",
    "DR1TP184",
    "DR1TP204",
    "DR1TP205",
    "DR1TP225",
    "DR1TP226",
    "DR1_300",
    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ",
    "DR1TWS",
    "DRD340",
    "DRD350A",
    "DRD350AQ",
    "DRD350B",
    "DRD350BQ",
    "DRD350C",
    "DRD350CQ",
    "DRD350D",
    "DRD350DQ",
    "DRD350E",
    "DRD350EQ",
    "DRD350F",
    "DRD350FQ",
    "DRD350G",
    "DRD350GQ",
    "DRD350H",
    "DRD350HQ",
    "DRD350I",
    "DRD350IQ",
    "DRD350J",
    "DRD350JQ",
    "DRD350K",
    "DRD360",
    "DRD370A",
    "DRD370AQ",
    "DRD370B",
    "DRD370BQ",
    "DRD370C",
    "DRD370CQ",
    "DRD370D",
    "DRD370DQ",
    "DRD370E",
    "DRD370EQ",
    "DRD370F",
    "DRD370FQ",
    "DRD370G",
    "DRD370GQ",
    "DRD370H",
    "DRD370HQ",
    "DRD370I",
    "DRD370IQ",
    "DRD370J",
    "DRD370JQ",
    "DRD370K",
    "DRD370KQ",
    "DRD370L",
    "DRD370LQ",
    "DRD370M",
    "DRD370MQ",
    "DRD370N",
    "DRD370NQ",
    "DRD370O",
    "DRD370OQ",
    "DRD370P",
    "DRD370PQ",
    "DRD370Q",
    "DRD370QQ",
    "DRD370R",
    "DRD370RQ",
    "DRD370S",
    "DRD370SQ",
    "DRD370T",
    "DRD370TQ",
    "DRD370U",
    "DRD370UQ",
    "DRD370V"
]


if df.columns.tolist() != EXPECTED_COLUMNS:

    raise ValueError(
        "DR1TOT_G column names/order differ from expected.\n"
        f"Actual: {df.columns.tolist()}"
    )


if len(df) != 9338:

    raise ValueError(
        f"Unexpected row count: {len(df):,}"
    )


if len(df.columns) != 166:

    raise ValueError(
        f"Unexpected column count: {len(df.columns)}"
    )


print(
    "PASS: Structure = 9,338 rows x 166 columns."
)


# ============================================================
# 4. DEFINE GENUINE DECIMAL VARIABLES
# ============================================================

#
# CRITICAL DIETARY RULE
# ---------------------
#
# These variables contain genuine fractional values in the
# attached XPT.
#
# DO NOT:
#
#   - round them
#   - convert them to Int64
#   - specify float_format during CSV export
#
# A value such as:
#
#       6.0
#       0.0
#       1.25
#
# may be a legitimate representation in a Dietary decimal
# variable and should remain that way.
#

DECIMAL_COLUMNS = [
    "WTDRD1",
    "WTDR2D",

    "DR1TPROT",
    "DR1TCARB",
    "DR1TSUGR",
    "DR1TFIBE",
    "DR1TTFAT",
    "DR1TSFAT",
    "DR1TMFAT",
    "DR1TPFAT",

    "DR1TATOC",
    "DR1TATOA",

    "DR1TVB1",
    "DR1TVB2",
    "DR1TNIAC",
    "DR1TVB6",

    "DR1TCHL",
    "DR1TVB12",
    "DR1TB12A",
    "DR1TVC",
    "DR1TVD",
    "DR1TVK",

    "DR1TIRON",
    "DR1TZINC",
    "DR1TCOPP",
    "DR1TSELE",

    "DR1TALCO",
    "DR1TMOIS",

    "DR1TS040",
    "DR1TS060",
    "DR1TS080",
    "DR1TS100",
    "DR1TS120",
    "DR1TS140",
    "DR1TS160",
    "DR1TS180",

    "DR1TM161",
    "DR1TM181",
    "DR1TM201",
    "DR1TM221",

    "DR1TP182",
    "DR1TP183",
    "DR1TP184",
    "DR1TP204",
    "DR1TP205",
    "DR1TP225",
    "DR1TP226",

    "DR1_320Z",
    "DR1_330Z",
    "DR1BWATZ"
]


if len(DECIMAL_COLUMNS) != 50:

    raise ValueError(
        "Expected exactly 50 genuine decimal variables."
    )


INTEGER_COLUMNS = [
    col
    for col in EXPECTED_COLUMNS
    if col not in DECIMAL_COLUMNS
]


if len(INTEGER_COLUMNS) != 116:

    raise ValueError(
        "Expected exactly 116 whole-number variables."
    )


print("\n--- VARIABLE TYPE PLAN ---")

print(
    "Genuine decimal variables:",
    len(DECIMAL_COLUMNS)
)

print(
    "Whole-number variables:",
    len(INTEGER_COLUMNS)
)


# ============================================================
# 5. CHECK KNOWN XPORT TINY-ZERO ARTIFACT
# ============================================================

#
# Known pandas/SAS XPORT zero-decoding artifact:
#
#     5.397605346934028e-79
#
# IMPORTANT:
#
# Never perform:
#
#     df.replace(TINY_VALUE, 0)
#
# across the entire dataset without first checking the exact
# variable/count pattern.
#

TINY_VALUE = np.float64(
    5.397605346934028e-79
)


numeric_cols = df.select_dtypes(
    include=[np.number]
).columns.tolist()


tiny_counts = (
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
)


tiny_total = int(
    tiny_counts.sum()
)


true_zero_total = int(
    df[numeric_cols]
    .eq(0)
    .sum()
    .sum()
)


print("\n--- ORIGINAL XPT ZERO CHECK ---")


print(
    "Ordinary numeric zeros before restoration:",
    f"{true_zero_total:,}"
)


print(
    "Tiny artifact occurrences:",
    f"{tiny_total:,}"
)


print("\nTiny values by variable:")


print(
    tiny_counts[
        tiny_counts > 0
    ]
    .sort_values(
        ascending=False
    )
    .to_string()
)


# ============================================================
# 6. VALIDATE EXACT TINY-ZERO PATTERN
# ============================================================

#
# Exact pattern found in attached DR1TOT_G.xpt.
#
# CDC permits legitimate zero values for these variables.
#
# This includes nutrient values and water amounts.
#
# CDC also explicitly notes that complete recalls may
# legitimately contain zero nutrient intake.
#

EXPECTED_TINY_COUNTS = {

    "DR1BWATZ": 5372,

    "DR1DBIH": 309,

    "DR1TACAR": 1191,
    "DR1TALCO": 7215,

    "DR1TATOA": 6848,
    "DR1TATOC": 3,

    "DR1TB12A": 5508,
    "DR1TBCAR": 119,

    "DR1TCAFF": 2186,
    "DR1TCARB": 1,

    "DR1TCHL": 1,
    "DR1TCHOL": 92,

    "DR1TCRYP": 857,

    "DR1TFA": 191,
    "DR1TFDFE": 2,
    "DR1TFF": 77,
    "DR1TFIBE": 90,
    "DR1TFOLA": 2,

    "DR1TIRON": 1,

    "DR1TKCAL": 1,

    "DR1TLYCO": 2285,
    "DR1TLZ": 97,

    "DR1TM161": 19,
    "DR1TM181": 4,
    "DR1TM201": 59,
    "DR1TM221": 1590,

    "DR1TMFAT": 3,

    "DR1TNIAC": 1,

    "DR1TP182": 3,
    "DR1TP183": 3,
    "DR1TP184": 3395,
    "DR1TP204": 147,
    "DR1TP205": 1148,
    "DR1TP225": 886,
    "DR1TP226": 1083,

    "DR1TPFAT": 3,
    "DR1TPHOS": 2,
    "DR1TPOTA": 1,
    "DR1TPROT": 1,

    "DR1TRET": 50,

    "DR1TS040": 377,
    "DR1TS060": 455,
    "DR1TS080": 127,
    "DR1TS100": 27,
    "DR1TS120": 11,
    "DR1TS140": 5,
    "DR1TS160": 3,
    "DR1TS180": 3,

    "DR1TSELE": 2,
    "DR1TSFAT": 3,
    "DR1TSUGR": 2,

    "DR1TTFAT": 2,
    "DR1TTHEO": 4048,

    "DR1TVARA": 12,
    "DR1TVB1": 2,
    "DR1TVB12": 13,
    "DR1TVB2": 2,
    "DR1TVB6": 2,

    "DR1TVC": 31,
    "DR1TVD": 126,
    "DR1TVK": 3,

    "DR1TZINC": 1,

    "DR1_320Z": 1911,
    "DR1_330Z": 4575
}


actual_tiny_counts = (
    tiny_counts[
        tiny_counts > 0
    ]
    .astype(int)
    .to_dict()
)


if actual_tiny_counts != EXPECTED_TINY_COUNTS:

    raise ValueError(
        "STOP: Tiny-value pattern differs from "
        "validated DR1TOT_G pattern.\n"
        f"Expected: {EXPECTED_TINY_COUNTS}\n"
        f"Actual:   {actual_tiny_counts}"
    )


if tiny_total != 52589:

    raise ValueError(
        f"Unexpected total tiny-value count: {tiny_total:,}"
    )


if len(EXPECTED_TINY_COUNTS) != 64:

    raise ValueError(
        "Expected tiny artifacts in exactly 64 variables."
    )


print(
    "\nPASS: Exact 52,589-value tiny-zero "
    "pattern across 64 variables validated."
)


# ============================================================
# 7. RESTORE ONLY VALIDATED ZERO VALUES
# ============================================================

#
# Restore ONLY variables in EXPECTED_TINY_COUNTS.
#
# No blanket/global zero replacement.
#

print("\n--- VALIDATED ZERO RESTORATION ---")


correction_total = 0


for col, expected_count in EXPECTED_TINY_COUNTS.items():

    mask = (
        df[col]
        .eq(TINY_VALUE)
    )


    actual_count = int(
        mask.sum()
    )


    if actual_count != expected_count:

        raise ValueError(
            f"{col}: expected {expected_count:,} tiny values, "
            f"found {actual_count:,}."
        )


    df.loc[
        mask,
        col
    ] = 0


    correction_total += actual_count


    print(
        f"{col}: "
        f"{actual_count:,} restored to 0"
    )


if correction_total != 52589:

    raise ValueError(
        "Incorrect total zero-restoration count."
    )


print(
    "\nPASS:",
    f"{correction_total:,}",
    "validated zeros restored."
)


# ============================================================
# 8. CREATE EXPECTED CORRECTED SOURCE COPY
# ============================================================

#
# This copy represents exactly what the corrected numeric
# XPT should contain.
#
# The ONLY intended numeric changes are validated:
#
#     TINY_VALUE -> 0
#

expected_df = source_df.copy()


for col in EXPECTED_TINY_COUNTS:

    expected_df.loc[
        expected_df[col].eq(TINY_VALUE),
        col
    ] = 0


# ============================================================
# 9. CONFIRM NO TINY VALUES REMAIN
# ============================================================

remaining_tiny = int(
    df[numeric_cols]
    .eq(TINY_VALUE)
    .sum()
    .sum()
)


print("\nTiny values after correction:", remaining_tiny)


if remaining_tiny != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain after correction."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 10. DETECT GENUINE DECIMAL VARIABLES
# ============================================================

#
# CRITICAL DIETARY FIDELITY CHECK.
#
# This occurs AFTER validated zero restoration but BEFORE
# any Int64 conversion.
#
# We expect exactly the 50 variables listed above.
#

print("\n--- GENUINE DECIMAL CHECK ---")


detected_decimal_columns = []


for col in numeric_cols:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) == 0:

        continue


    fractional_mask = ~np.isclose(
        values,
        np.round(values),
        rtol=0,
        atol=1e-12
    )


    if fractional_mask.any():

        detected_decimal_columns.append(
            col
        )


        print(
            f"{col}: "
            f"{int(fractional_mask.sum()):,} "
            "fractional observations"
        )


if detected_decimal_columns != DECIMAL_COLUMNS:

    raise ValueError(
        "STOP: Decimal-variable pattern changed.\n"
        f"Expected: {DECIMAL_COLUMNS}\n"
        f"Actual:   {detected_decimal_columns}"
    )


print(
    "\nPASS: Exactly 50 genuine decimal "
    "variables detected."
)


# ============================================================
# 11. CLEAN AND VALIDATE SEQN
# ============================================================

seqn = (
    df["SEQN"]
    .dropna()
)


if not np.all(
    np.isclose(
        seqn,
        np.round(seqn),
        rtol=0,
        atol=1e-12
    )
):

    raise ValueError(
        "SEQN contains unexpected fractional values."
    )


df["SEQN"] = (
    df["SEQN"]
    .astype("Int64")
)


print("\n--- SEQN ---")


print(
    "Minimum:",
    df["SEQN"].min()
)


print(
    "Maximum:",
    df["SEQN"].max()
)


print(
    "Missing:",
    int(
        df["SEQN"]
        .isna()
        .sum()
    )
)


print(
    "Duplicates:",
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    )
)


if not (
    df["SEQN"].min() == 62161
    and
    df["SEQN"].max() == 71916
    and
    int(
        df["SEQN"]
        .isna()
        .sum()
    ) == 0
    and
    int(
        df["SEQN"]
        .duplicated()
        .sum()
    ) == 0
):

    raise ValueError(
        "SEQN validation failed."
    )


print(
    "PASS: SEQN validated."
)


# ============================================================
# 12. FORMAT ONLY PROVEN WHOLE-NUMBER COLUMNS
# ============================================================

#
# IMPORTANT:
#
# Only variables proven to contain whole-number values are
# converted to nullable Int64.
#
# NONE of the 50 DECIMAL_COLUMNS are changed.
#
# NO .round() is used for conversion.
#

for col in INTEGER_COLUMNS:

    values = (
        df[col]
        .dropna()
        .to_numpy(
            dtype=float
        )
    )


    if len(values) > 0:

        whole_mask = np.isclose(
            values,
            np.round(values),
            rtol=0,
            atol=1e-12
        )


        if not whole_mask.all():

            raise ValueError(
                f"{col} unexpectedly contains "
                "genuine decimal values."
            )


    df[col] = (
        df[col]
        .astype("Int64")
    )


print(
    "\nPASS: 116 whole-number variables converted "
    "to Int64."
)


print(
    "PASS: All 50 Dietary decimal variables "
    "remain unchanged as float."
)


# ============================================================
# 13. HELPER - EXACT CDC CATEGORICAL CHECK
# ============================================================

def check_frequency(
    column,
    expected_values,
    expected_missing
):

    actual_values = (
        df[column]
        .value_counts()
        .sort_index()
        .to_dict()
    )


    actual_missing = int(
        df[column]
        .isna()
        .sum()
    )


    if actual_values != expected_values:

        raise ValueError(
            f"{column} frequencies differ from CDC.\n"
            f"Expected: {expected_values}\n"
            f"Actual:   {actual_values}"
        )


    if actual_missing != expected_missing:

        raise ValueError(
            f"{column} missing differs from CDC.\n"
            f"Expected: {expected_missing}\n"
            f"Actual:   {actual_missing}"
        )


    print(
        f"PASS: {column} matches CDC."
    )


# ============================================================
# 14. CDC CHECK - DIETARY DAY 1 WEIGHT
# ============================================================

#
# CDC:
#
# WTDRD1
#
# Range:
#     1852.3028731 - 443805.73786
#
# Non-missing:
#     8519
#
# Missing:
#     819
#
# Because CDC displays rounded decimal endpoints, use
# tolerance for endpoint comparison only.
#

s = df["WTDRD1"]


if int(
    s.notna()
    .sum()
) != 8519:

    raise ValueError(
        "WTDRD1 non-missing count differs from CDC."
    )


if int(
    s.isna()
    .sum()
) != 819:

    raise ValueError(
        "WTDRD1 missing count differs from CDC."
    )


if not np.isclose(
    s.min(),
    1852.3028731,
    rtol=0,
    atol=1e-6
):

    raise ValueError(
        "Unexpected WTDRD1 minimum."
    )


if not np.isclose(
    s.max(),
    443805.73786,
    rtol=0,
    atol=1e-5
):

    raise ValueError(
        "Unexpected WTDRD1 maximum."
    )


print(
    "PASS: WTDRD1 matches CDC."
)


# ============================================================
# 15. CDC CHECK - TWO-DAY DIETARY WEIGHT
# ============================================================

#
# CDC:
#
# WTDR2D
#
# 1531.3308619 - 499271.52935
#
# n = 7605
# missing = 1733
#

s = df["WTDR2D"]


if int(
    s.notna()
    .sum()
) != 7605:

    raise ValueError(
        "WTDR2D non-missing count differs from CDC."
    )


if int(
    s.isna()
    .sum()
) != 1733:

    raise ValueError(
        "WTDR2D missing count differs from CDC."
    )


if not np.isclose(
    s.min(),
    1531.3308619,
    rtol=0,
    atol=1e-6
):

    raise ValueError(
        "Unexpected WTDR2D minimum."
    )


if not np.isclose(
    s.max(),
    499271.52935,
    rtol=0,
    atol=1e-5
):

    raise ValueError(
        "Unexpected WTDR2D maximum."
    )


print(
    "PASS: WTDR2D matches CDC."
)


# ============================================================
# 16. CDC CHECK - DR1DRSTZ
# ============================================================

#
# Dietary recall status:
#
# 1 = Reliable               8,389
# 2 = Not reliable              99
# 4 = Breast milk              130
# 5 = Not done                 720
#

check_frequency(
    "DR1DRSTZ",
    {
        1: 8389,
        2: 99,
        4: 130,
        5: 720
    },
    0
)


# ============================================================
# 17. CDC CHECK - NUMBER OF INTAKE DAYS
# ============================================================

#
# 1 = Day 1 only       914
# 2 = Day 1 + Day 2   7605
# Missing              819
#

check_frequency(
    "DRDINT",
    {
        1: 914,
        2: 7605
    },
    819
)


# ============================================================
# 18. CDC CHECK - DR1DBIH
# ============================================================

#
# Days between intake day and household interview.
#
# CDC:
#
# -46 to 70 = 8,589
# Missing   =   749
#
# Attached XPT includes:
#
# zero = 309
#
# These were encoded as the tiny XPORT zero artifact.
#

s = df["DR1DBIH"]


print("\n--- DR1DBIH ---")


print(
    "Minimum:",
    s.min()
)


print(
    "Maximum:",
    s.max()
)


print(
    "Zero:",
    int(
        s.eq(0)
        .sum()
    )
)


if not (
    s.min() == -46
    and
    s.max() == 70
    and
    int(
        s.notna()
        .sum()
    )
    == 8589
    and
    int(
        s.isna()
        .sum()
    )
    == 749
    and
    int(
        s.eq(0)
        .sum()
    )
    == 309
):

    raise ValueError(
        "DR1DBIH differs from CDC / "
        "validated zero restoration."
    )


print(
    "PASS: DR1DBIH matches CDC."
)


# ============================================================
# 19. CDC CHECK - DAY OF WEEK
# ============================================================

check_frequency(
    "DR1DAY",
    {
        1: 1378,
        2: 639,
        3: 741,
        4: 817,
        5: 884,
        6: 2126,
        7: 2033
    },
    720
)


# ============================================================
# 20. CDC CHECK - NUMBER OF FOODS
# ============================================================

#
# CDC:
#
# 1-61 = 8,519
# Missing = 819
#

s = df["DR1TNUMF"]


if not (
    s.min() == 1
    and
    s.max() == 61
    and
    int(
        s.notna()
        .sum()
    )
    == 8519
    and
    int(
        s.isna()
        .sum()
    )
    == 819
):

    raise ValueError(
        "DR1TNUMF differs from CDC."
    )


print(
    "PASS: DR1TNUMF matches CDC."
)


# ============================================================
# 21. CDC CHECK - ENERGY
# ============================================================

#
# CDC:
#
# DR1TKCAL:
#
# 0-13687 = 8,389
# Missing = 949
#
# Attached file contains exactly one zero.
#

s = df["DR1TKCAL"]


if not (
    s.min() == 0
    and
    s.max() == 13687
    and
    int(
        s.notna()
        .sum()
    )
    == 8389
    and
    int(
        s.isna()
        .sum()
    )
    == 949
    and
    int(
        s.eq(0)
        .sum()
    )
    == 1
):

    raise ValueError(
        "DR1TKCAL differs from CDC / zero restoration."
    )


print(
    "PASS: DR1TKCAL matches CDC."
)


# ============================================================
# 22. CDC CHECK - PROTEIN
# ============================================================

#
# DR1TPROT is a genuine Dietary decimal.
#
# CDC:
#
# 0 - 387.37
# n = 8389
# missing = 949
#
# The zero is retained numerically as 0.0 because this is
# a DECIMAL variable.
#

s = df["DR1TPROT"]


if not (
    np.isclose(
        s.min(),
        0,
        rtol=0,
        atol=0
    )
    and
    np.isclose(
        s.max(),
        387.37,
        rtol=0,
        atol=1e-10
    )
    and
    int(
        s.notna()
        .sum()
    )
    == 8389
    and
    int(
        s.isna()
        .sum()
    )
    == 949
    and
    int(
        s.eq(0)
        .sum()
    )
    == 1
):

    raise ValueError(
        "DR1TPROT differs from CDC."
    )


print(
    "PASS: DR1TPROT matches CDC."
)


# ============================================================
# 23. CDC CHECK - ALCOHOL
# ============================================================

#
# DR1TALCO is a genuine decimal variable.
#
# CDC:
#
# 0 - 550.1
# n = 8389
# missing = 949
#
# Attached XPT:
#
# 7,215 validated zero values.
#

s = df["DR1TALCO"]


print("\n--- DR1TALCO ---")


print(
    "Zero:",
    int(
        s.eq(0)
        .sum()
    )
)


if not (
    np.isclose(
        s.min(),
        0,
        rtol=0,
        atol=0
    )
    and
    np.isclose(
        s.max(),
        550.1,
        rtol=0,
        atol=1e-12
    )
    and
    int(
        s.notna()
        .sum()
    )
    == 8389
    and
    int(
        s.isna()
        .sum()
    )
    == 949
    and
    int(
        s.eq(0)
        .sum()
    )
    == 7215
):

    raise ValueError(
        "DR1TALCO differs from CDC / "
        "validated zero restoration."
    )


print(
    "PASS: DR1TALCO matches CDC."
)


# ============================================================
# 24. CDC CHECK - AMOUNT VS USUAL
# ============================================================

check_frequency(
    "DR1_300",
    {
        1: 724,
        2: 6264,
        3: 1565,
        7: 3,
        9: 62
    },
    720
)


# ============================================================
# 25. CDC CHECK - PLAIN WATER
# ============================================================

#
# Genuine decimal Dietary/water variable.
#
# CDC:
#
# 0 - 12886.88
# n = 8519
# missing = 819
#
# Attached XPT:
#
# zero = 1911
#

s = df["DR1_320Z"]


if not (
    np.isclose(
        s.min(),
        0,
        rtol=0,
        atol=0
    )
    and
    np.isclose(
        s.max(),
        12886.88,
        rtol=0,
        atol=1e-10
    )
    and
    int(
        s.notna()
        .sum()
    )
    == 8519
    and
    int(
        s.isna()
        .sum()
    )
    == 819
    and
    int(
        s.eq(0)
        .sum()
    )
    == 1911
):

    raise ValueError(
        "DR1_320Z differs from CDC."
    )


print(
    "PASS: DR1_320Z matches CDC."
)


# ============================================================
# 26. CDC CHECK - TAP WATER
# ============================================================

s = df["DR1_330Z"]


if not (
    np.isclose(
        s.min(),
        0,
        rtol=0,
        atol=0
    )
    and
    np.isclose(
        s.max(),
        12886.88,
        rtol=0,
        atol=1e-10
    )
    and
    int(
        s.notna()
        .sum()
    )
    == 8519
    and
    int(
        s.isna()
        .sum()
    )
    == 819
    and
    int(
        s.eq(0)
        .sum()
    )
    == 4575
):

    raise ValueError(
        "DR1_330Z differs from CDC."
    )


print(
    "PASS: DR1_330Z matches CDC."
)


# ============================================================
# 27. CDC CHECK - BOTTLED WATER
# ============================================================

s = df["DR1BWATZ"]


if not (
    np.isclose(
        s.min(),
        0,
        rtol=0,
        atol=0
    )
    and
    np.isclose(
        s.max(),
        10000,
        rtol=0,
        atol=0
    )
    and
    int(
        s.notna()
        .sum()
    )
    == 8519
    and
    int(
        s.isna()
        .sum()
    )
    == 819
    and
    int(
        s.eq(0)
        .sum()
    )
    == 5372
):

    raise ValueError(
        "DR1BWATZ differs from CDC."
    )


print(
    "PASS: DR1BWATZ matches CDC."
)


# ============================================================
# 28. CDC CHECK - TAP WATER SOURCE
# ============================================================

check_frequency(
    "DR1TWS",
    {
        1: 5490,
        2: 645,
        3: 87,
        4: 1879,
        99: 517
    },
    720
)


# ============================================================
# 29. CDC CHECK - SHELLFISH
# ============================================================

check_frequency(
    "DRD340",
    {
        1: 3880,
        2: 4306,
        7: 20,
        9: 34
    },
    1098
)


# ============================================================
# 30. VALUE FIDELITY BEFORE EXPORT
# ============================================================

#
# CRITICAL:
#
# At this point the ONLY expected numeric differences from
# pandas' original XPT are the 52,589 validated tiny-value
# restorations to zero.
#
# No nutrient decimal is allowed to change.
#

print("\n--- VALUE FIDELITY BEFORE EXPORT ---")


fidelity_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    current_values = (
        pd.to_numeric(
            df[col],
            errors="coerce"
        )
        .to_numpy(
            dtype=float,
            na_value=np.nan
        )
    )


    same = np.allclose(
        expected_values,
        current_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        fidelity_failures.append(
            col
        )


if fidelity_failures:

    raise ValueError(
        "Unexpected value changes before export:\n"
        + str(fidelity_failures)
    )


print(
    "PASS: All 166 variables match the "
    "expected corrected XPT exactly."
)


# ============================================================
# 31. EXPORT CSV
# ============================================================

#
# THIS IS IMPORTANT FOR DIETARY.
#
# DO NOT USE:
#
#     float_format=
#
# DO NOT ROUND.
#
# Pandas will write the genuine decimal variables naturally.
#
# Whole-number fields are Int64 and therefore do not receive
# unnecessary ".0".
#

df.to_csv(
    csv_file,
    index=False,
    na_rep=""
)


print("\nCSV created:")

print(
    csv_file
)


# ============================================================
# 32. RE-READ CSV WITH ROUND-TRIP FLOAT PARSER
# ============================================================

#
# float_precision="round_trip" is important here.
#
# It ensures the text representation is parsed back to the
# identical binary floating-point value where possible.
#

roundtrip_df = pd.read_csv(
    csv_file,
    low_memory=False,
    float_precision="round_trip"
)


if roundtrip_df.shape != df.shape:

    raise ValueError(
        "CSV shape changed after export."
    )


if (
    roundtrip_df.columns.tolist()
    != EXPECTED_COLUMNS
):

    raise ValueError(
        "CSV columns/order changed."
    )


# ============================================================
# 33. EXACT CORRECTED XPT -> CSV NUMERIC FIDELITY
# ============================================================

print("\n--- EXACT XPT -> CSV FIDELITY ---")


csv_failures = []


for col in EXPECTED_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        csv_failures.append(
            col
        )


if csv_failures:

    raise ValueError(
        "Corrected XPT -> CSV numeric differences found:\n"
        + str(csv_failures)
    )


print(
    "PASS: All 166 CSV variables round-trip "
    "to the exact corrected XPT numeric values."
)


# ============================================================
# 34. MISSING VALUE FIDELITY
# ============================================================

missing_mismatches = 0


for col in EXPECTED_COLUMNS:

    expected_missing = (
        expected_df[col]
        .isna()
        .to_numpy()
    )


    exported_missing = (
        roundtrip_df[col]
        .isna()
        .to_numpy()
    )


    missing_mismatches += int(
        np.sum(
            expected_missing
            !=
            exported_missing
        )
    )


print("\n--- MISSING VALUE FIDELITY ---")


print(
    "Missing-position mismatches:",
    missing_mismatches
)


if missing_mismatches != 0:

    raise ValueError(
        "Missing-value positions changed."
    )


print(
    "PASS: Missing values preserved exactly."
)


# ============================================================
# 35. RE-READ RAW CSV AS TEXT FOR FORMAT CHECKS
# ============================================================

#
# Do this separately from numeric round-trip verification.
#

raw_csv_df = pd.read_csv(
    csv_file,
    dtype=str,
    keep_default_na=False
)


# ============================================================
# 36. FINAL TINY-VALUE CHECK
# ============================================================

tiny_csv_count = 0


for col in EXPECTED_COLUMNS:

    values = pd.to_numeric(
        raw_csv_df[col]
        .where(
            raw_csv_df[col] != "",
            np.nan
        ),
        errors="coerce"
    )


    tiny_csv_count += int(
        values
        .eq(TINY_VALUE)
        .sum()
    )


print("\n--- FINAL TINY VALUE CHECK ---")


print(
    "Tiny values remaining:",
    tiny_csv_count
)


if tiny_csv_count != 0:

    raise ValueError(
        "Tiny XPORT artifacts remain in CSV."
    )


print(
    "PASS: No tiny XPORT artifacts remain."
)


# ============================================================
# 37. .0 FORMAT CHECK - WHOLE-NUMBER COLUMNS ONLY
# ============================================================

#
# VERY IMPORTANT FOR DIETARY:
#
# DO NOT search the entire CSV for ".0".
#
# Many genuine Dietary decimal variables can legitimately
# contain:
#
#     0.0
#     1.0
#     6.0
#     etc.
#
# Only proven whole-number variables are checked.
#

print("\n--- WHOLE-NUMBER .0 FORMAT CHECK ---")


dot_zero_errors = {}


for col in INTEGER_COLUMNS:

    tokens = raw_csv_df[col]


    bad_mask = (
        tokens.ne("")
        &
        tokens.str.endswith(".0")
    )


    count = int(
        bad_mask.sum()
    )


    if count > 0:

        dot_zero_errors[col] = count


if dot_zero_errors:

    raise ValueError(
        "Unexpected .0 formatting in "
        "whole-number fields:\n"
        + str(dot_zero_errors)
    )


print(
    "PASS: No unnecessary .0 in "
    "whole-number fields."
)


# ============================================================
# 38. CONFIRM DECIMAL .0 VALUES ARE ALLOWED
# ============================================================

#
# This is informational.
#
# A large number of genuine Dietary decimal values
# legitimately end in ".0".
#
# They SHOULD NOT be stripped.
#

decimal_dot_zero_counts = {}


for col in DECIMAL_COLUMNS:

    count = int(
        (
            raw_csv_df[col].ne("")
            &
            raw_csv_df[col].str.endswith(".0")
        )
        .sum()
    )


    if count > 0:

        decimal_dot_zero_counts[col] = count


print(
    "\nDecimal variables containing "
    "legitimate '.0' values:",
    len(decimal_dot_zero_counts)
)


print(
    "PASS: Decimal .0 values were left AS IS."
)


# ============================================================
# 39. EXACT DECIMAL FIDELITY
# ============================================================

print("\n--- DIETARY DECIMAL FIDELITY ---")


decimal_failures = []


for col in DECIMAL_COLUMNS:

    expected_values = (
        expected_df[col]
        .to_numpy(
            dtype=float
        )
    )


    exported_values = (
        roundtrip_df[col]
        .to_numpy(
            dtype=float
        )
    )


    same = np.allclose(
        expected_values,
        exported_values,
        rtol=0,
        atol=0,
        equal_nan=True
    )


    if not same:

        decimal_failures.append(
            col
        )


if decimal_failures:

    raise ValueError(
        "Dietary decimal fidelity failed:\n"
        + str(decimal_failures)
    )


print(
    "PASS: All 50 genuine Dietary decimal "
    "variables preserved exactly."
)


# ============================================================
# 40. SCIENTIFIC NOTATION CHECK
# ============================================================

with open(
    csv_file,
    "r",
    encoding="utf-8"
) as f:

    csv_text = f.read()


scientific_count = (
    csv_text.lower().count("e-")
    +
    csv_text.lower().count("e+")
)


print(
    "\nScientific notation occurrences:",
    scientific_count
)


if scientific_count != 0:

    raise ValueError(
        "Unexpected scientific notation found."
    )


print(
    "PASS: No scientific notation."
)


# ============================================================
# 41. FINAL CSV SEQN CHECK
# ============================================================

final_seqn = pd.to_numeric(
    raw_csv_df["SEQN"],
    errors="raise"
)


if not (
    final_seqn.min() == 62161
    and
    final_seqn.max() == 71916
    and
    int(
        final_seqn.duplicated()
        .sum()
    ) == 0
):

    raise ValueError(
        "Final CSV SEQN validation failed."
    )


if (
    raw_csv_df["SEQN"]
    .str.endswith(".0")
    .any()
):

    raise ValueError(
        "Final CSV SEQN still contains .0."
    )


print(
    "PASS: Final CSV SEQN is valid "
    "and contains no .0."
)


# ============================================================
# 42. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 80
)


print(
    "DR1TOT_G CONVERSION COMPLETE"
)


print(
    "=" * 80
)


print(
    "Rows:",
    f"{len(df):,}"
)


print(
    "Columns:",
    len(df.columns)
)


print(
    "Original tiny-value artifacts:",
    f"{tiny_total:,}"
)


print(
    "Variables containing tiny artifacts:",
    len(EXPECTED_TINY_COUNTS)
)


print(
    "Total validated zero restorations:",
    f"{correction_total:,}"
)


print(
    "Genuine decimal variables:",
    len(DECIMAL_COLUMNS)
)


print(
    "Whole-number variables:",
    len(INTEGER_COLUMNS)
)


print(
    "XPT -> CSV unexpected numeric differences: "
    "0 expected"
)


print(
    "Missing-position differences: "
    "0 expected"
)


print(
    "Tiny values remaining: "
    "0 expected"
)


print(
    "Unnecessary .0 in whole-number fields: "
    "0 expected"
)


print(
    "Dietary decimal values altered: "
    "0 expected"
)


print(
    "Scientific notation: "
    "0 expected"
)
```
