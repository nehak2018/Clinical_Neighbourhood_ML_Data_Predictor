Purpose
-------
Run source-engineering QC on a unified NHANES file before harmonization,
cohort construction, statistical analysis, or machine-learning work.

This script checks:
1. File shape and duplicate headers
2. SEQN integrity and raw serialization
3. NHANES cycle integrity
4. Required study variables and component coverage
5. Missingness and variable-level profiling
6. Transformed-zero / near-zero artifacts
7. Character-code fidelity (e.g., SMDUPCA)
8. Exact duplicate columns
9. All-missing and constant columns
10. Audit / engineering columns
11. Expected NHANES counts for configured cycle profiles
12. Final PASS / REVIEW / FAIL engineering gate

```python
# =============================================================================
# 2. USER-EDITABLE CONFIGURATION
# =============================================================================

# Known transformed-zero value found during SCHARE source review.
ZERO_SENTINEL = 5.397605346934028e-79

# Also flag any non-zero numeric value with absolute magnitude below this level.
# This is intentionally conservative for participant-level NHANES analytic data.
NEAR_ZERO_THRESHOLD = 1e-50

# Study-critical variables that should generally be present in the unified file.
BASE_REQUIRED_VARIABLES = [
    "SEQN",
    "SDDSRVYR",
    "DIQ010",
    "HUQ030",
    "HUQ071",
]

# Component presence rules. At least one matching column must exist.
BASE_COMPONENT_RULES = {
    "Physical Activity (PAQ)": [r"^PAQ"],
    "Occupation (OCQ)": [r"^OCQ"],
    "Housing (HOQ)": [r"^HOQ"],
}

# Character/code variables where fixed-width preservation matters.
# Extend this dictionary when additional code fields are introduced.
BASE_CODE_WIDTHS = {
    "SMDUPCA": 12,
}

# Engineering/audit columns that should not enter scientific models.
AUDIT_PATTERNS = [
    r"^SEQN_IN_",
    r"^PRESENT_",
    r"RAW_SOURCE",
    r"SOURCE_FILE",
    r"SOURCE_DOMAIN",
]

# Built-in reference profile for the current 2003–2004 validation.
# These values are used only when --profile 2003-2004 is selected.
# Add later cycles after they are independently verified.
PROFILES: Dict[str, Dict[str, Any]] = {
    "2003-2004": {
        "expected_rows": 10122,
        "expected_sddsrvyr": 3,
        "expected_start_year": 2003,
        "expected_end_year": 2004,
        "required_variables": [
            "LBXGH",
            "BMXBMI",
            "MCQ220",
            "DIQ010",
            "DID040G",
            "DID040Q",
            "HUQ030",
            "HUQ050",
            "HUQ071",
            "HID010",
        ],
        "component_rules": {
            "Physical Activity (PAQ)": [r"^PAQ"],
            "Dietary Day 1": [r"^DR1", r"^WTDRD1$"],
            "Dietary Day 2": [r"^DR2", r"^WTDR2D$"],
            "Occupation (OCQ)": [r"^OCQ"],
            "Housing (HOQ)": [r"^HOQ"],
        },
        "expected_nonmissing": {
            "LBXGH": 6601,
            "BMXBMI": 8687,
            "WTDRD1": 9034,
            "WTDR2D": 8354,
        },
        "expected_value_counts": {
            "MCQ220": {
                "1": 478,
                "2": 4551,
                "9": 11,
            },
            "HUQ050": {
                "0": 1250,
            },
        },
        "code_widths": {
            "SMDUPCA": 12,
        },
        "expected_absent_components": {
            "Sleep Disorders (SLQ)": [r"^SLQ"],
        },
    }
}
```

```python
# =============================================================================
# 3. HELPERS
# =============================================================================

def now_stamp() -> str:
    return datetime.now().strftime("%Y%m%d_%H%M%S")


def safe_filename(path: str) -> str:
    return os.path.basename(path.rstrip("/")) or "dataset"


def resolve_column(df: pd.DataFrame, requested: str) -> Optional[str]:
    lookup = {str(c).upper(): c for c in df.columns}
    return lookup.get(requested.upper())


def matching_columns(columns: List[str], patterns: List[str]) -> List[str]:
    out = []
    for col in columns:
        u = str(col).upper()
        if any(re.search(p, u, flags=re.I) for p in patterns):
            out.append(col)
    return out


def add_issue(
    issues: List[Dict[str, Any]],
    severity: str,
    check: str,
    variable: str,
    finding: str,
    action: str = "",
) -> None:
    issues.append({
        "severity": severity,
        "check": check,
        "variable": variable,
        "finding": finding,
        "action": action,
    })


def read_dataset(path: str) -> pd.DataFrame:
    lower = path.lower()
    if lower.endswith((".csv", ".csv.gz")):
        return pd.read_csv(path, low_memory=False)
    if lower.endswith((".xpt", ".xport")):
        return pd.read_sas(path, format="xport")
    raise ValueError("Supported input formats: .csv, .csv.gz, .xpt, .xport")


def get_raw_header(path: str) -> List[str]:
    """Read true CSV header before pandas name mangling when possible."""
    if not path.lower().endswith((".csv", ".csv.gz")):
        return []
    if path.startswith("gs://"):
        # pandas can read cloud paths if gcsfs is available, but Python csv cannot.
        return list(pd.read_csv(path, nrows=0).columns)
    if path.lower().endswith(".gz"):
        import gzip
        with gzip.open(path, "rt", encoding="utf-8-sig", newline="") as f:
            return next(csv.reader(f))
    with open(path, "r", encoding="utf-8-sig", newline="") as f:
        return next(csv.reader(f))


def read_columns_as_text(path: str, columns: List[str]) -> pd.DataFrame:
    if not columns or not path.lower().endswith((".csv", ".csv.gz")):
        return pd.DataFrame()
    return pd.read_csv(
        path,
        usecols=columns,
        dtype=str,
        keep_default_na=False,
        low_memory=False,
    )


def normalize_expected_value(value: str) -> Any:
    try:
        f = float(value)
        return f
    except Exception:
        return value


def count_expected_value(series: pd.Series, expected_value: str) -> int:
    target = normalize_expected_value(expected_value)
    if isinstance(target, float):
        numeric = pd.to_numeric(series, errors="coerce")
        return int((numeric == target).sum())
    return int((series.astype(str) == str(target)).sum())


def series_fingerprint(series: pd.Series) -> str:
    h = pd.util.hash_pandas_object(series, index=False, categorize=True).values
    return hashlib.blake2b(h.tobytes(), digest_size=16).hexdigest()
```

```python
# =============================================================================
# 4. THIS IS THE VALIDATION MODULES
# =============================================================================

def profile_variables(df: pd.DataFrame) -> pd.DataFrame:
    rows = []
    n = len(df)
    for col in df.columns:
        s = df[col]
        numeric = pd.api.types.is_numeric_dtype(s)
        rows.append({
            "variable": col,
            "dtype": str(s.dtype),
            "row_count": n,
            "non_missing_count": int(s.notna().sum()),
            "missing_count": int(s.isna().sum()),
            "missing_pct": round(float(s.isna().mean() * 100), 3) if n else np.nan,
            "unique_nonmissing": int(s.nunique(dropna=True)),
            "numeric_min": float(s.min(skipna=True)) if numeric and s.notna().any() else np.nan,
            "numeric_max": float(s.max(skipna=True)) if numeric and s.notna().any() else np.nan,
            "numeric_mean": float(s.mean(skipna=True)) if numeric and s.notna().any() else np.nan,
            "example_values": " | ".join(
                s.dropna().astype(str).drop_duplicates().head(5).tolist()
            ),
        })
    return pd.DataFrame(rows)


def check_identifier(
    df: pd.DataFrame,
    raw_text: pd.DataFrame,
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    rows = []
    col = resolve_column(df, "SEQN")
    if col is None:
        add_issue(
            issues, "CRITICAL", "Identifier integrity", "SEQN",
            "SEQN is missing.", "Do not use dataset until participant identifier is restored."
        )
        return pd.DataFrame([{"variable": "SEQN", "present": False}])

    s = df[col]
    numeric = pd.to_numeric(s, errors="coerce")
    integer_like = int((numeric.dropna() % 1 == 0).sum())
    missing = int(s.isna().sum())
    duplicates = int(s.duplicated(keep=False).sum())
    unique = int(s.nunique(dropna=True))
    raw_dot_zero = None

    if not raw_text.empty and col in raw_text.columns:
        r = raw_text[col].astype(str)
        raw_dot_zero = int(r.str.match(r"^\d+\.0$").sum())

    if missing:
        add_issue(
            issues, "CRITICAL", "Identifier integrity", "SEQN",
            f"{missing} SEQN values are missing.", "Restore valid participant identifiers."
        )
    if duplicates:
        add_issue(
            issues, "CRITICAL", "Identifier integrity", "SEQN",
            f"{duplicates} rows participate in duplicate SEQN values.",
            "Resolve duplicate participant records before analysis."
        )
    if numeric.notna().sum() and integer_like != int(numeric.notna().sum()):
        add_issue(
            issues, "HIGH", "Identifier integrity", "SEQN",
            "One or more SEQN values are not integer-like.",
            "SEQN should be preserved as an integer/string identifier."
        )
    if raw_dot_zero:
        add_issue(
            issues, "MODERATE", "Identifier serialization", "SEQN",
            f"{raw_dot_zero} SEQN values are serialized with a trailing '.0'.",
            "Write SEQN as integer or character, not floating point."
        )

    rows.append({
        "variable": "SEQN",
        "present": True,
        "dtype": str(s.dtype),
        "row_count": len(s),
        "missing_count": missing,
        "unique_count": unique,
        "duplicate_row_count": duplicates,
        "integer_like_nonmissing": integer_like,
        "raw_trailing_dot_zero_count": raw_dot_zero,
    })
    return pd.DataFrame(rows)


def check_cycle(
    df: pd.DataFrame,
    profile: Dict[str, Any],
    profile_name: Optional[str],
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    rows = []

    sdd = resolve_column(df, "SDDSRVYR")
    expected_sdd = profile.get("expected_sddsrvyr")
    if sdd:
        vals = sorted(pd.to_numeric(df[sdd], errors="coerce").dropna().unique().tolist())
        rows.append({"field": sdd, "observed": str(vals), "expected": str(expected_sdd)})
        if expected_sdd is not None and vals != [float(expected_sdd)]:
            add_issue(
                issues, "HIGH", "Cycle integrity", sdd,
                f"Observed SDDSRVYR values {vals}; expected {expected_sdd}.",
                "Correct cycle assignment before replication."
            )
    else:
        add_issue(
            issues, "HIGH", "Cycle integrity", "SDDSRVYR",
            "SDDSRVYR is missing.", "Retain the official NHANES survey-cycle variable."
        )

    cyc = resolve_column(df, "CYCLE")
    if cyc:
        vals = df[cyc].dropna().astype(str).drop_duplicates().head(20).tolist()
        rows.append({"field": cyc, "observed": " | ".join(vals), "expected": profile_name or ""})
        source_token = any(re.match(r"^[A-Z0-9]+_[A-Z]$", v.upper()) for v in vals)
        if source_token:
            add_issue(
                issues, "MODERATE", "Cycle labeling", cyc,
                f"CYCLE contains source-file style values: {vals}.",
                "Keep source component separately from the analytic cycle label."
            )

    for field, expected_key in [("START_YEAR", "expected_start_year"),
                                ("END_YEAR", "expected_end_year")]:
        actual = resolve_column(df, field)
        if actual:
            numeric = pd.to_numeric(df[actual], errors="coerce")
            invalid = int(df[actual].notna().sum() - numeric.notna().sum())
            vals = sorted(numeric.dropna().unique().tolist())
            rows.append({
                "field": actual,
                "observed": str(vals),
                "expected": str(profile.get(expected_key, "")),
                "invalid_non_numeric_count": invalid,
            })
            if invalid:
                add_issue(
                    issues, "HIGH", "Cycle year parsing", actual,
                    f"{invalid} non-numeric year values found.",
                    "Derive years from SDDSRVYR or an explicit cycle mapping."
                )
            expected = profile.get(expected_key)
            if expected is not None and vals and vals != [float(expected)]:
                add_issue(
                    issues, "HIGH", "Cycle year parsing", actual,
                    f"Observed {vals}; expected {expected}.",
                    "Correct year derivation."
                )
    return pd.DataFrame(rows)


def check_required_variables(
    df: pd.DataFrame,
    profile: Dict[str, Any],
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    required = list(dict.fromkeys(
        BASE_REQUIRED_VARIABLES + profile.get("required_variables", [])
    ))
    rows = []
    for req in required:
        actual = resolve_column(df, req)
        if actual is None:
            rows.append({"required_variable": req, "present": False})
            add_issue(
                issues, "CRITICAL", "Required variable", req,
                "Required variable is absent.",
                "Verify the original NHANES component and rebuild the unified file."
            )
        else:
            s = df[actual]
            rows.append({
                "required_variable": req,
                "present": True,
                "actual_column": actual,
                "non_missing_count": int(s.notna().sum()),
                "missing_count": int(s.isna().sum()),
                "missing_pct": round(float(s.isna().mean() * 100), 3),
                "dtype": str(s.dtype),
            })
    return pd.DataFrame(rows)


def check_components(
    df: pd.DataFrame,
    profile: Dict[str, Any],
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    rules = dict(BASE_COMPONENT_RULES)
    rules.update(profile.get("component_rules", {}))
    rows = []
    cols = list(df.columns)

    for component, patterns in rules.items():
        matched = matching_columns(cols, patterns)
        rows.append({
            "component": component,
            "expected": "Present",
            "matched_column_count": len(matched),
            "example_columns": " | ".join(matched[:15]),
        })
        if not matched:
            add_issue(
                issues, "HIGH", "Component coverage", component,
                "No matching columns found.",
                "Verify the source component before replicating the build."
            )

    for component, patterns in profile.get("expected_absent_components", {}).items():
        matched = matching_columns(cols, patterns)
        rows.append({
            "component": component,
            "expected": "Absent",
            "matched_column_count": len(matched),
            "example_columns": " | ".join(matched[:15]),
        })
        if matched:
            add_issue(
                issues, "MODERATE", "Component coverage", component,
                f"Component expected to be absent but {len(matched)} matching columns were found.",
                "Verify cycle-specific component availability."
            )

    return pd.DataFrame(rows)


def check_zero_artifacts(
    df: pd.DataFrame,
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    rows = []
    total_sentinel = 0
    total_near_zero = 0
    total_true_zero = 0

    for col in df.select_dtypes(include=[np.number]).columns:
        arr = pd.to_numeric(df[col], errors="coerce").to_numpy(dtype=float)
        finite = np.isfinite(arr)

        true_zero = int(np.sum(finite & (arr == 0)))
        sentinel = int(np.sum(
            finite
            & np.isclose(arr, ZERO_SENTINEL, rtol=1e-6, atol=0.0)
        ))
        near_zero = int(np.sum(
            finite
            & (arr != 0)
            & (np.abs(arr) < NEAR_ZERO_THRESHOLD)
        ))

        if true_zero or sentinel or near_zero:
            rows.append({
                "variable": col,
                "true_zero_count": true_zero,
                "zero_sentinel_count": sentinel,
                "nonzero_near_zero_count": near_zero,
                "numeric_min": np.nanmin(arr) if finite.any() else np.nan,
                "numeric_max": np.nanmax(arr) if finite.any() else np.nan,
            })

        total_true_zero += true_zero
        total_sentinel += sentinel
        total_near_zero += near_zero

    if total_sentinel:
        add_issue(
            issues, "CRITICAL", "Transformed-zero artifact", "ALL NUMERIC VARIABLES",
            f"{total_sentinel:,} occurrences of the known transformed-zero sentinel were found.",
            "Correct zero conversion at the earliest source-import/serialization step and rebuild."
        )
    elif total_near_zero:
        add_issue(
            issues, "HIGH", "Near-zero artifact", "ALL NUMERIC VARIABLES",
            f"{total_near_zero:,} suspicious non-zero values below {NEAR_ZERO_THRESHOLD:g} were found.",
            "Review whether these are legitimate values or transformed zeros."
        )

    if total_true_zero == 0:
        add_issue(
            issues, "MODERATE", "Zero distribution", "ALL NUMERIC VARIABLES",
            "No true numeric zeros were found anywhere in the dataset.",
            "Review together with the transformed-zero report."
        )

    return pd.DataFrame(rows)


def check_character_codes(
    df: pd.DataFrame,
    path: str,
    profile: Dict[str, Any],
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    widths = dict(BASE_CODE_WIDTHS)
    widths.update(profile.get("code_widths", {}))

    requested = ["SEQN"] + list(widths.keys())
    actual_cols = [resolve_column(df, c) for c in requested]
    actual_cols = [c for c in actual_cols if c is not None]

    raw = read_columns_as_text(path, actual_cols)
    rows = []

    for requested_name in requested:
        actual = resolve_column(df, requested_name)
        if actual is None:
            continue

        width = widths.get(requested_name)
        dtype = str(df[actual].dtype)
        raw_series = raw[actual].astype(str) if not raw.empty and actual in raw.columns else None

        trailing_dot_zero = 0
        sci_notation = 0
        wrong_width = 0
        leading_zero = 0
        nonblank = 0

        if raw_series is not None:
            vals = raw_series.str.strip()
            vals = vals[~vals.isin(["", "nan", "NaN", "NA", "N/A", "."])]
            nonblank = len(vals)
            trailing_dot_zero = int(vals.str.match(r"^[+-]?\d+\.0$").sum())
            sci_notation = int(vals.str.contains(r"[eE][+-]?\d+", regex=True).sum())
            leading_zero = int(vals.str.match(r"^0\d+").sum())

            if width:
                cleaned = vals.str.replace(r"\.0$", "", regex=True)
                digit_only = cleaned.str.match(r"^\d+$")
                wrong_width = int(
                    (digit_only & (cleaned.str.len() != int(width))).sum()
                )

        rows.append({
            "variable": requested_name,
            "actual_column": actual,
            "expected_width": width,
            "pandas_dtype": dtype,
            "nonblank_raw_count": nonblank,
            "trailing_dot_zero_count": trailing_dot_zero,
            "scientific_notation_count": sci_notation,
            "wrong_width_digit_count": wrong_width,
            "leading_zero_present_count": leading_zero,
        })

        if requested_name != "SEQN" and pd.api.types.is_numeric_dtype(df[actual]):
            add_issue(
                issues, "HIGH", "Character-code fidelity", requested_name,
                f"{requested_name} was inferred as numeric ({dtype}).",
                "Preserve fixed-width code fields as character data."
            )
        if trailing_dot_zero and requested_name != "SEQN":
            add_issue(
                issues, "HIGH", "Character-code fidelity", requested_name,
                f"{trailing_dot_zero} raw values end in '.0'.",
                "Prevent numeric coercion during merge/export."
            )
        if wrong_width and width:
            add_issue(
                issues, "HIGH", "Character-code fidelity", requested_name,
                f"{wrong_width} digit values do not have expected width {width}.",
                "Preserve leading zeros and fixed-width formatting."
            )
        if sci_notation and requested_name != "SEQN":
            add_issue(
                issues, "HIGH", "Character-code fidelity", requested_name,
                f"{sci_notation} values use scientific notation.",
                "Preserve code fields as character data."
            )

    return pd.DataFrame(rows)


def check_duplicate_columns(
    df: pd.DataFrame,
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    fingerprints: Dict[str, List[str]] = {}
    for col in df.columns:
        try:
            fp = series_fingerprint(df[col])
            fingerprints.setdefault(fp, []).append(col)
        except Exception:
            continue

    rows = []
    for cols in fingerprints.values():
        if len(cols) < 2:
            continue
        base = cols[0]
        for other in cols[1:]:
            if df[base].equals(df[other]):
                rows.append({
                    "column_1": base,
                    "column_2": other,
                    "exact_duplicate": True,
                })

    if rows:
        add_issue(
            issues, "MODERATE", "Duplicate columns", "MULTIPLE",
            f"{len(rows)} exact duplicate column pairs were found.",
            "Retain the authoritative source field and document/remove redundant engineered copies."
        )
    return pd.DataFrame(rows)


def check_column_quality(
    df: pd.DataFrame,
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    rows = []
    for col in df.columns:
        s = df[col]
        nonmissing = int(s.notna().sum())
        unique = int(s.nunique(dropna=True))
        all_missing = nonmissing == 0
        constant = unique <= 1 and nonmissing > 0
        if all_missing or constant:
            rows.append({
                "variable": col,
                "all_missing": all_missing,
                "constant_nonmissing": constant,
                "non_missing_count": nonmissing,
                "unique_nonmissing": unique,
                "constant_value": (
                    str(s.dropna().iloc[0]) if constant and nonmissing else ""
                ),
            })

    if any(r["all_missing"] for r in rows):
        count = sum(r["all_missing"] for r in rows)
        add_issue(
            issues, "MODERATE", "All-missing columns", "MULTIPLE",
            f"{count} columns are entirely missing.",
            "Confirm whether they are structurally unavailable or engineering remnants."
        )
    return pd.DataFrame(rows)


def check_audit_columns(df: pd.DataFrame) -> pd.DataFrame:
    rows = []
    for col in df.columns:
        matches = [p for p in AUDIT_PATTERNS if re.search(p, str(col), flags=re.I)]
        if matches:
            rows.append({
                "variable": col,
                "matched_pattern": " | ".join(matches),
                "modeling_status": "Exclude from scientific feature matrix",
            })
    return pd.DataFrame(rows)


def check_expectations(
    df: pd.DataFrame,
    profile: Dict[str, Any],
    issues: List[Dict[str, Any]],
) -> pd.DataFrame:
    rows = []

    expected_rows = profile.get("expected_rows")
    if expected_rows is not None:
        passed = len(df) == int(expected_rows)
        rows.append({
            "check_type": "row_count",
            "variable": "",
            "expected": expected_rows,
            "observed": len(df),
            "pass": passed,
        })
        if not passed:
            add_issue(
                issues, "HIGH", "Reference expectation", "ROW COUNT",
                f"Observed {len(df):,} rows; expected {expected_rows:,}.",
                "Reconcile against the official cycle denominator/source merge."
            )

    for var, expected in profile.get("expected_nonmissing", {}).items():
        actual = resolve_column(df, var)
        observed = int(df[actual].notna().sum()) if actual else None
        passed = observed == int(expected)
        rows.append({
            "check_type": "nonmissing_count",
            "variable": var,
            "expected": expected,
            "observed": observed,
            "pass": passed,
        })
        if actual and not passed:
            add_issue(
                issues, "HIGH", "Reference expectation", var,
                f"Observed {observed} nonmissing values; expected {expected}.",
                "Verify source component, eligibility, and merge."
            )

    for var, expected_counts in profile.get("expected_value_counts", {}).items():
        actual = resolve_column(df, var)
        if not actual:
            continue
        for value, expected in expected_counts.items():
            observed = count_expected_value(df[actual], value)
            passed = observed == int(expected)
            rows.append({
                "check_type": "value_count",
                "variable": var,
                "value": value,
                "expected": expected,
                "observed": observed,
                "pass": passed,
            })
            if not passed:
                add_issue(
                    issues, "HIGH", "Reference expectation", var,
                    f"Value {value}: observed {observed}; expected {expected}.",
                    "Review coding/serialization against the original NHANES source."
                )

    return pd.DataFrame(rows)

```

```python
# =============================================================================
# 5. MAIN VALIDATION CODE
# =============================================================================

def validate(
    input_path: str,
    output_dir: str,
    profile_name: Optional[str] = None,
) -> Dict[str, Any]:

    stamp = now_stamp()
    out = Path(output_dir)
    out.mkdir(parents=True, exist_ok=True)

    print("=" * 78)
    print("NIH SCHARE RESEARCH — NHANES ENGINEERING VALIDATION")
    print("=" * 78)
    print(f"Input:   {input_path}")
    print(f"Output:  {out}")
    print(f"Profile: {profile_name or 'Generic'}")

    profile = PROFILES.get(profile_name or "", {})
    df = read_dataset(input_path)

    print(f"Loaded:  {len(df):,} rows x {len(df.columns):,} columns")

    issues: List[Dict[str, Any]] = []

    # -------------------------------------------------------------------------
    # Header integrity
    # -------------------------------------------------------------------------
    raw_header = get_raw_header(input_path)
    header_rows = []

    if raw_header:
        counts = pd.Series(raw_header).value_counts()
        duplicate_headers = counts[counts > 1]

        for name, count in duplicate_headers.items():
            header_rows.append({
                "header": name,
                "occurrences": int(count),
            })

            add_issue(
                issues,
                "HIGH",
                "Duplicate header",
                str(name),
                f"Header occurs {int(count)} times in the raw file.",
                "Resolve duplicate field names before pandas import.",
            )

    duplicate_header_df = pd.DataFrame(header_rows)

    # -------------------------------------------------------------------------
    # Read identifier/code fields as raw text
    # -------------------------------------------------------------------------
    code_fields = ["SEQN"] + list(
        dict(BASE_CODE_WIDTHS, **profile.get("code_widths", {})).keys()
    )

    actual_code_cols = [
        resolve_column(df, c)
        for c in code_fields
        if resolve_column(df, c)
    ]

    raw_text = read_columns_as_text(
        input_path,
        actual_code_cols
    )

    # -------------------------------------------------------------------------
    # Run QC modules
    # -------------------------------------------------------------------------
    variable_profile = profile_variables(df)

    identifier_qc = check_identifier(
        df,
        raw_text,
        issues
    )

    cycle_qc = check_cycle(
        df,
        profile,
        profile_name,
        issues
    )

    required_qc = check_required_variables(
        df,
        profile,
        issues
    )

    component_qc = check_components(
        df,
        profile,
        issues
    )

    zero_qc = check_zero_artifacts(
        df,
        issues
    )

    character_qc = check_character_codes(
        df,
        input_path,
        profile,
        issues
    )

    duplicate_columns_qc = check_duplicate_columns(
        df,
        issues
    )

    column_quality_qc = check_column_quality(
        df,
        issues
    )

    audit_columns_qc = check_audit_columns(df)

    expectation_qc = check_expectations(
        df,
        profile,
        issues
    )

    # -------------------------------------------------------------------------
    # Consolidate issues
    # -------------------------------------------------------------------------
    issues_df = pd.DataFrame(issues)

    if issues_df.empty:
        issues_df = pd.DataFrame(
            columns=[
                "severity",
                "check",
                "variable",
                "finding",
                "action",
            ]
        )

    # -------------------------------------------------------------------------
    # Engineering gate
    # -------------------------------------------------------------------------
    severity_counts = (
        issues_df["severity"].value_counts().to_dict()
        if not issues_df.empty
        else {}
    )

    blocking = int(
        severity_counts.get("CRITICAL", 0)
        + severity_counts.get("HIGH", 0)
    )

    if severity_counts.get("CRITICAL", 0) > 0:
        gate = "FAIL"

    elif severity_counts.get("HIGH", 0) > 0:
        gate = "FAIL"

    elif severity_counts.get("MODERATE", 0) > 0:
        gate = "REVIEW"

    else:
        gate = "PASS"

    seqn_col = resolve_column(df, "SEQN")

    summary = {
        "project": "NIH SCHARE Research – Team 4",
        "script": "priority4a_nhanes_engineering_validation.py",
        "run_timestamp": stamp,
        "input_file": safe_filename(input_path),
        "input_path": input_path,
        "profile": profile_name or "Generic",
        "row_count": int(len(df)),
        "column_count": int(len(df.columns)),
        "unique_seqn": (
            int(df[seqn_col].nunique(dropna=True))
            if seqn_col
            else None
        ),
        "issue_counts": severity_counts,
        "blocking_issue_count": blocking,
        "engineering_gate": gate,
        "gate_definition": {
            "PASS":
                "No critical/high/moderate engineering issues detected.",

            "REVIEW":
                "No blocking issues, but moderate findings require review.",

            "FAIL":
                "One or more critical/high engineering issues detected.",
        },
    }

    # -------------------------------------------------------------------------
    # Organize QC outputs
    # -------------------------------------------------------------------------
    outputs = {
        "Variable_Profile": variable_profile,
        "Identifier_QC": identifier_qc,
        "Cycle_QC": cycle_qc,
        "Required_Variables": required_qc,
        "Component_Coverage": component_qc,
        "Zero_Artifact_Summary": zero_qc,
        "Character_Code_QC": character_qc,
        "Duplicate_Columns": duplicate_columns_qc,
        "Column_Quality": column_quality_qc,
        "Audit_Columns": audit_columns_qc,
        "Expectation_Checks": expectation_qc,
        "Duplicate_Headers": duplicate_header_df,
        "Issues": issues_df,
    }

    # -------------------------------------------------------------------------
    # Write individual CSV QC files
    # -------------------------------------------------------------------------
    written = {}

    for label, table in outputs.items():

        file_path = (
            out
            / f"NHANES_Engineering_QC_{label}_{stamp}.csv"
        )

        table.to_csv(
            file_path,
            index=False
        )

        written[label] = str(file_path)

    # -------------------------------------------------------------------------
    # Write JSON summary
    # -------------------------------------------------------------------------
    summary_path = (
        out
        / f"NHANES_Engineering_QC_Summary_{stamp}.json"
    )

    summary_path.write_text(
        json.dumps(
            summary,
            indent=2,
            default=str
        ),
        encoding="utf-8"
    )

    written["Summary"] = str(summary_path)

    # -------------------------------------------------------------------------
    # Write human-readable TXT report
    # -------------------------------------------------------------------------
    report_path = (
        out
        / f"NHANES_Engineering_QC_Report_{stamp}.txt"
    )

    report_lines = [
        "NIH SCHARE RESEARCH — NHANES ENGINEERING VALIDATION",
        "=" * 70,
        f"Input file: {summary['input_file']}",
        f"Profile: {summary['profile']}",
        f"Rows: {summary['row_count']:,}",
        f"Columns: {summary['column_count']:,}",
        f"Unique SEQN: {summary['unique_seqn']}",
        f"Engineering gate: {gate}",
        "",
        "Issue counts:",
        f"  CRITICAL: {severity_counts.get('CRITICAL', 0)}",
        f"  HIGH:     {severity_counts.get('HIGH', 0)}",
        f"  MODERATE: {severity_counts.get('MODERATE', 0)}",
        f"  INFO:     {severity_counts.get('INFO', 0)}",
        "",
        "Findings:",
    ]

    if issues:

        for i, row in enumerate(
            issues,
            1
        ):

            report_lines.append(
                f"{i}. "
                f"[{row['severity']}] "
                f"{row['check']} | "
                f"{row['variable']}: "
                f"{row['finding']}"
            )

            if row.get("action"):

                report_lines.append(
                    f"   Action: {row['action']}"
                )

    else:

        report_lines.append(
            "No engineering issues detected."
        )

    report_path.write_text(
        "\n".join(report_lines),
        encoding="utf-8"
    )

    written["Report"] = str(report_path)

    # -------------------------------------------------------------------------
    # Build consolidated Excel QC validation workbook
    # -------------------------------------------------------------------------
    profile_tag = (
        profile_name.replace("-", "_")
        if profile_name
        else "Generic"
    )

    excel_path = (
        out
        / f"NHANES_{profile_tag}_Processed_v2_QC_Validation.xlsx"
    )

    summary_df = pd.DataFrame(
        [
            [
                "Project",
                summary["project"],
            ],
            [
                "Input File",
                summary["input_file"],
            ],
            [
                "Profile",
                summary["profile"],
            ],
            [
                "Rows",
                summary["row_count"],
            ],
            [
                "Columns",
                summary["column_count"],
            ],
            [
                "Unique SEQN",
                summary["unique_seqn"],
            ],
            [
                "Engineering Gate",
                summary["engineering_gate"],
            ],
            [
                "Critical Issues",
                severity_counts.get(
                    "CRITICAL",
                    0
                ),
            ],
            [
                "High Issues",
                severity_counts.get(
                    "HIGH",
                    0
                ),
            ],
            [
                "Moderate Issues",
                severity_counts.get(
                    "MODERATE",
                    0
                ),
            ],
            [
                "Info Issues",
                severity_counts.get(
                    "INFO",
                    0
                ),
            ],
            [
                "Blocking Issues",
                summary[
                    "blocking_issue_count"
                ],
            ],
            [
                "Run Timestamp",
                summary[
                    "run_timestamp"
                ],
            ],
        ],
        columns=[
            "Metric",
            "Value",
        ],
    )

    # -------------------------------------------------------------------------
    # Excel worksheet names
    # -------------------------------------------------------------------------
    excel_sheet_map = {
        "Variable_Profile":
            "Variable Profile",

        "Identifier_QC":
            "Identifier QC",

        "Cycle_QC":
            "Cycle QC",

        "Required_Variables":
            "Required Variables",

        "Component_Coverage":
            "Component Coverage",

        "Zero_Artifact_Summary":
            "Zero Artifact",

        "Character_Code_QC":
            "Character Code QC",

        "Duplicate_Columns":
            "Duplicate Columns",

        "Column_Quality":
            "Column Quality",

        "Audit_Columns":
            "Audit Columns",

        "Expectation_Checks":
            "Expectation Checks",

        "Duplicate_Headers":
            "Duplicate Headers",

        "Issues":
            "Issues",
    }

    # -------------------------------------------------------------------------
    # Write consolidated Excel workbook
    # -------------------------------------------------------------------------
    with pd.ExcelWriter(
        excel_path,
        engine="openpyxl"
    ) as writer:

        summary_df.to_excel(
            writer,
            sheet_name="Summary",
            index=False
        )

        for label, table in outputs.items():

            sheet_name = excel_sheet_map.get(
                label,
                label[:31]
            )

            table.to_excel(
                writer,
                sheet_name=sheet_name,
                index=False
            )

        workbook = writer.book

        # ---------------------------------------------------------------------
        # Basic workbook formatting
        # ---------------------------------------------------------------------
        for worksheet in workbook.worksheets:

            worksheet.freeze_panes = "A2"

            # Bold header row
            for cell in worksheet[1]:

                cell.font = Font(
                    bold=True
                )

            # Auto-filter
            if (
                worksheet.max_row > 1
                and worksheet.max_column > 0
            ):

                worksheet.auto_filter.ref = (
                    worksheet.dimensions
                )

            # Adjust column widths
            for column_cells in worksheet.columns:

                max_length = 0

                column_letter = get_column_letter(
                    column_cells[0].column
                )

                for cell in column_cells:

                    value_length = (
                        len(str(cell.value))
                        if cell.value is not None
                        else 0
                    )

                    max_length = max(
                        max_length,
                        value_length
                    )

                worksheet.column_dimensions[
                    column_letter
                ].width = min(
                    max(
                        max_length + 2,
                        12
                    ),
                    50
                )

    written["Excel_Report"] = str(
        excel_path
    )

    # -------------------------------------------------------------------------
    # Final console output
    # -------------------------------------------------------------------------
    print("")
    print("-" * 78)

    print(
        f"ENGINEERING GATE: {gate}"
    )

    print(
        "Issues: "
        f"CRITICAL="
        f"{severity_counts.get('CRITICAL', 0)}, "
        f"HIGH="
        f"{severity_counts.get('HIGH', 0)}, "
        f"MODERATE="
        f"{severity_counts.get('MODERATE', 0)}"
    )

    print("-" * 78)

    print(
        f"Text report: {report_path}"
    )

    print(
        f"JSON summary: {summary_path}"
    )

    print(
        f"Excel QC report: {excel_path}"
    )

    # -------------------------------------------------------------------------
    # Return validation results
    # -------------------------------------------------------------------------
    return {
        "summary": summary,
        "outputs": written,
        "issues": issues_df,
    }
```

```python
# =============================================================================
# 6. RUN VALIDATION AND OUTPUT REPORT GENERATIONS
# =============================================================================

if __name__ == "__main__":

    results = validate(
        input_path=INPUT_FILE,
        output_dir=OUTPUT_DIR,
        profile_name=PROFILE_NAME,
    )

    print("")
    print("Validation complete.")

    print(
        f"Input file: "
        f"{INPUT_FILE}"
    )

    print(
        f"Output directory: "
        f"{OUTPUT_DIR}"
    )

    print(
        f"Profile: "
        f"{PROFILE_NAME}"
    )
```
