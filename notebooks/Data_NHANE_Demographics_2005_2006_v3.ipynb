```python
# =================================================================================================
# PART 1 - Demographics part one will aggregate all data into 1 file
# Raw data : This is All Demographics data from NHANES
# Previously Done : To run this file, the list of CSVs below have processed datasets yearly into saperate notebooks
# Purpose : This Aggregator program combining all Demographics data 2004-2014 into one as below will get all yearly csvs and create one dataset file 2003-2014 
# Specification : Each NHANE dataset has SEQN "Participent ID" which is uniqud id for labs. 
# Files Used :  DEMO_C.csv, DEMO_D.csv, DEMO_E.csv,DEMO_F.csv, DEMO_G.csv, DEMO_H.csv
# Data Availability Matrix : https://docs.google.com/spreadsheets/d/1E_zIzHZsEs0o-XWx8HbTN7ZPODge7EEC/edit?gid=1721397761#gid=1721397761
# Source Site : https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public
# Developer Notes :  The notebook had created for windows desktop environment with Python3.
# File Creator : Neha Khairnar
# ===================================================================================================
# 📦 Step 1: Imports
# =========================
import pandas as pd
from pathlib import Path

# =========================
# 📂 Step 2: Define Paths
# =========================
BASE_DIR = Path().resolve().parent
DATA_DIR = BASE_DIR / "data" / "2005_2006" / "Demographics"

# List all demographic files (update names as per your files)
FILES = [
    "DEMO_D_corrected.csv",
    # "DEMO_D.csv",
    # "DEMO_E.csv",
    # "DEMO_F.csv",
    # "DEMO_G.csv",
    # "DEMO_H.csv"
]


# =========================
# 📥 Step 3: Load & Append
# =========================
dfs = []
for file in FILES:
    path = DATA_DIR / file
    #df = pd.read_csv(path)  
    
    df = pd.read_csv(
        path,
        dtype=str,
        keep_default_na=False
    )

    # Standardize column names
    df.columns = df.columns.str.upper()
    # Optional: Add survey cycle column
    if file == "DEMO_D_corrected.csv":
        df["CYCLE"] = "2005-2006"
        df["Source"] = "DEMO_C"
    print(file, "shape:", df.shape)
    dfs.append(df)


# =========================
# 🔗 Step 4: Concatenate (Append)
# =========================
demo_all = pd.concat(dfs, axis=0, ignore_index=True)

# =========================
# 👀 Step 5: Preview
# =========================
print("Combined Demographics Preview:")
display(demo_all.head(10))

# =========================
# 📊 Step 6: Validation
# =========================
print("Final shape:", demo_all.shape)

# Check duplicate SEQN (should be unique across cycles)
duplicates = demo_all["SEQN"].duplicated().sum()
print("Duplicate SEQN:", duplicates)

# =========================
# 🧼 Step 7: Optional Cleaning
# =========================
# Drop exact duplicates if any
demo_all = demo_all.drop_duplicates()

# =========================
# 💾 Step 8: Save Output
# =========================
OUTPUT_DIR = BASE_DIR / "data" / "processed_v3"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

demo_all.to_csv(OUTPUT_DIR / "neighbourhood_nhanes_demographics_2005_2005_raw_corrected.csv", index=False)

print("Saved to:")
print(OUTPUT_DIR / "neighbourhood_nhanes_demographics_2005_2006_raw_corrected.csv")
```
